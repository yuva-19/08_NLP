# 21.1 Query (Q)

## 1. What is a Query?

The simplest mental model:

> **Query = What information am I looking for right now?**

In encoder–decoder attention, the **decoder creates the Query**.

```text
Decoder's current state
          │
          ▼
      Query (Q)
          │
          ▼
"What information do
 I need from the encoder?"
```

This matches the standard encoder–decoder attention interpretation: the decoder produces a query at each output position to determine which input positions deserve attention.

---

## 2. Example: Translation

Suppose the encoder receives:

```text
The cat eats fish
```

and produces:

$$
h_1,h_2,h_3,h_4
$$

Now the decoder is generating a target word.

Its current hidden representation contains information about **what it has generated so far and what it needs next**.

That representation is projected into a Query:

$$
\boxed{Q=s_tW_Q}
$$

where:

- $s_t$ = decoder representation at timestep $t$
- $W_Q$ = learned Query projection
- $Q$ = resulting Query vector

Conceptually:

```text
Decoder state sₜ
      │
      │ WQ
      ▼
   Query Q
```

In Transformer cross-attention, the decoder's representations are projected to form $Q$, while the encoder output is projected to form $K$ and $V$. 

---

## 3. Why Does the Decoder Need a Query?

Remember the problem from **Module 20**:

The decoder shouldn't have to use the same information for every output token.

Instead, it should be able to ask:

> **"Which part of the source sentence is useful for me right now?"**

For example:

```text
Source:
The | cat | eats | fish
          ↑
          │
      relevant now
```

At another decoding step:

```text
Source:
The | cat | eats | fish
                    ↑
                    │
               relevant now
```

The decoder's state changes → the Query can change → the attention distribution can change.

$$
\boxed{
s_t \rightarrow Q_t
}
$$

Therefore:

$$
\boxed{
Q_1,Q_2,Q_3,\ldots
}
$$

can be different for different decoder positions.

---

## 4. Query Is a "Search Request"

A very useful analogy:

| Attention component | Intuition |
|---|---|
| **Query (Q)** | What am I looking for? 🔎 |
| **Key (K)** | What kind of information do you represent? |
| **Value (V)** | What information do you actually provide? |

So attention works roughly like:

```text
             "What do I need?"
                    │
                    ▼
                 Query
                    │
                    ▼
              compare with
                    │
          ┌─────────┴─────────┐
          ▼         ▼         ▼
         K₁        K₂        K₃ ...
          │         │         │
          └─────────┬─────────┘
                    ▼
             Relevance scores
```

The Query itself doesn't contain the answer.

It helps determine **where the answer should be retrieved from**.

---

## 5. Query → Key → Attention Score

This is the next important connection.

We compare the Query against the Keys:

$$
\boxed{
QK^T
}
$$

This produces **attention scores**.

Conceptually:

```text
Query Q
  │
  ├──── compare → Key₁ → score₁
  ├──── compare → Key₂ → score₂
  ├──── compare → Key₃ → score₃
  └──── compare → Key₄ → score₄
```

A higher score means:

> **This encoder position is more relevant to the current Query.**

The syllabus's next step is precisely **Attention Scores → Softmax → Weights → Weighted Values**. syllabus(20261018-070535)

# 6. Self-Attention vs Cross-Attention

This distinction is **Must Know**.

### Self-Attention

Q, K and V originate from the **same sequence**:

```text
             Same sequence
                  │
             ┌────┼────┐
             ▼    ▼    ▼
             Q    K    V
```

Your existing ViT knowledge fits here.

### Cross-Attention

Q and K/V originate from **different sides**:

```text
Decoder                 Encoder
   │                       │
   ▼                       ├── K
   Q                       └── V
   │
   └──────────┐
              ▼
         Cross-Attention
```

So:

$$
\boxed{
Q=\text{Decoder}
}
$$

$$
\boxed{
K,V=\text{Encoder}
}
$$

This is the key NLP connection your syllabus wants you to understand. syllabus(20261018-070535)

## 7. One Equation to Remember

For Transformer cross-attention:

$$
\boxed{
Q = X_{\text{decoder}}W_Q
}
$$

The Query is therefore a **learned projection of the decoder representation**.

Then it interacts with encoder Keys:

$$
QK^T
$$

to determine where the decoder should focus.

Later, the complete mechanism becomes:

$$
\boxed{
\operatorname{Attention}(Q,K,V)
=
\operatorname{softmax}
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
}
$$

which is the scaled dot-product attention formulation in your syllabus. syllabus(20261018-070535)

## 🧠 The Mental Model

Don't memorize:

> **Q = matrix**

Instead remember:

> **Q = the decoder's current information request.**

Then:

$$
\boxed{
\text{Query}
\rightarrow
\text{"What am I looking for?"}
}
$$

$$
\boxed{
\text{Key}
\rightarrow
\text{"What do you represent?"}
}
$$

$$
\boxed{
\text{Value}
\rightarrow
\text{"What information can you give me?"}
}
$$

That's the conceptual foundation of QKV attention.

### What you should remember

1. **Query represents what the current decoder position is looking for.**
2. In cross-attention, **Q comes from the decoder**.
3. The decoder representation is projected using $W_Q$.
4. Query is compared with encoder **Keys** to calculate relevance.
5. Different decoder positions can produce different Queries.
6. **Q asks; K matches; V provides the information.**
