# 2. Docker Fundamentals

Docker is a containerization platform used to package, build, run, and distribute applications in isolated environments.

In this section, we'll understand the core Docker components and how they relate to deploying our AI/ML applications.

## 2.1 Dockerfile

A **Dockerfile** is a text file containing instructions used to build a Docker image.

It defines how the application environment should be created, including:

- Base operating system or runtime image
- Working directory
- Dependencies
- Application files
- Environment variables
- Commands to execute when the container starts

### Example Dockerfile

```dockerfile
FROM python:3.10

WORKDIR /app

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["python", "app.py"]
```

### How Docker processes this

| Instruction | Purpose |
|---|---|
| `FROM` | Selects the base image |
| `WORKDIR` | Sets the working directory inside the image |
| `COPY` | Copies files from host to image |
| `RUN` | Executes commands during image building |
| `CMD` | Specifies the default command when the container starts |

The Dockerfile acts as a **blueprint for creating the Docker image**.

## 2.2 Docker Image

A **Docker image** is an immutable, packaged template used to create containers.

It contains the application code, runtime environment, dependencies, and configuration needed to run the application.

An image is created from instructions defined in a Dockerfile.

### Example

```text
Dockerfile
    ↓
Docker Build
    ↓
Docker Image
    ↓
Create Container
```

For example, an AI model-serving image may contain:

```text
AI Model Serving Image
│
├── Python 3.10
├── FastAPI
├── Uvicorn
├── PyTorch
├── Transformers
├── Application Code
└── Configuration
```

The image itself is not the running application. It is the packaged template from which containers are created.

### Key characteristics

- Immutable template
- Contains application environment
- Can be reused
- Can be versioned
- Can be distributed through registries

## 2.3 Docker Container

A **Docker container** is a running (or stopped) instance of a Docker image.

When Docker creates a container from an image, it provides an isolated execution environment in which the application runs.

### Image vs Container

```text
               Docker Image
           (Application Template)
                   │
           ┌───────┴────────┐
           ▼                ▼
      Container 1      Container 2
       Instance         Instance
```

The same image can be used to create multiple containers.

For example, the same FastAPI image could be used to start multiple API instances.

### Image

- Template
- Packaged environment
- Immutable layers

### Container

- Instance created from an image
- Running application process
- Writable container layer
- Can be started, stopped, restarted, and removed

## 2.4 Ports

Containers have their own networking environment. Applications running inside a container may listen on a port, but that port isn't automatically accessible from the host machine.

**Port mapping** connects a host machine port to a container port.

### Example

Suppose FastAPI runs inside a container and listens on:

```text
Container Port: 8000
```

We expose it to the host using:

```bash
docker run -p 8000:8000 my-api
```

### Port mapping syntax

```bash
-p HOST_PORT:CONTAINER_PORT
```

| Port | Meaning |
|---|---|
| Host port | Port accessible from your host machine |
| Container port | Port where the application listens inside the container |

### Example: Different Ports

```bash
docker run -p 8080:8000 my-api
```

Here:

```text
Host
http://localhost:8080
        │
        ▼
Docker Port Mapping
8080 → 8000
        │
        ▼
Container
FastAPI listening on port 8000
```

You access the API from your host machine through:

```text
http://localhost:8080
```

Even though the application is listening on `8000` inside the container.

## 2.5 Environment Variables

**Environment variables** are key-value pairs used to configure applications without hardcoding configuration directly into the source code.

They are commonly used for:

- API URLs
- Database connection settings
- Application environment
- Debug configuration
- Model configuration
- Secrets and credentials

### Example in Python

```python
import os

APP_ENV = os.getenv("APP_ENV", "development")

print(APP_ENV)
```

If the environment variable `APP_ENV` exists, Python retrieves its value.

Otherwise, it uses `"development"` as the default.

### Passing environment variables to a container

```bash
docker run -e APP_ENV=production my-api
```

Inside the container:

```python
import os

APP_ENV = os.getenv("APP_ENV")

print(APP_ENV)
```

Output:

```text
production
```

### Why use environment variables?

Instead of changing your source code for different environments:

```text
Development → Development configuration
Testing     → Testing configuration
Production  → Production configuration
```

You can use the same application code with different environment variables.

**Security note:** Environment variables are convenient for configuration, but they are not automatically a secure secrets-management solution. Sensitive credentials require appropriate handling and access controls.

## 2.6 Volumes

A **Docker volume** is a mechanism for persisting data outside the container's writable layer.

Normally, when a container is removed, files written only inside its container filesystem may be lost.

Volumes allow data to survive container removal and recreation.

### Without a volume

```text
Container
│
└── Application
    └── Data files

Remove Container
       ↓
Data may be lost ❌
```

### With a volume

```text
Container
│
└── Application
       │
       ▼
Docker Volume
       │
       └── Persistent Data

Remove Container
       ↓
Volume data remains ✅
```

### Example

```bash
docker run -v app_data:/app/data my-api
```

Here:

- `app_data` → Docker-managed volume
- `/app/data` → path inside the container

The application writes data to `/app/data`, while Docker stores that data in the persistent volume.

### Common uses of volumes

- Database files
- Uploaded files
- Logs
- Persistent application data
- Model storage

For AI/ML applications, volumes can be particularly useful for storing model weights, datasets, cache files, and uploaded content without baking all of that data into the container's writable layer.

### Volume vs Container Filesystem

| Container filesystem | Volume |
|---|---|
| Associated with container lifecycle | Independent of container lifecycle |
| Data can disappear when container is removed | Data persists when container is removed |
| Suitable for temporary files | Suitable for persistent data |

## 2.7 `.dockerignore`

The `.dockerignore` file specifies files and directories that should be excluded from the Docker build context.

When building an image, Docker sends the build context—files available to the build process—to the builder.

Without `.dockerignore`, unnecessary files may be included in the build context.

This can lead to:

- Larger build contexts
- Slower builds
- Unnecessary image contents
- Accidentally including local development files
- Potentially exposing sensitive files

### Example `.dockerignore`

```dockerignore
# Git
.git
.gitignore

# Python cache
__pycache__
*.pyc
*.pyo

# Virtual environments
.venv
venv

# Environment variables and secrets
.env
.env.*

# Jupyter
.ipynb_checkpoints

# IDE
.vscode
.idea

# OS files
.DS_Store
Thumbs.db

# Tests and local artifacts (optional)
tests
```

### Why exclude `.venv`?

Your local virtual environment may contain hundreds of megabytes of Python packages, including libraries already installed in your local environment.

We generally don't want to copy the local `.venv` into the Docker image.

Instead, Docker should install the dependencies specified in `requirements.txt` inside the image's own environment.

### Why exclude `.env`?

A `.env` file may contain:

- API keys
- Database credentials
- Tokens
- Secret configuration

It should not accidentally be copied into the Docker image.

**Important:** `.dockerignore` excludes files from the build context, but it is not a complete security mechanism. Secrets should never be intentionally baked into an image.

### 🎯 Interview question

**What is the difference between a Docker image and a Docker container?**

A Docker image is an immutable template containing the application and its required environment. A Docker container is an instance created from that image that runs the application in an isolated environment.
