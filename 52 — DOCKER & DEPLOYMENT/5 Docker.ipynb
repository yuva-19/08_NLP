# 📦 MODULE 1 — HOW APPLICATIONS ACTUALLY WORK

## 1.1 What Is an Application?

An **application** is a software system designed to perform a task for a user.

Examples:

- Instagram — social media
- Amazon — e-commerce
- ChatGPT — AI-powered conversation
- Banking apps — financial services
- Sentiment Analysis application — analyzes text and predicts sentiment

Modern applications are often not a single program. They consist of multiple components working together.

For example, an AI Sentiment Analysis application might contain:

| Component | Responsibility |
|---|---|
| Frontend | User interface |
| Backend | Application logic |
| API | Communication interface |
| AI Model | Predicts sentiment |
| Database | Stores information |

---

## 1.2 Frontend vs Backend

### 🖥️ Frontend

The **frontend** is the part of the application users interact with.

It handles:

- User interface (UI)
- Input forms
- Buttons
- Displaying results
- User interactions

Example:

A user enters:

> "This product is amazing!"

The frontend collects this text and sends it to the backend for analysis.

Frontend technologies include:

- HTML
- CSS
- JavaScript
- React
- Vue
- Angular

### ⚙️ Backend

The **backend** is the server-side part of the application.

It handles:

- Processing requests
- Business logic
- Authentication
- Validating user input
- Communicating with AI models
- Database operations
- Returning responses

For our application, the backend receives text from the frontend and decides what should happen next.

For example:

1. Receive text.
2. Validate the input.
3. Send the text to the AI model.
4. Receive the prediction.
5. Store the result if required.
6. Return the response to the frontend.

### Frontend vs Backend

| Frontend | Backend |
|---|---|
| User-facing interface | Server-side logic |
| Collects input | Processes input |
| Displays results | Generates results |
| Runs primarily in browser/client | Runs on server |
| React, HTML, CSS, JS | Python, FastAPI, Django, etc. |

---

## 1.3 What Is a Server?

A **server** is a computer or software process that provides services or resources to other computers or programs.

A server doesn't necessarily mean a huge computer in a data center.

Your laptop can run a server.

For example, when you run a FastAPI application locally:

```bash
uvicorn main:app --port 8000
```

Your computer is running a server process that can receive requests.

The server:

1. Listens for incoming requests.
2. Accepts requests.
3. Processes them.
4. Sends responses.

### Client vs Server

- **Client:** Requests a service.
- **Server:** Provides a service.

Example:

```text
Browser → Request → Server
Browser ← Response ← Server
```

A client and server can communicate over a network—even when both are running on the same computer.

---

## 1.4 What Is an API?

**API (Application Programming Interface)** is a defined interface that allows different software components to communicate with each other.

Think of an API as a communication contract.

The frontend doesn't need to know how the AI model works internally. It only needs to know:

- Where to send the request
- What data to send
- What response it can expect

### Example API Request

```http
POST /predict
Content-Type: application/json
```

Request body:

```json
{
  "text": "This product is amazing!"
}
```

The backend receives the request and processes it.

### Example API Response

```json
{
  "text": "This product is amazing!",
  "sentiment": "positive",
  "confidence": 0.98
}
```

### Why APIs Matter

APIs allow independent components to communicate without needing to understand each other's internal implementation.

For example:

- Frontend can be written in React.
- Backend can be written in Python.
- AI model can use PyTorch.
- Database can use PostgreSQL.

They can still work together because they communicate through defined interfaces.

---

## 1.5 How HTTP Request and Response Work

Most web applications use HTTP or HTTPS for communication between clients and servers.

### Request

A client sends a request containing information such as:

- HTTP method
- URL
- Headers
- Body

Common HTTP methods:

| Method | Purpose |
|---|---|
| GET | Retrieve information |
| POST | Send data or create/process something |
| PUT | Replace a resource |
| PATCH | Partially update a resource |
| DELETE | Delete a resource |

### Response

The server returns:

- Status code
- Headers
- Response body

Common status codes:

| Code | Meaning |
|---|---|
| 200 | Success |
| 201 | Created |
| 400 | Bad request |
| 401 | Unauthorized |
| 404 | Not found |
| 500 | Internal server error |

### Example Request Flow

```text
User
  │
  ▼
Frontend
  │
  │ HTTP POST request
  ▼
Backend API
  │
  │ Process request
  ▼
AI Model
  │
  │ Prediction
  ▼
Backend API
  │
  │ HTTP response
  ▼
Frontend
  │
  ▼
User sees result
```

## 1.6 AI Inference Service

An AI model is often integrated into an application through an inference service.

### What is inference?

**Inference** means using a trained model to generate a prediction from new input data.

Training:

```text
Training Data → Model Learning → Trained Model
```

Inference:

```text
New Input → Trained Model → Prediction
```

Example:

```text
Input:
"This phone has excellent battery life."

Model:
Sentiment Analysis

Output:
Positive
```

The AI model could be:

- A PyTorch model
- A TensorFlow model
- A Hugging Face Transformer
- A scikit-learn model

The model doesn't necessarily need to run inside the backend's process.

It can be exposed as a separate service.

### Why Separate an AI Model Service?

AI inference may have different requirements:

- GPU access
- Large memory requirements
- Different Python libraries
- Different scaling requirements
- Longer processing times
- Independent model updates

Separating it allows the backend and model to be managed independently.

## 1.7 What Is a Database?

A **database** is an organized system for storing, retrieving, and managing data.

Our application may store:

- User input
- Predicted sentiment
- Confidence score
- Timestamp
- User information
- Prediction history

Example:

| ID | Input | Sentiment | Confidence |
|---|---|---|---|
| 1 | Great product! | Positive | 0.98 |
| 2 | Very disappointing | Negative | 0.95 |
| 3 | It is okay | Neutral | 0.72 |

### Why Not Store Everything in Python Variables?

Python variables exist within a running process.

If the application stops, those variables generally disappear.

A database provides persistent storage.

**Persistence** means data remains available even after an application or system restarts, subject to how the storage is configured.

Common databases:

| Database | Type |
|---|---|
| PostgreSQL | Relational |
| MySQL | Relational |
| SQLite | Relational |
| MongoDB | Document-oriented |

For this course, we will use PostgreSQL as our example database.

# ⚙️ MODULE 2 — SOFTWARE ENVIRONMENTS & DEPENDENCY PROBLEMS

## 2.1 What Is a Software Environment?

A **software environment** is the collection of software, libraries, versions, configuration, and runtime components needed for an application to execute.

A Python application may require:

- Python 3.12
- FastAPI
- PyTorch
- NumPy
- Transformers
- PostgreSQL driver
- Environment variables
- System libraries

The application depends on a specific combination of these components.

---

## 2.2 What Is a Runtime?

A **runtime** is the software environment or execution engine that allows a program to execute.

Examples:

| Language | Runtime / Execution Environment |
|---|---|
| Python | Python interpreter |
| Java | JVM |
| JavaScript | Node.js or browser engine |
| .NET | CLR |

For our application:

```text
Python Code
     │
     ▼
Python Interpreter
     │
     ▼
Program Execution
```

If our application requires Python 3.12 but another system uses an incompatible version, the behavior or compatibility may differ.

---

## 2.3 What Are Dependencies?

A **dependency** is software that your application relies on to perform its work.

Example:

```python
from fastapi import FastAPI
import numpy as np
import torch
from transformers import pipeline
```

Our application depends on:

- FastAPI
- NumPy
- PyTorch
- Transformers

These packages may themselves depend on other packages.

This creates a dependency tree.

```text
Sentiment Analysis Application
│
├── FastAPI
│   ├── Starlette
│   └── Pydantic
│
├── Transformers
│   ├── Tokenizers
│   └── Hugging Face Hub
│
├── PyTorch
│   └── Supporting libraries
│
└── NumPy
```

Installing one library may install many additional dependencies automatically.

---

## 2.4 What Is a Dependency Conflict?

A dependency conflict occurs when different software requirements cannot be satisfied by the same installed versions or configuration.

### Example

Project A requires:

```text
numpy == 1.26
```

Project B requires:

```text
numpy >= 2.0
```

If both projects use the same global Python environment, installing packages for one project may break the other.

### Another Example

```text
Project A:
Python 3.10
PyTorch version X
Transformers version X

Project B:
Python 3.12
PyTorch version Y
Transformers version Y
```

The projects may need different dependency versions.

This is why developers isolate environments.

---

## 2.5 Python Virtual Environments

A **virtual environment** creates an isolated Python environment for a project.

It allows projects to maintain their own installed packages instead of sharing every dependency globally.

### Without Virtual Environments

```text
System Python
│
├── Project A
│   └── Packages installed globally
│
└── Project B
    └── Packages installed globally
```

Both projects may interfere with one another.

### With Virtual Environments

```text
                 System
                   │
         ┌─────────┴─────────┐
         │                   │
         ▼                   ▼
     Project A           Project B
     venv                venv
     │                   │
     ├─ NumPy 1.x        ├─ NumPy 2.x
     ├─ Torch X          ├─ Torch Y
     └─ FastAPI          └─ Django
```

![Python virtual environments separate dependencies between projects](https://miro.medium.com/v2/resize%3Afit%3A1024/1%2ATi9ZMnCYNofb-0fG7tYONQ.png)

### Creating a Virtual Environment

```bash
python -m venv .venv
```

Activate it on Windows:

```powershell
.venv\Scripts\Activate.ps1
```

Activate it on Linux/macOS:

```bash
source .venv/bin/activate
```

Install packages:

```bash
pip install fastapi torch transformers
```

The packages are installed into that environment.

### What Virtual Environments Solve

- Dependency isolation
- Package version conflicts
- Cleaner project management
- Easier reproducibility
- Safer experimentation

### What They Don't Fully Solve

A virtual environment does not automatically isolate:

- Operating system packages
- System-level libraries
- OS differences
- Hardware drivers
- All environment variables
- External services such as databases

Virtual environments primarily isolate the Python package environment.

---

## 2.6 Version Problems

Software versions matter because different versions may contain:

- Different APIs
- Different features
- Bug fixes
- Breaking changes
- Different dependencies
- Different runtime requirements

Example:

```text
Developer's Laptop
Python 3.12
FastAPI 0.x
PyTorch X
```

But the production server has:

```text
Production Server
Python 3.10
FastAPI older version
PyTorch Y
```

The application may behave differently or fail.

### Version Pinning

Version pinning means specifying the versions you intend to use.

Example:

```text
fastapi==0.115.0
numpy==1.26.4
```

Instead of allowing package installation to select any version, you define known versions.

A dependency file might look like:

```text
fastapi
uvicorn
torch
transformers
numpy
```

A more controlled environment records specific versions:

```text
fastapi==0.115.0
uvicorn==0.30.6
numpy==1.26.4
```

> Exact versions above are illustrative examples—not recommendations for a new project. Compatibility should be verified before choosing versions.

---

## 2.7 Environment Variables

An **environment variable** is a configuration value provided to a running process through its environment.

Applications use them to avoid hardcoding configuration into source code.

Examples:

```text
DATABASE_URL
API_KEY
MODEL_NAME
PORT
DEBUG
```

Python example:

```python
import os

model_name = os.getenv("MODEL_NAME", "default-model")
```

### Why Use Environment Variables?

Consider changing the database:

Development:

```text
localhost
```

Production:

```text
production-database-server
```

Instead of changing Python code, configuration can be supplied externally.

This separates:

- Application code
- Environment-specific configuration

### Important

Environment variables are not automatically a secure secret-management system.

Passwords and API keys must be handled carefully. Avoid committing secret values into Git repositories.

---

## 2.8 The "Works on My Machine" Problem

This is one of the most common problems in software development.

### Developer Machine

```text
Operating System: Windows
Python: 3.12
Packages: Latest versions
Database: Local PostgreSQL
Environment Variables: Configured
```

The application works.

### Another Developer's Machine

```text
Operating System: Linux
Python: 3.10
Packages: Different versions
Database: Not installed
Environment Variables: Missing
```

The application fails.

### Production Server

```text
Operating System: Different
Python: Different
Dependencies: Different
Configuration: Different
```

The application may fail again.

### Why Does This Happen?

The application depends on more than source code.

It depends on the complete execution environment:

```text
Application Code
       +
Runtime Version
       +
Libraries
       +
System Dependencies
       +
Configuration
       +
External Services
       │
       ▼
Application Behavior
```

Changing any important part may affect whether the application runs correctly.

---

## 2.9 What Docker Will Eventually Solve

Docker helps package an application with its required environment into a standardized image.

Conceptually:

```text
Application
   +
Python Runtime
   +
Required Libraries
   +
Application Files
   +
Configuration Requirements
          │
          ▼
     Docker Image
          │
          ▼
   Docker Container
```

The container can run on different systems that support the required container runtime.

Docker doesn't magically guarantee identical behavior in every environment. Hardware, kernel, external services, configuration, networking, and permissions can still matter.

But it significantly reduces environment inconsistency.

---

## 🎯 MODULE 2 — KEY TAKEAWAYS

- Applications depend on runtimes, libraries, versions, and configuration.
- Dependencies are external software required by your application.
- Dependency conflicts can break projects.
- Virtual environments isolate Python package installations.
- Environment variables separate configuration from code.
- Different environments are a major source of deployment failures.
- Docker takes environment isolation and packaging further than Python virtual environments.

# 🏗️ MODULE 3 — APPLICATION ARCHITECTURE

## 3.1 What Is Application Architecture?

**Application architecture** describes how an application is structured.

It defines:

- Components
- Responsibilities
- Communication
- Data flow
- Dependencies
- Boundaries between services

Architecture helps us understand how an application should be built and how its components should interact.

---

## 3.2 Monolithic Architecture

A **monolithic application** is an application in which multiple functionalities are built and deployed together as one main application unit.

For example:

```text
┌───────────────────────────────────────┐
│           APPLICATION               │
│                                     │
│  ┌──────────┐  ┌──────────┐        │
│  │ Backend  │  │ AI Model │        │
│  └──────────┘  └──────────┘        │
│                                     │
│  ┌──────────┐  ┌──────────┐        │
│  │Database  │  │Auth Logic│        │
│  └──────────┘  └──────────┘        │
│                                     │
│ Everything deployed together          │
└───────────────────────────────────────┘
```

### Advantages

- Easier to begin development
- Simpler deployment initially
- Fewer communication boundaries
- Straightforward debugging in small systems

### Disadvantages

- Components become tightly coupled
- Updating one component may require deploying the entire application
- Scaling individual components may be difficult
- Large codebases can become difficult to maintain

---

## 3.3 Multi-Service Architecture

A **multi-service architecture** separates application components into independently managed services.

For our sentiment analysis system:

```text
┌─────────────────┐
│    Frontend    │
└────────┬────────┘
         │
         ▼
┌─────────────────┐
│  Backend API   │
└──────┬──────┬───┘
       │      │
       ▼      ▼
┌──────────┐ ┌──────────┐
│ AI Model│ │ Database │
│ Service │ │ Service  │
└──────────┘ └──────────┘
```

Each component has a distinct responsibility.

### Backend Service

Handles:

- Request validation
- Business logic
- API endpoints
- Authentication
- Coordination between services

### AI Model Service

Handles:

- Model loading
- Tokenization
- Inference
- Prediction generation

### Database

Handles:

- Data persistence
- Queries
- Prediction history
- User records

---

## 3.4 Why Separate Backend, AI Model and Database?

Each component has different operational requirements.

| Component | Main Requirement |
|---|---|
| Backend | Request handling, CPU, network |
| AI Model | RAM, GPU, model files, inference compute |
| Database | Persistent storage, reliability, data integrity |

### Example Scenario

Suppose 1,000 users send sentiment requests simultaneously.

The AI model may become the bottleneck because inference requires more compute.

You may want to:

- Increase AI model capacity
- Run multiple model instances
- Optimize inference
- Reduce model size

But the database might be handling the load perfectly.

With separate services, you can potentially scale the model service without scaling the database or the entire backend.

---

## 3.5 How Services Communicate

Different components need to exchange information.

Common communication methods:

- HTTP APIs
- REST
- gRPC
- Message queues
- Shared storage (where appropriate)

For our learning application, we'll primarily use HTTP-based communication.

### Example

Backend sends a request to the AI Model Service:

```json
{
  "text": "The delivery was fast and excellent."
}
```

The AI Model Service responds:

```json
{
  "sentiment": "positive",
  "confidence": 0.96
}
```

The backend then stores the result in PostgreSQL.

---

## 3.6 Service Dependencies

A **service dependency** means one service needs another service to perform some operation.

For example:

```text
Frontend
   │
   ▼
Backend API
   │
   ├──────► AI Model
   │
   └──────► Database
```

The backend depends on the AI Model Service when it needs predictions.

It depends on the database when it needs to save or retrieve records.

### Dependency Failure Example

What happens if the AI Model Service crashes?

```text
Frontend
   │
   ▼
Backend API ────────X──── AI Model
   │
   ▼
Database
```

Possible consequences:

- Backend cannot generate predictions.
- Prediction requests may fail.
- Existing stored results may still be accessible.
- Other backend endpoints might continue functioning.

A well-designed application should handle these failures rather than crashing completely.

We'll study health checks, failure handling, restart policies, and service reliability in later phases.

---

## 3.7 Synchronous vs Asynchronous Communication

### Synchronous Communication

The requester waits for the response.

```text
Backend
   │
   │ Request
   ▼
AI Model
   │
   │ Prediction
   ▼
Backend continues
```

Example:

The user submits text and waits for the sentiment prediction.

### Asynchronous Communication

The requester doesn't need to wait for the entire task to finish immediately.

```text
Backend
   │
   ▼
Queue
   │
   ▼
Worker
   │
   ▼
Long-running task
```

Useful for:

- Large AI tasks
- Background processing
- Batch predictions
- Sending emails
- Processing uploaded files

For our initial Docker application, we'll keep the architecture simple and use synchronous inference.

## 3.9 Architecture vs Infrastructure

These two concepts are different.

### Application Architecture

Describes:

- Frontend
- Backend
- AI Model
- Database
- APIs
- Service communication

### Infrastructure

Describes where and how those components run.

Examples:

- Laptop
- Virtual machine
- Linux server
- Cloud server
- Networking
- Storage
- Containers

### Example

Application architecture:

```text
Frontend → Backend → AI Model
                 └→ Database
```

Infrastructure:

```text
Server
│
├── Container 1 → Backend
├── Container 2 → AI Model
└── Container 3 → PostgreSQL
```

Docker belongs to the infrastructure and deployment side, but it directly supports how we package and run the components of our application.

---

## 3.10 Why Docker Fits This Architecture

Imagine we have the following services:

```text
Backend
- Python 3.12
- FastAPI
- Pydantic

AI Model
- Python 3.12
- PyTorch
- Transformers
- CUDA libraries (if GPU deployment is used)

Database
- PostgreSQL
```

Installing everything on a single system can create:

- Dependency conflicts
- Different Python requirements
- Complex configuration
- Difficult upgrades
- Deployment problems

Instead, Docker allows us to package and run each service in a controlled environment.

### Future Docker Architecture

```text
                AI SENTIMENT APPLICATION
                         │
                 ┌───────▼────────┐
                 │    FRONTEND    │
                 └───────┬────────┘
                         │
                         ▼
                 ┌────────────────┐
                 │ BACKEND API    │
                 │  Container     │
                 └──────┬─────────┘
                        │
              ┌─────────┴─────────┐
              │                   │
              ▼                   ▼
       ┌──────────────┐   ┌──────────────┐
       │ AI MODEL    │   │ POSTGRESQL   │
       │ CONTAINER   │   │ CONTAINER    │
       └──────────────┘   └──────────────┘

       Each service has its own runtime
       and controlled environment.
```

This architecture is the foundation for the remaining Docker phases.

## **4.4 Docker vs Virtual Machines (VMs)**

This is one of the most important Docker concepts.

![10 Things to Know About Docker\<!-- --> | Spantree](https://images.openai.com/static-rsc-4/wWD0FUKj2dX3dPPKkHNCzWMYYVAPb0kAw8gWHDBZorFisARVJiYMO1GKNuUKLqPJoOw-a1goCq3D5GVEnSS1YNxDb2z_mIB1EnMtOmo4R2Mou9Uqlvx9OyfCHa752c7DWTOm_gu6YjUjxEy4QZ6sWhX1B7KcRsVfDxaykVluR-k?purpose=inline)

| Virtual Machine                           | Container                               |
| ----------------------------------------- | --------------------------------------- |
| Virtualizes hardware through a hypervisor | Isolates processes at OS level          |
| Includes a guest operating system         | Shares host kernel                      |
| Generally heavier                         | Generally lightweight                   |
| Often takes more resources                | Often starts faster with lower overhead |
| VM isolation boundary                     | Container isolation boundary            |

### **Virtual Machine Architecture**

```
Physical Hardware
       ↓
Hypervisor
       ↓
┌────────────────────────┐
│ VM 1                  │
│ Guest OS              │
│ Application            │
└────────────────────────┘

┌────────────────────────┐
│ VM 2                  │
│ Guest OS              │
│ Application            │
└────────────────────────┘
```

Each VM typically has its own guest OS and kernel.

### **Container Architecture**

```
Physical Machine
       ↓
Host Operating System Kernel
       ↓
Container Runtime
       ↓
┌──────────────┐  ┌──────────────┐
│ Container A │  │ Container B │
│ App + Files │  │ App + Files │
└──────────────┘  └──────────────┘
```

Containers share the host kernel while isolating processes and resources using operating-system mechanisms.

### **Important**

A container is **not a tiny VM**. It is an isolated process environment with its own filesystem view, resources, and configuration.

**When to use what?**

- VM → need a separate operating system or stronger infrastructure-level isolation.
- Container → package and run applications efficiently with process-level isolation.

## **5.2 Docker Client (CLI)**

The **Docker Client** is the interface through which we communicate with Docker.

Usually, we use the terminal.

Example:

```
docker run nginx
```

We type this command into the Docker CLI.

But the CLI itself isn't necessarily the component that creates the container. It sends the request to the Docker daemon.

Common commands:

```
docker run
docker build
docker pull
docker push
docker ps
docker stop
```

Think of the CLI as the **command interface**.

## **5.3 Docker Daemon (**`dockerd`**)**

The Docker Daemon is the background service that performs Docker operations.

It manages:

- Images
- Containers
- Networks
- Volumes
- Container lifecycle

When you request a container, the daemon handles the actual work.

For example:

```
docker run nginx
```

The daemon may:

1. Check whether the image exists locally.
2. Pull the image if required.
3. Create a container.
4. Configure its filesystem and networking.
5. Start the configured process.

The daemon is commonly called:

```
dockerd
```

## **5.4 Docker API**

The Docker Client and Daemon communicate through the **Docker API**.

The API defines how commands and requests are communicated.

```
User
 │
 │ docker run nginx
 ▼
Docker CLI
 │
 │ Docker API request
 ▼
Docker Daemon
 │
 ├── Pull Image (if needed)
 ├── Create Container
 ├── Configure Network
 └── Start Container
```

The client and daemon may run on the same machine, or the client may connect to a remote daemon.

```
Docker CLI
   │
   ▼
Docker API
   │
   ▼
Docker Daemon
```

### **Step 3 — Image lookup**

The daemon checks whether the `nginx` image exists locally.

```
Local Image Store
       │
       ├── Image exists → Use it
       │
       └── Image missing → Pull from Registry
```

### **Step 5 — Container creation**

The daemon creates a new container from the image.

### **Step 6 — Container startup**

Docker starts the process configured by the image, subject to the container's configuration.

## **6.4 Image Layers**

Docker images are composed of layers.

Each layer represents filesystem changes that are part of the image.

Conceptually:

```
┌────────────────────────────┐
│ Application Code Layer   │
├────────────────────────────┤
│ Python Dependencies      │
├────────────────────────────┤
│ Python Runtime Layer     │
├────────────────────────────┤
│ Base Filesystem Layer   │
└────────────────────────────┘
```

The actual image layer structure depends on how the image is built.

### **Why Layers Matter**

Layers help with:

- Reuse
- Storage efficiency
- Build caching
- Faster rebuilds

If two images use compatible shared layers, Docker can reuse stored layers rather than downloading or storing duplicates unnecessarily.

## **6.5 Image Immutability**

Images are considered **immutable**.

Once an image has been built, its contents don't change in place as a running container modifies files.

If you want a changed application version:

```
Old Source Code
      │
      ▼
Image v1
      │
      ▼
Container v1
```

You modify the source and build a new image:

```
Updated Source Code
      │
      ▼
Image v2
      │
      ▼
Container v2
```

The old image remains unchanged unless explicitly removed.

This makes images useful for reproducible releases and deployment.

## **7.2 How a Container Is Created**

```
Docker Image
    │
    │ Docker creates a container
    ▼
Container Filesystem
    │
    ├── Read-only image layers
    │
    └── Writable container layer
             │
             ▼
        Application Process
             │
             ▼
        Running Container
```

A container is not simply a copy of the image.

It uses the image's layers and adds a container-specific writable layer.

## **7.3 The Container Writable Layer**

When a container starts, Docker provides a writable layer above the image layers.

Conceptually:

```
┌──────────────────────────────┐
│ Container Writable Layer  │ ← Runtime changes
├──────────────────────────────┤
│ Application Image Layer    │
├──────────────────────────────┤
│ Dependency Image Layer     │
├──────────────────────────────┤
│ Base Image Layer          │
└──────────────────────────────┘
```

If the application creates or modifies files inside the container filesystem, those changes generally go into its writable layer.

### **Example**

Suppose the image contains:

```
/app/main.py
```

The running container creates:

```
/app/log.txt
```

That new file is a container-level change, not a modification to the original image.

When the container is removed, its writable layer is removed too.

> Important: container writable-layer data is not the right place for data that must survive container replacement. Persistent storage will be covered in Phase 4.

## **7.4 Container Isolation**

Containers use operating-system mechanisms to isolate processes and resources.

On Linux, important mechanisms include:

- **Namespaces** — provide isolated views of resources such as processes, networking, and mounts.
- **Control groups (cgroups)** — help manage and limit resource usage.
- **Capabilities and security controls** — restrict certain privileges.

### **Namespace Example**

Imagine two containers:

```
Host System
│
├── Container A
│   └── Processes A
│
└── Container B
    └── Processes B
```

Processes in one container are isolated from the other according to the configured namespace boundaries.

Isolation is not the same as absolute security. Container security depends on the runtime, configuration, kernel, privileges, and other controls.

## **7.5 Container Lifecycle**

Containers have a lifecycle.

```
         CREATE
           │
           ▼
        CREATED
           │
           ▼
         START
           │
           ▼
        RUNNING
       /       \
      ▼         ▼
   STOPPED    EXITED
      │
      ▼
   RESTART
      │
      ▼
   RUNNING
```

A container can be:

- Created
- Started
- Running
- Stopped
- Restarted
- Removed

A container that has stopped is not automatically deleted.

The container object may remain on the machine until you remove it.

We'll learn the exact lifecycle commands in Phase 3.

## **7.6 One Image → Multiple Containers**

This is a very important concept.

```
                 IMAGE
            sentiment-api:v1
                    │
          ┌─────────┼─────────┐
          ▼         ▼         ▼
      Container  Container  Container
         A          B          C
```

All containers can originate from the same image.

But they are separate instances.

They can have different:

- Names
- Ports
- Environment variables
- Networks
- Resource limits

They share the image's layers but maintain separate container runtime state.

### **Why Create Multiple Containers?**

- Run multiple copies of a service.
- Handle more requests.
- Test different configurations.
- Separate workloads.
- Improve availability (when combined with appropriate architecture).

## **7.7 Multiple Replicas vs Multiple Services**

Don't confuse these two.

### **Multiple Replicas**

Multiple instances of the same service.

```
Backend Image
    │
    ├── Backend Container 1
    ├── Backend Container 2
    └── Backend Container 3
```

All three perform the backend role.

### **Multiple Services**

Different components of an application.

```
Application
    │
    ├── Backend Container
    ├── AI Model Container
    └── Database Container
```

| Multiple Replicas                  | Multiple Services               |
| ---------------------------------- | ------------------------------- |
| Same role                          | Different responsibilities      |
| Same application image may be used | Different images typically used |
| Scale one component                | Separate application components |
| Backend × 3                        | Backend + Model + Database      |

## **7.8 Container vs Image — Final Comparison**

| Feature            | Image                          | Container                            |
| ------------------ | ------------------------------ | ------------------------------------ |
| Nature             | Template/package               | Runnable instance                    |
| Filesystem         | Image layers                   | Image layers + writable layer        |
| State              | Immutable package              | Runtime state                        |
| Created by         | Build or pull                  | Create/run from image                |
| Multiple instances | Can create many                | Each is an instance                  |
| Lifecycle          | Build, tag, pull, push, remove | Create, start, stop, restart, remove |

Eventually, all three services will run together, communicate with one another, and form a complete application.

# 📦 MODULE 8 — DOCKERFILE

## 8.1 What Is a Dockerfile?

A **Dockerfile** is a text file containing instructions Docker uses to build a Docker image.

It defines:

- Which base image to use
- Where to work inside the image
- Which application files to copy
- Which dependencies to install
- Environment configuration
- The default command to run when a container starts

### Dockerfile Workflow

```text
Application Files
      +
Dockerfile
      │
      ▼
  Docker Build
      │
      ▼
 Docker Image
      │
      ▼
 Docker Container
```

A Dockerfile does not run the application directly. It defines how to construct the image.

---

## 8.2 Our First Dockerfile

Imagine our project has this structure:

```text
sentiment-app/
│
├── main.py
├── requirements.txt
└── Dockerfile
```

### `main.py`

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Sentiment API is running"}

@app.get("/health")
def health():
    return {"status": "healthy"}
```

### `requirements.txt`

```text
fastapi
uvicorn
```

### `Dockerfile`

```dockerfile
FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install --no-cache-dir -r requirements.txt

COPY . .

EXPOSE 8000

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

This Dockerfile tells Docker how to prepare the environment and start our API.

---

## 8.3 Understanding Every Dockerfile Instruction

### `FROM`

```dockerfile
FROM python:3.12-slim
```

Defines the base image.

We start with an image containing the Python runtime, rather than installing Python manually from scratch.

```text
Python Base Image
       │
       ▼
Our Custom Image
```

---

### `WORKDIR`

```dockerfile
WORKDIR /app
```

Sets the working directory inside the image for subsequent instructions.

If `/app` doesn't exist, Docker creates it.

The following commands operate relative to this directory unless another path is specified.

```text
Container Filesystem

/
├── bin
├── usr
├── etc
└── app  ← Working directory
```

---

### `COPY`

```dockerfile
COPY requirements.txt .
```

Copies a file from the build context into the image.

Here:

- Source: `requirements.txt` on the build context
- Destination: current working directory `/app`

Another example:

```dockerfile
COPY . .
```

Copies project files into `/app`.

---

### `RUN`

```dockerfile
RUN pip install --no-cache-dir -r requirements.txt
```

Executes a command during the image build process.

This installs the Python dependencies into the image.

> **RUN happens during image building.**

It does not mean the application server starts here.

---

### `CMD`

```dockerfile
CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

Defines the default command to execute when a container starts from this image.

In this example, the command starts the FastAPI application using Uvicorn.

### `main:app` means

- `main` → Python file named `main.py`
- `app` → FastAPI instance defined inside that file

---

### `EXPOSE`

```dockerfile
EXPOSE 8000
```

Documents the port the application is intended to listen on inside the container.

> **Important:** `EXPOSE` alone does not publish the port to your host machine.

Actual host access requires port publishing when the container is run.

---

## 8.4 Dockerfile — Build Time vs Runtime

This is an important distinction.

### 🏗️ BUILD TIME

```text
Docker builds the image
        │
        ▼
FROM
WORKDIR
COPY
RUN
EXPOSE metadata
CMD configuration
        │
        ▼
       IMAGE
```

### 🚀 RUNTIME

```text
Docker starts a container
        │
        ▼
Container starts
        │
        ▼
CMD executes
        │
        ▼
Application runs
```

| Instruction | When it matters |
|---|---|
| `FROM` | Build |
| `WORKDIR` | Build configuration and runtime working directory |
| `COPY` | Build |
| `RUN` | Build |
| `EXPOSE` | Image metadata / port documentation |
| `CMD` | Runtime default process |

---

## 8.5 Docker Image Build Process

To build an image from our project:

```bash
docker build -t sentiment-api:v1 .
```

### Command breakdown

| Part | Meaning |
|---|---|
| `docker build` | Build an image |
| `-t` | Assign a tag |
| `sentiment-api` | Image name |
| `v1` | Tag |
| `.` | Build context — current directory |

### What happens?

```text
Project Directory
       │
       ▼
 Docker Build Context
       │
       ▼
Dockerfile Instructions
       │
       ├── Select Base Image
       ├── Set Working Directory
       ├── Copy Requirements
       ├── Install Dependencies
       └── Copy Application
       │
       ▼
sentiment-api:v1
```

---

## 8.6 What Is the Build Context?

The **build context** is the set of files and directories made available to Docker during the build.

In:

```bash
docker build -t sentiment-api:v1 .
```

The `.` means the current directory is the build context.

Docker can use files from that context in instructions such as `COPY`.

```text
sentiment-app/
│
├── main.py
├── requirements.txt
├── Dockerfile
└── README.md
        │
        ▼
   Build Context
```

### Why Build Context Matters

Sending unnecessary files to the builder can:

- Increase build time
- Increase data transfer
- Include files you don't need
- Accidentally expose sensitive files if they get copied into the image

Later, we'll use `.dockerignore` to control what enters the build context.

---

## 8.7 What Is Docker Build Cache?

Docker can reuse results from previous builds.

For example:

```dockerfile
FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install --no-cache-dir -r requirements.txt

COPY . .

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

Imagine we modify only `main.py`.

If the dependency file hasn't changed, Docker may reuse the cached dependency-installation step and only rebuild the later affected steps.

### Why Order Matters

❌ Less efficient structure:

```dockerfile
COPY . .
RUN pip install -r requirements.txt
```

Any source code change may invalidate the `COPY` step, causing the dependency-installation step to run again.

✅ More cache-friendly structure:

```dockerfile
COPY requirements.txt .
RUN pip install -r requirements.txt
COPY . .
```

Now a change to application source code that doesn't change `requirements.txt` can allow the dependency installation layer to be reused.

### Build Cache Principle

> Put relatively stable and expensive steps earlier; put frequently changing source files later.

Cache reuse depends on the instructions and their inputs remaining valid.

# 🔄 MODULE 9 — CONTAINER LIFECYCLE & MANAGEMENT

## 9.1 Container Lifecycle

A container is not permanently running.

It can be created, started, stopped, restarted, and removed.

```text
                 ┌──────────┐
                 │  IMAGE   │
                 └────┬─────┘
                      │
                      ▼
                 ┌──────────┐
                 │ CREATED  │
                 └────┬─────┘
                      │ Start
                      ▼
                 ┌──────────┐
                 │ RUNNING  │
                 └────┬─────┘
                      │
          ┌───────────┴───────────┐
          ▼                       ▼
       STOPPED                  EXITED
          │                       │
          └───────────┬───────────┘
                      │
                    Restart
                      │
                      ▼
                  RUNNING
```

---

## 9.2 `docker run`

```bash
docker run nginx
```

Creates a new container from the image and starts it.

If the image isn't locally available, Docker may pull it first.

### Conceptually

```text
Image
  │
  ▼
Create Container
  │
  ▼
Start Container
```

`docker run` is commonly used as the combined create-and-start operation.

---

## 9.3 `docker create`

```bash
docker create --name my-nginx nginx
```

Creates the container without starting its main process.

```text
Image
  │
  ▼
Container Created
  │
  ▼
Container remains stopped
```

You can start it later.

---

## 9.4 `docker start`

```bash
docker start my-nginx
```

Starts an existing stopped container.

Unlike `docker run`, this does not create a brand-new container.

```text
Existing Container
      │
      ▼
     Start
      │
      ▼
   Running
```

---

## 9.5 `docker stop`

```bash
docker stop my-nginx
```

Stops a running container.

Docker generally sends a termination signal first and allows the process time to shut down gracefully before forcefully terminating it if necessary.

```text
Running
   │
   ▼
Stop Request
   │
   ▼
Graceful Shutdown
   │
   ▼
Stopped
```

This is preferable to abruptly killing processes when they need cleanup time.

---

## 9.6 `docker restart`

```bash
docker restart my-nginx
```

Restarts an existing container.

```text
Running
  │
  ▼
Stop
  │
  ▼
Start
  │
  ▼
Running
```

Useful when you need to restart an application after certain configuration or operational changes.

---

## 9.7 `docker rm`

```bash
docker rm my-nginx
```

Removes a stopped container.

> **Important:** Removing a container is not the same as removing its image.

```text
Image: nginx
   │
   ├── Container A → Removed
   │
   ├── Container B → Still exists
   │
   └── Container C → Still exists
```

The image can remain available to create new containers.

---

## 9.8 `docker ps`

```bash
docker ps
```

Lists running containers.

To see all containers, including stopped ones:

```bash
docker ps -a
```

| Command | Shows |
|---|---|
| `docker ps` | Running containers |
| `docker ps -a` | All containers |

---

## 9.9 `docker exec`

Allows you to execute a command inside a running container.

Example:

```bash
docker exec -it my-nginx bash
```

### What do the flags mean?

| Flag | Meaning |
|---|---|
| `-i` | Keep standard input open |
| `-t` | Allocate a terminal |

Together, `-it` are commonly used for interactive shell access.

Depending on the image, `bash` may not exist. A minimal image may only provide `sh`.

Example alternative:

```bash
docker exec -it my-nginx sh
```

### What is it used for?

- Inspecting files
- Debugging
- Running commands
- Checking installed software
- Examining configuration

### `exec` vs `run`

| `docker run` | `docker exec` |
|---|---|
| Creates and starts a new container | Runs an additional command inside an already running container |
| Starts the container's primary process | Executes another process in the container |

# 🌐 MODULE 10 — PORTS & PORT MAPPING

## 10.1 What Is a Port?

A **port** is a logical communication endpoint used by networked applications.

An IP address identifies a network interface or host in a network context. Ports help identify which application or service should receive traffic.

For example, one machine can run:

```text
Web Server       → Port 80
Backend API      → Port 8000
Database         → Port 5432
```

The same machine can host multiple network services because they can listen on different ports.

---

## 10.2 Why Do Containers Need Port Mapping?

Containers are isolated network environments.

A service running inside a container is not automatically accessible from your host computer using a host port.

Suppose our FastAPI application is listening inside the container on:

```text
Port 8000
```

But the host machine doesn't have a published port forwarding rule.

```text
Host Machine
│
│ Browser
│ localhost:8000
│
X  No port publishing
│
▼
Container
└── FastAPI :8000
```

The service may be running, but it isn't published to the host in the way the user expects.

---

## 10.3 Container Port vs Host Port

These are two different ports.

### Container Port

The port where the application listens inside its container network environment.

Example:

```text
FastAPI → 8000
```

### Host Port

The port on your own computer through which you access the application.

Example:

```text
Browser → localhost:8080
```

Docker can connect these two through port publishing.

---

## 10.4 Port Mapping

Port mapping connects a host port to a container port.

```bash
docker run -p 8080:8000 sentiment-api:v1
```

### Syntax

```text
-p HOST_PORT:CONTAINER_PORT
```

In our command:

| Port | Value |
|---|---:|
| Host port | 8080 |
| Container port | 8000 |

### Request Flow

```text
Browser
   │
   │ http://localhost:8080
   ▼
HOST PORT 8080
   │
   │ Docker port forwarding
   ▼
CONTAINER PORT 8000
   │
   ▼
FastAPI Application
```

You access the service through:

```text
http://localhost:8080
```

Docker forwards traffic to port `8000` inside the container.

---

## 10.5 Why Does the Application Need `0.0.0.0`?

Look at our startup command:

```bash
uvicorn main:app --host 0.0.0.0 --port 8000
```

### `127.0.0.1` (localhost)

Generally refers to the loopback interface of the current network namespace.

If the application inside the container listens only on its own loopback interface, it may not accept connections arriving through the container's network interface.

### `0.0.0.0`

For a listening socket, means bind to all IPv4 interfaces in that network namespace.

This allows the application to listen on the container's network interfaces, so Docker's published-port forwarding can reach it.

> **Common Docker networking mistake 🚨**
>
> Running FastAPI with `--host 127.0.0.1` inside the container may prevent published-port traffic from reaching it. Port publishing alone may not make the service reachable if the process only listens on the container loopback interface.

---

## 10.6 Complete Example — Run Our API

Assume our image is already built:

```bash
docker build -t sentiment-api:v1 .
```

Now run it:

```bash
docker run --name sentiment-container -p 8080:8000 sentiment-api:v1
```

### What's happening?

```text
Docker Image
sentiment-api:v1
       │
       ▼
Create Container
sentiment-container
       │
       ▼
FastAPI starts inside container
       │
       ▼
Container listens on 8000
       │
       ▼
Docker publishes host port 8080
       │
       ▼
Browser accesses localhost:8080
```

### Test it

Open:

```text
http://localhost:8080
```

Expected response:

```json
{
  "message": "Sentiment API is running"
}
```

Open:

```text
http://localhost:8080/health
```

Expected response:

```json
{
  "status": "healthy"
}
```

## 10.7 Port Mapping Variations

### Map Host 8080 → Container 8000

```bash
docker run -p 8080:8000 sentiment-api:v1
```

Access:

```text
localhost:8080
```

### Map Host 8000 → Container 8000

```bash
docker run -p 8000:8000 sentiment-api:v1
```

Access:

```text
localhost:8000
```

### Map to a Specific Host Interface

```bash
docker run -p 127.0.0.1:8080:8000 sentiment-api:v1
```

This restricts the published port to the host's loopback address, instead of publishing on all host interfaces by default.

---

## 10.8 What Does `EXPOSE` Actually Do?

In Dockerfile:

```dockerfile
EXPOSE 8000
```

This documents the port the containerized application is expected to listen on.

It does **not** create a host port mapping by itself.

| `EXPOSE` | `-p` |
|---|---|
| Image metadata / documentation | Publishes and forwards a host port |
| Declares intended container port | Connects host and container ports |
| Does not make the service directly accessible from host | Enables host access through the published port |

Example:

```dockerfile
EXPOSE 8000
```

Then:

```bash
docker run -p 8080:8000 sentiment-api:v1
```

Both serve different purposes.

---

## 10.9 Port Mapping in a Multi-Service Application

Later, our application will have:

```text
Frontend
   │
   ▼
Backend API
   │
   ├────► AI Model Service
   │
   └────► PostgreSQL
```

Ports might be:

| Service | Internal Port |
|---|---:|
| Frontend | 3000 |
| Backend | 8000 |
| AI Model | 8001 |
| PostgreSQL | 5432 |

Not every service needs to be published to the host.

For example, PostgreSQL may only need to be accessible to the backend through the Docker network.

```text
HOST
 │
 │ Published port
 ▼
Backend :8000
 │
 ├── Docker Network ──► AI Model :8001
 │
 └── Docker Network ──► PostgreSQL :5432
```

This is the foundation of multi-container networking, which we'll study in Phase 4.

### **The problem**

```text
Container
│
├── Application code
├── Python packages
├── Model files
└── User data  ❌
```

When the container is deleted, its writable layer disappears.

### **Solution: Persistent Storage**

Docker provides storage mechanisms that allow data to exist independently of a container's lifecycle.

```text
┌──────────────────────┐
│    Container       │
│                    │
│   /app              │
│   /data ─────────────┼──────┐
└──────────────────────┘      │
                            ▼
                    Persistent Storage
                    ┌─────────────────┐
                    │ Dataset files   │
                    │ Model files     │
                    │ Database data   │
                    └─────────────────┘
```

## **11.2 Docker Storage Types**

Docker storage can be broadly understood through three important mechanisms.

| Storage type | Purpose | Typical use |
|---|---|---|
| Writable container layer | Temporary container data | Temporary files |
| Bind mount | Connect host directory to container | Development |
| Docker volume | Docker-managed persistent storage | Databases, production data |

## **11.3 Writable Container Layer**

Every running container has a writable layer above its image layers.

```text
┌─────────────────────────────┐
│ Container Writable Layer   │ ← Changes made at runtime
├─────────────────────────────┤
│ Image Layer 3             │
├─────────────────────────────┤
│ Image Layer 2             │
├─────────────────────────────┤
│ Image Layer 1             │
└─────────────────────────────┘
```

For example:

```bash
docker run -it ubuntu bash
```

Inside the container:

```bash
echo "Hello Docker" > /hello.txt
```

The file is stored in the container's writable layer.

If you remove the container:

```bash
docker rm my-container
```

The writable layer is removed along with that container.

> ⚠️ Container writable layers are not suitable for important persistent application data.

The volume survives the removal of Container 1.

## **11.6 Named Volumes vs Anonymous Volumes**

### **Named volume**

You explicitly provide the volume name.

```bash
docker run -it \
  --mount type=volume,source=ml-data,target=/data \
  ubuntu bash
```

The volume is named:

```text
ml-data
```

Advantages:

- Easy to identify
- Easy to reuse
- Easy to inspect
- Easy to manage

### **Anonymous volume**

Docker creates a volume without you explicitly naming it.

```bash
docker run -it \
  --mount type=volume,target=/data \
  ubuntu bash
```

Docker assigns a generated volume identifier.

| Feature | Named | Anonymous |
|---|---|---|
| Custom name | Yes | No |
| Easy to reuse | Yes | Less convenient |
| Explicit management | Yes | Less convenient |
| Persistent | Yes | Yes |

| Portability | Host path dependent | More Docker-managed |

## **11.9 Read-Only Mounts**

Sometimes containers should be able to read data but should not modify it.

```bash
docker run -it \
  --mount type=bind,source="$(pwd)/models",target=/models,readonly \
  python:3.12-slim bash
```

The container can access model files but cannot write to that mounted directory.

Useful for:

- Production model inference
- Configuration files
- Read-only datasets
- Security-sensitive files

## **13.1 What Are Environment Variables?**

**Environment variables** are key-value pairs used to configure applications without hardcoding configuration directly into source code.

Example:

```text
APP_ENV=production
PORT=8000
DEBUG=False
MODEL_PATH=/models/sentiment_model.pkl
```

Instead of hardcoding:

```python
port = 8000
debug = True
```

We can read configuration from the environment.

```python
import os

port = os.getenv("PORT", "8000")
debug = os.getenv("DEBUG", "False")

print(port)
print(debug)
```

This allows the same application code to run in different environments.

## **13.2 Why Environment Variables Are Important**

Imagine your AI model API runs in three environments:

| Environment | Configuration |
|---|---|
| Development | Debug enabled |
| Testing | Test database |
| Production | Production database, debug disabled |

We don't want to rewrite our code every time the environment changes.

Instead:

```text
Same Application Code
        │
        ├── Development Environment
        │       └── DEBUG=True
        │
        ├── Testing Environment
        │       └── APP_ENV=test
        │
        └── Production Environment
                └── DEBUG=False
```

## **13.3 Dockerfile `ENV` Instruction**

The Dockerfile `ENV` instruction sets environment variables in the image configuration.

### **Example**

```dockerfile
FROM python:3.12-slim

WORKDIR /app

ENV APP_ENV=production
ENV PORT=8000

CMD ["python", "app.py"]
```

The application can access these variables at runtime.

```python
import os

print(os.getenv("APP_ENV"))
print(os.getenv("PORT"))
```

Expected:

```text
production
8000
```

### **Syntax**

```dockerfile
ENV KEY=value
```

Multiple variables can also be declared in one instruction:

## **14.6 What Is a Compose File?**

## **15.1 Understanding YAML**

YAML is a human-readable data serialization format used for configuration files.

Docker Compose uses YAML to define services and their configuration.

Example:

```yaml
services:
  backend:
    image: python:3.12-slim
```

YAML relies on indentation.

⚠️ Incorrect indentation can cause YAML parsing errors.

## **15.2 YAML Basics**

### **Key-value pairs**

```yaml
name: sentiment-app
version: 1
```

### **Nested structure**

```yaml
application:
  name: sentiment-app
  environment:
    debug: false
    port: 8000
```

### **Lists**

```yaml
ports:
  - "8080:8000"
  - "8081:8001"
```

### **Important rules**

- Use consistent indentation.
- Usually use spaces, not tabs.
- Indentation defines hierarchy.
- Strings containing port mappings should generally be quoted.

## **15.3 Compose File Structure**

```yaml
services:
  frontend:
    image: nginx

  backend:
    build: .

volumes:
  app-data:

networks:
  app-network:
```

The major sections:

- `services` → application containers
- `volumes` → persistent storage resources
- `networks` → communication networks

Additional Compose features include configuration and secret definitions in supported contexts.

## **15.4 Using `image`**

The `image` property tells Docker which image to use.

```yaml
services:
  api:
    build: .
```

```bash
docker build -t api-image .
```

## **15.9 Networking in Compose**

Compose automatically creates a default network for an application project.

Services within that network can communicate with each other using service names.

```yaml
services:
  backend:
    image: python:3.12-slim

  redis:
    image: redis:alpine
```

The backend can use Redis hostname:

```text
redis
```

Instead of:

```text
127.0.0.1
```

Inside the backend container, `localhost` refers to the backend container itself, not the Redis container.

## **15.10 `depends_on`**

`depends_on` defines service dependency relationships.

```yaml
services:
  backend:
    image: my-backend
    depends_on:
      - redis

  redis:
    image: redis:alpine
```

This specifies that Redis is a dependency of the backend for Compose startup ordering.

Important: basic `depends_on` does not guarantee that the dependency is fully ready to accept connections.

For readiness-based startup, use health checks with supported `depends_on` conditions.

## **15.11 Health Checks**

A health check tests whether a service is functioning as expected.

```yaml
services:
  redis:
    image: redis:alpine
    healthcheck:
      test: ["CMD", "redis-cli", "ping"]
      interval: 5s
      timeout: 3s
      retries: 5
```

A dependent service:

```yaml
services:
  backend:
    build: .
    depends_on:
      redis:
        condition: service_healthy
```

This lets Compose wait for the Redis service's health check to pass before starting the dependent backend.

```text
Redis Starts
     │
     ▼
Health Check
     │
     ├── Unhealthy → Continue Checking
     │
     └── Healthy
           │
           ▼
       Backend Starts
```

## **15.12 Restart Policies**

```yaml
services:
  backend:
    image: my-backend
    restart: unless-stopped
```

Common restart policies:

| Policy | Meaning |
|---|---|
| `no` | Do not automatically restart |
| `always` | Restart whenever the container stops |
| `on-failure` | Restart when process exits unsuccessfully |
| `unless-stopped` | Restart unless explicitly stopped |

These policies control container restart behavior.

## **17.1 Why Is Debugging Important?**

A container can fail even when its image builds successfully.

## **17.2 Understanding Docker Logs**

Applications running inside containers typically write messages to two output streams:

| Stream | Purpose |
|---|---|
| `STDOUT` | Normal application output |
| `STDERR` | Error messages and diagnostic output |

Docker's logging system can capture these streams through its configured logging driver. `docker logs` normally displays both.

![Docker container logging architecture](https://s3.amazonaws.com/media-p.slid.es/uploads/229737/images/2116174/Log2Syslog1.png)

### **Basic commands**

```bash
# List running containers
docker ps

# List all containers
docker ps -a

# View container logs
docker logs <container_name>

# Follow logs continuously
docker logs -f <container_name>

# Show last 100 lines
docker logs --tail 100 <container_name>

# Display timestamps
docker logs -t <container_name>

# Show logs from the last 10 minutes
docker logs --since 10m <container_name>
```

You can also combine options:

```bash
docker logs -f --tail 50 -t sentiment-api
```

This displays the last 50 log lines, timestamps them, and continues streaming new output.

## **17.3 Application Logging in Python**

For an AI/ML API, use Python's logging module instead of relying only on `print()`.

### `main.py`

```python
import logging
from fastapi import FastAPI

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logger = logging.getLogger(__name__)
app = FastAPI()

@app.get("/")
def home():
    logger.info("Home endpoint accessed")
    return {"message": "AI API running"}

@app.get("/predict")
def predict():
    logger.info("Prediction request received")
    try:
        result = "positive"
        logger.info("Prediction completed successfully")
        return {"prediction": result}
    except Exception:
        logger.exception("Prediction failed")
        raise
```

Run the container and inspect:

```bash
docker logs -f sentiment-api
```

Example output:

```text
2026-11-21 10:00:00 - INFO - Home endpoint accessed
2026-11-21 10:00:05 - INFO - Prediction request received
2026-11-21 10:00:06 - INFO - Prediction completed successfully
```

**Important:** Production applications should use structured, useful logs and avoid recording passwords, API keys, tokens, or sensitive user data.

## **17.5 Understanding Container Exit Codes**

When the main process inside a container exits, Docker records its exit status.

| Exit code | General interpretation |
|---|---|
| `0` | Process completed successfully |
| `1` | General application error |
| `2` | Commonly incorrect command or usage |
| `125` | Docker itself encountered an error running the container |
| `126` | Command could not be invoked |
| `127` | Command not found |
| `137` | Process received SIGKILL; possible OOM kill, but not proof by itself |
| `143` | Process terminated following SIGTERM |

These are common conventions, not universal diagnoses. Always inspect the actual error and process state.

## **17.8 Docker Logging Drivers**

Docker uses logging drivers to manage container log output.

Common drivers:

| Driver | Purpose |
|---|---|
| `json-file` | Stores logs in JSON files |
| `local` | Docker-managed local logging with rotation-oriented storage |
| `syslog` | Sends logs to syslog |
| `journald` | Sends logs to systemd journal |
| `fluentd` | Sends logs to Fluentd |
| `awslogs` | Sends logs to Amazon CloudWatch Logs |
| `gcplogs` | Sends logs to Google Cloud Logging |

Docker's default logging driver is commonly `json-file`, but it can be changed through daemon configuration. The `local` driver is worth considering to help avoid unbounded local log-file growth.

Example: configure a container with log rotation options.

```bash
docker run -d \
  --name web \
  --log-driver=json-file \
  --log-opt max-size=10m \
  --log-opt max-file=3 \
  nginx
```

This configures a maximum log-file size and number of retained files for that container.

# **📦 MODULE 18 — DOCKER IMAGE REGISTRIES**

## **18.1 What Is a Docker Registry?**

A **Docker registry** is a service that stores and distributes container images.

Think of it as a centralized repository for your Docker images.

Your local machine stores images locally. A registry makes those images available to other machines and deployment environments.

## **19.4 Deployment Using Docker Compose**

Real applications usually have multiple services.

For example:

```text
AI APPLICATION
│
├── Backend API
├── Model Service
├── PostgreSQL
└── Redis
```

Instead of running every container manually, deploy the stack with Compose.

### **Production Compose configuration**

`compose.production.yaml`

```yaml
services:

  backend:
    image: YOUR_DOCKER_USERNAME/backend:v1
    restart: unless-stopped
    ports:
      - "8000:8000"
    environment:
      APP_ENV: production
      MODEL_SERVICE_URL: http://model-service:8001
    depends_on:
      - model-service

  model-service:
    image: YOUR_DOCKER_USERNAME/model-service:v1
    restart: unless-stopped
    expose:
      - "8001"

  postgres:
    image: postgres:16
    restart: unless-stopped
    environment:
      POSTGRES_DB: appdb
      POSTGRES_USER: appuser
      POSTGRES_PASSWORD: ${POSTGRES_PASSWORD}
    volumes:
      - postgres-data:/var/lib/postgresql/data

volumes:
  postgres-data:
```

Create a `.env` file containing the deployment configuration:

```env
POSTGRES_PASSWORD=replace_with_a_secure_secret
```

Protect this file with suitable permissions and never commit it to Git.

Start:

```bash
docker compose \
  -f compose.yaml \
  -f compose.production.yaml \
  up -d
```

This merges the base configuration with the production overrides. Docker's official production guidance describes this approach.

**Note:** This is an illustrative deployment configuration. A real deployment should also configure health checks, secret injection, resource limits, backups, TLS, and access controls.

## **19.5 Restart Policies**

What happens if your application process crashes?

Docker provides restart policies to control automatic restarts.

| Policy | Behavior |
|---|---|
| `no` | No automatic restart |
| `on-failure` | Restart after an unsuccessful exit |
| `always` | Always restart when stopped |
| `unless-stopped` | Restart unless explicitly stopped |

Example:

```bash
docker run -d \
  --name sentiment-api \
  --restart unless-stopped \
  -p 8000:8000 \
  sentiment-api:v1
```

Compose example:

```yaml
services:
  backend:
    image: sentiment-api:v1
    restart: unless-stopped
```

Docker recommends restart policies for container restart management rather than adding an unrelated process manager just to restart containers.

**Important:** A restart policy does not fix an application bug. If a container continually crashes, Docker may continually restart it. You still need logs, monitoring, and a proper fix.

## **19.6 Health Checks in Production**

A container being in the `running` state does not necessarily mean its application is ready.

For example:

```text
Container: Running
Application: Database connection failed
```

This is why health checks matter.

Dockerfile:

```dockerfile
HEALTHCHECK \
  --interval=30s \
  --timeout=5s \
  --start-period=10s \
  --retries=3 \
  CMD python -c "import urllib.request; urllib.request.urlopen('http://localhost:8000/health')"
```

Inspect:

```bash
docker inspect sentiment-api
```

Find the health status:

```bash
docker inspect \
  --format='{{.State.Health.Status}}' sentiment-api
```

Possible states:

- `starting`
- `healthy`
- `unhealthy`

Health checks report application health. They do not automatically guarantee that an unhealthy standalone container will be restarted.

## **19.7 Resource Limits**

Production applications should not be allowed to consume unlimited host resources.

Example:

```bash
docker run -d \
  --name sentiment-api \
  --memory=1g \
  --cpus=1.5 \
  sentiment-api:v1
```

This configures:

- Memory limit: 1 GB
- CPU limit: 1.5 CPUs

Compose example:

```yaml
services:
  backend:
    image: sentiment-api:v1
    mem_limit: 1g
    cpus: 1.5
```

Resource constraints are especially important for ML workloads because model loading and inference can consume substantial memory and CPU.

## **19.8 Production Security Fundamentals**

Before deploying a Docker application, consider:

1. **Use trusted base images.** Avoid unnecessary packages and outdated dependencies.
2. **Do not run as root unnecessarily.** Use a dedicated non-root application user.
3. **Do not hardcode secrets.** Use secure secret-management mechanisms.
4. **Expose only necessary ports.** Keep internal services private.
5. **Use image versions or digests.** Avoid depending blindly on mutable tags.
6. **Keep images updated.** Rebuild and test when security updates are available.
7. **Limit resources.** Prevent a container from exhausting host resources.
8. **Monitor application health.** Logs and health checks are essential.

# **📈 MODULE 20 — SCALING & LOAD BALANCING**

## **20.1 What Is Scaling?**

Scaling means increasing or decreasing application capacity according to demand.

Imagine your AI API normally receives 100 requests per minute.

Suddenly, it receives 5,000 requests per minute.

One backend container may not be able to handle the workload efficiently.

You may need additional instances.

### **Two types of scaling**

| Type | Meaning |
|---|---|
| Vertical scaling | Increase resources assigned to a machine or container |
| Horizontal scaling | Add more application instances |

## **20.2 Vertical Scaling**

Vertical scaling increases the resources available to an existing application instance.

Before:

```text
Backend Container
CPU: 1
RAM: 1 GB
```

After:

```text
Backend Container
CPU: 4
RAM: 8 GB
```

Example:

```bash
docker run -d \
  --name backend \
  --cpus=4 \
  --memory=8g \
  backend:v1
```

Advantages:

- Straightforward for some workloads
- No additional application replicas required
- Useful when an application needs more resources per instance

Limitations:

- Host hardware capacity
- Higher infrastructure cost
- Single-instance failure risk remains
- Application may not efficiently use additional resources

## **20.3 Horizontal Scaling**

Horizontal scaling adds multiple instances of the same service.

Before:

```text
          Requests
              │
              ▼
          Backend 1
```

After:

```text
              Requests
                 │
                 ▼
             Load Balancer
             /     |     \
            ▼      ▼      ▼
        Backend 1 Backend 2 Backend 3
```

Advantages:

- More capacity
- Better distribution of requests
- Potentially improved availability
- Easier to add or remove instances

Limitations:

- Requires routing or load balancing
- Shared state must be handled
- More complex monitoring
- More infrastructure consumption

## **20.4 What Is Load Balancing?**

A **load balancer** distributes incoming requests across multiple application instances.

Imagine three AI inference containers.

Instead of sending every request to one container, the load balancer distributes requests among the available instances.

![Load balancing across multiple Docker Swarm nodes](https://miro.medium.com/v2/resize%3Afit%3A1400/1%2A6emcn8efz3au3No68yP9yA.png)

### **Example**

```text
             Client Requests
                    │
                    ▼
               Load Balancer
                    │
          ┌─────────┼─────────┐
          │         │         │
          ▼         ▼         ▼
      Backend 1 Backend 2 Backend 3
          │         │         │
          └─────────┴─────────┘
                    │
                    ▼
                PostgreSQL
```

## **20.5 Common Load Balancing Algorithms**

| Algorithm | How it works |
|---|---|
| Round Robin | Sends requests to instances in rotation |
| Weighted Round Robin | Sends more traffic to higher-weight instances |
| Least Connections | Prefers the instance with fewer active connections |
| IP Hash | Uses the client IP to select an instance |
| Least Response Time | Prefers instances with lower response times, depending on implementation |

Example round robin:

```text
Request 1 → Backend 1
Request 2 → Backend 2
Request 3 → Backend 3
Request 4 → Backend 1
Request 5 → Backend 2
```

## **20.6 Scaling with Docker Compose**

Docker Compose can create multiple instances of a service.

Example:

```yaml
services:
  backend:
    image: sentiment-api:v1
    expose:
      - "8000"
```

Scale:

```bash
docker compose up -d --scale backend=3
```

Check:

```bash
docker compose ps
```

Compose now creates three backend containers.

**Important distinction:** Creating three replicas does not automatically provide a complete external load-balancing setup. You still need to decide how incoming traffic reaches them.

Also, avoid a fixed host-port mapping such as `8000:8000` on a service that you want to scale into multiple containers on the same host. Multiple containers cannot all bind the same host IP and port.

## **20.7 Service Discovery and Dynamic Ports**

When replicas are created, each container has its own network endpoint.

Compose networking supports service-name discovery between services.

For a scaled service, Docker Compose can assign dynamic published host ports:

```yaml
services:
  backend:
    image: sentiment-api:v1
    ports:
      - "8000"
```

Start three replicas:

```bash
docker compose up -d --scale backend=3
```

Inspect assigned host ports:

```bash
docker compose port --index=1 backend 8000
docker compose port --index=2 backend 8000
docker compose port --index=3 backend 8000
```

Each replica can receive a different host port. This is useful for testing, but a reverse proxy or load balancer is normally a better way to provide one stable public endpoint.

## **20.8 Reverse Proxy vs Load Balancer**

A reverse proxy receives client requests and forwards them to backend services.

It can also provide:

- TLS termination
- Request routing
- Access control
- Compression
- Caching
- Load balancing, depending on its configuration

Common technologies include:

- NGINX
- HAProxy
- Traefik

Example architecture:

```text
                   INTERNET
                      │
                      ▼
                 NGINX / PROXY
                      │
              ┌────────┼────────┐
              ▼        ▼        ▼
          Backend 1 Backend 2 Backend 3
```

A proxy can route traffic to backend replicas without exposing each replica directly to the internet.

## **20.9 Docker Swarm — Introduction**

Docker Compose is useful for managing multi-container applications, especially on a single Docker host.

When you need multi-host orchestration using Docker's native clustering technology, you can learn **Docker Swarm**.

Docker Swarm provides:

- Cluster management
- Manager and worker nodes
- Desired-state service management
- Replicated services
- Overlay networking
- Built-in service discovery
- Routing mesh and ingress load balancing

## **20.13 Docker Compose vs Docker Swarm**

| Feature | Docker Compose | Docker Swarm |
|---|---|---|
| Main purpose | Multi-container application definition and management | Cluster orchestration |
| Single-host workflows | Excellent | Supported |
| Multi-host orchestration | Not provided by Compose itself | Yes |
| Service replicas | Yes, with limitations | Yes |
| Built-in cluster management | No | Yes |
| Built-in routing mesh | No | Yes |
| Overlay networking | Not as a standalone Compose cluster feature | Yes |
| Manager/worker nodes | No | Yes |

**Important:** Compose and Swarm are related but not interchangeable. Docker Compose's `--scale` command does not turn a group of containers into a multi-node Swarm cluster.

## **20.14 Scaling AI/ML Inference Workloads**

Consider a production sentiment analysis system.

```text
                 CLIENT REQUESTS
                       │
                       ▼
                  LOAD BALANCER
                       │
             ┌──────────┼──────────┐
             ▼          ▼          ▼
          Model 1    Model 2    Model 3
             │          │          │
             └──────────┼──────────┘
                       │
                       ▼
                  SHARED STORAGE
```

Before scaling, consider:

- Model size
- CPU and GPU requirements
- Memory per model instance
- Concurrent inference requests
- Average and peak latency
- Batch inference vs real-time inference
- Database connection limits
- Model-loading time

For example, three replicas of a model that each consume 2 GB of memory could require around 6 GB just for model memory, before accounting for application overhead and other processes.

Scaling model services is therefore a resource-planning problem, not simply a matter of increasing replica count.

# **🔐 MODULE 21 — DOCKER SECURITY & BEST PRACTICES**

## **21.1 Why Docker Security Matters**

Docker containers provide isolation, but they are not automatically secure.

A poorly configured container can introduce security risks through:

- Running applications as root
- Vulnerable base images
- Exposed network ports
- Hardcoded credentials
- Excessive Linux capabilities
- Unrestricted resource usage
- Untrusted image sources
- Misconfigured mounts

Imagine deploying a machine-learning API that contains:

```text
AI APPLICATION
│
├── Python application
├── Model weights
├── API credentials
├── Database credentials
└── User data
```

If credentials are baked into the image or the application runs with unnecessary privileges, a compromised container can create serious problems.

## **21.2 Run Containers as a Non-Root User**

By default, many container images run their processes as root unless configured otherwise.

Root inside a container is not identical to unrestricted host root, but it still grants more privileges than most applications need.

## **21.3 Avoid Running Privileged Containers**

A privileged container receives significantly expanded access to host capabilities and devices.

Avoid this unless a specific, well-understood requirement makes it necessary:

```bash
docker run --privileged ubuntu
```

Prefer granting only the specific capabilities needed.

For example:

```bash
docker run \
  --cap-drop=ALL \
  --cap-add=NET_BIND_SERVICE \
  nginx
```

This drops the default Linux capabilities and adds back one capability. Whether an application needs additional capabilities depends on its workload.

## **21.4 Read-Only Container Filesystem**

A read-only root filesystem prevents the container from modifying its root filesystem.

```bash
docker run -d \
  --name secure-api \
  --read-only \
  sentiment-api:v1
```

However, applications may need temporary writable locations.

You can provide a temporary filesystem:

```bash
docker run -d \
  --name secure-api \
  --read-only \
  --tmpfs /tmp \
  sentiment-api:v1
```

This is useful when an application needs temporary files but should not modify its image filesystem.

For an AI application, model files can be mounted read-only while logs or temporary files are placed in explicitly configured writable locations.

## **22.1 Why Optimize Docker Images?**

A Docker image may work correctly but still be inefficient.

For example:

```text
Application Image
│
├── Python runtime
├── Application dependencies
├── Source code
├── Build tools
├── Test dependencies
├── Temporary files
├── Development packages
└── Unnecessary system packages
```

A production image should ideally contain only what the application needs to run.

### **Benefits of optimization**

- Smaller image size
- Faster image downloads
- Faster deployment
- Reduced storage usage
- Better build performance
- Smaller attack surface
- Easier maintenance

Docker's official recommendations emphasize multi-stage builds, cache optimization, smaller build contexts, and regular rebuilding.

![Docker Documentation](https://www.google.com/s2/favicons?domain=https://docs.docker.com&sz=32)

Docker Docs+1

## **22.2 Understanding Docker Image Layers**

Docker images are composed of layers.

For example:

```dockerfile
FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["python", "main.py"]
```

Conceptually:

```text
┌──────────────────────────────┐
│ CMD                          │
├──────────────────────────────┤
│ COPY application files       │
├──────────────────────────────┤
│ RUN install dependencies     │
├──────────────────────────────┤
│ COPY requirements.txt        │
├──────────────────────────────┤
│ WORKDIR                      │
├──────────────────────────────┤
│ Python base image            │
└──────────────────────────────┘
```

Docker can reuse cached build steps when their instructions and dependencies have not changed.

Inspect image layers:

```bash
docker history sentiment-api:v1
```

Inspect image size:

```bash
docker images
```

## **22.3 Optimize Dockerfile Layer Ordering**

Consider this Dockerfile:

Less efficient approach ❌

```dockerfile
FROM python:3.12-slim

WORKDIR /app

COPY . .

RUN pip install -r requirements.txt

CMD ["python", "main.py"]
```

The entire project is copied before dependencies are installed.

If a source file changes, the `COPY . .` layer changes, which can invalidate the subsequent dependency installation step.

### **Improved approach**

Better cache organization ✅

```dockerfile
FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install --no-cache-dir -r requirements.txt

COPY . .

CMD ["python", "main.py"]
```

Why?

```text
requirements.txt unchanged
        │
        ▼
Dependency installation cached
        │
        ▼
Source code changes
        │
        ▼
Only later layers need rebuilding
```

This is particularly useful when dependencies are large or expensive to install.

### **Single-stage approach**

```text
One Image
│
├── Compiler
├── Build dependencies
├── Test dependencies
├── Application
└── Runtime
```

### **Multi-stage approach**

```text
BUILD STAGE
│
├── Build tools
├── Development dependencies
├── Application build
└── Generated artifacts
         │
         ▼
RUNTIME STAGE
│
├── Runtime
├── Required dependencies
└── Application artifacts
```

### **Example: Python application**

```dockerfile
# Build stage
FROM python:3.12 AS builder

WORKDIR /app

COPY requirements.txt .

RUN pip install --prefix=/install \
    --no-cache-dir \
    -r requirements.txt

COPY . .

# Runtime stage
FROM python:3.12-slim

WORKDIR /app

COPY --from=builder /install /usr/local

COPY --from=builder /app /app

CMD ["python", "main.py"]
```

`COPY --from=builder` transfers selected artifacts from the earlier stage into the runtime stage.

## **22.6 BuildKit Cache Mounts**

BuildKit supports cache mounts that retain package-manager cache data between builds.

For example, a Python dependency installation step can use a pip cache:

```dockerfile
# syntax=docker/dockerfile:1

FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .

RUN --mount=type=cache,target=/root/.cache/pip \
    pip install -r requirements.txt

COPY . .

CMD ["python", "main.py"]
```

The cache mount can speed up subsequent builds by reusing downloaded packages.

The cached files are not included in the final image layer.

Docker reference: [Optimize build cache usage](https://docs.docker.com/build/cache/optimize/?utm_source=chatgpt.com).

## **22.7 Build Secrets**

Sometimes an application build needs access to a private package repository.

Never use this:

```dockerfile
ARG PRIVATE_TOKEN
```

or:

```dockerfile
ENV PRIVATE_TOKEN=my-secret
```

as a way to store build credentials.

Build arguments and environment variables are not appropriate secret-storage mechanisms because secrets can persist in image metadata or layers.

Instead, use BuildKit secret mounts.

![Docker Documentation](https://www.google.com/s2/favicons?domain=https://docs.docker.com&sz=32)

Docker Docs

Example Dockerfile:

```dockerfile
# syntax=docker/dockerfile:1

FROM python:3.12-slim

WORKDIR /app

RUN --mount=type=secret,id=pip_config \
    pip install --no-cache-dir \
    --index-url "$(cat /run/secrets/pip_config)" \
    private-package
```

Build:

```bash
docker build \
  --secret id=pip_config,src=./pip_config.txt \
  -t private-app:v1 .
```

This exposes the secret temporarily to the build instruction instead of copying it into the final image.

The example assumes `pip_config.txt` contains a suitable private package index URL. Protect that file and never commit it.

## **23.9 Step 7 — Create Docker Compose Configuration**

Create:

`compose.yaml`

```yaml
services:

  api:
    build: .
    container_name: sentiment-api

    ports:
      - "8000:8000"

    environment:
      MODEL_PATH: /app/models/sentiment.joblib
      REDIS_URL: redis://redis:6379/0

    depends_on:
      redis:
        condition: service_healthy

    restart: unless-stopped

    healthcheck:
      test:
        - CMD
        - python
        - -c
        - "import urllib.request; urllib.request.urlopen('http://localhost:8000/health')"

      interval: 30s
      timeout: 5s
      retries: 3
      start_period: 10s

  redis:
    image: redis:7-alpine

    restart: unless-stopped

    healthcheck:
      test: ["CMD", "redis-cli", "ping"]
      interval: 5s
      timeout: 3s
      retries: 5
```

### **Understand the configuration**

**API service**

- Builds the application image.
- Publishes port 8000.
- Receives the model and Redis configuration.
- Waits for Redis health status before starting.
- Uses a restart policy.
- Provides an API health check.

**Redis service**

- Uses the official Redis image.
- Runs as a separate container.
- Checks its own health with `redis-cli ping`.
- Communicates with the API using the hostname `redis`.

Compose automatically creates a default network for the application.

## **23.15 Step 13 — Production Deployment Considerations**

Before exposing this application to real users, consider:

| Area | Production improvement |
|---|---|
| Model | Train and evaluate on a representative dataset |
| API | Add request validation, authentication and rate limiting |
| Redis | Configure persistence or treat it explicitly as disposable cache |
| Security | Use secrets and restricted network access |
| Deployment | Put a reverse proxy in front of the API |
| HTTPS | Configure TLS certificates |
| Monitoring | Add metrics, logs and alerting |
| Storage | Back up important artifacts |
| Reliability | Test recovery after service failures |
| Scaling | Measure load before increasing replicas |

**One important distinction:** This is a working educational deployment design, not a fully hardened public production service. The sample model is deliberately tiny, and the project has not been load-tested.
