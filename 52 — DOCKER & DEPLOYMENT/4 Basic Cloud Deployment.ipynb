# **4. Basic Cloud Deployment**

The goal of cloud deployment is to make our containerized AI inference application accessible outside our local machine so that other applications or users can send requests to it.

We will focus only on the concepts required to deploy a containerized inference application to a managed cloud/container service—not deep cloud architecture.

---

## **4.1 Why Cloud Deployment?**

Currently, our AI application runs locally:

```text
Your Laptop
│
├── FastAPI
├── DistilBERT
├── YOLO
└── Streamlit
        │
        ▼
  localhost:8000
```

The application is accessible primarily from your own machine.

Cloud deployment allows us to run the application on a remotely hosted machine or service.

```text
User / Client
      │
      ▼
   Internet
      │
      ▼
  Cloud Service
      │
      ▼
 Docker Container
      │
      ├── FastAPI
      ├── NLP Model
      └── CV Model
```

Instead of accessing:

```text
http://127.0.0.1:8000
```

A deployed API may be accessible through a public endpoint such as:

```text
https://your-api.example.com
```

---

## **4.2 What Is Cloud Deployment?**

**Cloud deployment** is the process of making an application run on infrastructure provided by a cloud platform so that it can be accessed remotely.

The cloud provider manages some or all of the underlying infrastructure, depending on the service selected.

For our containerized application, the workflow is:

```text
Application Code
      ↓
Dockerfile
      ↓
Build Docker Image
      ↓
Push Image to Registry
      ↓
Cloud Container Service
      ↓
Run Container
      ↓
Public API Endpoint
```

The cloud service handles the hosting environment required to run the container.

---

## **4.3 Managed Container Services**

A **managed container service** allows developers to deploy and run containerized applications without manually managing all the underlying server infrastructure.

Depending on the service, the cloud provider may handle:

- Infrastructure provisioning
- Container execution
- Application health monitoring
- Restarting failed instances
- Networking
- Scaling
- Deployment management

Instead of manually configuring an entire server, we provide the container image and deployment configuration.

### **Deployment Model**

```text
Developer
   │
   ▼
Docker Image
   │
   ▼
Container Registry
   │
   ▼
Managed Container Service
   │
   ├── Runs container
   ├── Monitors application
   └── Provides network endpoint
            │
            ▼
        API Clients
```

### **Why Use Managed Services?**

- Less infrastructure administration
- Easier deployment
- Simplified container hosting
- Built-in operational features
- Faster deployment of small applications and APIs

---

## **4.4 Container Registry**

A **container registry** is a repository used to store, manage, and distribute container images.

After building an image locally, we can push it to a registry.

Examples include:

- Docker Hub
- GitHub Container Registry
- Cloud provider container registries

### **Workflow**

```text
Local Machine
    │
    ├── Dockerfile
    │
    ▼
Build Image
    │
    ▼
Container Image
    │
    ▼
Push to Registry
    │
    ▼
Cloud Service Pulls Image
    │
    ▼
Deploy Container
```

### **Why Is a Registry Needed?**

The cloud service needs access to the Docker image in order to run it.

Instead of transferring your entire source code and manually installing packages on the server, the cloud service can pull the packaged image from a registry.

---

## **4.5 Deployment Configuration**

A container may need configuration values that differ between local development and cloud deployment.

Examples:

- Port
- Environment variables
- API configuration
- Model identifier
- Memory allocation
- CPU allocation
- Instance count
- Storage configuration

### **Example**

**Local Environment**

```env
APP_ENV=development
PORT=8000
```

**Cloud Environment**

```env
APP_ENV=production
PORT=8000
```

The same application image can be deployed with different configurations.

### **Important Deployment Considerations**

| Configuration | Purpose |
|---|---|
| Port | Allows external requests to reach the application |
| Environment variables | Configure application behavior |
| CPU | Provides computational resources |
| RAM | Supports model loading and inference |
| Storage | Stores application or model data |
| Instance count | Determines how many application instances run |

For AI/ML applications, memory allocation is especially important because the model must fit into the available memory.

---

## **4.6 Public API Endpoint**

After deployment, the cloud service typically provides an endpoint through which users can access the application.

**Local API**

```text
http://localhost:8000/predict
```

**After Cloud Deployment**

```text
https://my-ai-api.example.com/predict
```

The client sends requests through the internet.

```text
Client Application
       │
       │ HTTPS Request
       ▼
  Cloud Endpoint
       │
       ▼
   Container
       │
       ▼
   FastAPI
       │
       ▼
 Model Inference
       │
       ▼
  JSON Response
```

The API contract remains the same. The main difference is where the application runs and how clients reach it.

## **4.7 Deployment Workflow for an AI Inference Application**

### **1️⃣ Develop Application**

FastAPI + Model + Preprocessing

### **2️⃣ Containerize**

Dockerfile + Dependencies

### **3️⃣ Build Image**

Create Docker image

### **4️⃣ Push to Registry**

Store image remotely

### **5️⃣ Deploy**

Managed container service runs image

### **6️⃣ Access API**

Remote clients send inference requests

### **Step-by-Step**

1. Develop and test the model-serving application locally.
2. Create a Dockerfile.
3. Build the Docker image.
4. Test the container locally.
5. Push the image to a container registry.
6. Configure the managed cloud container service.
7. Deploy the image.
8. Obtain the public endpoint.
9. Test the deployed API.

---

## **4.8 AI/ML Deployment Considerations**

Deploying an AI model requires more consideration than deploying a lightweight web application.

### **Resource Requirements**

**CPU**

Inference computation, preprocessing and application processes

**RAM**

Model loading, runtime memory and request processing

**Storage**

Model weights, libraries and application files

**GPU (Optional)**

Accelerates compatible inference workloads

### **Key Considerations**

- Model size
- Container image size
- Model startup time
- CPU inference speed
- RAM consumption
- GPU availability
- Concurrent request handling
- API response latency
- Hosting cost

A small model such as a lightweight classifier may run on a CPU-based cloud instance, whereas larger transformer or computer vision models may require higher memory or GPU resources depending on workload and latency requirements.

---

## **4.9 Basic Cloud Deployment vs. Full Cloud Architecture**

For our AI/ML engineer roadmap, we only need the practical deployment fundamentals.

| Learn Now | Not Necessary to Specialize in Now |
|---|---|
| Docker image deployment | Deep cloud networking |
| Container registry | Advanced VPC design |
| Managed container service | Complex Kubernetes administration |
| Environment variables | Infrastructure architecture specialization |
| CPU/RAM configuration | Multi-region architecture |
| Public API endpoint | Advanced distributed systems |
| Basic logs and monitoring | Cloud platform engineering deep dive |

The goal is to be able to **containerize, deploy, expose, and troubleshoot an AI inference application**.

### **Interview Question 🎯**

**How would you deploy a Dockerized ML inference API to the cloud?**

> First, I would containerize the inference application using a Dockerfile and test it locally. I would then build the Docker image and push it to a container registry. Next, I would deploy that image using a managed container service, configure the required port, environment variables, and CPU/RAM resources, and verify the public endpoint. Finally, I would test the inference API and monitor logs, resource utilization, and response latency.
