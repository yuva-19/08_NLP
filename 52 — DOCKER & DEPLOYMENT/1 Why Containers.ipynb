# 1. Why Containers?

## 1.1 The Problem Before Containers

Imagine you develop an AI application on your laptop. Your application works perfectly because your system has:

- Python 3.10
- PyTorch 2.7
- Transformers 5.x
- Specific libraries and dependencies
- CUDA configuration
- Particular environment variables

Now you send your project to another developer or deploy it on a server.

Their environment might have:

- Python 3.12 instead of Python 3.10
- Different package versions
- Missing dependencies
- Different operating systems
- Different system configurations

Your application may fail even though the code itself is correct.

This is the classic problem:

> **"It works on my machine!"** 😭

## 1.2 What Is a Container?

A **container** is an isolated execution environment that packages an application and its required dependencies so that it can run consistently across different systems.

A container typically includes:

- Application code
- Runtime dependencies
- Required libraries
- Configuration
- Environment settings

Containers share the host operating system's kernel but isolate application processes and their environments.

## 1.3 Reproducibility

**Reproducibility** means being able to recreate the same application environment and execution conditions reliably.

In machine learning, reproducibility is particularly important because an application depends on multiple components:

- Python version
- ML framework version
- Model library versions
- CUDA compatibility
- System libraries
- Application configuration

### Without containers

You manually configure each environment.

```
Developer Laptop
├── Python 3.10
├── PyTorch 2.7
├── Transformers 5.x
└── FastAPI

Production Server
├── Python 3.12
├── PyTorch 2.5
├── Transformers 4.x
└── FastAPI
```

The environments are different, so the application may behave differently or fail.

### With containers

You define the environment once and package it into an image.

```
Docker Image
├── Python Runtime
├── Dependencies
├── Application Code
└── Configuration
       │
       ├── Developer Laptop
       ├── Testing Server
       └── Production Server
```

The same image can be used across multiple environments, reducing environment-related inconsistencies.

### Benefits

- Consistent software versions
- Reproducible deployments
- Easier debugging
- Reduced environment configuration problems
- More predictable application behavior

### The dependency problem

Suppose your application requires:

```
Application
    ↓
Transformers 5.2
    ↓
PyTorch 2.7
    ↓
Python 3.10
```

If the deployment server has incompatible versions, dependency conflicts can occur.

For example:

- Package A requires version 2.0 of a library.
- Package B requires version 3.0.
- Your system already contains version 1.5.

Installing dependencies manually can become difficult.

### Containers solve this by defining dependencies inside the environment

```
Container
│
├── Python 3.10
├── FastAPI
├── Uvicorn
├── PyTorch
├── Transformers
└── Application
```

The application doesn't need to depend on whatever Python packages happen to be installed globally on the host machine.

> Containers don't eliminate dependency management. They make the dependency environment easier to define, package, and reproduce.

## 1.5 Runtime Isolation

**Runtime isolation** means an application runs within a separated environment where its processes, filesystem view, networking configuration, and resource usage can be managed independently from other applications.

Imagine a server running three applications:

```
Physical Server
│
├── Application A → Container A
│   ├── Python 3.10
│   └── FastAPI
│
├── Application B → Container B
│   ├── Python 3.12
│   └── Django
│
└── Application C → Container C
    ├── Node.js
    └── Express
```

Each application can have its own dependencies and runtime configuration.

Application A can use Python 3.10, while Application B uses Python 3.12, without requiring both applications to use the same Python installation on the host.

### What containers isolate

Containers provide isolation mechanisms for:

- Processes
- Filesystem views
- Network namespaces
- Environment variables
- Resource limits (when configured)
- Application dependencies

### Why isolation matters

Without isolation, multiple applications may interfere with one another.

For example:

- One application upgrades a shared library.
- Another application breaks because the library version changed.
- Applications compete for memory and CPU.
- Application configurations conflict.

Containers help reduce these conflicts by separating application environments.

**Important:** Containers provide isolation, but they are not equivalent to full virtual machines. Containers share the host kernel, so their isolation boundary is different from a VM's.

## 1.6 Portable Deployment

**Portability** means an application can be moved and executed across different supported environments with minimal changes.

For example, a Dockerized AI application can potentially run on:

- Windows with Docker Desktop
- Linux servers

### Traditional deployment

```
Developer
    ↓
Copy source code
    ↓
Install Python
    ↓
Install dependencies
    ↓
Configure environment
    ↓
Fix compatibility issues
    ↓
Run application
```

This process may differ on every server.

### Containerized deployment

```
Developer
    ↓
Build Docker Image
    ↓
Transport Image
    ↓
Run Container
    ↓
Application Starts
```

The deployment target doesn't need to recreate every dependency manually because the environment is packaged into the image.

### Example

You develop a FastAPI AI application locally.

```
Local Development
       ↓
Docker Image
       ↓
┌──────────────┬──────────────┬──────────────┐
│ Linux Server│ Cloud VM     │ Test Server  │
└──────────────┴──────────────┴──────────────┘
```

The same image can be deployed to these environments, provided the target platform supports the image's architecture, required runtime, and hardware needs.

Containers are especially useful in AI/ML because model-serving applications have complex environments.

Consider our **Project 53 — Model Serving with FastAPI & Streamlit**.

- Production AI systems

## 1.8 Containers vs Virtual Machines

| Feature | Containers | Virtual Machines |
|---|---|---|
| Virtualization | OS-level | Hardware-level |
| Operating system | Share host kernel | Guest OS included |
| Startup | Generally fast | Generally slower |
| Size | Usually lightweight | Usually larger |
| Isolation | Process/environment isolation | Stronger OS boundary |
| Resource usage | Usually lower overhead | Usually higher overhead |
| Use case | Applications, APIs, microservices | Full OS environments |

## 1.9 Key Takeaways

| Concept | Meaning |
|---|---|
| **Reproducibility** | Recreate the same application environment consistently |
| **Dependencies** | Packages, libraries and software required by an application |
| **Runtime Isolation** | Separate applications and their environments from one another |
| **Portability** | Run an application across different supported environments |
| **Container** | A packaged application running in an isolated environment |

### 🧠 One-line revision

> **Containers package an application and its dependencies into an isolated, portable environment, making deployment more consistent and reproducible across systems.**

### 🎯 Interview question

**Why do we use Docker in AI/ML deployment?**

We use Docker to package an AI/ML application with its runtime environment, dependencies, and configuration into a portable container image. This helps maintain consistency between development, testing, and production environments, reduces dependency conflicts, and simplifies deployment across different systems.
