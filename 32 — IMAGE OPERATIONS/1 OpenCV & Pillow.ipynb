````python
from PIL import Image
import cv2
import numpy as np
import matplotlib.pyplot as plt
````

````python
image_path = "images/image.jpg"
````

````python
image_pillow = Image.open(image_path)
print(image_pillow)
````

````python
print("Format:", image_pillow.format)
print("Size:", image_pillow.size)
print("Mode:", image_pillow.mode)
````

````python
plt.imshow(image_pillow)
plt.axis("off")
plt.show()
````

````python
# Convert Pillow Image → NumPy

image_np = np.array(image_pillow)

print(image_np.shape) #now this in shape of H W C 
print(image_np.dtype)
````

````python
# Load with OpenCV

image_cv = cv2.imread(image_path)

print(image_cv.shape)
print(image_cv.dtype)
````

## difference between pillow and opencv
```
Pillow:
[R, G, B]

OpenCV:
[B, G, R]

````python
# plotting opencv image directly 
plt.imshow(image_cv)
plt.axis('off')
plt.show()
````

````python
# now image looks wrong coz 
# OpenCV in [B, G, R]
# so we have to convert 

image_rgb = cv2.cvtColor(image_cv, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.axis("off")
plt.show()
````

````python
# Compare Both Pillow and OpenCV 

print("Pillow:")
print("Format:", image_pillow.format)
print("Size:", image_pillow.size)
print("Mode:", image_pillow.mode)

print("\nOpenCV:")
print("Shape:", image_cv.shape)
print("Dtype:", image_cv.dtype)
````

````python
# save the pillow 

output_path = "images/output_pillow.jpg"
image_pillow.save(output_path)
````

````python
# to verify the that 

saved_image = Image.open(output_path)

print(saved_image.format)
print(saved_image.size)
````

````python
# saving the OpenCV

output_path_cv = "images/output_opencv.jpg"

cv2.imwrite(output_path_cv, image_cv)
````
