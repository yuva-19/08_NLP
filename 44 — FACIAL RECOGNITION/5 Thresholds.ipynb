## 46.5 — Thresholds in Face Recognition

🔥 **Must Know Conceptually**

We've already learned how face verification works:

1. Capture two face images.
2. Generate their embeddings.
3. Calculate similarity.
4. Compare the similarity against a threshold.
5. Decide whether they belong to the same person.

Now let's understand the most important part of that decision: **How do we choose the threshold, and what happens when we choose it incorrectly?**

## 3. Similarity Threshold

The similarity threshold is a predefined value used to decide whether two face embeddings belong to the same person.

Let's assume our model produces cosine similarity scores between -1 and +1.

For this illustrative example, let's choose:

$$
T = 0.80
$$

### Decision rule

$$
\text{Decision} =
\begin{cases}
\text{MATCH}, & S \geq T\\
\text{NO MATCH}, & S < T
\end{cases}
$$

Where:

- \(S\) = cosine similarity score
- \(T\) = similarity threshold

### Example

| Pair | Similarity | Decision |
|---|---:|---|
| Genuine pair A | 0.94 | MATCH |
| Genuine pair B | 0.87 | MATCH |
| Impostor pair A | 0.45 | NO MATCH |
| Impostor pair B | 0.72 | NO MATCH |

The threshold separates the two decision regions.

### Interactive threshold experiment

Change the threshold and observe how the decisions change.

**Illustrative scores:**

| Pair | Type | Score |
|---|---|---:|
| Genuine A | Genuine | 0.94 |
| Genuine B | Genuine | 0.87 |
| Genuine C | Genuine | 0.76 |
| Impostor A | Impostor | 0.72 |
| Impostor B | Impostor | 0.45 |

At an initial threshold of 0.75:

- Genuine pairs correctly accepted: 3/3
- Impostor pairs incorrectly accepted: 0/2

*Illustrative scores only — not outputs from a trained face recognition model.*

**Important:** A higher threshold makes the system stricter. A lower threshold makes it more permissive.

## 4. False Acceptance (FAR)

False acceptance happens when the system incorrectly accepts two different people as the same person.

### Example

- Person A submits their face.
- Person B attempts verification.
- The system incorrectly accepts B as A.

This is a **security failure**.

### False Acceptance Rate (FAR)

$$
FAR =
\frac{\text{Number of false acceptances}}
{\text{Total impostor attempts}}
$$

**Example:**

- Total impostor attempts = 1,000
- False acceptances = 20

$$
FAR = \frac{20}{1000} = 0.02
$$

$$
\boxed{FAR = 2\%}
$$

Lower FAR generally means stronger protection against impostor attempts.

---

## 5. False Rejection (FRR)

False rejection happens when the system incorrectly rejects two images belonging to the same person.

### Example

- You registered your face in your smartphone.
- You try unlocking it under poor lighting.
- The system fails to recognize you.

This is a **usability failure**.

### False Rejection Rate (FRR)

$$
FRR =
\frac{\text{Number of false rejections}}
{\text{Total genuine attempts}}
$$

**Example:**

- Total genuine attempts = 1,000
- False rejections = 50

$$
FRR = \frac{50}{1000} = 0.05
$$

$$
\boxed{FRR = 5\%}
$$

Lower FRR generally means fewer genuine users are incorrectly rejected.

---

## 6. Relationship Between Threshold, FAR and FRR

This is one of the most important concepts in face verification.

For cosine similarity:

- Increasing the threshold makes acceptance harder.
- Decreasing the threshold makes acceptance easier.

| Threshold change | FAR | FRR |
|---|---|---|
| Increase threshold | Generally decreases | Generally increases |
| Decrease threshold | Generally increases | Generally decreases |

### Think about a smartphone face-unlock system

**Case A — Very low threshold**

The system accepts faces easily.

- Convenient for genuine users.
- Greater chance of accepting an impostor.

**Case B — Very high threshold**

The system accepts faces only when similarity is very high.

- Better resistance to impostors.
- Greater chance of rejecting the genuine user.

The engineering challenge is to find a suitable balance.

## 7. ROC-Style Analysis

🔥 **Important for interviews**

ROC stands for **Receiver Operating Characteristic**.

It is a graph used to evaluate how a classification system behaves at different thresholds.

For face verification, we commonly plot:

- X-axis: False Positive Rate (FPR), equivalent to FAR in this setting.
- Y-axis: True Positive Rate (TPR), also called genuine acceptance rate.

### Mathematical formulas

$$
TPR = \frac{TP}{TP+FN}
$$

$$
FPR = \frac{FP}{FP+TN}
$$

Where:

| Term | Meaning |
|---|---|
| TP | True Positives |
| FN | False Negatives |
| FP | False Positives |
| TN | True Negatives |

As the threshold changes, the system produces different combinations of TPR and FPR.

Original source: [Wikimedia Commons — ROC curve](https://commons.wikimedia.org/wiki/File%3ARoc_curve.svg?utm_source=chatgpt.com)

### How to interpret the ROC curve

- A curve closer to the top-left corner generally indicates better discrimination.
- A diagonal line represents random-classifier performance.
- AUC (Area Under the ROC Curve) summarizes the ROC curve into one number.
- Higher AUC generally indicates better separation between genuine and impostor pairs.

### Why do we need ROC?

Suppose we are developing face authentication for:

1. A smartphone.
2. A high-security access-control system.

Both systems may use the same recognition model, but their acceptable FAR and FRR may differ.

ROC analysis helps engineers evaluate the trade-off and choose an operating threshold based on the application.

### EER — Equal Error Rate

Another useful metric is **Equal Error Rate (EER)**.

It is the operating point where:

$$
FAR = FRR
$$

A lower EER generally indicates better verification performance.

However, a real application does not necessarily use the EER threshold. A security-sensitive system may prioritize a much lower FAR.

## 8. How Engineers Select the Threshold

A practical evaluation process:

1. Collect genuine and impostor face pairs from a representative validation dataset.
2. Generate embeddings using the same face recognition model.
3. Calculate similarity scores for all pairs.
4. Test different thresholds.
5. Calculate FAR and FRR at each threshold.
6. Plot the ROC curve.
7. Select a threshold according to the application's security and usability requirements.
8. Validate the selected threshold on separate test data.

**Critical engineering point:** A threshold such as 0.80 is not universally correct. It depends on the model, preprocessing, dataset, demographic and environmental conditions, and evaluation protocol.

---

## What you should remember 🧠

1. **Genuine pair:** Two images of the same person.
2. **Impostor pair:** Two images of different people.
3. **Threshold:** Controls the MATCH / NO MATCH decision.
4. **FAR:** Different people incorrectly accepted.
5. **FRR:** Same person incorrectly rejected.
6. **ROC:** Evaluates performance across multiple thresholds.
7. **EER:** Point where FAR and FRR are equal.
