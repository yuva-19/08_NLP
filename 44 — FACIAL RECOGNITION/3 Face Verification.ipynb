## 46.3 Face Verification

**Must Know**

*Computer Vision · Biometrics*

The central question in face verification is:

> "Are these two faces the same person?"

Face verification is a **1:1 biometric matching process**. It compares two facial images and determines whether they belong to the same identity.

For example:

- A person registers their face on a smartphone.
- Later, they try to unlock the phone using their face.
- The system compares the newly captured face with the registered face.

## 2. Step-by-step explanation

### Step 1 — Registered face

The system stores a reference representation of an enrolled user's face.

```text
Registered Image
       ↓
Face Recognition Model
       ↓
Embedding A
```

Example:

$$
A=[0.12,\ 0.45,\ 0.78,\ 0.23]
$$

### Step 2 — Captured face

The camera captures a new image when the person attempts authentication.

```text
New Camera Image
       ↓
Face Recognition Model
       ↓
Embedding B
```

Example:

$$
B=[0.14,\ 0.43,\ 0.76,\ 0.25]
$$

### Step 3 — Compare embeddings

The system calculates the similarity between A and B.

Two common approaches:

- Cosine similarity
- Euclidean distance

### Step 4 — Apply threshold

Suppose the cosine similarity is:

$$
S(A,B)=0.94
$$

And the illustrative matching threshold is:

$$
T=0.85
$$

Since:

$$
0.94\geq0.85
$$

**MATCH**

The system accepts the biometric comparison.

If the similarity falls below the threshold, it returns NO MATCH.

**Important:** These threshold values are examples, not universal face verification settings.

---

## 3. Four possible verification outcomes

This is an important concept in biometric systems.

| Actual situation | System decision | Outcome |
|---|---|---|
| Same person | Match | True Accept |
| Different people | No Match | True Reject |
| Different people | Match | False Accept |
| Same person | No Match | False Reject |

Let's understand the two important errors.

### A. False Acceptance Rate (FAR)

A different person is incorrectly accepted as the enrolled person.

$$
FAR=
\frac{\text{False Accepts}}
{\text{All Impostor Attempts}}
$$

### B. False Rejection Rate (FRR)

The genuine person is incorrectly rejected.

$$
FRR=
\frac{\text{False Rejects}}
{\text{All Genuine Attempts}}
$$

![Biometric authentication error rates illustration](https://images.openai.com/static-rsc-4/FSeMgD2SDttL8llfNX8xIza_o-wxHQcM_sf_IfuHDLOZrdc1VN4WTWjs5FfdsqFzHAl4MwxmkPzpCBqmgNWT_8MkWsAkSSgngFkM6Qjs26lUIOqt1n56VxI3d7DlYoiW9rcI2cATJl7HvlRqq2aSQxF_kjcUCHkN7F3PBJHuY9I?purpose=inline)

[Original image source — F1000Research](https://f1000research.com/articles/15-5?utm_source=chatgpt.com)

### Threshold trade-off

- A stricter matching threshold can reduce false acceptance but increase false rejection.
- A more permissive threshold can reduce false rejection but increase false acceptance.

The appropriate threshold depends on the application's security and usability requirements.

## 4. Practical Python implementation

Let's simulate a verification system using PyTorch embeddings.

```python
import torch
import torch.nn.functional as F

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Example face embeddings
registered_face = torch.tensor(
    [[0.12, 0.45, 0.78, 0.23]],
    device=device
)

captured_face = torch.tensor(
    [[0.14, 0.43, 0.76, 0.25]],
    device=device
)

# Calculate cosine similarity
similarity = F.cosine_similarity(
    registered_face,
    captured_face,
    dim=1
)

threshold = 0.85

if similarity.item() >= threshold:
    print("MATCH")
else:
    print("NO MATCH")

print("Similarity:", similarity.item())
```

This is a verification example using manually created vectors, not a complete facial recognition application. In a real system, both embeddings would come from a trained face recognition model.

## 6. Face verification vs face identification

| Feature | Verification | Identification |
|---|---|---|
| Question | Is this the same person? | Who is this person? |
| Comparison | 1:1 | 1:N |
| Input | Two face representations | One face and a database |
| Output | Match / No Match | Candidate identity or no match |
| Example | Smartphone unlock | Searching a face database |

---

## 7. What you should remember

1. Face verification is a 1:1 biometric comparison.
2. It compares embeddings generated from two face images.
3. Cosine similarity or Euclidean distance can measure similarity.
4. A threshold determines the matching decision.
5. FAR measures incorrect acceptance of an impostor.
6. FRR measures incorrect rejection of a genuine user.
7. Changing the threshold creates a trade-off between security and usability.
