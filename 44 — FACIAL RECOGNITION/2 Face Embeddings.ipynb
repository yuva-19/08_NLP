## 1. What is a face embedding?

## 2. Understanding embedding vectors

A vector is an ordered collection of numerical values.

For example:

$$
\mathbf{A}=[0.2,\ 0.5,\ 0.8]
$$

This vector has three dimensions.

Real face recognition models commonly produce much larger vectors, such as 128, 256 or 512 dimensions. The exact dimension depends on the architecture.

Mathematically:

$$
f(I)=\mathbf{e}
$$

Where:

- \(I\) = Input face image.
- \(f\) = Trained neural network.
- \(\mathbf{e}\) = Generated embedding vector.

### How does the model learn these vectors?

Think about the CNN concepts you've already studied.

```text
Face Image
    ↓
Convolution Layers
    ↓
Feature Maps
    ↓
Feature Extraction
    ↓
Fully Connected / Projection Layer
    ↓
Embedding Vector
```

The network learns useful feature representations through training.

The goal is to organize the embedding space so that faces belonging to the same identity are generally closer together than faces belonging to different identities.

### Important distinction

The individual numbers in an embedding are not necessarily directly interpretable as specific facial measurements.

For example:

- Dimension 1 does not necessarily mean eye width.
- Dimension 2 does not necessarily mean nose length.
- Dimension 3 does not necessarily mean mouth shape.

They are learned numerical features that work together to represent identity-related information.

## 3. What is distance in an embedding space?

Distance measures how far apart two embedding vectors are.

Think of each embedding as a point in a mathematical space.

- Same person's embeddings should ideally be close.
- Different people's embeddings should ideally be farther apart.

![Contrastive learning and embedding space illustration](https://images.openai.com/static-rsc-4/7_GNpVWde5B8e4Fml_jtbVjZ58ohKFlOxC01D3aT2Uod2Hh0b4xYNDpgdtpTd9eYrA-dMUx7zm8fkyiW0v-VJ9Kcy-7yWt5uQRSCA3mbMI9kBKlNYIT9MwTCwBQBpq0JdsSzRTdgRLiC3yU_EA2T7p4JrYnokXK9gc4ht9-HLt0?purpose=inline)

[Original image source — MetricGate](https://metricgate.com/blogs/contrastive-vs-supervised-learning/?utm_source=chatgpt.com)

For example:

```text
Person A — Image 1 → Embedding A
Person A — Image 2 → Embedding B
Person B — Image 1 → Embedding C
```

We compare:

- Distance(A, B): Ideally small.
- Distance(A, C): Ideally larger.

There are different ways to calculate this distance or similarity. Two important methods are Euclidean distance and Cosine similarity.

## 4. Euclidean distance

**Must Know**

Euclidean distance measures the straight-line distance between two points in a vector space.

You may already know this concept from coordinate geometry.

For two-dimensional points:

$$
A=(x_1,y_1),\quad B=(x_2,y_2)
$$

The distance is:

$$
d(A,B)=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

For face embeddings with \(n\) dimensions, we extend the formula:

$$
\boxed{d(A,B)=\sqrt{\sum_{i=1}^{n}(A_i-B_i)^2}}
$$

### Practical numerical example

Consider two embeddings:

$$
A=[0.2,\ 0.5,\ 0.8]
$$

$$
B=[0.3,\ 0.4,\ 0.7]
$$

**Step 1: Subtract corresponding values.**

$$
A-B=[-0.1,\ 0.1,\ 0.1]
$$

**Step 2: Square each difference.**

$$
[0.01,\ 0.01,\ 0.01]
$$

**Step 3: Add the squared differences.**

$$
0.01+0.01+0.01=0.03
$$

**Step 4: Take the square root.**

$$
d(A,B)=\sqrt{0.03}
$$

$$
\boxed{d(A,B)\approx0.1732}
$$

| Large distance | Vectors are far apart |

Remember: For Euclidean distance, smaller values indicate greater similarity.

## 5. Cosine similarity

**Must Know**

Cosine similarity measures the angle between two vectors rather than directly measuring their physical distance.

Its formula is:

$$
\boxed{
\text{Cosine Similarity}(A,B)=
\frac{A\cdot B}{\|A\|\|B\|}
}
$$

Where:

- \(A\cdot B\) = Dot product of two vectors.
- \(\|A\|\) = Magnitude (length) of vector A.
- \(\|B\|\) = Magnitude (length) of vector B.

### Understanding the concept visually

![Cosine similarity and vector angle illustration](https://images.openai.com/static-rsc-4/3n_O3eGn89L19ryMkX7uryjXz18CjzLUJ7csvDauotD9x8Y0cVnzms5pPISXHlPzP9InG1o1xOreUi_SAEW3RWp-C03nPyuOr3-UsQLGDAaXqjNuPJI9deuZSk_u2NhFD26YkYBneop4IXGzV0NlnBxKkgrQ2sjyOP8JeudUQ14?purpose=inline)

[Original image source — reinteractive](https://reinteractive.com/articles/reinteractive-company-news-and-information/understanding-ai-semantic-search?utm_source=chatgpt.com)

Consider three situations:

**Case 1 — Same direction**

Angle = 0°

$$
\cos(0^\circ)=1
$$

Similarity = 1

The vectors point in the same direction.

**Case 2 — Perpendicular vectors**

Angle = 90°

$$
\cos(90^\circ)=0
$$

Similarity = 0

The vectors are perpendicular.

**Case 3 — Opposite directions**

Angle = 180°

$$
\cos(180^\circ)=-1
$$

Similarity = -1

The vectors point in opposite directions.

### Numerical example

Consider:

$$
A=[1,2]
$$

$$
B=[2,4]
$$

**Step 1: Calculate dot product.**

$$
A\cdot B=(1\times2)+(2\times4)=10
$$

**Step 2: Calculate vector magnitudes.**

$$
\|A\|=\sqrt{1^2+2^2}=\sqrt5
$$

$$
\|B\|=\sqrt{2^2+4^2}=\sqrt{20}
$$

**Step 3: Substitute into the formula.**

$$
\text{Similarity}=
\frac{10}{\sqrt5\times\sqrt{20}}
$$

$$
\boxed{\text{Similarity}=1}
$$

Both vectors have the same direction, so their cosine similarity is 1.

Remember: For cosine similarity, higher values indicate greater similarity.

Cosine similarity is undefined for a zero vector because its magnitude is zero. Face embedding models generally produce nonzero embeddings.

## 6. Cosine similarity vs Euclidean distance

This distinction is extremely important for interviews and practical implementation.

![Cosine similarity versus Euclidean distance comparison](https://images.openai.com/static-rsc-4/mhTLf6nL1pm0wk2umyNUIfTZpBpGKDhjfnpvKpKHFZEMpxADx4zkoAoZhwREeIegHvL3teEjttsPzrj92hIHgA3GCfKP3JuW7e5ysNgTwzRQXdfeqYkaJ6x7x_d7YNWNPbO62DKGlKUDwi7uKuY6nCCcbG-NJVOHeTcZUx7gmtE?purpose=inline)

[Original image source — Proptimise AI](https://proptimiseai.com/blog/cosine-similarity-vs-euclidean-distance-vector-search?utm_source=chatgpt.com)

| Property | Cosine similarity | Euclidean distance |
|---|---|---|
| Measures | Angle between vectors | Straight-line distance |
| Focus | Direction | Distance and magnitude |
| Best matching indication | Higher score | Lower distance |
| Identical vectors | 1 | 0 |
| Sensitive to vector magnitude | Less sensitive | Yes |
| Typical use | Embedding comparison | Embedding comparison |

### An important mathematical connection

Suppose both embeddings are L2-normalized, meaning their magnitudes are 1.

Then:

$$
\|A\|=\|B\|=1
$$

Euclidean distance squared becomes:

$$
d^2=\|A-B\|^2
$$

Expanding:

$$
d^2=\|A\|^2+\|B\|^2-2(A\cdot B)
$$

Since both magnitudes are 1:

$$
\boxed{d^2=2-2\cos(A,B)}
$$

This means that for unit-normalized embeddings:

- Higher cosine similarity corresponds to lower Euclidean distance.
- Both metrics produce equivalent rankings of similarity.

This is a particularly useful connection to understand.

## 7. Practical implementation using Python and PyTorch

Now let's calculate both metrics using embeddings.

We'll use PyTorch because the same operations can run on your CUDA GPU when available.

### Step 1 — Create sample embeddings

```python
import torch
import torch.nn.functional as F

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

embedding_A = torch.tensor(
    [[0.2, 0.5, 0.8]],
    device=device
)

embedding_B = torch.tensor(
    [[0.3, 0.4, 0.7]],
    device=device
)

print("Device:", device)
```

### Step 2 — Calculate Euclidean distance

```python
euclidean_distance = torch.norm(
    embedding_A - embedding_B,
    p=2,
    dim=1
)

print("Euclidean distance:", euclidean_distance.item())
```

### Step 3 — Calculate cosine similarity

```python
cosine_similarity = F.cosine_similarity(
    embedding_A,
    embedding_B,
    dim=1
)

print("Cosine similarity:", cosine_similarity.item())
```

### Step 4 — Normalize embeddings

```python
normalized_A = F.normalize(
    embedding_A,
    p=2,
    dim=1
)

normalized_B = F.normalize(
    embedding_B,
    p=2,
    dim=1
)
```

Now calculate both metrics again.

```python
cosine = F.cosine_similarity(
    normalized_A,
    normalized_B,
    dim=1
)

distance = torch.norm(
    normalized_A - normalized_B,
    p=2,
    dim=1
)

print("Cosine similarity:", cosine.item())
print("Euclidean distance:", distance.item())
```

You should observe the mathematical relationship we discussed earlier.

These examples compare manually created vectors. They do not perform actual face recognition because no trained face embedding model has been loaded.

## 8. How does a recognition system make the final decision?

Suppose the system has two face embeddings.

```text
Registered Face → Embedding A
Captured Face   → Embedding B
```

It calculates their similarity and compares the result against a threshold.

### For cosine similarity

```python
if similarity >= threshold:
    print("MATCH")
else:
    print("NO MATCH")
```

### For Euclidean distance

```python
if distance <= threshold:
    print("MATCH")
else:
    print("NO MATCH")
```

The threshold must be selected using validation data for the specific model and application.

### What happens during model training?

The recognition model learns to create a useful embedding space.

Its training objective is broadly:

```text
Same identity
      ↓
Embeddings should be closer

Different identities
      ↓
Embeddings should be more separated
```

Two important training approaches you'll encounter later:

- **Triplet Loss:** Learns by comparing anchor, positive and negative examples.
- **ArcFace Loss:** Encourages better separation between identity classes using an angular margin.

We'll study these when we reach face recognition architectures and loss functions.

1. A face embedding is a numerical representation generated by a trained neural network.
2. Embeddings are points in a multidimensional feature space.
3. Euclidean distance measures the straight-line distance between embeddings.
4. Cosine similarity measures the angle between embedding vectors.
5. Lower Euclidean distance generally means greater similarity.
6. Higher cosine similarity generally means greater similarity.
7. For L2-normalized vectors, cosine similarity and Euclidean distance have an equivalent ranking.
