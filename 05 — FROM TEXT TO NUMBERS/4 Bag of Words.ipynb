# Module 5.4 — Bag of Words (BoW) 📚

Bag of Words is a fundamental Natural Language Processing (NLP) technique that converts text documents into numerical vectors using word occurrences or frequencies.

Core idea: Represent a document by counting its words, generally ignoring grammar and word order.

## 1. How Bag of Words works

Consider three documents:

- D1: `I love NLP`
- D2: `I love Python`
- D3: `Python love NLP`

The BoW process has three steps:

1. Build a vocabulary of unique words.
2. Count each vocabulary word in every document.
3. Convert the counts into numerical vectors.



## 2. Vocabulary

A vocabulary is the collection of unique tokens identified across a corpus.

For our example:

\\[ V=[\text{I},\text{love},\text{NLP},\text{Python}] \\]

Vocabulary size:

\\[ |V|=4 \\]

Each word is assigned a fixed position in the vector. The same vocabulary order must be used for all documents.

## 3. Word Frequency

Word frequency is the number of times a word appears in a document.

Example:

`NLP is useful and NLP is interesting`

| Word        | Frequency |
| ----------- | --------- |
| NLP         | 2         |
| is          | 2         |
| useful      | 1         |
| and         | 1         |
| interesting | 1         |

In count-based BoW, each feature stores the raw word count. Other variants use binary occurrence (present or absent) or normalized counts.

## 4. Document-Term Matrix (DTM)

A Document-Term Matrix (DTM) represents a collection of documents as a matrix.

- Rows represent documents.
- Columns represent vocabulary terms.
- Each cell represents a word's frequency in a document.

For our example, using the vocabulary `[I, love, NLP, Python]`:

| Document | I | love | NLP | Python |
| -------- | - | ---- | --- | ------ |
| D1       | 1 | 1    | 1   | 0      |
| D2       | 1 | 1    | 0   | 1      |
| D3       | 0 | 1    | 1   | 1      |



For example, document D1 becomes:

\\[ D_1=[1,1,1,0] \\]

Document D2 becomes:

\\[ D_2=[1,1,0,1] \\]

Important: The vocabulary order determines the position of each value. Changing that order changes how the vector is interpreted.

## 5. Sparse Vectors

A sparse vector is a vector in which most elements are zero.

Suppose a corpus contains 10,000 unique words, but one document contains only 100 of them.

- Total dimensions: 10,000
- Nonzero entries: 100
- Zero entries: 9,900

Sparsity is calculated as:

\\[ \text{Sparsity}= \frac{\text{Number of zeros}}{\text{Total elements}}\times100 \\]

\\[ =\frac{9900}{10000}\times100 =\boxed{99\\%} \\]



### Why does sparsity matter?

- Large vocabularies create high-dimensional vectors.
- Storing every zero wastes memory.
- Sparse matrix formats store nonzero values and their positions, reducing storage requirements.
- Many traditional machine learning algorithms support sparse representations.

Sparse storage reduces memory usage, but it does not eliminate the high dimensionality of the vocabulary.

## 6. Limitations of Bag of Words

### 6.1 Loss of word order

Consider these sentences:

- `Dog bites man`
- `Man bites dog`

Both contain the same words with the same frequencies. Standard unigram BoW therefore produces identical vectors, even though their meanings differ.



### 6.2 No inherent semantic understanding

Consider `cat`, `kitten` and `airplane`.

BoW assigns each word a separate feature. It does not inherently understand that cat and kitten are semantically related.

### 6.3 High dimensionality

The vector dimension equals the vocabulary size. A corpus containing 100,000 unique words can produce document vectors with 100,000 features.

### 6.4 Sparsity

Most documents contain only a small fraction of the vocabulary, resulting in many zero-valued features.

### 6.5 Unseen words

A fitted BoW vectorizer uses its learned vocabulary. A new word that is absent from that vocabulary generally does not receive its own feature unless the vocabulary is updated or the vectorizer is retrained.

### 6.6 Frequency does not equal importance

A word that appears frequently is not necessarily useful for distinguishing documents. Common words such as `the` may occur often but provide limited discriminative information.

TF-IDF addresses this limitation by weighting terms according to their frequency within a document and across the corpus.

## 7. Bag of Words vs One-Hot Encoding

| Feature          | One-hot encoding           | Bag of Words                     |
| ---------------- | -------------------------- | -------------------------------- |
| Represents       | Individual tokens          | Entire documents                 |
| Values           | One 1, remaining values 0  | Word counts or binary occurrence |
| Main purpose     | Represent token identity   | Represent document content       |
| Word order       | Not inherently represented | Ignored by standard unigram BoW  |
| Vector dimension | Vocabulary size            | Vocabulary size                  |

One-hot encoding represents an individual token, whereas BoW aggregates token occurrences into a document-level vector.

## 8. Key Takeaways 📝

1. Vocabulary: Set of unique words extracted from a corpus.
2. Word frequency: Number of times a word occurs in a document.
3. Document-Term Matrix: Rows represent documents; columns represent vocabulary terms.
4. Sparse vectors: Most entries are zero because documents use only a fraction of the vocabulary.
5. Limitations: BoW loses word order, lacks inherent semantic understanding and can create high-dimensional vectors.
6. TF-IDF: Extends count-based representations by weighting terms according to their importance across documents.
