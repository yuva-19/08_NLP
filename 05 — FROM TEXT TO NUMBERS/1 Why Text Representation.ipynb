# 🧠 Module 5.1 — Why Text Representation?

## 1. The fundamental problem

Machine-learning models operate on numerical data, but human language consists of words, sentences, punctuation and meaning.

Consider this sentence:

> "I love machine learning."

A computer can store this as a string, but a typical ML algorithm cannot directly use the raw string as its numerical input. We need to convert the text into a representation the model can process.

## 2. What exactly is a numerical representation?

A numerical representation converts text into a vector or sequence of numbers while preserving information useful for a particular NLP task.

For example, suppose our vocabulary contains only three words:

\\[ V = [\text{cat},\text{dog},\text{fish}] \\]

We could represent each word using a one-hot vector:

| Word | Numerical representation |
| ---- | ------------------------ |
| cat  | \\([1,0,0]\\)            |
| dog  | \\([0,1,0]\\)            |
| fish | \\([0,0,1]\\)            |

Each position corresponds to a fixed vocabulary word. This allows a model to receive numerical features instead of raw text.










## 3. What makes a good representation?

A useful representation should provide the information needed by the model and the task.

- Machine-readable: represents text numerically.
- Consistent: the same feature position has the same meaning across samples.
- Informative: preserves useful signals, such as word frequency or semantic relationships.
- Computationally practical: can be processed efficiently for the dataset and model.

Not every representation preserves every property of language. For example, one-hot encoding identifies words, but does not express that cat and kitten have related meanings.

1\. One-Hot Encoding

Assigns each vocabulary word a vector with one `1` and all remaining values `0`.

Useful for understanding basic categorical representations; inefficient for large vocabularies.

2\. Bag of Words (BoW)

Represents a document using word counts or word-presence indicators.

Useful for classical ML tasks such as spam detection and document classification.

[From Words to Vectors: Understanding Word2Vec in NLP](https://images.openai.com/static-rsc-4/Pm4Zyx1Md5fGR_eBp_4lNoZh2e6SJyROeN68EBG3fyGrbTzCukRDK7TLdgNoldchLdl36LC4wM2VSohrKlbUTkgknY5GUl_dQWKvz1Zwtb7AEJ0sbV3d6TzLxNqbLqQ1-Yx-0veDThw07sdZQwm4emZl9rn9OdGK_6ZH3IGcpZ8?purpose=inline)

3\. Word Embeddings

Represent words as dense, learned vectors that can capture semantic relationships.

Important for neural NLP and understanding how modern language models represent words.

These methods form a progression from simple symbolic representations toward representations that can capture richer relationships.


## What you should remember

1. Raw text must be represented numerically before conventional ML models can process it.
2. A vector contains numerical features describing a word or document.
3. The vocabulary determines the dimensions of many classical text representations.
4. Different representations preserve different kinds of information.
5. BoW and TF-IDF are classical approaches; embeddings are learned representations.
