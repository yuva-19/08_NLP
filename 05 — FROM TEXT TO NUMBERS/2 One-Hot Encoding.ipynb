# Module 5.2 — One-Hot Encoding

## 1. Definition

One-hot encoding is a text representation technique that converts each word into a binary vector containing exactly one `1` and all other values `0`.

## 2. Vocabulary

A vocabulary is the set of unique tokens in a corpus. Each token is assigned a unique index.

Example corpus:

- "I love NLP"
- "I love Python"

Vocabulary:

\\[ V=[\text{I},\text{love},\text{NLP},\text{Python}] \\]

Vocabulary size:

\\[ |V|=4 \\]

## 3. One-hot vector construction

Each word receives a vector whose dimension equals the vocabulary size.

| Word   | One-hot vector |
| ------ | -------------- |
| I      | `[1, 0, 0, 0]` |
| love   | `[0, 1, 0, 0]` |
| NLP    | `[0, 0, 1, 0]` |
| Python | `[0, 0, 0, 1]` |



For a vocabulary of size \\(V\\), the one-hot vector for word \\(i\\) is:

\\[ x_j= \begin{cases} 1 & j=i\\\ 0 & j\ne i \end{cases} \\]

## 4. Sparse representation

A vector is sparse when most of its elements are zero.

For a vocabulary containing 10,000 words, each one-hot vector contains one non-zero value and 9,999 zeros.

\\[ \text{Sparsity}= \frac{\text{Number of zeros}}{\text{Total elements}}\times100 \\]

\\[ =\frac{9999}{10000}\times100 =\boxed{99.99\%} \\]



Sparse storage can reduce memory usage by avoiding the storage of zeros. However, the vector's logical dimension still depends on vocabulary size.

## 5. Dimensionality problem

The number of dimensions grows with the vocabulary size.

| Vocabulary size | Vector dimensions |
| --------------- | ----------------- |
| 100             | 100               |
| 10,000          | 10,000            |
| 50,000          | 50,000            |
| 100,000         | 100,000           |

High-dimensional input representations can increase computational and parameter costs in downstream models.

For example, a dense linear layer with 50,000 input features and 128 output units has:

\\[ 50,000\times128=6,400,000 \\]

trainable weights, excluding biases.

## 6. Lack of semantic relationships

One-hot encoding identifies words but does not inherently represent their meanings.

Example:

\\[ \begin{aligned} \text{cat}&=[1,0,0]\\\ \text{kitten}&=[0,1,0]\\\ \text{airplane}&=[0,0,1] \end{aligned} \\]



The dot product between any two different one-hot vectors is zero:

\\[ \text{cat}\cdot\text{kitten}=0 \\]

Thus, the vectors do not inherently express that cat and kitten are more closely related in meaning than cat and airplane.

## 7. One-hot encoding vs word embeddings

| Property            | One-hot encoding       | Word embeddings           |
| ------------------- | ---------------------- | ------------------------- |
| Values              | Binary                 | Learned numerical values  |
| Dimensions          | Vocabulary-sized       | Usually lower-dimensional |
| Representation      | Sparse                 | Usually dense             |
| Semantic similarity | Not inherently encoded | Can be learned            |
| Examples            | One-hot vectors        | Word2Vec, GloVe, FastText |

Word embeddings address many limitations of one-hot representations by learning compact vector representations from data.

## 8. Key takeaways

- One-hot encoding assigns a unique binary vector to each vocabulary token.
- Vector dimension equals vocabulary size.
- One-hot vectors are sparse.
- Different words have zero dot product, regardless of semantic similarity.
- Word embeddings provide compact, learned representations that can capture semantic relationships.
