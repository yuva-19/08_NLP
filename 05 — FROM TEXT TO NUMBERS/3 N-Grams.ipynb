# Module 5.3 — N-Grams in NLP 🧠

## 1. Introduction

An N-gram is a continuous sequence of \\(N\\) items extracted from text. The items can be words or characters.

Example sentence: `I love machine learning`

N-grams help NLP models capture local patterns and word order rather than treating every word independently.

## 2. Types of N-Grams

### 2.1 Unigram (\\(N=1\\))

A unigram contains one word at a time.

For `I love machine learning`:

- `I`
- `love`
- `machine`
- `learning`

Use: Measures individual word occurrences.

Limitation: Does not directly capture relationships between consecutive words.

### 2.2 Bigram (\\(N=2\\))

A bigram contains two consecutive words.

- `I love`
- `love machine`
- `machine learning`

Use: Captures short phrases and local word order. For example, `not good` can provide information that individual unigram features do not directly capture.

### 2.3 Trigram (\\(N=3\\))

A trigram contains three consecutive words.

- `I love machine`
- `love machine learning`

Use: Captures more local context than unigrams and bigrams, but generally requires more training data to learn reliable patterns.



### 2.4 Higher-Order N-Grams

Higher-order n-grams contain four or more consecutive items.

| N-gram  | Example                          |
| ------- | -------------------------------- |
| Unigram | `machine`                        |
| Bigram  | `machine learning`               |
| Trigram | `I love machine`                 |
| 4-gram  | `I love machine learning`        |
| 5-gram  | `I love machine learning models` |

As \\(N\\) increases, n-grams capture longer sequences. However, the number of possible features grows, and many sequences may occur rarely in the training data.

Number of n-gram occurrences:

For a sequence containing \\(L\\) items, the number of n-grams of size \\(N\\), without boundary padding, is:

\\[ \text{Count}=\max(0,L-N+1) \\]

For a five-word sentence:

- Unigrams: 5
- Bigrams: 4
- Trigrams: 3
- 4-grams: 2

This counts occurrences in a single sequence, not unique n-grams across an entire dataset.

## 3. Word N-Grams

Word n-grams group consecutive words together.

Example: `Natural language processing is useful`

Word bigrams:

- `Natural language`
- `language processing`
- `processing is`
- `is useful`


Advantages

- Capture phrases and local word order.
- Help distinguish expressions such as `not good` and `very good`.
- Provide useful features for sentiment analysis and text classification.

Limitations

- Large vocabularies can produce many possible features.
- Rare phrases may not occur often enough for reliable learning.
- Similar expressions such as `very good` and `really good` may become separate features.

## 4. Character N-Grams

Character n-grams are sequences of consecutive characters rather than whole words.

Consider the word `hello`.

Character unigrams

h

e

l

l

o

Character bigrams

he

el

ll

lo

Character trigrams

hel

ell

llo


Applications

- Spelling variations: `color` and `colour` share some character patterns.
- Morphology: Related forms such as `play`, `playing` and `played`.
- Noisy text: Helps handle misspellings and informal spellings.
- Text classification: Character patterns can be useful even when exact word matches are unavailable.

Limitations

- Some character sequences have little linguistic meaning.
- Short sequences can be ambiguous.
- Character n-grams do not automatically understand the meaning of a complete word or sentence.

## 5. Word N-Grams vs Character N-Grams

| Property              | Word n-grams                            | Character n-grams                                   |
| --------------------- | --------------------------------------- | --------------------------------------------------- |
| Basic unit            | Word                                    | Character                                           |
| Bigram example        | `machine learning`                      | `ma`, `ac`, `ch`                                    |
| Captures              | Phrases and local word order            | Spelling and subword patterns                       |
| Misspelling tolerance | Usually lower                           | Often higher                                        |
| Common applications   | Sentiment analysis, text classification | Noisy text, spelling variation, text classification |

## 6. N-Grams with BoW and TF-IDF

N-grams determine which features to extract. Bag of Words (BoW) counts these features, while TF-IDF weights them according to their frequency within a document and across documents.

For `I love machine learning`:

- Unigram features: `I`, `love`, `machine`, `learning`
- Bigram features: `I love`, `love machine`, `machine learning`
- Combined features: Both unigram and bigram features.

These features can be represented using BoW or TF-IDF for classical machine learning models.

N-grams are explicit text features, not word embeddings. They capture local patterns but do not inherently encode semantic similarity.

## 7. Key Takeaways 📝

1. An n-gram is a continuous sequence of \\(N\\) words or characters.
2. Unigrams contain one item, bigrams two, and trigrams three.
3. Larger n-grams capture more context but often create sparse feature spaces.
4. Word n-grams capture phrases; character n-grams capture subword patterns.
5. N-grams can be combined with BoW and TF-IDF.
6. N-grams capture local patterns rather than complete language understanding.
