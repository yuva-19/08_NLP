# Train BPE

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

bpe_tokenizer = Tokenizer(BPE(unk_token = "[UNK]"))

bpe_tokenizer.pre_tokenizer = Whitespace()

bpe_trainer = BpeTrainer(
    vocab_size=500,
    min_frequency=1,
    special_tokens=["[UNK]"]
)

bpe_tokenizer.train_from_iterator(corpus, trainer= bpe_trainer)
```

```python
print("BPE vocabulary Size: ", bpe_tokenizer.get_vocab_size())
bpe_vocab = bpe_tokenizer.get_vocab()
bpe_vocab
```

# Train WordPiece

```python
from tokenizers.models import WordPiece
from tokenizers.trainers import WordPieceTrainer

wordpiece_tokenizer = Tokenizer(
    WordPiece(
        unk_token="[UNK]",
        continuing_subword_prefix="##"
    )
)

wordpiece_tokenizer.pre_tokenizer = Whitespace()

wordpiece_trainer = WordPieceTrainer(
    vocab_size=500,
    min_frequency=1,
    special_tokens=["[UNK]"]
)

wordpiece_tokenizer.train_from_iterator(
    corpus,
    trainer=wordpiece_trainer
)
```

```python
print("WordPiece vocabulary size:", wordpiece_tokenizer.get_vocab_size())
wordpiece_tokenizer.get_vocab()
```

# Train SentencePiece + BPE

```python
import sentencepiece as spm
```

```python
with open("corpus.txt", "w", encoding="utf-8") as f:
    for sentence in corpus:
        f.write(sentence + "\n")
```

```python
spm.SentencePieceTrainer.train(
    input="corpus.txt",
    model_prefix="sp_bpe",
    vocab_size=500,
    model_type="bpe",
    character_coverage=1.0,
    unk_id=0,
    bos_id=-1,
    eos_id=-1,
    pad_id=-1
)
```

```python
sp_bpe = spm.SentencePieceProcessor(
    model_file="sp_bpe.model"
)
```

```python
print("SentencePiece BPE vocabulary size:", sp_bpe.get_piece_size())
```

# Train SentencePiece + Unigram

```python
spm.SentencePieceTrainer.train(
    input="corpus.txt",
    model_prefix="sp_unigram",
    vocab_size=500,
    model_type="unigram",
    character_coverage=1.0,
    unk_id=0,
    bos_id=-1,
    eos_id=-1,
    pad_id=-1
)
```

```python
sp_unigram = spm.SentencePieceProcessor(
    model_file="sp_unigram.model"
)
```

```python
print(
    "SentencePiece Unigram vocabulary size:",
    sp_unigram.get_piece_size()
)
```

```python
# Tokenization Efficiency

test_text = "The players discovered xylophone"

# Remove spaces because we want to measure actual text characters
num_chars = len(test_text.replace(" ", ""))

# Number of words
num_words = len(test_text.split())

# Token counts from our tokenizer experiment
token_counts = {
    "BPE": 5,
    "WordPiece": 5,
    "SentencePiece BPE": 5,
    "SentencePiece Unigram": 5
}

print("Total characters:", num_chars)
print("Total words:", num_words)
print()

for tokenizer_name, token_count in token_counts.items():

    chars_per_token = num_chars / token_count
    tokens_per_word = token_count / num_words

    print(tokenizer_name)
    print(f"  Token count:       {token_count}")
    print(f"  Characters/token:  {chars_per_token:.2f}")
    print(f"  Tokens/word:       {tokens_per_word:.2f}")
    print()
```
