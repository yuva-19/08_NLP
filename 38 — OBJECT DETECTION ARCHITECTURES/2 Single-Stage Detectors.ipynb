In the previous section, we studied two-stage detectors:

Their approach was:

1. Generate region proposals.

## 1. What are Single-Stage Detectors?

A single-stage detector predicts object classes and bounding boxes directly from image features in one detection pipeline.

It does not use a separate region-proposal network followed by a second-stage detection head.

![Single-stage object detection framework](https://images.openai.com/static-rsc-4/uywwNxc-1iSOT1d3fX36_L3dr0EUSdwZYpyEdC3DPpkNcqGiz_VOvxMbgHFSrt1BDM4NhL80Dd6V4WllfwzGnuFnopuHleAnnnGUc47iIaOF0A9kwi1ieDYXNwkdcHw8t90fyR6YhWjoPldhddZ3SFa2hyEy0OzwMKJTpi7oilY?purpose=inline)

[Source: MDPI — The YOLO Framework: A Comprehensive Review](https://www.mdpi.com/2073-431X/13/12/336)

### Basic architecture

```text
           Input Image
                |
                ▼
           CNN Backbone
                |
                ▼
         Feature Extraction
                |
                ▼
       Detection Prediction
          /          \
         ▼            ▼
   Classification  Bounding Box
                    Regression
         \            /
          ▼          ▼
         Final Detections
```

The model directly predicts:

- Object class
- Bounding box coordinates
- Confidence scores

Unlike Faster R-CNN, it doesn't first generate a separate set of region proposals for a second-stage classifier.

### Two-stage vs single-stage

**Two-stage detector**

![Two-stage object detection](https://images.openai.com/static-rsc-4/MV3_T2hbUKeTN0G-ZHkLcJekFYzviWbwJS6hvwdRVqhygAS4a1A7fKpkGHkReGVA84PsjMHHkJZWP5E3G3rkCfivkLbiGxqTTfpG5ALMhg0SFSdJkX7mVoWgMHv5nOgajIKhVHEuVlKDYXYS4VUX_D2f2UPGj9hKQGSso7gyTKA?purpose=inline)

[Source: Medium — Deep Learning Object Detection Revolution](https://krrishrebba.medium.com/part-6-deep-learning-object-detection-revolution-2014-2025-349425f9616b)

```text
Image → Feature Extraction
      → Region Proposals
      → Detection Head
      → Final Detections
```

Separate proposal and detection stages.

**Single-stage detector**

![Single-stage object detection](https://images.openai.com/static-rsc-4/Abc5Sz3QqqxWT06WH5LCqtHHBnWWYGjwHaHMMMxNxN2oCXriXWGiD9PEqlsIaNZHi1rkQ_iTYxXyeuEE2dv-1UnrRd05Lx3DtlJ6A5O5F1veFvS2SkzbwJMd7io714U1Uij6X32pt0TiQrjatdlVH4wv1VXmXCf1Db6Jx1QFh3E?purpose=inline)

[Source: Springer Nature — Floating Waste Detection Using Deep Learning](https://link.springer.com/article/10.1007/s00521-026-12051-w)

```text
Image → Feature Extraction
      → Direct Predictions
      → Final Detections
```

Direct object predictions without a separate proposal stage.

### Why are single-stage detectors important?

They are generally designed for faster inference, making them useful for real-time applications such as:

- Autonomous driving
- CCTV surveillance
- Robotics
- Industrial inspection
- Traffic monitoring

## 2. SSD — Single Shot MultiBox Detector

<badge color="success">Must Know</badge>

![SSD architecture](https://images.openai.com/static-rsc-4/vFWdA1rcPk7yx-_o9L72llA2otbPA_800OxJjXIvtsFPlu6ZxoIa7koYgRXj2TusLTWyhh74KBKAX5EHhWeIxB7vYm0g06eupn28u2vG2_9pTjm22J1Rwanil2RZRP4JwsNYkvX4gIG_4cVNZkFxsCqBzvw87YjvoAaX0DDdC8E?purpose=inline)

[Source: SSD — Single Shot Detector](https://woochan-autobiography.tistory.com/924)

SSD stands for **Single Shot MultiBox Detector**.

It was introduced in 2016 by Wei Liu and colleagues.

Its main idea:

**Detect objects of different sizes by making predictions from multiple feature maps at different resolutions.**

### 2.1 Why does SSD use multiple feature maps?

Imagine an image containing:

- A small bird on a tree.
- A medium-sized car.
- A large bus.

Can one feature map detect all these objects equally well?

It can be challenging.

- High-resolution feature maps preserve more spatial details, which helps detect small objects.
- Lower-resolution feature maps contain more high-level semantic information, which helps detect larger objects.

SSD uses multiple feature maps to detect objects at different scales.

![SSD multiple feature maps](https://images.openai.com/static-rsc-4/0IUiqdzolGxK4ce7RHe9yzIVOLEwF2czv0pz80PFuW-xIFjisEpWogMAPZ6hyHlAk7MVklwgaskLHFFrxCvL07rya_czX42SzIEgwEpYBThuZMyl0ib7nzggjTPQdK7nSO74xXwn4dz4fMpj4m4KGtbOhAgOmEzFfkTadMoAjKQ?purpose=inline)

[Source: SSD, RetinaNet, DETR, CornerNet and CenterNet](https://fayej1221.github.io/blog/dl/2025-04-30-object-detection-ssd-retinanet-detr-cornernet-centernet/)

### 2.2 SSD architecture

```text
              Input Image
                   |
                   ▼
               CNN Backbone
                   |
                   ▼
            Initial Feature Maps
                   |
           ┌───────┼────────┐
           ▼       ▼        ▼
        Feature  Feature  Feature
         Map 1    Map 2    Map 3
           |       |        |
           ▼       ▼        ▼
         Small   Medium   Large
        Objects  Objects  Objects
           |       |        |
           └───────┼────────┘
                   ▼
          Classification +
          Bounding Box Prediction
                   |
                   ▼
              NMS Filtering
                   |
                   ▼
             Final Detections
```

### 2.3 Default Boxes (Anchor Boxes)

SSD uses predefined reference boxes called **default boxes**.

These boxes have different:

- Scales
- Aspect ratios

For example:

![SSD default boxes](https://images.openai.com/static-rsc-4/7p_Z_dBDd0fKWCWgzH6uCKI1DLBVxit9cLT0QlRqNevaQ9VcTjSutPWMqsW-PEZn2THVMPPxn1KOZ1YKDPxw-mM5ZAFBn1_bKj7shHuviiVLMnuxsq6_QD2kurSz-BIeC0bL-P78AqQpnnlsVMiZawBqDtGmbkEeValqwkO5uF8?purpose=inline)

[Source: Understanding Single Shot Object Detection](https://manalelaidouni.github.io/Single%20shot%20object%20detection.html)

At different positions in the feature maps, SSD predicts:

1. Class probabilities for each default box.
2. Bounding-box coordinate adjustments.

The model learns which default boxes correspond to actual objects and how to adjust them.

### 2.4 How SSD makes predictions

Suppose the model detects a car.

A default box is initially placed at a particular feature-map location.

The network predicts:

```text
Default Box
     |
     ▼
Class Prediction
     |
     └── Car: 0.92

Bounding Box Adjustment
     |
     └── Refined coordinates
```

The adjusted box becomes a candidate detection.

After prediction, confidence filtering and NMS help produce the final detections.

### Key characteristics of SSD

| Feature | SSD |
|---|---|
| Architecture | Single-stage |
| Backbone | CNN |
| Detection | Multiple feature maps |
| Reference boxes | Default boxes |
| Object scales | Multiple scales |
| Prediction | Class + box regression |
| Post-processing | NMS |

**Remember:** SSD uses multiple feature maps to detect objects at different scales in a single detection pipeline.

## 3. RetinaNet

<badge color="info">Must Know Conceptually</badge>

![RetinaNet object detection](https://images.openai.com/static-rsc-4/APGVnBBNiElGmND_0bKGKlGFfkBBLLaS9iKmUznlChOV6nY2xRmImNKDYy-LIWpBrqNpRWTXn2hxgpojL56PUfDZMG41fdwHNQ-_SYqd0Io2tYmaAc6S5kRxz7KeGWevLnCkU-Wf2kvqhfq6kdbvq3IvUHbAbi2q1ekFxOo5TSY?purpose=inline)

[Source: Lil'Log — Object Detection Part 4: Fast Detection Models](https://lilianweng.github.io/posts/2018-12-27-object-recognition-part-4/)

RetinaNet was introduced in 2017 by Tsung-Yi Lin and colleagues.

It addresses an important problem encountered in single-stage object detection:

**Class imbalance.**

### 3.1 What is the class imbalance problem?

Imagine an image containing one car.

The detector examines thousands of possible locations or anchor boxes.

Most of them correspond to background rather than actual objects.

For example:

| Prediction type | Number of examples |
|---|---:|
| Background | 9,900 |
| Actual object | 100 |

The training data contains far more background examples than positive object examples.

This creates a class imbalance problem.

### Why is this a problem?

During training:

- Easy background examples can dominate the loss.
- The model spends considerable effort learning examples it already classifies correctly.
- Harder object examples may receive relatively less attention.

RetinaNet addresses this using **Focal Loss**.

### 3.2 Focal Loss

Focal Loss reduces the contribution of well-classified examples and gives relatively more importance to difficult examples.

Its commonly used formulation is:

$$
FL(p_t)=-\alpha_t(1-p_t)^\gamma\log(p_t)
$$

Where:

- \\(p_t\\) = Model probability assigned to the correct class.
- \\(\alpha_t\\) = Class-balancing factor.
- \\(\gamma\\) = Focusing parameter.

Let's understand intuitively.

**Easy example:**

The model predicts an object correctly with high confidence.

$$
p_t = 0.95
$$

The factor \\((1-p_t)^\gamma\\) becomes small, reducing its loss contribution.

**Hard example:**

The model struggles to classify an example correctly.

$$
p_t = 0.30
$$

The factor becomes larger, giving the example more influence during training.

![Focal Loss illustration](https://images.openai.com/static-rsc-4/jbbwvXYKFAB5a_lNMz0T7kuHdtKhN7m6Fao7KB4Y1-q0vDPBymgHuwRby12VculJWsHRbie_L_7XpAFUXVLwENKZsK7v6Ybf0z38E4lR_3TGxnod5SfriVajcyiCAwwwWZlm-k6IZkgCTvG9NQyKlLD6c4HTcbn0jkcwsYwYtcc?purpose=inline)

[Source: Lil'Log — Object Detection Part 4](https://lilianweng.github.io/posts/2018-12-27-object-recognition-part-4/)

**Important:** Focal Loss does not simply discard easy examples. It down-weights their contribution to the training loss.

### 3.3 RetinaNet architecture

RetinaNet combines three important components:

1. Backbone — extracts image features.
2. Feature Pyramid Network (FPN) — combines information at different spatial scales.
3. Detection subnets — predict object classes and bounding-box coordinates.

![RetinaNet architecture with FPN](https://images.openai.com/static-rsc-4/Nma9u2Nf7XUqIS-JNV7kud2lKU_fPaenyx848-513by5FqVaIIWHOjlMgjTwiMKedowpelPmapYLTUyU3nonqezQlmBg_2UI6Rp78xUNIPI3zBJqD-gBRkDMJ8PeMIIe03LPQpRUytbLAe_RfnF7BbiOyIi9O2cY5a6gq8Jbo8I?purpose=inline)

[Source: Jonathan Hui — SSD, YOLO, FPN and Focal Loss](https://jonathan-hui.medium.com/what-do-we-learn-from-single-shot-object-detectors-ssd-yolo-fpn-focal-loss-3888677c5f4d)

Its architecture:

```text
             Input Image
                  |
                  ▼
              CNN Backbone
                  |
                  ▼
         Feature Pyramid Network
                  |
            ┌─────┼─────┐
            ▼     ▼     ▼
           P3    P4    P5
            |     |     |
            └─────┼─────┘
                  |
           ┌──────┴───────┐
           ▼              ▼
    Classification    Box Regression
       Subnet             Subnet
           |              |
           └──────┬───────┘
                  ▼
             Final Predictions
                  |
                  ▼
                 NMS
```

The classification subnet predicts object categories.

The box regression subnet predicts bounding-box adjustments.

### 3.4 Why is RetinaNet important?

Before RetinaNet, two-stage detectors often had an accuracy advantage over single-stage detectors.

RetinaNet demonstrated that a carefully designed single-stage detector using FPN and Focal Loss could achieve strong detection accuracy while maintaining the efficiency of a one-stage approach.

**Remember:** RetinaNet is particularly important for understanding how Focal Loss addresses foreground-background class imbalance in dense object detection.

## 4. YOLO Family — You Only Look Once

<badge color="success">Very Important</badge>

![YOLO object detection](https://images.openai.com/static-rsc-4/b2UyEzVZkWWBSqezq7h9D_vW3tVDAqDKMD8lTAmAWTR24pZrD7JIJy_JDz1Hd_g5VN8LTNFbYagkXI24jVohhaebTg04MornUJYdCXuwWIA7LLHSbP12oxW8m9SmQt3K5O1mamG94eg0kChJyMMszT6FYACw1lHsUNgXhQ5hOm8?purpose=inline)

[Source: The Best Object Detection Methods for 2023](https://www.augmentedstartups.com/blog/the-best-object-detection-methods-for-2023-a-comprehensive-guide)

YOLO stands for **You Only Look Once**.

It is one of the most influential families of real-time object detection models.

Its central idea:

**Predict object classes and bounding boxes directly from image features in a single detection pipeline.**

Unlike traditional R-CNN approaches, YOLO does not require a separate region-proposal stage followed by a second-stage classifier.

### 4.1 Original YOLO concept

The original YOLO model, introduced in 2016, divided an input image into a grid.

For example, imagine a 7 × 7 grid:

![Original YOLO grid-based prediction](https://images.openai.com/static-rsc-4/zGA7_fXz3fppE6uon7uExS18e7QDSkn8lMtGJAOtC23SjBHUOXQ4Bnrop52ErJ-hJylca0TtmuHFZWfGrpH-K3jmyLRHef4YaIec0jxgqgb93OOuDHalLP5TUaIGZy4BoJbgJevnFUuhpbIkv7hrquT9Sc2S_GvLDqX2nmw06D8?purpose=inline)

[Source: You Only Look Once (YOLO)](https://chih-sheng-huang821.medium.com/%E6%B7%B1%E5%BA%A6%E5%AD%B8%E7%BF%92-%E7%89%A9%E4%BB%B6%E5%81%B5%E6%B8%AC-you-only-look-once-yolo-4fb9cf49453c)

Each grid cell was responsible for predicting objects whose centers fell within that cell.

The original YOLO formulation predicted:

- Bounding box coordinates.
- Bounding box confidence.
- Class probabilities.

The network generated predictions for the entire image in one forward pass.

### 4.2 Original YOLO architecture

```text
             Input Image
                  |
                  ▼
              CNN Backbone
                  |
                  ▼
           Feature Extraction
                  |
                  ▼
           Grid-based Predictions
                  |
           ┌──────┴───────┐
           ▼              ▼
     Bounding Boxes   Class Probabilities
           |              |
           └──────┬───────┘
                  ▼
          Confidence Filtering
                  |
                  ▼
                 NMS
                  |
                  ▼
            Final Detections
```

### 4.3 How does YOLO predict objects?

Suppose an image contains a car and a pedestrian.

The network predicts candidate bounding boxes and their associated confidence and class information.

Example:

```text
Prediction 1:
Class      → Car
Confidence → 0.95
Box        → (100, 120, 300, 280)

Prediction 2:
Class      → Pedestrian
Confidence → 0.91
Box        → (350, 100, 420, 300)
```

The model processes the image as a whole rather than separately extracting CNN features for thousands of proposed regions.

This makes YOLO particularly suitable for real-time detection applications.

### 4.4 Evolution of the YOLO family

YOLO is not just one architecture. It is a family of evolving detection models.

![YOLO family evolution](https://images.openai.com/static-rsc-4/5rohxsU5Og2IvDu2d5MmWAQEN5Yl4-umiD6KhAWUdF5H7yGIPk1_8hxv6gtK0ysoq-DFZpxyngePVpc_Ou4U6EYnZ4MaYcUDWBoJDxFRlSSH6HmvX2I5W3vldTAUB0XNkiU4wX333F269MCxQE-3a_f-nY_8cZl56vPaWux7rf0?purpose=inline)

[Source: YOLO Evolution Overview](https://mdnice.com/writing/8130a10a584a4ea485201e2acd80c4be)

Here is a conceptual overview of important generations:

| Generation | Important development |
|---|---|
| YOLOv1 | Original grid-based, single-pass detection |
| YOLOv2 / YOLO9000 | Anchor boxes, improved training and detection |
| YOLOv3 | Multi-scale predictions and improved feature extraction |
| YOLOv4 | Architectural and training improvements |
| YOLOv5 | Popular PyTorch-based implementation and ecosystem |
| YOLOv6 | Industrial and deployment-oriented optimizations |
| YOLOv7 | Trainable bag-of-freebies and architectural improvements |
| YOLOv8 | Anchor-free detection design and updated task framework |
| YOLOv9 | Programmable gradient information and GELAN |
| YOLOv10 | End-to-end detection design with consistent dual assignments and NMS-free inference |

This is a selected conceptual timeline, not an exhaustive list. YOLO implementations come from different research groups and maintainers; version numbers do not represent one universally maintained sequence.

### 4.5 What changed across YOLO generations?

The broad improvements include:

- Better feature extraction.
- Multi-scale object detection.
- Improved bounding-box prediction.
- Better training strategies.
- Faster inference.
- Improved detection of small objects.
- Different anchor-based and anchor-free prediction approaches.
- In some newer architectures, end-to-end prediction without traditional NMS.

**Important:** Not every YOLO generation uses the same architecture or detection mechanism. Original YOLO, later anchor-based YOLO models and newer anchor-free models have important differences.

## 6. Two-Stage vs Single-Stage: Final Comparison

This is a frequently asked computer vision interview question.

| Aspect | Two-stage | Single-stage |
|---|---|---|
| Detection process | Proposal generation + detection | Direct detection |
| Region proposal stage | Separate stage | No separate proposal stage |
| Typical examples | Faster R-CNN, Mask R-CNN | SSD, RetinaNet, YOLO |
| Inference speed | Generally slower | Generally faster |
| Accuracy | Often strong localization performance | Modern models can also achieve excellent accuracy |
| Applications | Detailed detection tasks | Real-time detection applications |

**Important:** Single-stage does not automatically mean less accurate. Modern single-stage detectors can achieve excellent accuracy, and the best architecture depends on the dataset, task, hardware and deployment requirements.
