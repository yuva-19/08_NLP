# Module 7.2 — Inverse Document Frequency (IDF) 🧠

## 1. What is IDF?

Inverse Document Frequency (IDF) measures how rare or distinctive a word is across an entire collection of documents (called a corpus).

Remember Module 7.1, Term Frequency (TF)?

- TF: How frequently does a word appear in one document?
- IDF: In how many documents does that word appear across the corpus, and how informative is it for distinguishing documents?

For example, the word `the` may appear in almost every document, whereas `transformer` may appear in only a few. IDF assigns a lower weight to widespread words and a higher weight to rarer words.

## 2. IDF formula

The basic textbook formula is:

\\[ \boxed{IDF(t)=\log\left(\frac{N}{DF(t)}\right)} \\]

Where:

- \\(N\\) = total number of documents in the corpus.
- \\(DF(t)\\) = number of documents containing the term \\(t\\), regardless of how many times it occurs in each document.
- \\(\log\\) = logarithm, commonly the natural logarithm \\(\ln\\).

## 3. Visual intuition 🖼️


Notice the relationship: as the number of documents containing a word increases, its IDF decreases.

## 4. Worked example

Suppose we have these five documents:

| Document | Text                 |
| -------- | -------------------- |
| D1       | `I love NLP`         |
| D2       | `I love Python`      |
| D3       | `NLP is useful`      |
| D4       | `Python is powerful` |
| D5       | `I love AI`          |

Let's calculate IDF for three words.

### Word A: `I`

Appears in D1, D2 and D5.

\\[ N=5,\quad DF(I)=3 \\]

\\[ IDF(I)=\ln(5/3)\approx0.511 \\]

### Word B: `NLP`

Appears in D1 and D3.

\\[ IDF(NLP)=\ln(5/2)\approx0.916 \\]

### Word C: `powerful`

Appears only in D4.

\\[ IDF(powerful)=\ln(5/1)\approx1.609 \\]

Compare the IDF scores

I

### 0.511

Appears in 3 documents

NLP

### 0.916

Appears in 2 documents

powerful

### 1.609

Appears in 1 document

Key insight: `powerful` has the highest IDF because it occurs in the fewest documents. This does not automatically mean it is the most important word for every task; IDF only measures how distinctive its document distribution is.

## 5. Important details

- Term frequency counts occurrences. Document frequency counts documents containing the word.
- If a word appears 20 times in one document, it contributes only 1 to that document's document frequency.
- If a term appears in every document, its IDF is zero under the basic formula.
- In practice, libraries such as scikit-learn commonly use smoothed IDF:

\\[ IDF(t)=\ln\left(\frac{1+N}{1+DF(t)}\right)+1 \\]

This avoids zero-valued IDF scores for words present in every document and provides more robust weighting.

## What you should remember

1. IDF measures how widely a word is distributed across documents.
2. \\(DF\\) counts documents, not total word occurrences.
3. Rare words receive higher IDF; common words receive lower IDF.
4. IDF is calculated across the corpus, whereas TF is calculated within an individual document.
5. TF and IDF combine to produce TF-IDF.
