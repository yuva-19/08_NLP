# Module 7.3 — TF-IDF (Term Frequency–Inverse Document Frequency) 🧠

## 1. Why combine TF and IDF?

In the previous two sections, you learned:

- TF: How frequently a word appears within a document.
- IDF: How distinctive that word is across the entire collection of documents.

TF-IDF combines both to estimate how useful a word is for distinguishing a document from others.

\\[ \boxed{TF\text{-}IDF(t,d)=TF(t,d)\times IDF(t)} \\]

The intuition is simple:

- A word that appears frequently in one document can be useful.
- A word appearing in nearly every document is less useful for distinguishing documents.
- A word that appears frequently in one document but rarely elsewhere can receive a high score.

## 2. Complete numerical example

Consider these three documents:

| Document | Text                |
| -------- | ------------------- |
| D1       | `NLP is useful NLP` |
| D2       | `NLP is powerful`   |
| D3       | `Python is useful`  |

We'll calculate the TF-IDF score for `NLP` in D1.

### Step 1 — Calculate TF

D1 contains 4 tokens, and `NLP` occurs twice.

\\[ TF(NLP,D1)=\frac{2}{4}=0.5 \\]

### Step 2 — Calculate IDF

There are 3 documents, and `NLP` appears in D1 and D2.

\\[ IDF(NLP)=\ln\left(\frac{3}{2}\right) \\]

\\[ IDF(NLP)\approx0.405 \\]

### Step 3 — Calculate TF-IDF

\\[ TF\text{-}IDF(NLP,D1)=0.5\times0.405 \\]

\\[ \boxed{TF\text{-}IDF\approx0.203} \\]

This is the unnormalized TF-IDF score using the basic formulas.

## 3. Why do frequent words receive lower importance?

Imagine a corpus containing 1,000 documents.


Word: `the`

Appears in 950 documents.

\\[ IDF=\ln(1000/950)\approx0.051 \\]

Its low IDF means it contributes relatively little to distinguishing documents.


Word: `blockchain`

Appears in 10 documents.

\\[ IDF=\ln(1000/10)\approx4.605 \\]

Its higher IDF means it may be more useful for identifying documents about blockchain.

Important: Frequent words are not automatically unimportant. A word can have high TF but low TF-IDF if it appears across most documents. A rare word can have high IDF but low TF-IDF if it barely appears in a particular document.

## 4. Document length effects

Suppose we have:

- D1: `NLP NLP useful`
- D2: `NLP NLP NLP NLP useful useful`

For `NLP`:

\\[ TF(D1)=\frac{2}{3}=0.667 \\]

\\[ TF(D2)=\frac{4}{6}=0.667 \\]

Both documents receive the same normalized TF for this word. Dividing by document length helps prevent longer documents from receiving larger TF values simply because they contain more tokens.

However, TF normalization alone does not eliminate every document-length effect. In practice, scikit-learn's `TfidfVectorizer` also applies L2 normalization by default, scaling each nonzero document vector to unit length. This helps compare documents with different vector magnitudes.

## 5. What is sparse representation?

Suppose your corpus contains 10,000 unique words. Each document gets a vector with 10,000 features, one for each vocabulary word.

But a short document might contain only 20 of those words. Almost all its vector entries will be zero.



For example:

| Document | NLP | Python | useful | powerful | database |
| -------- | --- | ------ | ------ | -------- | -------- |
| D1       | 0.8 | 0      | 0.6    | 0        | 0        |
| D2       | 0   | 0.7    | 0      | 0.7      | 0        |
| D3       | 0   | 0      | 0.5    | 0        | 0.9      |

The values are illustrative TF-IDF weights.

Sparse representation means most entries are zero. Sparse matrices store mainly the nonzero values and their positions, reducing memory usage for text data.


## 7. Must Know vs Good to Know

Must Know

TF-IDF formula, TF vs IDF, why common words are downweighted, document-length normalization, and sparse matrices.

Good to Know

Smoothed IDF, L2 normalization, and why TF-IDF is useful for search and classical text classification.

## What you should remember

1. TF-IDF combines local word frequency with corpus-wide word rarity.
2. Common words tend to receive lower weights because they distinguish documents less effectively.
3. Rare words can receive higher weights, but rarity alone doesn't guarantee usefulness.
4. Document length can affect raw counts; normalization helps make document vectors more comparable.
5. Text representations are usually sparse because each document contains only a small fraction of the full vocabulary.
