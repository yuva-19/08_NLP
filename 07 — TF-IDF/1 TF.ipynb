# Module 7.1 — Term Frequency (TF) 🧠

## 1. What is Term Frequency?

Term Frequency (TF) measures how frequently a particular word appears in a document.

Remember your Module 6, Bag of Words (BoW)? You counted how many times each word appeared. TF takes this a step further by measuring a word's frequency relative to the document's total number of words.

Intuition: If a word appears repeatedly in a document, it may be important to that document.

## 2. Formula

The simplest normalized Term Frequency formula is:

\\[ \boxed{TF(t,d)=\frac{f\_{t,d}}{N_d}} \\]

Where:

- \\(t\\) = the target term (word).
- \\(d\\) = the document.
- \\(f\_{t,d}\\) = number of times the term occurs in the document.
- \\(N_d\\) = total number of tokens in the document.

Some implementations use raw counts or other TF variants. Here, we'll use normalized TF.

## 3. Worked example

Consider this document:

> `NLP is useful and NLP is powerful`

There are 7 tokens.

| Word     | Count | Normalized TF   |
| -------- | ----- | --------------- |
| NLP      | 2     | \\(2/7=0.286\\) |
| is       | 2     | \\(2/7=0.286\\) |
| useful   | 1     | \\(1/7=0.143\\) |
| and      | 1     | \\(1/7=0.143\\) |
| powerful | 1     | \\(1/7=0.143\\) |

Notice that `NLP` appears twice, so its TF is twice that of `useful`.



## 4. Why divide by total words?

Compare these documents:

- Document A: `NLP NLP useful` — 3 words
- Document B: `NLP NLP NLP NLP useful useful` — 6 words

For the word `NLP`:

\\[ TF_A=\frac{2}{3}=0.667 \\]

\\[ TF_B=\frac{4}{6}=0.667 \\]

Both documents have the same normalized TF for `NLP`, even though Document B contains more occurrences. This normalization helps account for document length.

## 5. Important distinction

- Raw TF: number of occurrences of a word.
- Normalized TF: number of occurrences divided by total tokens.
- IDF: how widely the word occurs across documents.
- TF-IDF: TF multiplied by IDF.

TF alone cannot determine whether a word is distinctive across the corpus. That's why we combine it with IDF in the next part.&#x20;

### What you should remember

1. TF measures word frequency within a single document.
2. Normalized TF is between 0 and 1 for a nonempty document.
3. Repeated words have higher TF when comparing words in the same document.
4. Normalization helps compare documents of different lengths.
5. TF does not account for how common a word is across the corpus.
