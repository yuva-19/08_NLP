# Module 7.4 — Limitations of TF-IDF 🧠

TF-IDF is a powerful classical NLP technique, but it represents text mainly through weighted word frequencies, not a deep understanding of language.

Let's understand its five major limitations and why modern NLP moved toward word embeddings and contextual Transformer models.

## 1. No deep semantic understanding

TF-IDF treats words as separate vocabulary features. It does not inherently understand that two different words can have similar meanings.

Consider:

- Sentence A: `The movie was excellent`
- Sentence B: `The movie was fantastic`

A human recognizes that excellent and fantastic have similar meanings. TF-IDF assigns them separate features.

Why it matters: Two sentences can express the same idea but have low lexical overlap.

How embeddings help: Word embeddings learn numerical representations in which words used in similar contexts can have similar vectors.

## 2. No word order beyond n-grams

TF-IDF with unigram features ignores the order in which words appear.

Compare:

- `Dog bites man`
- `Man bites dog`

With unigram features, both sentences contain the same words with the same counts. Their TF-IDF vectors will therefore be identical, assuming the same vocabulary and weighting.

Adding bigrams or trigrams can capture some local word order, but it does not provide a complete understanding of sentence structure.

How modern NLP helps: Sequence models and Transformers process token relationships and contextual information.

## 3. Sparse vectors

Suppose your vocabulary contains 20,000 unique words. Each document is represented by a vector with 20,000 dimensions, but a short document uses only a tiny fraction of those words.


For example:

| Representation  | Example                          |
| --------------- | -------------------------------- |
| TF-IDF          | `[0.8, 0, 0, 0.4, 0, 0, ...]`    |
| Dense embedding | `[0.12, -0.31, 0.67, 0.24, ...]` |

- TF-IDF: Usually high-dimensional and sparse.
- Embeddings: Usually lower-dimensional and dense.

Sparse matrices can be stored efficiently, so sparsity isn't always a disadvantage. However, TF-IDF does not inherently learn semantic relationships between features.

## 4. The same word generally receives the same representation

TF-IDF does not dynamically change a word's meaning according to its surrounding context.

Consider the word `bank`:

- `I deposited money at the bank.`
- `We sat on the river bank.`

TF-IDF assigns `bank` the same IDF weight in both documents. Its overall TF-IDF value can differ because the word frequency and document normalization differ, but the representation does not distinguish the two meanings of the word itself.

How contextual embeddings help: Models such as BERT generate representations that depend on the surrounding tokens, helping distinguish the financial meaning from the geographical meaning.

## 5. Weak contextual understanding

TF-IDF relies on word-frequency statistics rather than learning how words interact to express meaning.

Consider:

- `I like this product.`
- `I do not like this product.`

The second sentence contains a negation that reverses the sentiment. Unigram TF-IDF may capture the word `not`, but it does not inherently understand that `not like` expresses the opposite sentiment of `like`.

Bigrams can capture the phrase `not like`, while contextual models can learn more complex relationships between words.


## 6. Summary: TF-IDF vs embeddings

| Limitation           | TF-IDF                                | Embeddings / Transformers                          |
| -------------------- | ------------------------------------- | -------------------------------------------------- |
| Semantic similarity  | No inherent semantic understanding    | Can learn semantic relationships                   |
| Word order           | Only when represented through n-grams | Can model token relationships                      |
| Representation       | Usually sparse                        | Usually dense                                      |
| Word meaning         | No contextual word representation     | Contextual models adapt representations to context |
| Negation and context | Limited, feature-dependent            | Better ability to learn contextual patterns        |

One important distinction: static embeddings such as Word2Vec also assign one vector per word. Context-dependent representations are a feature of contextual models such as BERT.

## What you should remember 🎯

1. TF-IDF represents weighted word importance, not deep meaning.
2. Unigram TF-IDF cannot distinguish sentences with identical word counts but different word orders.
3. Sparse vectors may have thousands of mostly zero features.
4. TF-IDF cannot assign different contextual meanings to the same word.
5. N-grams address some limitations, while contextual embeddings address others.
