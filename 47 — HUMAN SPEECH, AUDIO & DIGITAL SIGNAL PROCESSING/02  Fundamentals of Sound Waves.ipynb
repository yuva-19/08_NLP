# **MODULE 2 — FUNDAMENTALS OF SOUND WAVES 🎧**

Before we start processing speech using Python, CNNs, spectrograms, or deep learning models, we need to understand the physics behind sound.

A speech signal is not just a sequence of numbers. It originates as a physical vibration, travels through a medium, and eventually reaches a microphone, where it can be converted into an electrical and digital signal.

In this module, we'll connect **sound-wave physics → human perception → digital audio processing**.

## **2.1 Sound Wave Formation**

**Sound is a mechanical disturbance produced by a vibrating object and transmitted through an elastic medium.**

Examples:

- Vocal cords vibrating during speech.
- A guitar string vibrating after being plucked.
- A speaker diaphragm vibrating while playing music.
- A tuning fork vibrating after being struck.

Consider what happens when you speak:

1. Air from your lungs passes through your vocal tract.
2. Your vocal folds vibrate for voiced sounds.
3. This vibration creates pressure variations in the surrounding air.
4. These pressure variations travel through the air.
5. Your ears detect the pressure variations as sound.

**Important:** Sound does not normally transport air molecules from the speaker to the listener. The particles oscillate around their equilibrium positions, transferring energy through the medium.

## **2.2 Mechanical Waves**

A mechanical wave is a disturbance that requires a material medium to propagate.

Sound is a mechanical wave because it needs a medium such as:

- Air
- Water
- Steel
- Wood

Sound cannot propagate through an ideal vacuum because there are no particles to transmit the mechanical disturbance.

There are two fundamental categories of mechanical waves:

**Longitudinal waves**

Particle vibration is parallel to the direction of wave propagation.

![Transverse waves](https://images.openai.com/static-rsc-4/BSWPHjfdOiHTd-nPBxdeYhWdvsBJvhmLjMNmhivnpv2QLY_FuBrcfaRXJ6PWmJYR8aK-zeuWDpSHeNxdVjb0TVssgTTa3QPHh3a59mmgaJtC8V0FBknWSl9f5e9LzBoShD5CLeruozqja6yvS6t3kLbpEKXpYqbprZKNHPAL7og?purpose=inline)

**Transverse waves**

Particle vibration is perpendicular to the direction of wave propagation.

In air, ordinary sound waves are longitudinal waves.

## **2.3 Longitudinal Waves**

In a longitudinal sound wave, air particles move back and forth in the same direction as the wave travels.

For example, imagine pushing and pulling a spring.

- Push the spring coils together.
- Release them.
- A region of compression travels along the spring.
- A region of lower density follows.

![Compression and rarefaction in sound waves](https://images.openai.com/static-rsc-4/ODGf1tOAP3dYguJdJI96JLDuFT_uNHcSeVa3vK3V5E3Y8Sjw3-8cV112bH9pRaUsOOODpyCmucuEPbklCJIpHu57pZgujVG8SUUbmCebmAp10Wid6tR1TBERYAZioWaxXr5B3CDcKHUYabiJIhmr3uvYwApqc8meVabZXsxIjWE?purpose=inline)

[Source: MasteryMind](https://masterymind.co.uk/study-guides/physics-ocr-gcse-sound-waves?utm_source=chatgpt.com)

### **Compression**

A compression is a region where:

- Air particles are closer together.
- Pressure is higher than the equilibrium pressure.
- Density is higher.

### **Rarefaction**

A rarefaction is a region where:

- Air particles are farther apart.
- Pressure is lower than the equilibrium pressure.
- Density is lower.

**Remember:** Sound propagates through alternating compressions and rarefactions.

### **A. Amplitude**

Amplitude is the maximum displacement of a wave from its equilibrium position in a displacement-wave representation.

In a sound-pressure representation, amplitude describes the maximum pressure variation from the equilibrium pressure.

- Higher amplitude generally corresponds to louder sound.
- Lower amplitude generally corresponds to quieter sound.

### **B. Frequency**

Frequency is the number of complete wave cycles occurring per second.

It is measured in Hertz (Hz).

$$
f = \\frac{\\text{Number of cycles}}{\\text{Time in seconds}}
$$

For example:

- 100 Hz means 100 cycles per second.
- 500 Hz means 500 cycles per second.
- 1,000 Hz means 1,000 cycles per second.

![Audio measurement graph](https://images.openai.com/static-rsc-4/RkUaq_Eru5DeuzEfposX7kmOm77Sk6Z8HVwfIc8gA4vPMOccAbuWOvnObrTqGGOdlqMuKbnG9xNmcKc9rVsnq5lnrhJoj2sUwjp8iaezdppsHwv3FwTlywJHEpnTlEt_pwxbWXDRZagoRfNcCFqFQf9JHYo2cDs8E3jxgDe2Pvw?purpose=inline)

[Source: Audio Lab Insight](https://audiolabinsight.com/t/topic/229?utm_source=chatgpt.com)

Higher frequency generally produces a higher perceived pitch for a periodic sound.

### **C. Time Period**

The time period is the time required to complete one full cycle.

It is measured in seconds.

$$
T = \\frac{1}{f}
$$

And:

$$
f = \\frac{1}{T}
$$

Example:

If the frequency is 100 Hz:

$$
T = \\frac{1}{100} = 0.01 \\text{ seconds}
$$

Therefore, one complete cycle takes 10 milliseconds.

### **D. Wavelength**

Wavelength is the physical distance between two successive points in the same phase of a wave.

Examples:

- Compression to the next compression.
- Rarefaction to the next rarefaction.

It is represented by the Greek letter lambda:

$$
\\lambda
$$

Its SI unit is the metre (m).

### **E. Phase**

Phase describes the position of a wave within its cycle relative to a reference point.

It is commonly measured in:

- Degrees (°)
- Radians (rad)

One complete cycle corresponds to:

$$
360^\\circ = 2\\pi \\text{ radians}
$$

Consider two identical waves:

- If they reach their peaks at the same time, they are in phase.
- If one wave reaches its peak when the other reaches its trough, they are 180° out of phase.

![In-phase and out-of-phase waves](https://images.openai.com/static-rsc-4/K9eyFr95hBuaRezVvrfTWCNs0KS7jpJoxdLGc88ZyC0Wzg4irieykFLFmZiGcSPGybw7WYDAweMrxRP1i5jNjf-VKPibw3CZKNuj6uPSFhvqrvWfyJ1osgKCMqRU9IWRsov_x4JIEa3eQf9sCBxyKSJUUtg8mKFYle0x8CBZ-BDPQBy6N0VIcUAEFyx4I_md?purpose=inline)

Phase is important in audio because combining waves can produce constructive or destructive interference.

## **2.5 Speed of Sound**

The speed of sound is the distance travelled by a sound wave per unit time through a medium.

It is represented by:

$$
v = \\frac{\\text{Distance}}{\\text{Time}}
$$

The speed depends on the properties of the medium, especially its elasticity, density, and temperature.

### **Approximate speed in different media**

| Medium | Speed of sound |
|---|---|
| Air (20°C) | 343 m/s |
| Water (20°C) | 1,480 m/s |
| Steel | 5,000–6,000 m/s |

These are approximate values; actual speed varies with material properties and conditions.

### **Effect of temperature in air**

As air temperature increases, the speed of sound generally increases.

An approximate formula for dry air is:

$$
v \\approx 331 + 0.6T
$$

Where:

- \(v\) = speed of sound in m/s.
- \(T\) = air temperature in °C.

For example, at 20°C:

$$
v = 331 + (0.6 \\times 20)
$$

$$
\\boxed{v = 343\\text{ m/s}}
$$

## **2.6 Relationship Between Frequency, Wavelength and Velocity**

This is one of the most important equations in wave physics.

$$
\\boxed{v = f\\lambda}
$$

Where:

- \(v\) = wave velocity (m/s).
- \(f\) = frequency (Hz).
- \(\lambda\) = wavelength (m).

This equation connects the three fundamental properties of a travelling wave.

### **Example 1**

A sound wave travels through air at 343 m/s and has a frequency of 343 Hz. Find its wavelength.

Given:

$$
v = 343\\text{ m/s}
$$

$$
f = 343\\text{ Hz}
$$

Using:

$$
\\lambda = \\frac{v}{f}
$$

$$
\\lambda = \\frac{343}{343}
$$

$$
\\boxed{\\lambda = 1\\text{ m}}
$$

### **Example 2 — What happens when frequency increases?**

Assume sound travels through the same medium at a constant speed.

$$
v = f\\lambda
$$

Therefore:

$$
\\lambda = \\frac{v}{f}
$$

When frequency increases, wavelength decreases.

## **2.9 Decibels (dB)**

The decibel is a logarithmic unit used to express a ratio between two quantities.

In audio, decibels are widely used because sound levels can vary over a very large range.

### **Why use a logarithmic scale?**

Imagine comparing a very quiet sound with an extremely loud sound.

Their physical intensities may differ by millions or billions of times. A logarithmic scale makes such large ratios easier to express.

For sound intensity:

$$
\\boxed{L_I = 10\\log_{10}\\left(\\frac{I}{I_0}\\right)}
$$

Where:

- \(L_I\) = sound intensity level in dB.
- \(I\) = measured sound intensity.
- \(I_0\) = reference sound intensity.

The commonly used reference intensity in air is:

$$
I_0 = 10^{-12}\\text{ W/m}^2
$$

### **Decibels for amplitude or pressure**

For sound pressure, the corresponding equation is:

$$
\\boxed{L_p = 20\\log_{10}\\left(\\frac{p}{p_0}\\right)}
$$

Where:

- \(L_p\) = sound pressure level in dB.
- \(p\) = measured RMS sound pressure.
- \(p_0\) = reference RMS sound pressure.

Why is the multiplier 20 instead of 10?

Because sound intensity is proportional to the square of sound pressure under the same propagation conditions.

$$
I \\propto p^2
$$

Therefore:

$$
10\\log_{10}(p^2) = 20\\log_{10}(p)
$$

### **Useful decibel relationships**

| Change in level | Pressure amplitude ratio | Intensity ratio |
|---|---|---|
| +3 dB | Approximately 1.41× | Approximately 2× |
| +6 dB | Approximately 2× | Approximately 4× |
| +10 dB | Approximately 3.16× | 10× |
| +20 dB | 10× | 100× |

These relationships are useful when analysing audio signals and sound recordings.

## **2.10 Sound Pressure Level (SPL)**

Sound Pressure Level measures the effective sound pressure relative to a standard reference pressure.

It is expressed in decibels (dB SPL).

The reference pressure for sound in air is:

$$
\\boxed{p_0 = 20\\ \\mu Pa}
$$

Where:

$$
20\\ \\mu Pa = 0.00002\\ Pa
$$

The formula is:

$$
\\boxed{SPL = 20\\log_{10}\\left(\\frac{p_{RMS}}{20\\ \\mu Pa}\\right)}
$$

Here, \(p_{RMS}\) is the root-mean-square sound pressure.

### **Example**

Suppose the measured RMS sound pressure is 0.2 Pa.

$$
SPL = 20\\log_{10}\\left(\\frac{0.2}{0.00002}\\right)
$$

$$
=20\\log_{10}(10000)
$$

$$
\\boxed{SPL = 80\\text{ dB}}
$$

### **What is dB(A)?**

You may encounter **dB(A)** in environmental noise measurements.

The A-weighting filter adjusts measurements to approximate the frequency sensitivity of human hearing at moderate sound levels.

- dB SPL: physical sound pressure level relative to a reference.
- dB(A): sound level measured with A-frequency weighting.

This distinction matters when interpreting sound measurements.

## **What you should remember 🧠**

1. Sound is a mechanical wave that requires a medium to propagate.
2. Sound in air travels as a longitudinal wave through compressions and rarefactions.
3. Amplitude, frequency, wavelength, time period, and phase describe wave characteristics.
4. The fundamental wave equation is \(v=f\\lambda\).
