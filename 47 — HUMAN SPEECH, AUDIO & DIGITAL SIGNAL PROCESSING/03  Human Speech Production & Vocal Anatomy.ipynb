# **MODULE 3 — HUMAN SPEECH PRODUCTION & VOCAL ANATOMY 🗣️**

This is an important addition to your speech-processing syllabus. Before a machine can recognize words, detect emotions, or identify speakers, we should understand **how the human body creates the speech signal in the first place**.

The central idea is:

**Lungs provide airflow → vocal folds generate sound → the vocal tract shapes that sound → speech comes out through the mouth or nose.**

This process connects human physiology with the signal-processing concepts we'll study later, especially pitch (\(F_0\)), formants, spectrograms, MFCCs, and Speech Emotion Recognition.

# **A. RESPIRATORY SYSTEM**

## **3.2 Lungs**

The lungs are the primary organs responsible for supplying the airflow required for ordinary speech.

During speech, air is generally pushed out from the lungs and passes through the respiratory tract towards the larynx.

This airflow provides the energy needed to produce most speech sounds.

![Human respiratory system](https://images.openai.com/static-rsc-4/H9O8obUHGO2-WZDOJuU_890gQxSjITSAElAJSkrMUUo8Sn1bu-qG-WWi72EAgj_kqCNI8UEZgMcynaPmHCljoISwngjoG6OLFhhUjHz0HejncIyBaJ-7WPW2PHpbrZ0nq7Qyq8LvbAMke-xAW9iTHpiBcSPkqZO1258OLrWqdUc?purpose=inline)

[Source: Pinterest](https://www.pinterest.com/pin/717339046874830512/?utm_source=chatgpt.com)

**Role in speech production:**

- Supplies the air required for speech.
- Maintains airflow during speaking.
- Helps produce the pressure needed for vocal-fold vibration.
- Supports longer utterances through controlled exhalation.

## **3.3 Diaphragm**

The diaphragm is a dome-shaped muscle located beneath the lungs. It plays a major role in breathing.

During inhalation:

- The diaphragm contracts.
- It moves downward.
- The chest cavity expands.
- Air enters the lungs.

During relaxed exhalation:

- The diaphragm relaxes.
- It moves upward.
- The chest cavity becomes smaller.
- Air moves out of the lungs.

![Diaphragm and breathing mechanism](https://images.openai.com/static-rsc-4/BJN-tJR0KHfxvM4GLTF3wNRT5Qzg7ZH5eg9p3t06PupRulmU8gyfi8CRU_huPzlx1naDJvCxN0kaFIUFm0FFgt3MhBujO0dJy-1TY_L028w05ZVtBMgn3cXwOCYs6Hg-icV5q6yb9Xzl3myTYACaVOUBYKN0LXjzLAUKpAGrvMw?purpose=inline)

[Source: Pain Free at Sea](https://painfreeatsea.com/breath-training/?utm_source=chatgpt.com)

During speech, breathing is coordinated differently from quiet breathing. Speakers typically take a relatively quick breath and control the outward airflow to produce phrases.

## **3.4 Airflow and Air Pressure**

Air naturally moves from a region of higher pressure towards a region of lower pressure.

In speech production, the respiratory system creates pressure below the vocal folds, known as **subglottal pressure**.

This pressure helps drive vocal-fold vibration during voiced speech.

Think of it this way:

1. Air is pushed upwards from the lungs.
2. Pressure builds below the vocal folds.
3. Airflow interacts with the vocal folds.
4. The vocal folds vibrate under suitable conditions.
5. A periodic sound source is produced.

## **3.5 Pulmonic Egressive Airflow**

This is the dominant airflow mechanism used in ordinary human speech.

- *Pulmonic* refers to the lungs.
- *Egressive* means moving outward.

Therefore, pulmonic egressive airflow means **air flowing outward from the lungs**.

![Pulmonic egressive airflow](https://images.openai.com/static-rsc-4/YItPa4fVcDVMFa_hgDtqx-YHrHCn6CXxVaw39xLFnbEVYgslnboRzcMavIsGGwk-UMkyNcStHRPCsCJr9k72Ke54838jKWPIavyf-dG09rfT3Doi4oIgmSvi4PDTXyusa7z92OX2WfRsA5pMsTmT4WT_tuF5fYnQfNbPvlVDNlY?purpose=inline)

[Source: SlideShare](https://www.slideshare.net/BrianPereira5/communication-skills-47697003?utm_source=chatgpt.com)

Most sounds in languages such as English and Tamil use this airflow mechanism.

There are other airflow mechanisms in human languages, such as ejective and implosive sounds, but they are outside our current core scope.

### **Why is breathing important in speech?**

Compare ordinary breathing with speaking:

| Ordinary breathing | Speech breathing |
|---|---|
| Inhalation and exhalation are relatively automatic | Breathing is coordinated with language |
| Exhalation is not normally controlled for producing words | Exhalation is controlled to produce phrases |
| No requirement to create linguistic sounds | Airflow supports vocalization and articulation |

**AI connection:** Variations in breathiness, pauses, airflow, and vocal intensity can influence recorded speech features. They may also affect speech recognition performance.

# **B. PHONATORY SYSTEM**

## **3.6 Larynx**

The **larynx**, commonly called the voice box, is an organ located in the neck, above the trachea (windpipe).

It plays three important roles:

1. Produces voiced sound.
2. Allows air to pass during breathing.
3. Helps protect the airway during swallowing.

The vocal folds are located inside the larynx.

[Source: MDPI](https://www.mdpi.com/2075-4418/16/5/725?utm_source=chatgpt.com)

## **3.7 Vocal Folds (Vocal Cords)**

Vocal folds are two flexible tissue structures inside the larynx.

Although commonly called vocal cords, **vocal folds** is the more anatomically accurate term.

They can:

- Open during breathing.
- Come close together during normal voiced speech.
- Vibrate when suitable airflow and pressure conditions are established.

![Vocal folds open and closed](https://images.openai.com/static-rsc-4/-41GGXAhGD3yTRRJo017Yi2z5nPmGNXAjqePaLqdlOzbf7jKygAvZJDIwDAYIz3EBCLcbmd_MijfvKi06-I_yG_nr0pO4GgkC7O0b4aXofmw9Gt0TzEzvJxtaziUOVCp6V_EkG2wlxaOzoh-dnQfqgxyaLqj4_1vBmaEboqlXuo?purpose=inline)

[Source: Medillsb](https://www.medillsb.com/illustration_image_details.aspx?AID=13501&IID=394005&utm_source=chatgpt.com)

### **How do vocal folds vibrate?**

Let's understand the basic process:

1. The vocal folds are brought close together.
2. Air pressure builds below them.
3. Airflow pushes the folds apart.
4. Tissue elasticity and aerodynamic forces help bring them together again.
5. This opening-and-closing cycle repeats rapidly.

The repeated modulation of airflow produces a periodic sound source.

This process is called **phonation**.

The vibration is not simply the vocal folds being pushed open by air: it is a coordinated interaction between airflow, pressure, tissue properties, and vocal-fold movement.

## **3.8 Glottis**

The glottis is the opening between the two vocal folds.

Its configuration changes depending on what we are doing.

| Breathing | Open |
| Voiced speech | Close together and vibrate |
| Whispering | Narrow opening; no regular vocal-fold vibration |
| Complete glottal closure | Vocal folds come together to close the airway |

## **3.9 Voiced and Unvoiced Sounds**

This is one of the most important concepts in speech processing.

### **Voiced sounds**

Voiced sounds are produced with vocal-fold vibration.

Examples:

- Vowels: /a/, /i/, /u/
- Consonants: /b/, /d/, /g/, /m/, /n/, /z/

Try saying "aaaaa". You can feel vibration around your throat.

### **Unvoiced sounds**

Unvoiced sounds are produced without regular vocal-fold vibration.

Examples:

- /s/ as in *sun*
- /f/ as in *fish*
- /ʃ/ as in *shoe*
- /t/ as in *top*
- /k/ as in *kite*

These sounds are generated through airflow turbulence or the release of an airflow blockage.

### **A practical experiment**

Place your fingers gently on the front of your throat.

1. Say "zzzzzz".
2. Say "ssssss".

You should feel stronger vocal-fold vibration during /z/ than during /s/.

Notice that these two sounds have a similar tongue and airflow configuration, but differ in voicing.

| Sound | Example | Voicing |
|---|---|---|
| /z/ | Zoo | Voiced |
| /s/ | Sun | Unvoiced |
| /v/ | Van | Voiced |
| /f/ | Fish | Unvoiced |
| /b/ | Bat | Voiced |
| /p/ | Pat | Unvoiced |

The pairs /z/–/s/, /v/–/f/, and /b/–/p/ demonstrate how voicing can distinguish speech sounds.

## **3.10 Fundamental Frequency (F₀)**

When vocal folds vibrate periodically, the number of vibration cycles per second determines the **fundamental frequency**, represented by \(F_0\).

$$
\\boxed{F_0 = \\frac{1}{T_0}}
$$

Where:

- \(F_0\) = fundamental frequency in Hz.
- \(T_0\) = duration of one vocal-fold vibration cycle in seconds.

For example, if vocal folds complete 120 cycles per second:

$$
F_0 = 120\\text{ Hz}
$$

$$
T_0 = \\frac{1}{120} = 0.00833\\text{ s}
$$

So one vibration cycle takes approximately 8.33 milliseconds.

![Vocal fold vibration and speech production](https://images.openai.com/static-rsc-4/VaMvyGTwrFf80bBYQD3n4hUcVrE3zjr85crEbp_CB5CviMJqjSWkTwF6zS_ZOfro_ZGi1LegqYqTjL_NDm1kuue4E_L2pHIoFcJC19jcsmjQYrfHNo0SF4cxsZFmlCNhTW96Yw3fibv2zxv_C9l9PI6DzWbwO-6jBGh2iweTWXw?purpose=inline)

[Source: ResearchGate](https://www.researchgate.net/figure/Speech-production-general-principles-The-vocal-fold-vibration-by-the-lung-air-flow_fig2_259961609?utm_source=chatgpt.com)

### **What affects F₀?**

Fundamental frequency depends on several factors:

- Vocal-fold length.
- Vocal-fold tension or stiffness.
- Effective vibrating mass.
- Subglottal pressure and other aspects of vocal-fold dynamics.

In general:

- Increasing vocal-fold tension tends to increase \(F_0\).
- Increasing effective vibrating mass tends to decrease \(F_0\).
- Changes in airflow and pressure can also influence vocal-fold vibration.

### **F₀ and human voice**

The following are illustrative values, not strict biological limits:

| Voice category | Approximate speaking F₀ |
|---|---|
| Adult lower-pitched voice | 85–180 Hz |
| Adult higher-pitched voice | 165–255 Hz |
| Children's voices | Often higher, with considerable variation |

There is considerable overlap between individuals and populations.

## **3.11 Why F₀ Matters in Speech AI**

Imagine two people saying exactly the same sentence.

Their words may be identical, but their vocal characteristics can differ.

An AI model can analyse:

- Average \(F_0\).
- Minimum and maximum \(F_0\).
- \(F_0\) variation.
- Pitch contour.
- Voiced and unvoiced regions.

These features can contribute to:

- Speaker characterization.
- Speech emotion analysis.
- Voice quality analysis.
- Prosody and intonation analysis.

**Important:** \(F_0\) is not the same as formants. \(F_0\) relates to vocal-fold vibration; formants relate to vocal-tract resonance. We'll connect these two concepts in Section E.

# **C. ARTICULATORY SYSTEM**

## **3.12 What is the Articulatory System?**

The articulatory system consists of the structures that shape airflow and sound into recognizable speech sounds.

After the sound source is produced at the larynx, the vocal tract modifies it through movements of the tongue, lips, jaw, palate, and other structures.

Let's study each structure.

## **3.13 Tongue**

The tongue is one of the most important articulators.

It changes position and shape to create different speech sounds.

Its movements affect:

- The size of the oral cavity.
- The location of airflow constrictions.
- Vowel quality.
- Consonant production.

For example:

- /i/ as in *see*: tongue is relatively high and forward.
- /u/ as in *food*: tongue is relatively high and back.
- /a/ as in *father*: tongue is relatively low.

## **3.14 Lips**

The lips can:

- Close completely.
- Open partially.
- Round.
- Spread.
- Change the shape of the mouth opening.

Examples:

- /p/, /b/, /m/: both lips come together.
- /f/, /v/: lower lip approaches the upper teeth.
- /u/: lips are rounded.

Lip movements are especially important for identifying certain consonants and vowels.

## **3.15 Teeth**

The teeth provide contact surfaces for some speech sounds.

Examples:

- /θ/ as in *think*: tongue approaches or comes between the teeth.
- /ð/ as in *this*: a voiced dental sound.
- /f/ and /v/: lower lip approaches the upper teeth.

## **3.16 Hard Palate**

The hard palate is the bony front portion of the roof of the mouth.

It forms the upper surface of the oral cavity and acts as a contact region for certain tongue articulations.

For example, palatal sounds involve the tongue approaching the hard palate.

## **3.17 Soft Palate (Velum)**

The soft palate is the flexible rear portion of the roof of the mouth.

It controls whether airflow can pass into the nasal cavity.

![Speech organs and soft palate](https://images.openai.com/static-rsc-4/kmYNzWOkYWiEUGineNRYWLm0TEocdu63NTwPy1NXCuuhprRkFa1trkx3gu875bFWx0ei_uX2NhhfCoJ_SF5E5oTnQaH4HkQIZHo7qZwKFupS-2UrHa_ndkVsNMEy0S4vr56anSnaQDZdMjjl6ZOvKfGM2NnKAIXGHPodpypuXC8?purpose=inline)

[Source: SlideShare](https://www.slideshare.net/slideshow/speech-organ/40388304?utm_source=chatgpt.com)

- Raised velum → nasal passage is generally closed.
- Lowered velum → nasal passage is open.

This distinction is important for producing oral and nasal sounds.

## **3.18 Alveolar Ridge**

The alveolar ridge is the small raised region immediately behind the upper front teeth.

The tongue approaches or touches this region when producing several consonants.

Examples:

- /t/
- /d/
- /n/
- /s/
- /z/
- /l/

Sounds produced using this region are called **alveolar sounds**.

## **3.19 Nasal Cavity and Oral Cavity**

The oral cavity is the space inside the mouth.

The nasal cavity is the space inside the nose.

Both can act as resonating spaces during speech.

| Oral sounds | Nasal sounds |
|---|---|
| Air primarily exits through the mouth | Air exits through the nose |
| Velum generally raised | Velum lowered |
| Examples: /a/, /i/, /p/ | Examples: /m/, /n/, /ŋ/ |

The consonant /m/ is a good example: the lips close, the velum lowers, and air flows through the nose.

## **3.20 Jaw Movement**

The jaw helps control the opening of the mouth.

It works together with the tongue and lips to modify the vocal tract shape.

Examples:

- Wider jaw opening often accompanies low vowels.
- Smaller jaw opening often accompanies high vowels.
- Jaw movement also supports transitions between sounds.

# **D. SPEECH PRODUCTION MECHANISM**

## **3.21 How Does Airflow Become Speech?**

Now let's combine the three systems.

![Speech signal production model](https://images.openai.com/static-rsc-4/cntlF8pZbGEpukBDz3ir-oZf04yyLSJE-nGjgYCJDoPLFcjUQftAOlUDGqx9aHFkbzUemWALsa7xpD6UDdVUrOBF05Lz9RYPI1POMzyZT_zh8lqDlPH_Wz1j1A0mDdicm1cFnj-FDxw3vig1_VlmMO0ybMQ6o3Fs0WrdvdHF2I0?purpose=inline)

[Source: CircuitBread](https://www.circuitbread.com/textbooks/fundamentals-of-electrical-engineering-i/frequency-domain/modeling-the-speech-signal?utm_source=chatgpt.com)

The production process can be understood in five stages:

1. **Air generation:** The lungs provide outward airflow.
2. **Phonation:** The vocal folds may vibrate and generate a voiced sound source.
3. **Sound modification:** The vocal tract changes the sound through resonance and constrictions.
4. **Articulation:** The tongue, lips, teeth, palate, and jaw create particular speech sounds.
5. **Radiation:** The resulting acoustic signal travels out through the mouth, nose, or both.

Not every sound uses all these mechanisms in the same way. For example, /s/ is produced mainly through turbulent airflow, whereas /a/ uses vocal-fold vibration and an open vocal tract.

## **3.22 How Are Unvoiced Sounds Produced?**

Unvoiced sounds do not require regular vocal-fold vibration.

There are three important ways speech sounds can be generated.

**1. Voicing**

Vocal folds vibrate and produce periodic airflow modulation.

Example: /z/, /b/, /a/.

Example: /s/, /f/, /ʃ/.

## **3.23 Articulation**

**Articulation is the process of modifying airflow and sound using the movements and positions of the speech organs.**

For example, consider the sound /p/.

1. Both lips close.
2. Air pressure builds up behind the closure.
3. The lips suddenly open.
4. A short burst of sound is produced.

Now consider /s/.

1. The tongue approaches the alveolar region.
2. A narrow channel is formed.
3. Air passes through the constriction.
4. Turbulence produces the characteristic /s/ sound.

Different articulatory movements produce different phonemes.

## **3.24 Place of Articulation**

Place of articulation describes **where in the vocal tract the airflow is obstructed or narrowed**.

![Place of articulation diagram](https://images.openai.com/static-rsc-4/PZe_98OJwTFBMMPlrmMCGixG1QpFmxV04CFDBOiOmFpRt8SDyCvN4qvbKrLI9w7_Gg1pFr_oytKUnIZ4Q6Y2zq-2SrirbctA7bTwte0Z41UH5A2FIthtmYm6VlCDjKbh1sbm-LSIuF693BKjEHqExglS6gyochGdiCn4Caz62bE?purpose=inline)

[Source: Vaia](https://www.vaia.com/en-us/explanations/english/phonetics/place-of-articulation/?utm_source=chatgpt.com)

| Place | Articulators involved | Examples |
|---|---|---|
| Bilabial | Both lips | /p/, /b/, /m/ |
| Labiodental | Lower lip and upper teeth | /f/, /v/ |
| Dental | Tongue and teeth | /θ/, /ð/ |
| Alveolar | Tongue and alveolar ridge | /t/, /d/, /s/, /z/, /n/ |
| Postalveolar | Tongue behind alveolar ridge | /ʃ/, /ʒ/ |
| Palatal | Tongue and hard palate | /j/ |
| Velar | Back of tongue and soft palate | /k/, /g/, /ŋ/ |
| Glottal | Vocal folds/glottis | /h/, /ʔ/ |

These are common English phonetic examples; the sound inventory and exact articulation vary across languages.

**Tamil connection:** Tamil includes sounds produced at different places of articulation, including bilabial sounds such as ப, dental sounds such as த, and retroflex sounds such as ட. Retroflex sounds involve the tongue curling or bending back towards the postalveolar region. This is a useful example of why speech models must account for language-specific sound systems.

## **3.25 Manner of Articulation**

Manner of articulation describes **how the airflow is modified or obstructed**.

| Manner | How it is produced | Examples |
|---|---|---|
| Plosive (stop) | Complete closure followed by release | /p/, /t/, /k/ |
| Fricative | Narrow constriction creates turbulence | /f/, /s/, /ʃ/ |
| Affricate | Stop closure followed by frication | /tʃ/, /dʒ/ |
| Nasal | Oral closure with nasal airflow | /m/, /n/, /ŋ/ |
| Approximant | Articulators approach without strong turbulence | /w/, /j/ |
| Lateral approximant | Air flows around the sides of the tongue | /l/ |
| Trill | Repeated vibration of an articulator | /r/ in some languages |

**Place tells us where a sound is produced. Manner tells us how it is produced.**

For example:

- /p/: bilabial plosive.
- /s/: alveolar fricative.
- /m/: bilabial nasal.

## **3.26 Vowels and Consonants**

Speech sounds are broadly classified into vowels and consonants.

### **Vowels**

Vowels are generally produced with a relatively open vocal tract, without a major obstruction to airflow.

Examples:

- /a/
- /e/
- /i/
- /o/
- /u/

Vowel quality depends strongly on tongue position, jaw opening, and lip configuration.

Three important vowel dimensions:

1. **Height:** How high or low the tongue is.
2. **Backness:** Whether the tongue is positioned towards the front, centre, or back.
3. **Rounding:** Whether the lips are rounded.

### **Consonants**

Consonants are produced by creating a constriction or closure somewhere in the vocal tract.

Examples:

- /p/
- /t/
- /k/
- /s/
- /m/
- /n/

### **Comparison**

| Vowels | Consonants |
|---|---|
| Relatively open vocal tract | Usually involve a constriction or closure |
| Strongly characterized by formants | Characterized by bursts, noise, voicing, and transitions |
| Examples: /a/, /i/, /u/ | Examples: /p/, /s/, /m/ |

## **3.27 Oral and Nasal Sounds**

This distinction depends on whether the nasal cavity participates in the sound production.

### **Oral sounds**

The velum is generally raised, closing off the nasal passage. Air exits primarily through the mouth.

Examples: /p/, /t/, /s/, /a/.

### **Nasal sounds**

The velum is lowered, allowing air to pass through the nasal cavity.

Examples:

- /m/ — bilabial nasal.
- /n/ — alveolar nasal.
- /ŋ/ — velar nasal, as in the final sound of *sing*.

# **E. SOURCE-FILTER THEORY**

## **3.28 What Is Source-Filter Theory?**

<badge color="success">Core concept for speech processing</badge>

Source-filter theory is one of the most important acoustic models for understanding human speech production.

It explains speech as the combination of two major components:

1. **Source:** Generates the sound.
2. **Filter:** Shapes the sound based on the vocal tract's configuration.

![Source-filter theory](https://images.openai.com/static-rsc-4/VSpozBy_AdISAaKLcvLE8GACHgPSYTjAmtwm3d5czR57dx5R2GaVm3KWy0InY4s_qPjFHexnp8_aE6X0Slrhenrmm0uUpciCidfXXSecYTzkYBQfn4VQnBnLCHPBGiqqRcHOkM8bPtLB1aaJ2s-vcdWPdLLfS_1JMdz_CBRgs8o?purpose=inline)

[Source: SlideServe](https://www.slideserve.com/delu/flst-text-to-speech-synthesis?utm_source=chatgpt.com)

Think about a musical instrument.

- The vibrating string generates sound.
- The body of the instrument resonates and modifies that sound.

Human speech works somewhat similarly:

- Vocal folds generate a sound source.
- The vocal tract modifies the sound.

## **3.29 Glottal Source**

The glottal source is the sound excitation generated at the glottis.

During voiced speech:

1. Air flows from the lungs.
2. Vocal folds vibrate.
3. Airflow becomes periodically modulated.
4. A complex, harmonic-rich signal is generated.

This signal contains:

- Fundamental frequency (\(F_0\)).
- Harmonics at integer multiples of \(F_0\).

$$
f_n = nF_0
$$

Where:

- \(f_n\) = frequency of the nth harmonic.
- \(n\) = harmonic number.
- \(F_0\) = fundamental frequency.

### **Example**

Suppose a speaker produces a voiced sound with:

$$
F_0 = 100\\text{ Hz}
$$

Its harmonic frequencies are:

| Harmonic | Frequency |
|---|---|
| First | 100 Hz |
| Second | 200 Hz |
| Third | 300 Hz |
| Fourth | 400 Hz |
| Fifth | 500 Hz |
| Sixth | 600 Hz |

![Harmonic structure in speech](https://images.openai.com/static-rsc-4/ve6XOkHgU2oQKV37DvFDbG_sGcy-WCcToPmcfjK0LNvmQp1RAkuG4pNphPQSajVHm-uOJn1gnt5fc5t1kzWHVItQ6G3fgRpSiIpSRb0Lh3uSVorx0ZP9PhiDZHLYLG-yS0638KAH_bRL_Y3OENqZu7h48civSVSprQNAtEuzr_w?purpose=inline)

[Source: Maël Fabien](https://maelfabien.github.io/machinelearning/speech_reco/?utm_source=chatgpt.com)

The glottal source provides the underlying excitation, but it does not by itself determine the complete sound of a spoken vowel.

## **3.30 Vocal Tract as a Filter**

The vocal tract includes the air passages above the vocal folds, particularly the pharyngeal and oral cavities, and sometimes the nasal cavity.

It acts as an acoustic filter.

The tongue, lips, jaw, and velum continuously change its shape.

As a result, some frequency regions are amplified more strongly than others.

![Vocal tract acoustic filter](https://images.openai.com/static-rsc-4/wtXCIF5BUTNIlB3nHDSTI5F3eVolkCdKHZMNooKRxLCbRveR3UFKccf8wRu-7LiT1SEgcxgdEkk9YfHjNRPbNiNl-94315Ym4MIKarCInOlfhVuIYdibFXIGQ8mgBMtl9sFII5-asclbkDcY3XIhLRJCaF6ESnkPH8UG1kw7S4I?purpose=inline)

[Source: SlideServe](https://www.slideserve.com/coxb/acoustic-phonetics-powerpoint-ppt-presentation?utm_source=chatgpt.com)

For example, when you say:

- /a/
- /i/
- /u/

Your vocal folds may generate a similar type of periodic source, but your tongue and lips adopt different configurations.

Those configurations produce different vocal-tract resonances and therefore different vowel sounds.

## **3.31 Resonance**

**Resonance occurs when a system responds strongly to particular frequencies.**

The vocal tract contains air-filled cavities that have resonant frequencies.

These frequencies depend on:

- Vocal-tract length.
- Cavity shape.
- Cross-sectional area.
- Tongue position.
- Lip opening.
- Velum position.

For example:

- A larger resonating cavity tends to have lower resonant frequencies.
- A smaller resonating cavity tends to have higher resonant frequencies.

The vocal tract selectively amplifies frequency regions in the sound source.

These prominent resonant regions are associated with formants.

## **3.32 What Are Formants?**

Formants are prominent resonance frequencies of the vocal tract, observed as broad peaks in the spectral envelope of speech.

They are represented as:

$$
F_1,\\ F_2,\\ F_3,\\ F_4,\\ldots
$$

The first three formants are particularly important when studying vowel characteristics.

![Speech formants and spectral envelope](https://images.openai.com/static-rsc-4/74s8JE7V91xSyrbVVaL0cVzuZWsG_s1L_2jrJundLGX9Q7X_aNNVXLO1TE2CdROFIBV-kgdsztm15PNgVAL-dpHuDha8YF0B2NMW0cF4f0hPkLCxeTv0Lci2Nh4l74KUtixO5FPi_YAF1hSUIcVxDEKMhJFUISWZgQLug2b_FkU?purpose=inline)

[Source: CSDN Blog](https://blog.csdn.net/zj71hmvx/article/details/128191646?utm_source=chatgpt.com)

### **F1 — First Formant**

F1 is strongly related to vowel height and jaw opening.

General relationship:

- High vowels → lower F1.
- Low vowels → higher F1.

Example:

- /i/ generally has a low F1.
- /a/ generally has a higher F1.

### **F2 — Second Formant**

F2 is strongly related to tongue frontness.

General relationship:

- Front vowels → higher F2.
- Back vowels → lower F2.

Example:

- /i/ generally has a high F2.
- /u/ generally has a low F2.

### **F3 — Third Formant**

F3 provides additional information about vocal-tract shape and speech sound characteristics.

It can be particularly useful for studying:

- Certain rhotic sounds, such as English /r/.
- Speaker characteristics.
- Fine differences in articulation.

The relationships between F1, F2, and vowel articulation are useful general patterns, not absolute rules. Actual formant values vary with speaker, language, vocal-tract anatomy, and phonetic context.

## **3.34 Mathematical Representation of Source-Filter Theory**

Here is the mathematical foundation.

In a simplified linear source-filter model:

$$
\\boxed{s(t)=e(t)*h(t)}
$$

Where:

- \(s(t)\) = resulting speech signal.
- \(e(t)\) = excitation or source signal.
- \(h(t)\) = impulse response of the vocal-tract filter.
- \(*\) = convolution.

Taking the Fourier transform:

$$
\\boxed{S(f)=E(f)H(f)}
$$

Where:

- \(S(f)\) = speech spectrum.
- \(E(f)\) = source spectrum.
- \(H(f)\) = vocal-tract frequency response.

![Acoustic and physiological phonetics](https://images.openai.com/static-rsc-4/svOEjlWxP1p_umCLVICVw8l_g63zySkuBMhiipGgT2KOpslDSZgoPpqhndsUeyKqUM0XDz_YfvxIwFx61zq0XVG5AW1726L3haAnzi99dTjs-bbsExPl6PkQbhJ_FWjWm8qKZZ6RP54E_5hWkk6npw7TTcpv0mRL6NP_7Pn5LoE?purpose=inline)

[Source: SlideServe](https://www.slideserve.com/minnie/acoustic-and-physiological-phonetics?utm_source=chatgpt.com)

This tells us something important:

The resulting speech signal contains characteristics of both the sound source and the vocal tract.

**A useful distinction:**

| Property | Primarily associated with |
|---|---|
| Fundamental frequency \(F_0\) | Vocal-fold vibration |
| Harmonic structure | Periodic excitation |
| Formants \(F_1,F_2,F_3\) | Vocal-tract resonances |
| Vowel identity | Vocal-tract configuration and its acoustic output |

This is the simplified model. Real speech production can involve interactions between the vocal folds and vocal tract, so they are not always completely independent.

## **3.35 Formants vs Harmonics — Don't Confuse Them!**

This distinction is especially important for interviews and speech-processing work.

![Fundamental frequency and harmonic structure](https://images.openai.com/static-rsc-4/fBSg5ctJaXhu5ZXeBEkBBBGFyne030BgzaFCHq88Cu84vM_oZemcraaRp3_Qiz_8xjpGHorqvSj_QJeroeAHLVlCiVWg9hkYO3h_362IHkTWO66ZWFMtTg4JyikP7hbl12ihhtTdB8usYfx733BAmWNbzScNwo-ytgak9OsJimY?purpose=inline)

[Source: Introduction to Speech Processing](https://speechprocessingbook.aalto.fi/Representations/Fundamental_frequency_F0.html?utm_source=chatgpt.com)

| Harmonics | Formants |
|---|---|
| Associated with the sound source | Associated with vocal-tract resonance |
| Integer multiples of \(F_0\) in ideal periodic voicing | Resonant frequency regions |
| Determined by vocal-fold vibration rate | Determined by vocal-tract shape |
| Describe the fine spectral structure | Describe the broad spectral envelope |

For example, if:

$$
F_0=100\\text{ Hz}
$$

The harmonics occur at approximately 100, 200, 300, 400 Hz, and so on.

But the formants might occur around 500 Hz, 1,500 Hz, and 2,500 Hz.

The formants do not have to be exact multiples of \(F_0\).

**Remember: F0 tells us about the source; formants tell us about the filter.**

## **What you should remember 🧠**

1. Human speech production involves respiratory, phonatory, and articulatory systems.
2. The lungs supply airflow; vocal folds produce voiced sound; articulators shape speech.
3. \(F_0\) is the rate of vocal-fold vibration and is related to perceived pitch.
4. Place of articulation describes *where* a sound is produced; manner describes *how* it is produced.
5. Voiced sounds involve vocal-fold vibration, while unvoiced sounds lack regular vocal-fold vibration.
6. Source-filter theory models speech as a sound source shaped by the vocal tract.
7. \(F_0\) and harmonics describe the source structure; F1, F2, and F3 describe vocal-tract resonances.
