# **11.1 What Is Audio Noise?**

Audio noise refers to unwanted sound or unwanted signal components that interfere with the audio we want to capture or analyze.

For speech processing, we generally want:

- Clear speech.
- Accurate pronunciation information.
- Speaker characteristics.
- Useful acoustic features.

But noise can interfere with these characteristics.

A simplified audio recording model is:

$$
y(t)=s(t)+n(t)
$$

Where:

- $y(t)$ = Recorded noisy audio.
- $s(t)$ = Desired speech signal.
- $n(t)$ = Noise signal.

For example:

$$
\text{Recorded Audio}
=
\text{Human Speech}
+
\text{Background Noise}
$$

Let's understand the major types.

## **1. Background Noise**

Background noise refers to unwanted sounds present in the recording environment.

Examples:

- Fans.
- Air conditioners.
- People talking.
- Computers.
- Traffic.

## **2. White Noise**

White noise is a random signal with approximately equal power per unit bandwidth across its frequency range.

Think of it as a continuous noise containing many frequency components.

Examples:

- Electronic test noise.
- Static-like sound.
- Simulated noise used in signal-processing experiments.

**Important:** White noise does not mean equal amplitude at every frequency. It means approximately flat power spectral density.

## **3. Gaussian Noise**

Gaussian noise describes noise whose amplitude values follow a Gaussian (normal) probability distribution.

Its probability density function is:

$$
p(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- $\mu$ = Mean.
- $\sigma$ = Standard deviation.
- $x$ = Noise amplitude.

For zero-mean Gaussian noise:

$$
\mu=0
$$

Most values occur near zero, while larger positive and negative values occur less frequently.

### **White Noise vs Gaussian Noise**

These terms describe different properties.

| White Noise | Gaussian Noise |
|---|---|
| Describes frequency characteristics | Describes amplitude probability distribution |
| Approximately flat power spectral density | Amplitudes follow a normal distribution |
| Can have different amplitude distributions | Can have different spectral characteristics |

A noise signal can be both **white and Gaussian**.

This is commonly called **Additive White Gaussian Noise (AWGN)**.

It is widely used in communication-system simulations and audio-processing experiments.

## **4. Pink Noise**

Pink noise is a type of noise whose power spectral density decreases approximately inversely with frequency.

$$
P(f)\propto\frac{1}{f}
$$

This means lower frequencies contain more power per equal-width frequency interval than higher frequencies.

It is sometimes described as having equal power per octave.

Examples:

- Some natural environmental sound patterns.
- Certain acoustic test signals.
- Simulated ambient noise.

## **9. Reverberation**

Reverberation occurs when sound reflects from surfaces in an enclosed space.

Imagine speaking inside:

- A large empty hall.
- A classroom.
- A bathroom.
- A concrete room.

The microphone captures the direct sound and multiple reflected versions of it.

Reverberation causes sound energy to persist after the original sound has stopped.

It can make speech sound:

- Distant.
- Blurred.
- Less intelligible.
- Difficult to separate into individual words.

A simplified reverberant recording model is:

$$
y(t)=s(t)*h(t)
$$

Where:

- $s(t)$ = Original speech.
- $h(t)$ = Room impulse response.
- $*$ = Convolution.

Notice that reverberation is better understood as filtering the speech through the acoustic response of the room, rather than simply adding independent noise.

## **10. Echo**

Echo is a delayed reflection of the original sound that can be heard as a distinct repetition.

For example, shouting inside a mountain valley.

# **11.3 Signal Quality**

Now that we understand different noise sources, let's learn how to measure the quality of an audio signal.

The most important measurement in this section is **Signal-to-Noise Ratio (SNR)**.

## **1. Signal Power**

Signal power measures the average squared amplitude of a signal.

For a discrete-time signal:

$$
P_s=\frac{1}{N}\sum_{n=0}^{N-1}s[n]^2
$$

Where:

- $P_s$ = Average signal power.
- $s[n]$ = Signal samples.
- $N$ = Number of samples.

For real-valued audio, this is the mean square of the samples.

### **Python implementation**

```python
signal_power = np.mean(clean_speech ** 2)
```

## **2. Noise Power**

Noise power measures the average squared amplitude of the noise.

$$
P_n=\frac{1}{N}\sum_{n=0}^{N-1}n[n]^2
$$

### **Python implementation**

```python
noise_power = np.mean(noise ** 2)
```

## **3. Signal-to-Noise Ratio (SNR)**

SNR compares the strength of the desired signal with the strength of unwanted noise.

$$
\boxed{SNR=\frac{P_s}{P_n}}
$$

A higher SNR means the desired signal is stronger relative to the noise.

A lower SNR means noise is more significant relative to the desired signal.

### **SNR in decibels**

In audio processing, SNR is commonly expressed in decibels.

$$
\boxed{
SNR_{dB}=10\log_{10}
\left(\frac{P_s}{P_n}\right)
}
$$

Since power is proportional to squared amplitude, we can also calculate it using RMS amplitudes:

$$
SNR_{dB}=20\log_{10}
\left(\frac{RMS_s}{RMS_n}\right)
$$

### **Example 1 — High SNR**

Suppose:

$$
P_s=100
$$

$$
P_n=1
$$

Then:

$$
SNR=100
$$

$$
SNR_{dB}=10\log_{10}(100)
$$

$$
\boxed{SNR_{dB}=20\ dB}
$$

The signal is considerably stronger than the noise.

### **Example 2 — Equal signal and noise power**

Suppose:

$$
P_s=P_n
$$

Then:

$$
SNR_{dB}=10\log_{10}(1)
$$

$$
\boxed{SNR_{dB}=0\ dB}
$$

The signal and noise have equal average power.

### **Example 3 — Noise stronger than speech**

Suppose:

$$
P_s=1
$$

$$
P_n=10
$$

Then:

$$
SNR_{dB}=10\log_{10}(0.1)
$$

$$
\boxed{SNR_{dB}=-10\ dB}
$$

The noise power is ten times the speech power.

## **11.6 Clipping**

Clipping occurs when the audio signal exceeds the maximum amplitude that a recording system can represent.

For normalized digital audio, the usual amplitude range is approximately:

$$
-1\leq x[n]\leq1
$$

If the input exceeds the representable range, its peaks may be cut off.

![Audio clipping](https://images.openai.com/static-rsc-4/Uwsh0Kg5Fd4Q_1L2HNGBNBnVjBMzPBfavukCABnMEh8wdy4SpTzVYv7937KPIjWQSLhNZnEEhpU7qFndL6qtI1dh4C8vX6GzWjJb3nPksjIoC98CvuH5FHZvF-cG-J3F5Bd2r79KaxJU2ZEgPVGFRixf6AeEbWDvHxHxaXcbnqY?purpose=inline)

Reference: [Audio Clipping Explained — Music Production for Women](https://musicproductionforwomen.com/post/audio-clipping?utm_source=chatgpt.com)

### **Example**

Original signal:

$$
x[n]=1.3\sin(2\pi ft)
$$

If the recording system limits the signal to the range $[-1,1]$, the peaks are clipped.

This introduces distortion and unwanted frequency components.

### **Python implementation**

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 0.02, 1000)

original = 1.3 * np.sin(2 * np.pi * 200 * t)

clipped = np.clip(original, -1, 1)

plt.figure(figsize=(12, 4))

plt.plot(t, original, label="Original")
plt.plot(t, clipped, label="Clipped")

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Audio Clipping")

plt.legend()
plt.grid()
plt.show()
```

### **Why is clipping harmful?**

- Distorts the waveform.
- Introduces additional frequency components.
- Changes the original speech characteristics.
- Can affect ASR and speaker-recognition performance.

Clipping should ideally be prevented during recording by using appropriate microphone gain and input levels.

**Important:** Clipping is not the same as ordinary background noise. It is a nonlinear distortion, and once severe clipping occurs, the original signal generally cannot be perfectly recovered.

# **11.8 Noise Reduction Concepts**

Now we reach an important application of this module.

Suppose we have a noisy speech recording.

Can we recover cleaner speech?

Yes, to some extent.

There are several signal-processing techniques designed to reduce unwanted noise while preserving useful speech information.

The general process is called **noise reduction** or **noise suppression**.

## **1. Noise Suppression**

Noise suppression means reducing unwanted noise in an audio recording while attempting to preserve the desired speech.

For example:

- Removing background fan noise.
- Reducing traffic noise.
- Suppressing microphone hiss.
- Improving speech clarity during online meetings.

The goal is to improve speech quality without damaging important speech information.

Applications:

- Call-center recordings.
- Speech recognition.

## **2. Filtering**

Filtering is a signal-processing technique used to allow certain frequency ranges to pass while attenuating others.

Different filters perform different operations.

### **Low-pass filter**

Allows lower frequencies to pass and attenuates higher frequencies.

$$
H(f)\approx
\begin{cases}
1 & f<f_c\\
0 & f>f_c
\end{cases}
$$

Here, $f_c$ is the cutoff frequency.

### **High-pass filter**

Allows higher frequencies to pass and attenuates lower frequencies.

### **Band-pass filter**

Allows a selected frequency range to pass.

### **Band-stop filter**

Attenuates a selected frequency range.

![Band-pass filter circuit](https://images.openai.com/static-rsc-4/dzGecVhL7u9oSdp_49NtW6b3eUxfnfAIuWxFAKk8vXQsck8inStZSyJ59AUHISSunJ8RLSbjDm0Q_l8_108mzGR4u3xujMChkJvOkNnI17P6FaGtIl4DJzCOQxgIBLUXTJ9ZgP7UvGOiZG-l7irIgPvVFSu0kaelPRZWN9umJQQ?purpose=inline)

### **Speech-processing example**

Suppose a recording contains low-frequency wind rumble.

A high-pass filter may reduce some of that unwanted low-frequency energy.

But there is a problem:

Speech itself contains low-frequency information.

If the cutoff is too high, important speech components can also be removed.

**Important:** Filtering is useful when noise occupies a frequency region that can be separated from important speech components. It is not a universal solution for overlapping speech and noise.

## **3. Spectral Subtraction**

Spectral subtraction is a noise-reduction technique that operates in the frequency domain.

The basic idea:

1. Estimate the noise spectrum.
2. Calculate the noisy speech spectrum.
3. Subtract an estimate of the noise magnitude.
4. Reconstruct an enhanced signal.

A simplified magnitude equation is:

$$
|\hat S(k)|=
\max\left(
|Y(k)|-\alpha|\hat N(k)|,\ 0
\right)
$$

Where:

- $Y(k)$ = Noisy speech spectrum.
- $\hat N(k)$ = Estimated noise spectrum.
- $\hat S(k)$ = Estimated clean speech magnitude.
- $\alpha$ = Noise subtraction factor.

![Spectral processing](https://images.openai.com/static-rsc-4/91b8fYj3OO5l_qwBYSXZf1fRbb9pRZXUVrLmqFiYmNIWfvlT_FuEDh_AsFcKyUv_Md4-eReU3eyxdpgT23VPwTGfUar_XJYu2KB2425WnQCSDZMvq8CVOTYj2ndslxvZikthD5-mZwxDxrIbH59usZf8qFJDlVFWZuwfdLef8iU?purpose=inline)

Reference: [AudioLDM: Revolutionizing Text-to-Audio Generation Quality — Markovate](https://markovate.com/blog/audioldm/?utm_source=chatgpt.com)

### **Example**

Imagine:

- Noisy spectrum magnitude = 100.
- Estimated noise magnitude = 30.
- Subtraction factor = 1.

Then:

$$
\hat S=100-30=70
$$

The estimated speech magnitude becomes 70.

In practical implementations, spectral subtraction is performed frame by frame, often with additional smoothing and noise estimation.

### **Limitation**

If the noise estimate is inaccurate, the output can contain artificial musical tones known as *musical noise*.

## **4. Wiener Filtering**

Wiener filtering is a statistical signal-processing technique used to estimate a desired signal from a noisy observation.

Its objective is to minimize the expected mean squared error between the original signal and its estimate.

For an idealized additive-noise model with suitable statistical assumptions, a frequency-domain Wiener filter can be expressed as:

$$
H(f)=\frac{P_s(f)}{P_s(f)+P_n(f)}
$$

Where:

- $P_s(f)$ = Signal power spectral density.
- $P_n(f)$ = Noise power spectral density.
- $H(f)$ = Filter response.

The enhanced signal is approximately:

$$
\hat S(f)=H(f)Y(f)
$$

### **Intuition**

If a frequency contains strong speech and little noise, the filter allows more of it to pass.

If a frequency contains mostly noise, the filter attenuates it.

Wiener filtering is useful in speech enhancement, image restoration and other signal-estimation problems.

## **5. Noise Gates**

A noise gate is a dynamic audio-processing tool that reduces or mutes audio when its level falls below a selected threshold.

Imagine someone speaking into a microphone.

- When the speaker talks, the signal exceeds the threshold.
- When the speaker stops, the signal falls below the threshold.
- The gate reduces the low-level audio.

![Noise gate attack, hold and release](https://images.openai.com/static-rsc-4/b4RjVZFHJO8algz61q557Kl57P0zLwHY9klYkmhZXtw9r6SJ8aYNQFBOD1hYqvJIcNrLsMm8X1ptcMAXqMPLpv11OQZvw9WSe8NJoeM9CXioDRBIzPsDvFplRT2TQ7Q3w8qBWlHZBsGeGH-79kKQOXRAvqyV_hA349AZ4p8Tzoc?purpose=inline)

Reference: [Noise Gate Attack Hold Release — Wikimedia Commons](https://commons.wikimedia.org/wiki/File%3ANoise_Gate_Attack_Hold_Release.svg?utm_source=chatgpt.com)

Important parameters:

- Threshold.
- Attack time.
- Release time.
- Hold time.

### **Limitation**

A noise gate does not intelligently distinguish speech from noise.

If background noise is louder than the threshold, it may remain audible.

If quiet speech falls below the threshold, the gate may suppress it.

It is therefore more useful for controlling noise during pauses than for separating overlapping speech and noise.

## **6. Adaptive Noise Cancellation**

Adaptive noise cancellation is a technique that uses an adaptive filter to estimate and cancel unwanted noise.

A common setup uses two microphones or signal paths:

- Primary input: Desired speech mixed with noise.
- Reference input: A signal correlated with the unwanted noise.

The adaptive filter learns to estimate the noise component in the primary input.

That estimate is subtracted from the primary signal.

![Adaptive noise cancellation architecture](https://images.openai.com/static-rsc-4/Yf1jntHlQkqDBHAu9o9EZImsPsIm_v4fjrHlFSHQ6yzpqGvInJr5cueSh_qfBiSExE3dmi5-kwux9Vfw2FBumByIosrRDfPkjcJlXBn043wXX1DjrZwfjmN-OWPyWj4TyE4yoQZgeZ2AeHieeZQxEyV43nUpkAIKB-zWHTZFUsg?purpose=inline)

Reference: [Adaptive Noise Cancellation Using Reference Signal — GitHub](https://github.com/matousc89/Python-Adaptive-Signal-Processing-Handbook/issues/4?utm_source=chatgpt.com)

### **Basic architecture**

**Primary microphone**

Speech + noise

↓

**Reference microphone**

Correlated noise

↓

**Adaptive filter**

Estimates unwanted noise

↓

**Subtraction**

Primary signal − estimated noise

↓

**Output**

Enhanced speech

Common adaptive algorithms include:

- LMS — Least Mean Squares.
- NLMS — Normalized Least Mean Squares.
- RLS — Recursive Least Squares.

### **Where is it used?**

- Headphones with active noise cancellation.
- Hands-free communication.
- Vehicle communication systems.
- Industrial audio systems.

**Important distinction:** Active noise cancellation in headphones often uses acoustic anti-noise generated by speakers, whereas adaptive noise cancellation in signal processing estimates and subtracts a correlated noise signal. Both use cancellation principles, but their implementations differ.

## **11.9 Comparison of Noise-Reduction Techniques**

| Technique | Main idea | Useful situation |
|---|---|---|
| Filtering | Attenuate selected frequencies | Frequency-specific noise |
| Spectral subtraction | Subtract estimated noise spectrum | Relatively stationary background noise |
| Wiener filtering | Estimate desired signal statistically | Noise with estimable spectral characteristics |
| Noise gate | Reduce low-level audio | Noise during pauses |
| Adaptive cancellation | Estimate correlated noise and subtract it | Reference noise signal available |
| Noise suppression | General noise-reduction objective | Speech enhancement systems |

A modern deep-learning speech-enhancement model may also learn to separate speech from noise using training examples rather than relying entirely on hand-designed filters.
