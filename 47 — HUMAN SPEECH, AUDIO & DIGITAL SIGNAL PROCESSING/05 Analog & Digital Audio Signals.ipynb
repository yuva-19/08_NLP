# **MODULE 5 — ANALOG & DIGITAL AUDIO SIGNALS 🎧**

In the previous modules, we studied how sound is produced by the human body and how speech sounds are organized.

Now we reach one of the most important foundations of digital signal processing:

**How does a physical sound wave become a sequence of numbers that a computer can understand?**

When you speak into a microphone, the sound initially exists as a continuously varying pressure wave. A computer cannot directly store that physical wave in its original continuous form.

It must be converted into a digital representation.

This conversion involves:

1. Analog signal generation.
2. Analog-to-Digital Conversion (ADC).
3. Sampling.
4. Quantization.
5. Digital representation.

## **5.1 What Is an Analog Signal?**

An **analog signal** is a signal that varies continuously with time and can take a continuous range of amplitude values.

### Examples

- Human speech captured by a microphone.
- Music produced by an analog instrument.
- Electrical voltage from an analog sensor.

A microphone converts sound-pressure variations into corresponding electrical variations.

![Dynamic Microphone](https://images.openai.com/static-rsc-4/qmxOjOBMyCwY1k5XfAfJhmFdlcQcLXoMVkjx37RWKK3m3B3Iuvg2FBVjBBjWrfoRiN8yD0TREWdGMaiaJYNbX-Z42L8ZNNIf-CicN6QpQeY0hkb31ej6X8sNua0b-F7dmG5xsZi2Adr0PZTedgY0eXG5LhAAkXFjloyQjBj-vZY?purpose=inline)

**Reference:** [How Dynamic Microphones Create Audio Signal](https://www.rfvenue.com/blog/2015/01/05/how-dynamic-microphones-create-audio-signal)

### Characteristics of Analog Signals

- Continuous-time representation.
- Continuous amplitude values.
- Can represent continuously varying physical quantities.
- Can be affected by noise and interference.

### Mathematical Representation

An analog signal can be represented as:

$$
x(t)
$$

Where:

- $x$ = signal amplitude.
- $t$ = continuous time.

For example, a sinusoidal analog signal can be represented as:

$$
x(t)=A\sin(2\pi ft+\phi)
$$

Where:

- $A$ = amplitude.
- $f$ = frequency.
- $t$ = time.
- $\phi$ = phase.

This is a mathematical representation of a continuously varying signal.

## **5.2 What Is a Digital Signal?**

A **digital signal** represents information using discrete numerical values.

In digital audio, a continuous sound waveform is represented using a sequence of numerical samples.

### Example

```python
audio = [0.12, 0.35, 0.48, 0.21, -0.15, -0.42, -0.18]
```

Each number represents the amplitude of the signal at a particular sampling instant.

![Analog and Digital Signal Comparison](https://images.openai.com/static-rsc-4/A5hqmcMMQieFSSb93YwXr3hNFJIyuP935biLz3XeVVmoIIe8bRjCO2uDzyvzl3Vs8KvuLW8C7aXTHkppUl0PTYnpstuOSlakF6WrLeb5Gy_bXNAvN4JIm_1giKpzvL_QJir3aticq6HjeS6S7g4oB4ChRub0aW3k0tsKI1RgbgM?purpose=inline)

**Reference:** [Analog vs Digital Audio](https://note.com/michihito_takami/n/nf87fb5cafa6b)

### Characteristics of Digital Signals

- Discrete-time representation.
- Finite set of amplitude levels for fixed-bit-depth PCM.
- Can be stored as binary data.
- Can be processed using computers.
- Can be copied without accumulating analog-generation noise, although real playback and transmission systems still introduce noise and errors.

### Analog vs Digital Signals

| Analog | Digital |
|---|---|
| Continuous-time representation | Discrete-time representation |
| Continuous amplitude | Discrete amplitude levels in fixed-bit-depth PCM |
| Represented by continuously varying quantities | Represented by numerical values |
| Commonly used in microphones before conversion | Used by computers for audio processing |

## **5.3 Continuous-Time vs Discrete-Time Signals**

This distinction is extremely important in Digital Signal Processing (DSP).

### A. Continuous-Time Signal

A continuous-time signal is defined at every instant of time.

It is represented as:

$$
x(t)
$$

For example, an analog microphone voltage changes continuously as you speak.

![Analog to Digital Converter](https://images.openai.com/static-rsc-4/A1cS9pLhmep6lV_V8YB5LqUhjfM1XPeub8gi7WVEuASOWqgOcp2U-ZukrTM80z02drg-3uGTZaYx8ve_FtQ995uzmyTA01-r5GZDNC6bPnLVU66DjMJ9kI5aJ2jNL2daYrJ_pSGbkIZlHA8hEQYnnXoH9Rr5uqW9pxOpBw0cfJ4?purpose=inline)

**Reference:** [How Do Analog-to-Digital Converters (ADCs) Work?](https://www.circuitbread.com/ee-faq/how-do-analog-to-digital-converters-adcs-work)

### B. Discrete-Time Signal

A discrete-time signal is defined only at specific time instants.

It is represented as:

$$
x[n]
$$

Here, $n$ is an integer-valued sample index.

For example:

$$
x[0],x[1],x[2],x[3],\ldots
$$

Each value corresponds to one sample of the original signal.

![Discrete Time Signal](https://images.openai.com/static-rsc-4/bmvby2OJrdPTLRcSTffE4bxyVIq9h3i8QJ0MV0_moAdke9hOiJ4nVSxCZ0sru5zEHoA2ZZswBHaUf0syiDgT6moTdIsSpG1Vvz-NtJPwiLWwQBR1kMbI_mmPIqoKjgQxAywjMHy7woRaxKInFjEaCPHw6lvgN4-y9gigPvdPE9s?purpose=inline)

**Reference:** [How to Generate Discrete-Time Signals](https://danielegiardino.medium.com/how-to-generate-discrete-time-signals-5e3954c05628)

### Comparison

| Continuous-Time | Discrete-Time |
|---|---|
| Defined at every time instant | Defined at discrete time instants |
| Uses $x(t)$ | Uses $x[n]$ |
| Time variable is continuous | Sample index is discrete |
| Typical analog representation | Typical digital signal representation |

**Important:** Sampling makes time discrete. Quantization makes amplitude discrete. These are two different operations.

## **5.4 Continuous Amplitude vs Discrete Amplitude**

A signal has two independent dimensions:

1. Time.
2. Amplitude.

Understanding both will make ADC much easier.

### Continuous Amplitude

An analog signal can take any amplitude within its allowed range.

For example, its instantaneous voltage could be:

- 0.123 V
- 0.1234 V
- 0.12345 V

The amplitude is not restricted to a finite set of digital levels.

### Discrete Amplitude

In a fixed-bit-depth digital representation, the signal amplitude is mapped to one of a finite number of quantization levels.

For example, an ideal 8-bit unsigned quantizer has:

$$
2^8=256
$$

possible output codes.

## **5.5 Analog-to-Digital Conversion (ADC)**

**Core DSP Concept**

**Analog-to-Digital Conversion (ADC) is the process of converting an analog signal into a digital representation.**

A microphone captures sound and produces an analog electrical signal. An ADC converts that electrical signal into numerical samples.

![Audio Sampling](https://images.openai.com/static-rsc-4/ZTBG7LGxAszDe-8BHYifQJtfmQPWG-WOp0Ua0dysbCnI0GEqmuDPhbZcMsj1qN1J1YPc-AEdzo8Q470yu-WfHc0nopyn_4Y7BfyYH0waUOr3vuWenQPKjtf_OVirvZepAR5PIAtjs_j8uKCkrxfNZ_GxdbnLZBdoVPpY4IYlxYI?purpose=inline)

**Reference:** [Sample Rate Explained](https://acestudio.ai/blog/what-is-audio-sample-rate/)

### Main Stages of ADC

1. Anti-aliasing filtering.
2. Sampling.
3. Quantization.
4. Binary encoding.

Let's understand these operations individually.

### Step 1 — Anti-Aliasing Filtering

The analog input is filtered to remove or attenuate frequency components above the intended frequency limit before sampling.

### Step 2 — Sampling

The continuous-time signal is measured at regular time intervals.

### Step 3 — Quantization

Each sampled amplitude is mapped to one of a finite number of available levels.

### Step 4 — Encoding

Each quantized level is represented using a binary code.

The output is digital audio data.

## **5.6 Digital-to-Analog Conversion (DAC)**

**Digital-to-Analog Conversion (DAC) converts digital numerical samples into an analog electrical signal.**

### Why Do We Need DAC?

Because speakers and headphones ultimately require an electrical signal to drive their transducers.

![DSP System Block Diagram](https://images.openai.com/static-rsc-4/cHuKkFfYNd-EykQNUHORwZdvv0aJFByMqwwlguZpuIdoONVR3lgiyG9U6b5SDdMkmAoer7PtcfMjn4qJaALZIPLNU6Rf3z2poLk4p8Ufxv8erHEuNiIjaX2tJnmiGxfLTVt4sz-uCGSLrWauWZYXUFnVO_ELM48PaGaok0JjeiE?purpose=inline)

**Reference:** [Understanding the DSP System Block Diagram](https://elecsprout.com/block-diagram-of-dsp-system.html)

### DAC Process

1. Receives digital audio samples.
2. Converts numerical codes into corresponding analog signal levels.
3. Uses reconstruction filtering to smooth the output.
4. Produces an analog electrical signal.
5. Drives a speaker or headphone transducer.

### ADC vs DAC

| ADC | DAC |
|---|---|
| Analog to digital | Digital to analog |
| Used during recording | Used during playback |
| Produces numerical samples | Produces an analog electrical signal |
| Sampling and quantization | Reconstruction and filtering |

### Practical Example

**When you record your voice:**

Microphone → Analog electrical signal → ADC → Digital audio file.

**When you play the recording:**

Digital audio file → DAC → Amplifier → Speaker → Sound waves.

## **5.7 Sampling**

**Sampling is the process of measuring a continuous-time signal at specific, equally spaced time intervals.**

Imagine recording a person's voice. The analog waveform changes continuously, but the ADC measures its amplitude at regular intervals.

Each measurement is called a **sample**.

![Audio Sampling Illustration](https://images.openai.com/static-rsc-4/pZQaXZ-wV0X3NOpwIBc09TKzun8kCBXPJOwgjFIuQKNDJr5CICscvdmhZaQM-Z1WdRt4CZlVLCUoHSDmPfWP99nLuuAgNavQ3CL-Psa8kjbq7FOOT9aHRKLKi60F68e4i_q-W7HtQAmMbtWHc-7GhUpzo4K1UTt5iNKtCF1_6IM?purpose=inline)

**Reference:** [Why Is Audio Recorded at 24-bit/48 kHz?](https://note.com/tairamen/n/naad63fc7f282)

### Sampling Frequency

The sampling frequency (also called sampling rate) is the number of samples taken per second.

It is represented by:

$$
f_s
$$

Its unit is Hertz (Hz) or samples per second.

### Examples

| Sampling Rate | Meaning |
|---|---|
| 8,000 Hz | 8,000 samples per second |
| 16,000 Hz | 16,000 samples per second |
| 44,100 Hz | 44,100 samples per second |
| 48,000 Hz | 48,000 samples per second |

### Sampling Interval

The time between two consecutive samples is called the sampling interval.

$$
\boxed{T_s=\frac{1}{f_s}}
$$

Where:

- $T_s$ = sampling interval in seconds.
- $f_s$ = sampling frequency in Hz.

**Example:**

Suppose an audio signal is sampled at 16,000 Hz.

$$
T_s=\frac{1}{16000}
$$

$$
\boxed{T_s=62.5\ \mu s}
$$

Therefore, one sample is taken every 62.5 microseconds.

### Sampling a Speech Signal

Suppose you record 1 second of speech at 16 kHz.

$$
N=f_s\times t
$$

$$
N=16000\times1
$$

$$
\boxed{N=16000\text{ samples}}
$$

For a 5-second recording:

$$
N=16000\times5=80000
$$

You get 80,000 samples for one audio channel.

## **5.8 Nyquist-Shannon Sampling Theorem**

**Fundamental DSP Concept**

The Nyquist-Shannon sampling theorem explains the minimum sampling rate needed to reconstruct a band-limited signal from its samples.

**The theorem states that a band-limited signal can be perfectly reconstructed from its samples when the sampling frequency is greater than twice its highest frequency component, under ideal conditions.**

Mathematically:

$$
\boxed{f_s>2f_{\max}}
$$

Where:

- $f_s$ = sampling frequency.
- $f_{\max}$ = highest frequency present in the band-limited input signal.

The minimum theoretical boundary is:

$$
f_s=2f_{\max}
$$

This boundary is called the **Nyquist rate**, although practical systems generally use a higher sampling rate.

### Example 1

Suppose the highest frequency in a signal is 4 kHz.

What sampling frequency should we use?

$$
f_s>2\times4000
$$

$$
\boxed{f_s>8000\text{ Hz}}
$$

A practical choice could be 16 kHz.

### Example 2 — Speech Recording

Suppose the frequency range we want to preserve extends up to 7 kHz.

$$
f_s>2\times7000
$$

$$
f_s>14000\text{ Hz}
$$

A 16 kHz sampling rate can theoretically accommodate this band, provided the analog input is appropriately band-limited.

![Sampling Theorem](https://images.openai.com/static-rsc-4/OG7WghdDhN3SAJ2hp1gfbxDEYtir4QSwbcMIlqEqvS-p0-tVBQlqUWL-_UzHFFwZFGAtAv9FqrKn-JCPQz8xrWOHhlesVQwc54idADHjHv7UzbRt6qVWuFMNLKe0nQ8XrCaI_WBREmknX8bMrlrll4Yc8GpKNEIe6NwsvH-h4HI?purpose=inline)

**Reference:** [Understanding Digital Audio: Sampling, Quantization, and More](https://www.theaudiopod.com/blog/xank7qi4qjlgxsltcwz3wdkzjdnagy)

### Why Is This Important in Speech Processing?

Many speech-recognition systems use 16 kHz audio.

The Nyquist frequency at this sampling rate is:

$$
f_N=\frac{16000}{2}
$$

$$
\boxed{f_N=8000\text{ Hz}}
$$

This means the theoretically representable frequency band extends up to 8 kHz, assuming appropriate anti-aliasing and practical filter design.

## **5.9 Nyquist Frequency**

The **Nyquist frequency is half the sampling frequency**.

$$
\boxed{f_N=\frac{f_s}{2}}
$$

It represents the highest frequency that can theoretically be represented without aliasing in an ideally band-limited sampled system.

### Examples

| Sampling Frequency | Nyquist Frequency |
|---|---|
| 8 kHz | 4 kHz |
| 16 kHz | 8 kHz |
| 22.05 kHz | 11.025 kHz |
| 44.1 kHz | 22.05 kHz |
| 48 kHz | 24 kHz |

### Important Distinction

- Nyquist rate: twice the highest input frequency.
- Nyquist frequency: half the sampling frequency.

Don't confuse these two terms.

## **5.10 What Is Aliasing?**

**Aliasing is a distortion that occurs when a signal is sampled at an insufficient rate, causing different frequencies to become indistinguishable in the sampled representation.**

In simple terms, the sampling process can make a high-frequency signal appear to have a lower frequency.

This happens when the input contains frequency components above the Nyquist frequency.

![Aliasing Illustration](https://images.openai.com/static-rsc-4/EjJZA2Nan3f7RjdesmYbetv4cOvK0mjWCvyKSOMogLfZ5n7-Si3kLlMBPQ4qGmWIjcWuIxM9HLvI-1OgDqec3iRpuUvPyouOY3LhO_G5ca711TL3ZSMBoJdtQ_Oizo8-g2ctGW2w5VRlFu_hAtLEqBXbfgeizBn1xNjRYtCxbsM?purpose=inline)

**Reference:** [Speech Representations: From Waveforms to Mel Spectrograms](https://mbrenndoerfer.com/writing/speech-representations-mel-spectrograms)

### Example

Suppose:

- Input signal frequency = 7 kHz.
- Sampling frequency = 8 kHz.

The Nyquist frequency is:

$$
f_N=\frac{8000}{2}=4000\text{ Hz}
$$

The input frequency (7 kHz) is above the Nyquist frequency (4 kHz).

It will therefore alias to a lower frequency.

For this example:

$$
f_{\text{alias}}=|7000-8000|
$$

$$
\boxed{f_{\text{alias}}=1000\text{ Hz}}
$$

The 7 kHz input can appear as a 1 kHz component in the sampled signal.

### Why Is Aliasing Dangerous?

- It introduces incorrect frequency information.
- It can distort speech characteristics.
- It affects pitch and spectral analysis.
- It can damage the quality of audio recordings.
- It can mislead downstream machine-learning models.

**Important:** Once different frequencies have become indistinguishable due to aliasing, simply increasing the sampling rate afterwards cannot reliably recover the original signal.

## **5.11 Anti-Aliasing Filters**

**Important in Real-World ADC Systems**

An **anti-aliasing filter is an analog low-pass filter placed before sampling to attenuate frequency components that could cause aliasing.**

### Why Do We Need It?

Real-world audio signals may contain frequency components above the Nyquist frequency.

Before sampling, we must limit the input bandwidth.

![Anti-Aliasing Filter](https://images.openai.com/static-rsc-4/zHcc6yQzpAOPo4afrJGTEZdZmqKQ3w2FH5s0hX9B3Aj_MYGRk6gQnrQtWA0vp8K45GiXhab8jR1nYl2eOABP_IqIaytJONsNtdEPBuks1SqOw4Ws7U-vRtFdIaJh55vs4ks0n2HAirdHzRK1W7iaCEDFKmIjPVx9FIt8lzEPmJM?purpose=inline)

**Reference:** [What Is an Anti-Aliasing Filter?](https://mmjal.com/BLOG/?bmode=view&idx=170355275)

### How Does It Work?

1. The microphone produces an analog electrical signal.
2. The signal passes through an analog low-pass filter.
3. The filter attenuates unwanted high-frequency components.
4. The filtered signal enters the ADC.
5. The ADC samples the signal.

### Example

Suppose we want to record speech at 16 kHz.

The Nyquist frequency is:

$$
f_N=8\text{ kHz}
$$

The anti-aliasing filter should attenuate frequencies above the intended passband, with its transition band designed to provide sufficient attenuation before frequencies that would alias into the desired band.

A practical filter cannot change from fully passing to completely rejecting frequencies at exactly 8 kHz. Filter design requires a transition band.

### Why Not Use a Digital Filter After Sampling?

Because aliasing occurs during sampling.

Once unwanted high-frequency components have folded into lower frequencies, a digital filter cannot generally distinguish them from genuine low-frequency content.

Therefore, anti-aliasing filtering must happen before the ADC sampling operation.

## **5.12 Quantization**

Sampling makes time discrete, but the sampled amplitudes can still have values across a continuous range.

We need another operation to convert these amplitudes into finite numerical levels.

This process is called **quantization**.

**Quantization maps each sampled amplitude to the nearest available discrete amplitude level.**

![Digital Signal Quantization](https://images.openai.com/static-rsc-4/A-WW35lAZ0hM9dT6k6SfYcWqynzMG_DuqRiFQiA0269J3dxH407mycVFTAiHD_HwfNdpg2r05Tw0xog1DSynbWonIpzC8F2N0gZUNCWtnUAGdEEOch8NJ5ziWX_JSRYS3F2SdCFZ-uDMUkdx5xNJNDMZp0Z_xFYLqkMly208f-k?purpose=inline)

**Reference:** [AudioLabs Erlangen — Digital Signal Quantization](https://www.audiolabs-erlangen.de/resources/MIR/FMP/C2/C2S2_DigitalSignalQuantization.html)

### Example

Suppose an ADC uses 3-bit quantization.

The number of available levels is:

$$
L=2^B
$$

Where:

- $L$ = number of quantization levels.
- $B$ = bit depth.

For 3-bit quantization:

$$
L=2^3=8
$$

Therefore, the ADC can represent eight different quantization codes.

### Quantization Levels at Different Bit Depths

| Bit Depth | Number of Levels |
|---|---|
| 1 bit | 2 |
| 2 bits | 4 |
| 3 bits | 8 |
| 8 bits | 256 |
| 16 bits | 65,536 |
| 24 bits | 16,777,216 |

Higher bit depth provides more available amplitude levels and generally reduces quantization error.

---

## **5.13 Quantization Error**

Quantization cannot always represent the exact original sampled amplitude.

The difference between the original sample and its quantized value is called **quantization error**.

$$
\boxed{e[n]=x[n]-x_q[n]}
$$

Where:

- $x[n]$ = original sampled amplitude.
- $x_q[n]$ = quantized amplitude.
- $e[n]$ = quantization error.

### Example

Suppose:

Original sampled value:

$$
x[n]=0.73
$$

Quantized value:

$$
x_q[n]=0.75
$$

Then:

$$
e[n]=0.73-0.75
$$

$$
\boxed{e[n]=-0.02}
$$

The quantized value is not exactly equal to the original sample.

![Quantization Error](https://images.openai.com/static-rsc-4/7VzZq3PEfv9RXhznP3yFpvQ3aDwaXElQFJoCor10NrfyJ3NkPtl-4i3np_dxPsy1hFbdBZCAw7ZrgChh7l4QwJ02dS0X5xDrbUnnJT8B6SuncYBGkoaR8Zx3M8tzCTU3UEQzQZakiuJdDND1i2nlzoXW6UUy040AtaQD0MeXvh4?purpose=inline)

**Reference:** [Digital Communication — Quantization](https://www.tutorialspoint.com/digital_communication/digital_communication_quantization.htm)

### Quantization Step Size

For a uniform quantizer, the quantization step size is the amplitude difference between adjacent quantization levels.

For an ideal uniform quantizer spanning a range from $V_{\min}$ to $V_{\max}$:

$$
\boxed{\Delta=\frac{V_{\max}-V_{\min}}{L}}
$$

Where:

- $\Delta$ = quantization step size.
- $L$ = number of quantization intervals.

For rounding to the nearest level, the ideal quantization error is approximately bounded by:

$$
\boxed{-\frac{\Delta}{2}\leq e[n]\leq\frac{\Delta}{2}}
$$

This assumes the input is within the quantizer's supported range and there is no clipping.

**Important:** If the input exceeds the ADC's permitted range, clipping or overload distortion can occur. This is different from ordinary quantization error.

## **5.14 Quantization Noise**

Quantization error is the error associated with individual samples.

When viewed across a signal, the resulting quantization error can be treated as noise under suitable assumptions. This is called **quantization noise**.

![Audio Waveform and Noise](https://images.openai.com/static-rsc-4/-P7ZZF_pCiAU2O8SxjwZHe82stPvTcJmJqX41glm_Z23sSxC66z08F3caAd1fmu8hx5WKG5FdZh7eF0Rxckrz6nxMLflPXFo9EmuG6WZQaUhqQPPG8qeYhlTY7s-i0Crq4MeY9Q6DNv2oGhUjkRO6bSGODCVeLD8qcHo1x3sRMA?purpose=inline)

**Reference:** [Audio Is Scary #2 — Waveforms and Terms](https://www.bluelabpro.com/blog/audio2)

### What Happens When Bit Depth Increases?

Consider two ADC systems:

- System A: 8-bit ADC.
- System B: 16-bit ADC.

| 8-bit ADC | 16-bit ADC |
|---|---|
| 256 levels | 65,536 levels |
| Larger quantization steps | Smaller quantization steps |
| Generally higher quantization error | Generally lower quantization error |
| Lower theoretical quantization SNR | Higher theoretical quantization SNR |

For an ideal uniform quantizer with a full-scale sinusoidal input, the approximate signal-to-quantization-noise ratio is:

$$
\boxed{SQNR\approx6.02B+1.76\text{ dB}}
$$

Where $B$ is the number of bits.

### Example — 16-bit Audio

$$
SQNR\approx6.02(16)+1.76
$$

$$
\boxed{SQNR\approx98.08\text{ dB}}
$$

This is an ideal theoretical result, not a guarantee of the actual performance of every 16-bit recording system.

### Why Does This Matter in Speech Processing?

Quantization quality can influence:

- Recording quality.
- Low-level speech details.
- Background noise representation.
- Audio signal-to-noise ratio.
- Performance of downstream signal-processing systems.

Let's understand the complete audio journey.

### Recording

1. A person speaks.
2. Sound waves reach the microphone.
3. The microphone converts acoustic pressure variations into an analog electrical signal.
4. An anti-aliasing filter limits the input bandwidth.
5. The ADC samples and quantizes the signal.
6. Binary audio data is stored in a digital file.

### Playback

1. The computer reads the digital audio data.
2. The DAC converts the digital samples into an analog electrical signal.
3. A reconstruction filter smooths the output.
4. An amplifier drives the speaker.
5. The speaker converts electrical energy into sound waves.

## **5.17 Connecting This Module to Python and AI/ML**

Now let's see why these concepts matter when you start working with real audio datasets.

When you load a WAV file using Python, you typically obtain:

- Audio samples.
- Sampling rate.
- Number of channels.
- Data type and amplitude representation.

### Python Example

```python
import soundfile as sf

audio, sample_rate = sf.read("speech.wav")

print("Sampling Rate:", sample_rate)
print("Audio Shape:", audio.shape)
print("Data Type:", audio.dtype)
```

### Possible Output

```text
Sampling Rate: 16000
Audio Shape: (16000,)
Data Type: float64
```

This means:

- The audio contains 16,000 samples.
- It was sampled at 16 kHz.
- It contains one second of mono audio.
- Its samples are represented as floating-point values.

The actual output depends on the audio file and library configuration.

**Important:** A floating-point WAV representation does not necessarily mean the original recording was captured using a floating-point ADC. The file may have been converted from integer PCM.

### Why These Concepts Matter for AI Models

| Concept | Practical Relevance |
|---|---|
| Sampling rate | Determines the time resolution and available frequency bandwidth |
| Quantization | Affects amplitude precision |
| Bit depth | Influences quantization noise and dynamic range |
| Aliasing | Can introduce misleading frequency information |
| Anti-aliasing | Prevents unwanted frequency folding before sampling |
| ADC | Converts recorded sound into data |
| DAC | Converts digital audio back into a playable signal |

## **What You Should Remember 🧠**

1. Analog signals vary continuously in time and amplitude.
2. Digital audio represents sound using numerical samples.
3. Sampling discretizes time; quantization discretizes amplitude.
4. ADC converts analog signals into digital representations, while DAC performs the reverse conversion.
5. The Nyquist-Shannon theorem requires a sampling frequency greater than twice the highest frequency in a band-limited signal for ideal reconstruction.
6. Aliasing occurs when frequency components above the Nyquist frequency fold into lower frequencies.
7. Anti-aliasing filters attenuate unwanted high-frequency components before sampling.
8. Higher bit depth provides more quantization levels and generally reduces quantization noise.
