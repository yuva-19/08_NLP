# **A. SAMPLING RATE**

## **6.1 What is Sampling Rate?**

**Sampling rate is the number of audio samples captured per second from a continuous-time signal.**

It is represented by:

$$
f_s
$$

Its unit is Hertz (Hz), or samples per second.

For example:

- 8,000 Hz = 8,000 samples per second.
- 16,000 Hz = 16,000 samples per second.
- 44,100 Hz = 44,100 samples per second.

![Digital Sound Quality: Challenges & Solutions - Apptitude](https://images.openai.com/static-rsc-4/GsS2KczfXWdT3V8WERFfHGZYBZry9EJC4yBXpB6kYqAx0R86hzpNuTdghs2Uzf0FOkkKkq3Bi6lEKQFaBKDSjJ-0td6I4gxrVHJFDWe9UPAF9vS8T8umUSD14AzsPv_0X9AjlwUSngBi18mskOIMiY3hUSk_VsGm6w1Q3nVsy04?purpose=inline)

[Reference: Apptitude](https://apptitude.ch/en/development/digital-sound-quality-challenge-and-solutions/?utm_source=chatgpt.com)

### **Sampling rate vs Sampling frequency**

These terms are generally used interchangeably in digital audio.

For example:

$$
f_s=16000\text{ Hz}=16\text{ kHz}
$$

This means the system records 16,000 samples every second.

### **Samples per second**

The number of samples depends on both the sampling rate and recording duration.

$$
\boxed{N=f_s\times T}
$$

Where:

- $N$ = number of samples.
- $f_s$ = sampling rate.
- $T$ = duration in seconds.

**Example:**

A 10-second speech recording sampled at 16 kHz:

$$
N=16000\times10
$$

$$
\boxed{N=160000\text{ samples}}
$$

This is for one channel.

## **6.2 Nyquist Frequency**

The Nyquist frequency is half the sampling rate.

$$
\boxed{f_N=\frac{f_s}{2}}
$$

It represents the highest frequency that can theoretically be represented without aliasing in an ideally band-limited sampled system.

| Sampling rate | Nyquist frequency |
|---|---|
| 8 kHz | 4 kHz |
| 16 kHz | 8 kHz |
| 22.05 kHz | 11.025 kHz |
| 44.1 kHz | 22.05 kHz |
| 48 kHz | 24 kHz |

**Remember:** A 16 kHz recording does not preserve frequencies all the way up to 16 kHz. Its Nyquist frequency is 8 kHz.

## **6.3 Common Sampling Rates**

Different applications use different sampling rates depending on their requirements.

**8 kHz — Telephone-quality speech**

$$
\boxed{8000\text{ samples/s}}
$$

Nyquist frequency: 4 kHz

Commonly associated with traditional narrowband telephone audio.

**16 kHz — Speech processing**

$$
\boxed{16000\text{ samples/s}}
$$

Nyquist frequency: 8 kHz

Widely used in speech-recognition and speech-processing systems.

**22.05 kHz — Audio and music**

$$
\boxed{22050\text{ samples/s}}
$$

Nyquist frequency: 11.025 kHz

Used in some audio and music applications.

**44.1 kHz — CD audio**

$$
\boxed{44100\text{ samples/s}}
$$

Nyquist frequency: 22.05 kHz

Standard sampling rate for audio CDs.

**48 kHz — Video and professional audio**

$$
\boxed{48000\text{ samples/s}}
$$

Nyquist frequency: 24 kHz

Commonly used in video production, broadcasting, and professional audio workflows.

## **6.4 Why Do Speech Applications Commonly Use 16 kHz?**

This is an important practical question for your AI/ML projects.

Many speech-processing systems use 16 kHz because it offers a useful balance between:

1. Speech bandwidth.
2. Computational requirements.
3. Storage requirements.
4. Model compatibility.

### **Reason 1 — Sufficient speech information for many applications**

A 16 kHz sampling rate has a Nyquist frequency of 8 kHz.

This can preserve a useful portion of the speech spectrum for many speech-recognition applications.

### **Reason 2 — Lower computational requirements**

Compare two recordings of the same duration:

- 16 kHz → 16,000 samples per second.
- 48 kHz → 48,000 samples per second.

The 48 kHz recording has three times as many samples per second per channel.

Processing more samples generally requires more memory and computation.

### **Reason 3 — Model compatibility**

Many pretrained speech models specify an expected sampling rate, such as 16 kHz.

If a model expects 16 kHz audio, you should prepare your input accordingly.

**Important:** 16 kHz is not universally optimal. Some models require 8 kHz, 22.05 kHz, 44.1 kHz, or 48 kHz. Always check the model's expected input configuration.

## **6.5 Changing a Sampling-Rate Label vs Actual Resampling**

<text> </text>

**Common audio-processing mistake**

These are two completely different operations.

### **A. Changing the sampling-rate label**

Suppose you have a recording with:

- 44,100 samples per second.
- Duration = 1 second.
- Actual sampling rate = 44.1 kHz.

If you simply change the metadata to 16 kHz without modifying the samples, the recording is being interpreted at the wrong rate.

The playback duration becomes:

$$
T=\frac{44100}{16000}
$$

$$
\boxed{T=2.75625\text{ seconds}}
$$

The audio will play more slowly and its pitch will be lower.

This is not proper resampling.

### **B. Actual resampling**

Resampling converts audio from one sampling rate to another by performing appropriate signal-processing operations.

For example:

$$
44.1\text{ kHz}\rightarrow16\text{ kHz}
$$

The resampling process produces a new sequence of samples representing approximately the same underlying signal and duration.

![How to Change Sample Rate in Sound Forge Pro](https://images.openai.com/static-rsc-4/0wSLPBZtGl6bwy4y2PHR60U7vlq4sdFn6tF58fC_EmFschCtloMAa27FpJWfAcXIh4E_dwb40WogKWc0rkBRrccv3tjmG7bTygZRZhMKpWBJmdj1cWBUap7RP492nS5eeLq0QWH8RllsIkVJ9S9c69anq_UEx5I7oKz5h7zJfFg?purpose=inline)

[Reference: Sound Forge Pro](https://www.soundforgepro.com/how-to-change-sample-rate-sound-forge-pro/?utm_source=chatgpt.com)

### **Comparison**

| Changing label | Actual resampling |
|---|---|
| Changes the interpretation of sample timing | Changes the sample sequence |
| Does not calculate new samples | Calculates a new sampling representation |
| Alters playback speed and pitch if interpreted at the new rate | Preserves intended duration and pitch, subject to filtering and numerical limitations |
| Not suitable for model preprocessing | Used when matching model input requirements |

**AI/ML example:** If your pretrained ASR model expects 16 kHz, a 44.1 kHz recording must be properly resampled to 16 kHz rather than simply relabelled.

# **B. BIT DEPTH**

## **6.6 What is Bit Depth?**

Bit depth specifies the number of bits used to represent each digital audio sample in a fixed-bit-depth PCM representation.

It determines how many quantization levels are available.

The number of levels is:

$$
\boxed{L=2^B}
$$

Where:

- $L$ = quantization levels.
- $B$ = bit depth.

For example:

$$
B=16
$$

$$
L=2^{16}=65536
$$

Therefore, 16-bit PCM provides 65,536 possible codes.

[Reference: Teledynamics](https://info.teledynamics.com/blog/demystifying-codecs-part-1-digitizing-the-human-voice?utm_source=chatgpt.com)

## **6.7 Common Bit Depths**

### **8-bit audio**

$$
2^8=256\text{ levels}
$$

- Provides relatively low amplitude resolution.
- Has higher quantization noise than higher-bit-depth PCM, under comparable conditions.
- Common in older or low-quality audio applications.

### **16-bit audio**

$$
2^{16}=65,536\text{ levels}
$$

- Commonly used for CD-quality PCM audio.
- Provides good dynamic range for many recording and playback applications.
- Widely supported by audio software.

### **24-bit audio**

$$
2^{24}=16,777,216\text{ levels}
$$

- Provides much finer amplitude resolution.
- Common in professional recording and audio production.
- Offers greater theoretical dynamic range and more headroom for recording workflows.

| Bit depth | Quantization levels | Approximate ideal quantization SNR |
|---|---:|---:|
| 8-bit | 256 | 49.9 dB |
| 16-bit | 65,536 | 98.1 dB |
| 24-bit | 16,777,216 | 146.2 dB |

The SNR figures use the ideal full-scale sine-wave approximation:

$$
SQNR\approx6.02B+1.76\text{ dB}
$$

Real-world audio equipment has additional noise sources, so actual performance is lower.

## **6.8 Dynamic Range**

**Dynamic range is the range between the quietest and loudest signals that a system can represent or reproduce under specified conditions.**

In digital audio, bit depth strongly influences the theoretical dynamic range.

For ideal linear PCM, the approximate relationship is:

$$
\boxed{DR\approx6.02B+1.76\text{ dB}}
$$

Where $B$ is the bit depth.

For example, ideal 16-bit PCM has an approximate full-scale sine-wave quantization SNR of 98 dB.

### **Why is dynamic range important?**

Imagine recording someone speaking:

- The person whispers.
- Then speaks normally.
- Then shouts.

A recording system needs enough dynamic range to represent both quiet and loud portions without excessive noise or clipping.

A higher bit depth can help preserve low-level details while providing more room to record without quantization becoming a major limitation.

## **6.9 Bit Depth and Quantization Noise**

When a sample is mapped to the nearest available digital amplitude level, a small difference can occur between the original and represented values.

This difference produces quantization error.

Under suitable assumptions, the accumulated effect is described as quantization noise.

![Sampling rate & bit depth explained in simple terms](https://images.openai.com/static-rsc-4/4rxPMiTHYBJeaNGQ7iDbjNNZRPAZkO1jcOdEGX2f4WebL6GVUDHUzvNvOYI9GWIaIQH1muW_OPd-edGtR6p_x11cTyR498TzvNT_6Gb3WbUGz0QSZr1Oic3kNS320iNIl4MsBJkg9G5mCl1Ssp8MTTqmIb-wBxGY3NvcTPGeKIc?purpose=inline)

[Reference: Wood and Fire Studio](https://woodandfirestudio.com/en/sample-rate-bit-depth/?utm_source=chatgpt.com)

General relationship:

- Lower bit depth → fewer quantization levels.
- Fewer levels → larger quantization steps.
- Larger steps → potentially greater quantization error.
- Higher bit depth → finer amplitude representation.

**Important:** Increasing bit depth does not remove microphone noise, background noise, or room noise. It reduces the quantization limitations of the digital representation.

The number of channels determines how many separate audio signals are stored for each sampling instant.

**Stereo audio contains two separate channels: left and right.**

For example:

- Left channel may contain more sound from a source on the left.
- Right channel may contain more sound from a source on the right.

A stereo recording can be represented as:

$$

$$

Where:

- $x_R[n]$ = right-channel sample.

For a 16 kHz stereo recording, the total number of samples per second is:

$$
16000\times2=32000
$$

## **6.14 Channel Conversion**

In speech processing, audio datasets may contain mono or stereo recordings.

Sometimes a pretrained model expects mono audio.

You may therefore need to convert stereo audio into mono.

A simple averaging approach is:

$$
\boxed{x_M[n]=\frac{x_L[n]+x_R[n]}{2}}
$$

Where:

- $x_M[n]$ = mono sample.
- $x_L[n]$ = left-channel sample.
- $x_R[n]$ = right-channel sample.

[Reference: Learning Audacity](https://rgervacio.com/learning-audacity/managing-a-project/convert-a-mono-or-stereo-track/?utm_source=chatgpt.com)

### **Important practical consideration**

Averaging channels is not always the best choice.

If the channels contain phase differences, averaging can cause partial cancellation of some components.

For speech datasets, inspect the channels before conversion, particularly when using microphone arrays or recordings with multiple speakers.

# **D. OTHER DIGITAL AUDIO PARAMETERS**

## **6.15 Audio Duration**

**Audio duration is the total time represented by an audio recording.**

It is measured in seconds, minutes, or hours.

For a recording with a known number of samples:

$$
\boxed{T=\frac{N}{f_s}}
$$

Where:

- $T$ = duration in seconds.
- $N$ = number of samples per channel.
- $f_s$ = sampling rate.

### **Example**

Suppose a mono audio file contains 160,000 samples and has a sampling rate of 16 kHz.

$$
T=\frac{160000}{16000}
$$

$$
\boxed{T=10\text{ seconds}}
$$

For multichannel audio, use the number of samples in each channel, not the total number of sample values across all channels.

## **6.16 Bitrate**

**Bitrate is the amount of audio data represented or transmitted per second.**

It is generally measured in:

- bits per second (bps).
- kilobits per second (kbps).
- megabits per second (Mbps).

For uncompressed PCM audio:

$$
\boxed{R=f_s\times B\times C}
$$

Where:

- $R$ = bitrate in bits per second.
- $f_s$ = sampling rate.
- $B$ = bit depth.
- $C$ = number of channels.

### **Example**

Consider a recording with:

- Sampling rate = 16,000 Hz.
- Bit depth = 16 bits.
- Channels = 1.

$$
R=16000\times16\times1
$$

$$
\boxed{R=256000\text{ bps}=256\text{ kbps}}
$$

Now consider stereo audio with the same sampling rate and bit depth:

$$
R=16000\times16\times2
$$

$$
\boxed{R=512\text{ kbps}}
$$

Stereo requires twice the uncompressed PCM data rate of mono under these conditions.

## **6.17 Audio File Size**

Audio file size depends on:

- Sampling rate.
- Bit depth.
- Number of channels.
- Duration.
- Compression method.
- File headers and metadata.

For uncompressed PCM audio:

$$
\boxed{
\text{File Size (bytes)}
=
\frac{f_s\times B\times C\times T}{8}
}
$$

### **Example — 10-second speech recording**

Given:

| Parameter | Value |
|---|---:|
| Sampling rate | 16,000 Hz |
| Bit depth | 16 bits |
| Channels | 1 |
| Duration | 10 seconds |

Substitute:

$$
\text{Size}
=
\frac{16000\times16\times1\times10}{8}
$$

$$
=320000\text{ bytes}
$$

$$
\boxed{\text{Size}=320\text{ kB}}
$$

This is the theoretical PCM sample-data size, excluding file headers and metadata.

### **Python calculation**

```python
sample_rate = 16000
bit_depth = 16
channels = 1
duration = 10

file_size_bytes = (
    sample_rate * bit_depth * channels * duration
) / 8

print("File size:", file_size_bytes, "bytes")
```

## **6.27 Practical Python — Inspecting Digital Audio Parameters**

Now let's apply these concepts.

We'll use Python's `soundfile` library to inspect an audio file.

```python
import soundfile as sf

audio, sample_rate = sf.read("speech.wav")

# Basic audio information
print("Sampling rate:", sample_rate)
print("Audio shape:", audio.shape)
print("Data type:", audio.dtype)

# Number of channels
if audio.ndim == 1:
    channels = 1
else:
    channels = audio.shape[1]

# Number of samples per channel
num_samples = len(audio)

# Duration
duration = num_samples / sample_rate

print("Channels:", channels)
print("Samples per channel:", num_samples)
print("Duration:", duration, "seconds")
```

### **Calculate uncompressed PCM data size**

```python
bit_depth = 16  # Assumed PCM bit depth

file_size = (
    sample_rate * bit_depth * channels * duration
) / 8

print("Estimated PCM data size:", file_size, "bytes")
```

**Important:** The bit depth in this calculation is assumed. A floating-point array returned by `soundfile.read()` does not necessarily reveal the original file's PCM bit depth.

You can inspect the file's subtype:

```python
info = sf.info("speech.wav")

print("Format:", info.format)
print("Subtype:", info.subtype)
print("Channels:", info.channels)
print("Sample rate:", info.samplerate)
print("Duration:", info.duration)
```

This is the kind of inspection you'll perform when preparing datasets for speech-recognition and emotion-classification projects.

## **What you should remember 🧠**

1. Sampling rate determines the number of samples captured per second; 16 kHz is widely used in speech processing.
2. Changing a sampling-rate label is not resampling — actual resampling generates a new sample sequence.
3. Bit depth determines the number of available quantization levels and influences theoretical dynamic range.
4. Mono contains one channel; stereo contains left and right channels.
5. Uncompressed PCM file size depends on sampling rate, bit depth, channels, and duration.
6. Lossless compression preserves the original digital samples; lossy compression permanently discards information.
7. WAV is a container, PCM is a digital audio representation, and MP3 is a lossy audio codec.
