## **1. Introduction to Audio & Speech Processing 🎧**

Let's build the foundation first. Before learning Speech Recognition, Speech Emotion Recognition (SER), or speaker identification, you need to understand what sound actually is, how computers represent it, and how speech-processing systems work.

## **1.1 What is Sound?**

**Sound is a mechanical wave produced by vibrations that travel through a medium**, such as air, water, or solids.

For example, when you speak:

1. Air from your lungs passes through your vocal cords.
2. Your vocal cords vibrate.
3. These vibrations disturb nearby air molecules.
4. The disturbance travels through the air as sound waves.
5. Your ears detect these waves, and your brain interprets them.

### **Important properties of sound**

| Property | Meaning |
|---|---|
| Amplitude | Relates to the strength of a sound wave; perceived as loudness |
| Frequency | Number of cycles per second, measured in Hertz (Hz) |
| Wavelength | Distance covered by one complete wave cycle |
| Phase | Position of a wave within its cycle |
| Timbre | Quality that helps distinguish sounds with similar pitch and loudness |

### **Frequency and amplitude**

![Frequency and amplitude representation of sound waves](https://images.openai.com/static-rsc-4/dax1cJ2lJV9H9ufoUcTVteOR6ufTr_pxWp7SY_lvJQeMs7DjPCdjT_Pb82sx8_NLttBjFEqHPfGQvBdx2Xd_l9GmyDfD53WoZc6AbbkjIRL8I25ysKASxuwDHC5QJlJ9qWpoWPya4r1N0g_jRCCDLmKdkmduGf04qbaysWR26FM?purpose=inline)

[Source: teachComputing](https://www.teachwithict.com/binary-representation-of-sound1.html?utm_source=chatgpt.com)

**Frequency:** Determines the physical pitch of a periodic sound.

- Low frequency → low-pitched sound.
- High frequency → high-pitched sound.

**Amplitude:** Relates to the strength of the wave.

- Higher amplitude → generally louder sound.
- Lower amplitude → generally quieter sound.

The relationship between frequency and period is:

$$
f = 1 // T
$$

Where:

- \(f\) = frequency in Hertz.
- \(T\) = time period in seconds.

## **1.2 What is Audio?**

**Audio is the representation of sound in a form that can be recorded, stored, transmitted, processed, or played back.**

Audio can contain:

- Human speech
- Music
- Environmental sounds
- Animal sounds
- Mechanical sounds
- Noise

For example, a recording of a person speaking, a song, and a car engine are all audio signals.

### **Audio vs Sound**

| Sound | Audio |
|---|---|
| Physical phenomenon | Representation of sound |
| Exists as vibrations in a medium | Can be stored as a signal or digital file |
| Detected by ears or microphones | Processed by computers and audio devices |

A microphone converts sound waves into an electrical signal. An audio recording system can then convert that signal into digital data.

## **1.3 What is Speech?**

**Speech is the vocal production of language by humans.**

It is a specialized type of audio containing linguistic information.

Speech is produced using several parts of the human vocal system:

| Component | Function |
|---|---|
| Lungs | Provide airflow |
| Vocal cords | Produce voiced sounds through vibration |
| Tongue | Shapes sounds and pronunciation |
| Lips | Help produce sounds such as /p/, /b/, and /m/ |
| Teeth | Assist in articulation |
| Nasal cavity | Contributes to nasal sounds |

### **Example**

Consider the spoken sentence:

> "I am learning Artificial Intelligence."

This speech signal contains multiple types of information:

- Linguistic information — the words being spoken.
- Speaker information — who is speaking.
- Emotional information — whether the speaker sounds excited, sad, or angry.
- Paralinguistic information — speaking rate, pauses, emphasis, and pitch variation.

This is why speech is particularly interesting in AI: one signal can contain several layers of information.

## **1.5 What is Digital Audio Processing?**

Computers cannot directly process continuous physical sound waves in their original form. They represent audio using digital numbers.

The process generally involves three stages:

![Audio sampling rate representation](https://images.openai.com/static-rsc-4/ZTBG7LGxAszDe-8BHYifQJtfmQPWG-WOp0Ua0dysbCnI0GEqmuDPhbZcMsj1qN1J1YPc-AEdzo8Q470yu-WfHc0nopyn_4Y7BfyYH0waUOr3vuWenQPKjtf_OVirvZepAR5PIAtjs_j8uKCkrxfNZ_GxdbnLZBdoVPpY4IYlxYI?purpose=inline)

[Source: ACE Studio](https://acestudio.ai/blog/what-is-audio-sample-rate/?utm_source=chatgpt.com)

### **Stage 1 — Sampling**

Sampling converts a continuous-time signal into discrete-time measurements.

Imagine recording a sound wave and measuring its amplitude at regular intervals.

For example, at a sampling rate of 16,000 Hz, the system takes:

$$
16,000 \text{ samples per second}
$$

### **Stage 2 — Quantization**

Each sampled amplitude is assigned to one of a finite number of available amplitude levels.

For example, 16-bit audio provides:

$$
2^{16} = 65,536
$$

possible amplitude levels.

### **Stage 3 — Encoding**

The quantized values are represented in binary form so that they can be stored and processed digitally.

### **Important audio terminology**

| Term | Meaning |
|---|---|
| Sampling rate | Number of samples per second |
| Bit depth | Number of bits used to represent each sample |
| Channels | Number of audio channels |
| Mono | Single audio channel |
| Stereo | Two audio channels |

Common sampling rates:

| Sampling rate | Common usage |
|---|---|
| 8 kHz | Traditional telephone speech |
| 16 kHz | Common in speech recognition |
| 44.1 kHz | CD-quality audio |
| 48 kHz | Video and professional audio |

**Nyquist principle:** To represent a signal without aliasing, the sampling frequency must be greater than twice the highest frequency component of the signal, assuming appropriate band-limiting.

$$
f_s > 2f_{\max}
$$

We'll explore sampling, quantization, PCM, and digital signal representation in greater depth later.

### **Important distinction: Speaker Recognition vs Speaker Verification**

| Speaker Recognition | Speaker Verification |
|---|---|
| Answers: Who is speaking? | Answers: Is this the claimed person? |
| Can compare against multiple speakers | Usually compares against one claimed identity |
| Identification task | Verification task |

## **What you should remember 🧠**

1. Sound is a physical vibration; audio is its recorded or represented form.
2. Speech is a specialized form of audio that carries linguistic and speaker-related information.
3. Audio processing handles general sound, while speech processing focuses on human speech.
4. Digital audio uses sampling, quantization, and encoding.
5. ASR converts speech into text; TTS converts text into speech.
6. Speaker recognition identifies a speaker, while speaker verification checks a claimed identity.
7. Speech Emotion Recognition attempts to infer emotion from vocal characteristics.
