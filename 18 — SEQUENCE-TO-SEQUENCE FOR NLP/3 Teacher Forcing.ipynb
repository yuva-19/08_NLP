# 18.3 Teacher Forcing — NLP Role 🧠


## 1. Example: Training a translation model

Suppose the target sentence is:

`I love NLP`

During training, the decoder learns to predict the next token using the actual previous target tokens, rather than its own previous predictions.

DECODER TRAINING EXAMPLE

Decoder input

`<BOS> I love`

Expected output

`I love NLP`

BOS means beginning of sequence. The decoder input is shifted right by one token relative to the target sequence.

At each step:

| Decoder input | Target to predict |
| ------------- | ----------------- |
| `<BOS>`       | `I`               |
| `I`           | `love`            |
| `I love`      | `NLP`             |
| `I love NLP`  | `<EOS>`           |

During teacher-forced training, the correct target prefix is supplied at every step—even if the model's previous prediction would have been wrong.

## 2. Why does teacher forcing matter in NLP?

- Faster learning: The decoder receives reliable previous tokens instead of accumulating errors from its own predictions during training.
- Stable sequence learning: The model learns to predict the next word given the correct preceding words.
- Parallel training in Transformers: With causal masking, many next-token predictions can be computed in parallel during training, while preventing each position from accessing future target tokens.

## 3. Training vs. inference

Training

Correct previous tokens are supplied

`I → love → NLP`

The decoder learns from the ground-truth sequence.

Inference

The model uses its own generated tokens

`<BOS> → I → love → NLP → <EOS>`

Each prediction becomes input for the next step.

This difference introduces a potential issue called exposure bias: during training, the decoder usually sees correct prefixes, but during inference, it may encounter its own mistakes and continue generating from them.

## What you should remember

1. In Seq2Seq NLP, teacher forcing supplies the ground-truth previous target tokens to the decoder during training.
2. The decoder learns next-token prediction from correctly aligned input and target sequences.
3. Inference generally uses the model's own generated tokens.
4. The difference between training and inference can contribute to exposure bias.
