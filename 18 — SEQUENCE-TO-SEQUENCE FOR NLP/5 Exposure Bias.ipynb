# 18.5 Exposure Bias — Seq2Seq NLP 🧠


## 1. What is Exposure Bias?

Exposure bias is the mismatch between training and inference: during training, the decoder typically receives correct previous tokens, but during inference, it must rely on its own previously generated tokens.

## 2. How does it happen?

Suppose we're training a Seq2Seq model to generate:

`The cat is sleeping`

During training

Teacher forcing supplies the correct previous tokens.

The

cat

is

sleeping

The model learns to predict each next token from the correct preceding tokens.

During inference

The model feeds its own predictions back into the decoder.

The

dog

is

running

A wrong early prediction can change the context used for later predictions.

The example illustrates the mechanism: the model generates dog instead of cat. It then predicts the next token using its own generated prefix, which may lead to further errors.

## 3. Why is exposure bias a problem?

- Error propagation: One incorrect token can influence subsequent predictions.
- Unfamiliar contexts: During inference, the decoder may encounter prefixes it rarely or never saw during teacher-forced training.
- Lower output quality: Generated translations, summaries, or chatbot responses may become less coherent or drift away from the intended meaning.

Importantly, an incorrect token does not always cause a cascade of errors. It simply increases the risk.

## 4. Exposure Bias vs. Context Vector Bottleneck

| Context vector bottleneck                      | Exposure bias                                                                |
| ---------------------------------------------- | ---------------------------------------------------------------------------- |
| A problem with compressing input information   | A mismatch between training and inference                                    |
| One fixed-size vector must summarize the input | The decoder uses its own predictions at inference                            |
| Attention helps address the bottleneck         | Training strategies such as scheduled sampling can help address the mismatch |

These are two different problems in Seq2Seq models.

## What you should remember 🧠

1. During teacher-forced training, the decoder usually sees correct previous target tokens.
2. During inference, it consumes its own generated tokens.
3. This mismatch is called exposure bias.
4. Early mistakes can propagate and reduce output quality.
5. Scheduled sampling is one proposed approach that mixes correct tokens with model-generated tokens during training; it is not a universal solution.
