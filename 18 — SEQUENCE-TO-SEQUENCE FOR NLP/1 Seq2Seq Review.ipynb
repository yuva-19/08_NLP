# 18.1 Seq2Seq Review — Encoder, Decoder & Context Vector 🧠


Unlike ordinary text classification, the output can contain a different number of tokens from the input.

Examples:

- Translation: English → Tamil
- Summarization: Long document → Short summary
- Conversational NLP: User message → Response

The central idea is simple:

Input sequence → Encoder → Context → Decoder → Output sequence

## 2. The five core concepts

### 1. Input sequence

The input sequence is the text or tokens provided to the model.

Example:

`I love machine learning`

After tokenization, it might become:

`[I, love, machine, learning]`

The number of input tokens is called the input sequence length.

### 2. Encoder

The encoder reads the input sequence and transforms it into numerical representations that capture information about the input.

In a classic RNN-based Seq2Seq model, the encoder processes tokens sequentially and updates its hidden state at each step.

Think of the encoder as a reader that understands the input before generating a response.

### 3. Context vector

The context vector carries information from the encoder to the decoder.

In the original, basic Seq2Seq architecture, the encoder's final hidden state often serves as the context vector. It is intended to summarize the input's relevant information.

For example:

`I love machine learning` → Context representation

The context vector is not another sentence. It is a numerical representation learned by the model.

Important limitation: A single fixed-size context vector can become a bottleneck when the input is long, because it must carry information about the entire sequence. Attention mechanisms were introduced to help address this limitation; we'll study them later.

### 4. Decoder

The decoder uses the context information to generate the output sequence, usually one token at a time in classic Seq2Seq models.

For example, it might generate:

`எனக்கு` → `இயந்திரக்` → `கற்றல்` → `பிடிக்கும்`

The decoder conditions each prediction on the information available from the encoder and the tokens generated so far.

### 5. Output sequence

The output sequence is the sequence produced by the decoder.

It can have a different length from the input.

| Task          | Input sequence   | Output sequence    |
| ------------- | ---------------- | ------------------ |
| Translation   | English sentence | Tamil sentence     |
| Summarization | Long article     | Short summary      |
| Chatbot       | User message     | Generated response |

The decoder typically stops when it generates a special end-of-sequence token.


## What you should remember 🧠

1. Encoder: Processes the input.
2. Context vector: Carries the encoded information.
3. Decoder: Generates the output.
4. Input and output sequences can have different lengths.
5. A single context vector can be a bottleneck for long inputs; attention helps the decoder access more relevant input information.
