# 9.3 Distributional Hypothesis 🧠

> Words appearing in similar contexts tend to have related meanings.

The Distributional Hypothesis is one of the fundamental ideas behind learning word embeddings. Instead of manually defining what a word means, we learn about its meaning from the words that frequently appear around it.

## 1. Understand with an example

Consider these sentences:

- “The cat drinks milk.”
- “The dog drinks milk.”
- “The cat chases the mouse.”
- “The dog chases the ball.”


Notice that cat and dog appear in similar contexts:

| Context around the word | Cat                  | Dog                  |
| ----------------------- | -------------------- | -------------------- |
| drinks milk             | ✓                    | ✓                    |
| chases something        | ✓                    | ✓                    |
| common nearby words     | drinks, milk, chases | drinks, milk, chases |

Because their surrounding words overlap, an NLP model may learn that cat and dog have related meanings. They are both animals and can participate in similar actions.

Important: The hypothesis does not mean the words are identical. It means their usage patterns provide clues about their meanings.

## 2. How does the model learn this?

Sentences in a text corpus

“The cat drinks milk” · “The dog drinks milk”

Observe surrounding words

cat → the, drinks, milk

dog → the, drinks, milk

Learn word representations

Similar usage patterns can produce similar embedding vectors.

The model learns statistical patterns from large amounts of text. In embedding methods, these patterns help determine where words are positioned in a vector space.

## 3. Why is this important for Word Embeddings?

The Distributional Hypothesis is the key intuition behind methods such as Word2Vec and GloVe.

For example, Word2Vec learns by predicting words from their contexts or predicting contexts from words.

- CBOW: Uses surrounding words to predict the target word.
- Skip-gram: Uses the target word to predict surrounding words.

As the model trains, it adjusts its parameters to capture patterns in word usage.

## 4. Limitations ⚠️

The hypothesis is powerful, but it is not perfect.

- Polysemy: “Bank” can mean a financial institution or the side of a river.
- Antonyms: “Hot” and “cold” can appear in similar contexts despite having opposite meanings.
- Context dependence: A word’s meaning can change with the sentence or domain.

Traditional static embeddings generally assign one vector per word, which makes handling multiple meanings difficult.

## 5. Key takeaway

Remember this chain:

Words → Context patterns → Learned relationships → Embedding vectors

The model does not directly understand meaning as a human does. It learns useful representations from patterns in language.
