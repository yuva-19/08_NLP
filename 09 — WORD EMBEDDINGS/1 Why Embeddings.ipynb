# Module 9 — Word Embeddings 🧠

## 9.1 Why do we need embeddings?

Yuva, you've already learned One-Hot Encoding, Bag of Words (BoW), and TF-IDF. All three convert text into numbers, making it possible for machine-learning models to process language.

But they have a fundamental limitation: they represent words using their identity or statistical importance, not their meaning directly.

Word embeddings address this by representing words as dense, low-dimensional vectors that can capture semantic and syntactic relationships.

## 1. The evolution of text representation



1\. One-Hot Encoding

Represents each word using a unique binary vector. The vector is mostly zeros.

Limitation: doesn't inherently capture similarity in meaning.



2\. Bag of Words (BoW)

Represents a document using word counts from a vocabulary.

Limitation: loses word order and deeper semantic relationships.


3\. TF-IDF

Weights words according to their importance in a document relative to a collection.

Limitation: still uses sparse vectors without inherently learning word meaning.



4\. Dense Word Embeddings

Represents words as compact, learned numerical vectors.

Advantage: similar usage and meanings can be reflected by nearby vectors.

## 2. The key difference: sparse vs. dense vectors

Imagine your vocabulary contains 10,000 unique words.

| Property           | TF-IDF                          | Word embeddings                |
| ------------------ | ------------------------------- | ------------------------------ |
| Vector size        | Up to 10,000 features           | E.g. 100 or 300 dimensions     |
| Values             | Mostly zeros                    | Usually real-valued numbers    |
| Meaning            | Word importance in a document   | Learned patterns of word usage |
| Similar words      | No inherent semantic similarity | Can have similar vectors       |
| Context-dependent? | Not inherently                  | Static embeddings: no          |

Dense embeddings are more compact and can help models generalize across words used in similar contexts.


### A simple example

Suppose our vocabulary is:

`["cat", "dog", "car", "banana"]`

A one-hot representation might look like this:

\\[ \text{cat}=[1,0,0,0] \\]

\\[ \text{dog}=[0,1,0,0] \\]

The vectors for cat and dog are just as different, mathematically, as the vectors for cat and car. The representation itself doesn't tell us that cats and dogs are semantically related.

An embedding model might instead learn illustrative vectors such as:

\\[ \text{cat}=[0.82,0.31,-0.12] \\]

\\[ \text{dog}=[0.79,0.35,-0.08] \\]

\\[ \text{car}=[-0.40,0.91,0.62] \\]

These are illustrative values, not real pretrained embeddings. Notice how cat and dog have similar coordinates, while car is different.



The embedding model learns these numerical relationships from training data; we don't manually assign semantic meanings to each dimension.
## 3. How are word embeddings learned?

This is the central intuition behind Word2Vec, which we'll study in Module 10.

Training text

"The cat drinks milk"

Learn from word contexts

Cat appears near words such as the, drinks and milk.

Update embedding vectors

Words appearing in similar contexts can develop similar representations.

This follows the distributional hypothesis:

> Words that occur in similar contexts tend to have related meanings

For example, if cat and dog frequently appear in similar sentence patterns, an embedding algorithm can learn representations that reflect their relationship.

One important distinction: embeddings capture statistical patterns in language, not guaranteed or complete human understanding. Static embeddings can also struggle with words that have multiple meanings, such as bank in "river bank" versus "bank account.

## 4. Where does the Embedding Layer fit in?

Since you've already studied PyTorch, connect this to your existing deep-learning knowledge:

- Word embedding: the numerical vector representing a word.
- Embedding layer: the learnable lookup table that maps token IDs to those vectors.

For a vocabulary of 10,000 words and an embedding dimension of 100, the embedding matrix has:

\\[ W\in\mathbb{R}^{10000\times100} \\]

Each row corresponds to one vocabulary token. Given a token ID, the embedding layer retrieves its corresponding 100-dimensional vector.


## What you should remember 🎯

1. One-hot encoding represents word identity; BoW represents word counts; TF-IDF weights document features.
2. Word embeddings provide compact, dense, learned vector representations.
3. Similar contexts can lead to similar embeddings.
4. An embedding layer is a lookup table whose rows contain word vectors.
5. Static embeddings assign one vector per word, regardless of its sentence context.
