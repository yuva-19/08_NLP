# 9.5 Important Limitation of Static Embeddings 🧠

The main limitation of static embeddings is that a word receives the same vector even when its meaning changes with context.

## 1. Example: The word “bank”



Financial institution

“I deposited money in the bank.”


River bank

“We sat on the bank of the river.”

A static embedding model stores one vector for the word `bank`.

\\[ \mathbf{v}\_{\text{bank}} = [0.31,\ 0.72,\ -0.15,\ldots] \\]

It uses this same vector in both sentences. Therefore, the representation does not explicitly distinguish the financial meaning from the geographical meaning.

## 2. Other important limitations

| Limitation                       | Explanation                                                                                                                                                   |
| -------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| Polysemy                         | A word has multiple related or unrelated meanings, such as bank.                                                                                              |
| Context independence             | The representation does not change with the surrounding sentence.                                                                                             |
| Out-of-vocabulary (OOV) words    | Traditional word-level models may struggle with words absent from their vocabulary. FastText partly addresses this using character-level subword information. |
| Limited contextual understanding | The vector alone cannot represent the specific meaning a word takes in a particular sentence.                                                                 |

## 3. How contextual embeddings solve this

Models such as BERT use the surrounding words to produce context-dependent representations.

Sentence A

“I deposited money in the bank.”

Bank representation A

Financial meaning

Sentence B

“We sat beside the river bank.”

Bank representation B

Geographical meaning

Same word, different contextual representations.

Unlike static embeddings, contextual models use the sentence to determine the representation of a word occurrence. The resulting vectors are not required to be different in every case, but they can capture different meanings when the context changes.

## 4. Key takeaway 🎯

- Static embeddings: One learned vector per word.
- Contextual embeddings: A representation that depends on the surrounding text.
- Why it matters: Context helps NLP systems interpret ambiguous words more accurately.
