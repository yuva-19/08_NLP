Yuva, now let's understand how word embeddings exist in a mathematical space, how we measure relationships between words, and why cosine similarity is one of the most useful tools in NLP.

## 1. Dense vectors

A word embedding represents a word as a vector of real-valued numbers.

For example, a simplified embedding might look like this:

\\[ \text{king}=[0.8,\ 0.3,\ -0.2,\ 0.6] \\]

\\[ \text{queen}=[0.7,\ 0.4,\ -0.1,\ 0.5] \\]

These four-dimensional vectors are illustrative. Real embeddings commonly contain dozens or hundreds of dimensions, depending on the model.



Unlike one-hot vectors, dense vectors generally have many nonzero values. Their dimensions are learned from data rather than manually assigned meanings.

For example, the first dimension does not necessarily mean "gender" or the second "royalty." The model learns distributed patterns across dimensions.

## 2. Dimensionality

Dimensionality is the number of values in an embedding vector.

| Representation   | Example dimensionality           |
| ---------------- | -------------------------------- |
| One-hot encoding | 10,000                           |
| TF-IDF           | 10,000 vocabulary features       |
| Word embedding   | 100, 300, or another chosen size |

These are examples, not fixed rules. TF-IDF dimensionality depends on the vocabulary, while embedding dimensionality depends on the model design.

Suppose we have 10,000 words and each embedding has 300 dimensions.

\\[ W\in\mathbb{R}^{10000\times300} \\]

The embedding matrix contains \\(10{,}000\times300=3{,}000{,}000\\) parameters if every entry is learned independently.

A smaller dimension reduces storage and computation, but may limit the information the model can represent. A larger dimension can represent richer patterns but requires more parameters and data.

## 3. Semantic similarity

Semantic similarity measures how closely related two words are in meaning or usage.

For example, a trained embedding model may place these words close together:

- `cat` and `dog`
- `king` and `queen`
- `car` and `vehicle`

But it may place `cat` and `volcano` farther apart.



This is a learned pattern, not a guarantee. Similarity depends on the training data and embedding model.&
Industry use: semantic similarity helps with document matching, duplicate detection, recommendation systems and semantic search.

## 4. Vector distance

Distance measures how far apart two vectors are.

### Euclidean distance (L2)

For two vectors \\(A\\) and \\(B\\):

\\[ d(A,B)=\sqrt{\sum\_{i=1}^{n}(A_i-B_i)^2} \\]

Consider:

\\[ A=[1,2],\quad B=[4,6] \\]

\\[ d(A,B)=\sqrt{(4-1)^2+(6-2)^2}=5 \\]

A smaller distance means the vectors are closer under this metric.



However, Euclidean distance depends on vector magnitude. Two vectors pointing in the same direction can have different Euclidean distances if their lengths differ.

That's why cosine similarity is also useful.

## 5. Cosine similarity ⭐

Cosine similarity measures the angle between two vectors, rather than their absolute distance.

\\[ \boxed{ \operatorname{cosine}(A,B)= \frac{A\cdot B}{\\|A\\|\\|B\\|} } \\]

Where:

- \\(A\cdot B\\): dot product.
- \\(\\|A\\|\\): magnitude (L2 norm) of vector \\(A\\).
- \\(\\|B\\|\\): magnitude of vector \\(B\\).



### Interpreting the score

| Cosine similarity | Geometric interpretation |
| ----------------- | ------------------------ |
| \\(+1\\)          | Same direction           |
| \\(0\\)           | Perpendicular vectors    |
| \\(-1\\)          | Opposite directions      |

These are mathematical interpretations. Real embedding models often produce scores within a narrower range.

### Worked example

Let:

\\[ A=[1,2],\quad B=[2,4] \\]

Step 1: Dot product

\\[ A\cdot B=(1)(2)+(2)(4)=10 \\]

Step 2: Magnitudes

\\[ \\|A\\|=\sqrt{1^2+2^2}=\sqrt5 \\]

\\[ \\|B\\|=\sqrt{2^2+4^2}=\sqrt{20} \\]

Step 3: Similarity

\\[ \frac{10}{\sqrt5\sqrt{20}}=1 \\]

Both vectors point in the same direction, so their cosine similarity is \\(1\\), even though their magnitudes differ.

The formula and its normalized dot-product interpretation are documented in scikit-learn's official metrics guide.


## What you should remember 🎯

- Dense vectors: compact, learned representations containing real-valued numbers.
- Dimensionality: the number of components in a vector.
- Semantic similarity: how related words are according to their learned representations.
- Euclidean distance: measures straight-line distance; smaller means closer.
- Cosine similarity: compares vector directions and is widely used in text similarity and retrieval.
