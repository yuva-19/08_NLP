# 9.4 Static Word Embeddings 🧠

Static word embeddings represent each word using one fixed vector, regardless of the context in which the word appears.

For example, Word2Vec, GloVe, and FastText are commonly used static embedding methods.

## 1. One vector per word

Suppose a model learns these illustrative embeddings:

| Word  | Embedding vector      |
| ----- | --------------------- |
| `cat` | `[0.21, 0.67, -0.14]` |
| `dog` | `[0.24, 0.62, -0.11]` |
| `car` | `[-0.53, 0.18, 0.72]` |

These vectors are simplified examples; actual embeddings usually have many more dimensions.

The vectors for `cat` and `dog` are relatively similar in this example, while `car` has a different representation.

The key idea: Every occurrence of `cat` retrieves the same learned vector.

## 2. Context-independent representation

Consider the word bank in these sentences:

- “I deposited money in the bank.”
- “We sat on the river bank.”


Humans understand that bank has different meanings in these sentences. However, a traditional static embedding assigns the word `bank` the same vector in both cases.

\\[ \text{bank}\_{\text{finance}} \rightarrow [0.31, 0.72, -0.15] \\]

\\[ \text{bank}\_{\text{river}} \rightarrow [0.31, 0.72, -0.15] \\]

These are illustrative values showing the same vector being reused. The embedding itself does not change based on the sentence.

## 3. How static embeddings work

Input word: `bank`

“I deposited money in the bank.”

Embedding lookup

Retrieve the stored vector for `bank`

Fixed vector

`[0.31, 0.72, -0.15, ...]`

The river-bank sentence retrieves the same vector.

## 4. Advantages and limitations

| Advantages ✅                          | Limitations ⚠️                                                   |
| ------------------------------------- | ---------------------------------------------------------------- |
| Compact numerical representations     | Cannot dynamically distinguish multiple meanings of a word       |
| Capture useful semantic relationships | Same vector in every context                                     |
| Efficient to store and retrieve       | Limited understanding of sentence-specific meaning               |
| Useful for traditional NLP tasks      | Unseen words can be difficult to handle, depending on the method |

## 5. Static vs contextual embeddings

| Feature         | Static embeddings         | Contextual embeddings                    |
| --------------- | ------------------------- | ---------------------------------------- |
| Representation  | One vector per word       | Representation changes with context      |
| Example methods | Word2Vec, GloVe, FastText | BERT, contextual Transformer models      |
| Word “bank”     | Same vector               | Different representations based on usage |

Remember: Static embeddings are not necessarily simplistic or poor-quality embeddings. They can capture useful semantic relationships, but their fixed representation limits their ability to handle context-dependent meanings.
