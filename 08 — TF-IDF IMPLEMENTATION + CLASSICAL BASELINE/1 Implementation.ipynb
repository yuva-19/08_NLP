```python
# TF 

documents = [
    "who love NLP",
    "who love Python",
    "Python is powerful"
]
unique_words = list(set(" ".join(documents).split()))
```

```python
def calculate_tf(document):
    tf = {}
    words = document.split()
    for word in words:
        tf[word] = words.count(word)/len(words)
    return tf 
```

```python
for document in documents:
    tf = calculate_tf(document)
    print(document,"---->",tf)
```

```python
import math

def calculate_idf(documents):
    idf = {}
    unique_words = list(set(" ".join(documents).split()))
    for word in unique_words:
        count = 0 
        for document in documents:
            if word in document.split():
                count+=1
        idf_score = math.log(len(documents)/count)
        idf[word] = idf_score  

    return idf
```

```python
idf_scores = calculate_idf(documents)
print(idf_scores)
```

```python
def calculate_tfidf(document,idf):
    tfidf ={}
    tf = calculate_tf(document)
    for word in tf:
        tfidf[word] = tf[word] * idf[word]

    return tfidf        
```

```python
for document in documents:
    tfidf = calculate_tfidf(document,idf_scores)
    print(document,"--->",tfidf)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(X.toarray().round(3))
```
