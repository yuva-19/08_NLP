```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd
```

```python

positive_texts = [
    "I love this product",
    "This is an amazing experience",
    "The quality is excellent",
    "I am very happy with this",
    "Absolutely wonderful service",
    "This product works perfectly",
    "The food tastes delicious",
    "The delivery was quick",
    "Customer support was helpful",
    "I highly recommend this",
    "The performance is outstanding",
    "I am satisfied with the purchase",
    "The design looks beautiful",
    "This app is easy to use",
    "The battery life is impressive",
    "The staff were friendly",
    "Everything worked as expected",
    "The experience was fantastic",
    "This exceeded my expectations",
    "The price is reasonable",
    "I enjoyed using this product",
    "The service was professional",
    "The instructions were clear",
    "Setup was simple and easy",
    "The material feels premium",
    "This is worth the money",
    "The product is reliable",
    "I received excellent service",
    "The results are impressive",
    "I would buy this again"
]

negative_texts = [
    "I hate this product",
    "This is a terrible experience",
    "The quality is very poor",
    "I am very disappointed",
    "Absolutely horrible service",
    "This product barely works",
    "The food tastes awful",
    "The delivery was very slow",
    "Customer support was unhelpful",
    "I would not recommend this",
    "The performance is terrible",
    "I regret this purchase",
    "The design looks ugly",
    "This app is difficult to use",
    "The battery life is disappointing",
    "The staff were rude",
    "Nothing worked as expected",
    "The experience was frustrating",
    "This failed my expectations",
    "The price is unreasonable",
    "I disliked using this product",
    "The service was unprofessional",
    "The instructions were confusing",
    "Setup was complicated and difficult",
    "The material feels cheap",
    "This is a waste of money",
    "The product is unreliable",
    "I received terrible service",
    "The results are disappointing",
    "I would never buy this again"
]

texts = positive_texts + negative_texts
labels = (
    ["positive"] * len(positive_texts)
    + ["negative"] * len(negative_texts)
)

df = pd.DataFrame({
    "text": texts,
    "sentiment": labels
})

print("Dataset shape:", df.shape)
print("\nClass distribution:")
print(df["sentiment"].value_counts())

df.sample(5, random_state=42)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Naive Bayes": MultinomialNB(),
    "Linear SVM": LinearSVC()
}

pipelines = {}

for name, classifier in models.items():
    pipelines[name] = Pipeline([
        ("tfidf", TfidfVectorizer()),
        ("classifier", classifier)
    ])
```

```python
results = []

for name, pipeline in pipelines.items():
    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    accuracy = accuracy_score(y_test, predictions)

    results.append({
        "Model": name,
        "Accuracy": accuracy
    })

    print(f"\n{'=' * 40}")
    print(name)
    print(f"Accuracy: {accuracy:.2%}")

    print(classification_report(
        y_test,
        predictions,
        zero_division=0
    ))

results_df = pd.DataFrame(results)
results_df.sort_values("Accuracy", ascending=False)
```

```python
new_texts = [
    "The product is excellent and works perfectly",
    "Terrible quality and disappointing service",
    "The product is okay"
]

for name, pipeline in pipelines.items():
    print(f"\n{name}")

    predictions = pipeline.predict(new_texts)

    for text, prediction in zip(new_texts, predictions):
        print(f"{text} --> {prediction}")
```
