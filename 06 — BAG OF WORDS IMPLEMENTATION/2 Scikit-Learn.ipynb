```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love NLP",
    "I love Python"
]

vectorizer = CountVectorizer()
```

```python
X = vectorizer.fit_transform(documents)

print(vectorizer.vocabulary_)
print(X)

# count vectorizer by  default it avoids the one character words so it avoids 'I' here 
# we have two documents here thats why 1st column contains only 0s 1s 

# (0, 0)	1   love 
# (0, 1)	1   NLP 
# (1, 0)	1   love (again love thats why same 0)
# (1, 2)	1   Python 
```

```python
print(X.toarray())
# ['Love NLP Python']
```

```python
# fit() learns the vocabulary from the supplied documents.
    
# transform() uses an already learned vocabulary to convert documents into vectors.

# fit_transform() learns the vocabulary and transforms those same documents in one operation.

vectorizer = CountVectorizer()

vectorizer.fit(documents)

print(vectorizer.vocabulary_)
```

```python
new_documents = ["Python NLP"]

X_new = vectorizer.transform(new_documents)

print(X_new.toarray())
```
