# 17.4 Token Classification 🤖

Token classification is an NLP task in which a model predicts a label for each token in an input sequence.

Instead of assigning one label to an entire sentence, the model assigns labels to individual tokens.

## 1. How token classification works

Consider this sentence:

> Sundar Pichai visited Chennai.

The model processes the sentence and predicts a label for each token.

INPUT TEXT

### Sundar Pichai visited Chennai.

Token classification model

Contextual representation of each token

Sundar

B-PER

Pichai

I-PER

visited

O

Chennai

B-LOC

Example predictions for Named Entity Recognition (NER).

The output contains one predicted label per token. Here, the labels are BIO-style NER tags.

## 2. What tasks use token classification?

Token classification is not limited to NER. It can solve several token-level NLP problems.


| Task        | Example                   | Output                                 |
| ----------- | ------------------------- | -------------------------------------- |
| NER         | `Google opened an office` | `ORG`, `O`, `O`, `O`                   |
| POS tagging | `Students learn quickly`  | `NOUN`, `VERB`, `ADV`                  |
| BIO tagging | `Sundar Pichai`           | `B-PER`, `I-PER`                       |
| Chunking    | `the young student`       | Labels identifying grammatical phrases |

The key idea: the model predicts a label for each token, and the label meanings depend on the task.

## 3. How modern Transformer models perform token classification




A typical Transformer-based token classification model has two main components:

1. Transformer encoder: Processes the input and creates contextual representations for each token. For example, it uses surrounding words to understand whether a name refers to a person or an organization.
2. Token classification head: A small prediction layer takes each token's representation and predicts a label from the task's label set.

During training, the model's predicted labels are compared with the correct labels from an annotated dataset. The error is used to update the model parameters.

For NER, the model learns from examples labeled with tags such as `B-PER`, `I-PER`, `B-ORG`, and `O`.

## 4. Token classification vs. text classification

This distinction is especially important because you have already studied document classification.

| Text classification                        | Token classification              |
| ------------------------------------------ | --------------------------------- |
| One label for the entire text              | One label per token               |
| Example: review → positive                 | Example: `Chennai` → `B-LOC`      |
| Often uses a sentence-level representation | Uses token-level representations  |
| Predicts sentiment, topic, spam, etc.      | Predicts NER tags, POS tags, etc. |

## 5. One important real-world challenge: subword tokenization

Modern models may split one word into multiple subword tokens. For example, a tokenizer might represent an unfamiliar word as several pieces.

This creates a challenge: how do we align the original word-level labels with the model's subword tokens?

During training, the dataset's labels must be aligned with the tokenizer output. Special tokens and padding positions are generally excluded from the loss calculation. The exact alignment strategy depends on the model and dataset.


This is an important practical concept for your upcoming Hugging Face implementation modules.

## What you should remember 🧠

1. Token classification predicts a label for each token.
2. NER and POS tagging are both token classification tasks.
3. Modern approaches commonly combine a pretrained Transformer with a token classification head.
4. Training requires token-level labels aligned with the model's tokenization.
5. Unlike document classification, token classification produces a sequence of labels.
