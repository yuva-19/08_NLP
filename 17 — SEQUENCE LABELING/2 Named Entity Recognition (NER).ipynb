# 17.2 Named Entity Recognition (NER) 🏷️


## 1. What is NER?

Named Entity Recognition (NER) is an NLP task that detects and classifies entities in text into predefined categories, such as people, organizations, locations, dates, money, and products.



For example:

> Sundar Pichai visited Chennai on 15 August 2026 and bought an iPhone for ₹80,000 from Apple.

An NER model could extract:

| Text span      | Entity label |
| -------------- | ------------ |
| Sundar Pichai  | PERSON       |
| Chennai        | LOCATION     |
| 15 August 2026 | DATE         |
| iPhone         | PRODUCT      |
| ₹80,000        | MONEY        |
| Apple          | ORGANIZATION |

Notice that NER identifies what a text span refers to, rather than its grammatical role.

## 2. The six entity types you must know



1\. PERSON — People

Names of individuals.

Examples: Sundar Pichai, A. R. Rahman, Virat Kohli


2\. ORGANIZATION — Companies and institutions

Businesses, government bodies, universities, and other organizations.

Examples: Google, ISRO, Kumaraguru College of Technology



3\. LOCATION — Places

Cities, countries, states, and geographical locations.

Examples: Chennai, Tamil Nadu, India



4\. DATE — Dates and time references

Calendar dates, months, years, and sometimes relative expressions.

Examples: 15 August 2026, January 2025, tomorrow



5\. MONEY — Monetary amounts

Amounts of money, often including currencies.

Examples: ₹80,000, $500, 2 million euros



6\. PRODUCT — Products and named items

Names of specific products, devices, or commercial offerings when the model's label scheme supports them.

Examples: iPhone 17, PlayStation 5, Samsung Galaxy

Important: Different NER datasets use different label sets. Some distinguish products, while others classify them under broader categories or do not label them at all.

## 3. How NER works as a sequence-labeling task

NER usually processes a sentence and assigns labels to its tokens or text spans.

INPUT SENTENCE

### Google opened an office in Chennai.

RECOGNIZED ENTITIES

#### Google

ORGANIZATION

#### Chennai

LOCATION

NER must recognize multi-word entities too. For example, “Sundar Pichai” is one person entity, not two separate people.

In token-level labeling systems, BIO tags can represent this:

| Token   | BIO tag |
| ------- | ------- |
| Sundar  | B-PER   |
| Pichai  | I-PER   |
| visited | O       |
| Chennai | B-LOC   |

- B-: Beginning of an entity.
- I-: Inside an entity.
- O: Outside any entity.

We'll study BIO tagging properly in Module 17.3.

## 4. NER vs POS tagging

| POS Tagging                                   | NER                                  |
| --------------------------------------------- | ------------------------------------ |
| Identifies grammatical roles                  | Identifies real-world entity types   |
| `Chennai → PROPN`                             | `Chennai → LOCATION`                 |
| `Google → PROPN`                              | `Google → ORGANIZATION`              |
| Focuses on how a word functions grammatically | Focuses on what an entity represents |

A word can be a proper noun in POS tagging and an organization or location in NER. The tasks answer different questions.

## 5. Where NER is used in industry 🚀

- Search engines: Recognize people, places, and companies in queries.
- Recruitment: Extract names, companies, skills, and dates from resumes.
- Finance: Extract company names, currencies, and monetary amounts from reports.
- Customer support: Identify product names and organizations in support tickets.
- Healthcare: Extract relevant entities such as medication names and dates, using specialized label schemes.
- RAG and information extraction: Extract entities to help organize documents and retrieve relevant information.

Modern Transformer models can perform NER using a token-classification head, predicting a label for each token. We'll connect this to Transformer-based sequence labeling later.

## What you should remember 🧠

1. NER identifies and classifies entities in text.
2. Know the six labels: PERSON, ORGANIZATION, LOCATION, DATE, MONEY, PRODUCT.
3. Entity names can contain multiple words.
4. NER is a sequence-labeling task; BIO tags are one way to mark entity boundaries.
5. POS tagging identifies grammatical roles; NER identifies entity types.
6. Entity labels depend on the dataset and task.
