# 17.3 BIO Tagging 🏷️

## 1. What does BIO mean?

BIO is a token-labeling scheme used to mark entity boundaries in text.

| Tag     | Meaning                   | Purpose                                         |
| ------- | ------------------------- | ----------------------------------------------- |
| `B-PER` | Beginning of Person       | First token of a person's name                  |
| `I-PER` | Inside Person             | Subsequent token(s) of that person's name       |
| `B-ORG` | Beginning of Organization | First token of an organization's name           |
| `I-ORG` | Inside Organization       | Subsequent token(s) of that organization's name |
| `O`     | Outside                   | Token that does not belong to a labeled entity  |

- B = Beginning
- I = Inside
- O = Outside

## 2. Example: Identifying people and organizations

Consider this sentence:

> Sundar Pichai works at Google India.

TOKEN-BY-TOKEN LABELING

Sundar

B-PER

Pichai

I-PER

works

O

at

O

Google

B-ORG

India

I-ORG

Here, Google India is treated as one organization name. Entity annotations can vary by dataset.

Why not label both `Sundar` and `Pichai` as `B-PER`?

Because `B-PER` marks the start of an entity. Since Pichai continues the same person's name, it receives `I-PER`.

## 3. Why not use only PER and ORG?

Suppose we use only these labels:

```
Sundar → PER
Pichai → PER
Google → ORG
India  → ORG
```

We can identify the entity types, but we don't explicitly mark the boundaries between entities.

BIO solves this by distinguishing the first token from subsequent tokens:

```
Sundar → B-PER
Pichai → I-PER
Google → B-ORG
India  → I-ORG
```

This allows a model to represent multi-token entities while preserving their boundaries.

## 4. What happens with single-word entities?

A single-word entity starts and ends on the same token, so it receives a `B-` tag.

Example:

> Rahul joined Google.

| Token  | BIO tag |
| ------ | ------- |
| Rahul  | `B-PER` |
| joined | `O`     |
| Google | `B-ORG` |

There is no `I-PER` for Rahul because his name has only one token.

## 5. What about locations and dates?

The same pattern extends to other entity types. For example:

> Chennai hosted the event on 15 August.

| Token   | BIO tag  |
| ------- | -------- |
| Chennai | `B-LOC`  |
| hosted  | `O`      |
| the     | `O`      |
| event   | `O`      |
| on      | `O`      |
| 15      | `B-DATE` |
| August  | `I-DATE` |

Here, `LOC` represents location and `DATE` represents date. These are additional entity types beyond the five tags listed in your syllabus snippet.

## What you should remember

1. `B-` marks the beginning of an entity.
2. `I-` marks subsequent tokens belonging to that entity.
3. `O` marks tokens outside the labeled entities.
4. Single-token entities receive a `B-` tag.
5. BIO tagging makes entity boundaries explicit and is widely used in token-level NER.
