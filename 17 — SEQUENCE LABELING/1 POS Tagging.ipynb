# Module 17.1 — Part-of-Speech (POS) Tagging 🧠


## 1. What is POS tagging?

Part-of-Speech tagging is the process of assigning a grammatical category to each word in a sentence.

Consider:

> The clever student writes quickly.

| Word    | POS tag          | Meaning                        |
| ------- | ---------------- | ------------------------------ |
| The     | Determiner (DET) | Specifies the noun             |
| clever  | Adjective (ADJ)  | Describes the noun             |
| student | Noun (NOUN)      | Person or thing                |
| writes  | Verb (VERB)      | Action                         |
| quickly | Adverb (ADV)     | Describes how an action occurs |



The input is a sequence of words, and the output is a corresponding sequence of grammatical labels.

\\[ (w_1,w_2,\ldots,w_T) \rightarrow (t_1,t_2,\ldots,t_T) \\]

Here, \\(w_i\\) is a word and \\(t_i\\) is its POS tag.

Unlike sentiment classification, which typically produces one label for an entire sentence, POS tagging produces a label for each token.

## 2. The important POS categories



1\. Noun (NOUN)

Names a person, place, thing, or concept.

Examples: student, Chennai, book, happiness.

Sentence: The student reads a book.


2\. Verb (VERB)

Expresses an action, event, or state.

Examples: run, write, think, is.

Sentence: She writes daily.



3\. Adjective (ADJ)

Describes or modifies a noun.

Examples: tall, beautiful, red, intelligent.

Sentence: She owns a beautiful house.



4\. Adverb (ADV)

Often modifies a verb, adjective, or another adverb.

Examples: slowly, quickly, very, carefully.

Sentence: He runs quickly.



5\. Pronoun (PRON)

Typically stands in for a noun phrase or refers to a participant or entity.

Examples: I, you, he, she, they, we.

Sentence: She is studying NLP.

### Other useful categories

| POS category                | Function                  | Examples                              |
| --------------------------- | ------------------------- | ------------------------------------- |
| Determiner (DET)            | Specifies a noun          | the, a, this                          |
| Adposition (ADP)            | Expresses relationships   | in, on, under                         |
| Conjunction (CCONJ / SCONJ) | Connects words or clauses | and, but, because                     |
| Interjection (INTJ)         | Expresses a reaction      | wow, oh                               |
| Numeral (NUM)               | Expresses a number        | three, 100                            |
| Particle (PART)             | Grammatical particle      | to in to run, in some tagging schemes |
| Punctuation (PUNCT)         | Punctuation marks         | ., !, ?                               |

Different tagging systems organize certain categories differently. For example, Universal Dependencies uses `ADP` rather than dividing all prepositions and related words into a single English-specific label.

## 3. Why context matters

A word does not always have the same POS tag.



Consider the word book:

- "I read a book." → Noun
- "Please book a ticket." → Verb

The spelling is identical, but the grammatical role changes with the sentence.

This is why POS tagging is not simply a dictionary lookup. A tagging system must use grammatical context to determine the appropriate label.

For NLP models, this is a sequence-labeling problem: the prediction for one token can depend on surrounding tokens.

## 4. How POS tagging is used in NLP

- Grammar analysis: Identifying sentence structure.
- Information extraction: Supporting the identification of noun phrases and entities.
- Search and text processing: Helping analyze queries and documents.
- Linguistic analysis: Studying how language is structured.
- Downstream NLP: Providing useful grammatical features for some pipelines.

Modern Transformer models can perform POS tagging through token-classification heads. Traditional rule-based and statistical taggers are also useful, depending on the application.

## What you should remember 🎯

1. POS tagging assigns a grammatical label to each token.
2. Nouns, verbs, adjectives, adverbs, and pronouns are core categories.
3. The same word can have different POS tags depending on context.
4. POS tagging is a sequence-labeling task, not whole-sentence classification.
5. Tag names and category boundaries depend on the tagging scheme.
