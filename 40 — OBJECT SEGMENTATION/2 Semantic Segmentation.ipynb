### 2.1 What is Semantic Segmentation?

Semantic Segmentation is a computer vision technique in which every pixel in an image is assigned a class label.

Unlike object detection, which identifies objects using bounding boxes, semantic segmentation classifies every pixel according to what it represents.

For example, consider a road image containing:

- Cars
- Pedestrians
- Roads
- Buildings
- Trees
- Sky

A semantic segmentation model examines the image and assigns a class to every pixel.

### 2.2 How Does It Work?

Consider an input image of dimensions 256 × 256.

The model predicts one class for each pixel.

```text
             Input Image
                  |
                  v
         Semantic Segmentation
                Model
                  |
                  v
          Pixel Classification
                  |
                  v
          Segmentation Map
                  |
                  v
       Class label for every pixel
```

Let's understand the mathematical representation.

Suppose we have five classes:

```python
classes = {
    0: "Background",
    1: "Road",
    2: "Car",
    3: "Person",
    4: "Building"
}
```

For an image of size 256 × 256:

$$
H = 256,\quad W = 256
$$

The ground-truth mask has dimensions:

$$
M \in \mathbb{Z}^{256 \times 256}
$$

Each pixel stores an integer representing its class.

For example:

```text
Input mask:

0 0 0 4 4
0 1 1 4 4
1 1 1 2 2
1 1 2 2 2
0 0 3 3 0
```

Here:

- `0` = Background
- `1` = Road
- `2` = Car
- `3` = Person
- `4` = Building

This is a simplified illustration of a segmentation mask.

### 2.3 What Happens Inside the Model?

This is where semantic segmentation becomes interesting.

The model does not directly output a colored image. It first predicts class scores for every pixel.

Suppose we have five classes.

For each pixel, the model generates five scores:

| Class | Predicted score |
|---|---:|
| Background | 0.05 |
| Road | 0.10 |
| Car | 0.75 |
| Person | 0.05 |
| Building | 0.05 |

The highest score belongs to `Car`.

Therefore, that pixel is classified as a car.

Mathematically:

$$
\hat{y}_{i,j} = \arg\max_{c} P(c \mid X_{i,j})
$$

Where:

- \\(X\\) = Input image.
- \\(i,j\\) = Pixel coordinates.
- \\(c\\) = Class.
- \\(P\\) = Predicted class probability.
- \\(\hat{y}_{i,j}\\) = Predicted class label.

For a five-class segmentation problem, the model's output logits typically have the shape:

$$
[B, C, H, W]
$$

Where:

- \\(B\\) = Batch size.
- \\(C\\) = Number of classes.
- \\(H\\) = Image height.
- \\(W\\) = Image width.

For example:

$$
[4,5,256,256]
$$

This means four images, five class scores per pixel, and a 256 × 256 output resolution.

Semantic segmentation assigns the same class label to all three cars.

```text
Car 1 → Class 2
Car 2 → Class 2
Car 3 → Class 2
```

It recognizes that all three belong to the `Car` category, but does not distinguish their individual identities.

This is an important distinction because we will study Instance Segmentation next.

### 2.5 Semantic Segmentation vs Classification vs Detection

| Task | What does the model predict? |
|---|---|
| Image classification | One or more image-level classes |
| Object detection | Class + bounding box |
| Semantic segmentation | Class for every pixel |

### 2.6 How Is It Used in Industry?

- **Autonomous driving:** Road, vehicles, pedestrians and sidewalks.
- **Medical AI:** Tumor, organ and tissue segmentation.
- **Satellite imaging:** Buildings, water bodies, forests and agricultural land.
- **Robotics:** Identifying traversable areas and obstacles.

### 2.7 Important Training Concept

Semantic segmentation is generally trained using pixel-wise Cross-Entropy Loss.

$$
L = -\frac{1}{N}\sum_{i=1}^{N}\log P(y_i \mid X_i)
$$

Here, \\(N\\) represents the number of valid pixels, and \\(y_i\\) is the ground-truth class of pixel \\(i\\).

The model is penalized when it predicts the wrong class for a pixel.

In PyTorch, the typical workflow uses:

```python
criterion = torch.nn.CrossEntropyLoss()

# logits: [B, C, H, W]
# masks:  [B, H, W] (integer class labels)

loss = criterion(logits, masks)
```

Notice that the target mask contains integer class IDs, not RGB colors.

## What You Should Remember 🧠

1. Semantic segmentation assigns a class to every pixel.
2. Its output is a pixel-level segmentation map.
3. Each class has an integer label.
4. Multiple objects belonging to the same category receive the same class label.
5. The typical model output shape is `[B, C, H, W]`.
6. Pixel-wise Cross-Entropy Loss is commonly used for training.
7. Semantic segmentation does not distinguish individual object instances.
