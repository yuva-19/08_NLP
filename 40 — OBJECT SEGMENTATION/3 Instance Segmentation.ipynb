### 3.1 What is Instance Segmentation?

Instance Segmentation identifies individual objects in an image and generates a separate pixel-level mask for each object.

Remember our previous topic, Semantic Segmentation?

Semantic segmentation identifies the class of every pixel but does not distinguish individual objects belonging to the same class.

Instance segmentation solves this limitation.

Let's take an image containing three cars.

The output can be represented as:

| Instance ID | Object class | Mask |
|---|---|---|
| 1 | Car | Mask 1 |
| 2 | Car | Mask 2 |
| 3 | Person | Mask 3 |
| 4 | Person | Mask 4 |
| 5 | Bicycle | Mask 5 |

Notice the difference:

- **Class** tells us what the object is.
- **Instance ID** tells us which individual object it is.
- **Mask** tells us which pixels belong to that object.

These three pieces of information are essential to understand instance segmentation.

**Semantic Segmentation**

All cars receive the same class label and mask color.

**Instance Segmentation**

Each individual car receives its own mask.

| Feature | Semantic | Instance |
|---|---|---|
| Pixel classification | Yes | Yes |
| Object class | Yes | Yes |
| Individual object identity | No | Yes |
| Separate mask per object | No | Yes |
| Handles overlapping instances | Cannot separate them | Designed to separate them |

### 3.4 How Are Individual Masks Represented?

Suppose the input image has dimensions:

$$
H \times W = 256 \times 256
$$

If the model detects three objects, it can produce three separate binary masks:

$$
M_1, M_2, M_3
$$

Each mask has dimensions:

$$
M_i \in \{0,1\}^{256 \times 256}
$$

For example:

```text
Mask 1 → First car
Mask 2 → Second car
Mask 3 → Third car
```

Each mask indicates whether a pixel belongs to that particular object.

In practical implementations, a model may also return:

- Bounding boxes
- Class labels
- Confidence scores
- Individual instance masks

### 3.5 How Does an Instance Segmentation Model Work?

A common architecture is Mask R-CNN.

**Original reference:**

- [What is Mask R-CNN? — SKY ENGINE AI](https://images.openai.com/static-rsc-4/uBDICO5Ug_X7NYfpIexVGDBaZC4xgx6-zZT-OS-cN766OQb68nagXAsmes7BkO1tB42UBEMPjLsYOXrRH0xx_qLcpNqm-t8Zz5joPheVsZtVbRWsRi_86mbbvzomVkRWFRXxi26oELC02buahLqFj1ez0Wa4JaOAk5z15Fk3FOg?purpose=inline)
- [skyengine.ai](https://www.skyengine.ai/blog/what-is-mask-r-cnn?utm_source=chatgpt.com)

Conceptually, its workflow is:

```text
Input Image
     |
     v
Feature Extraction
     |
     v
Region Proposal Network
     |
     v
Candidate Object Regions
     |
     v
ROI Align
     |
     +------------------+
     |                  |
     v                  v
Class Prediction   Bounding Box
     |                  |
     +--------+---------+
              |
              v
          Mask Branch
              |
              v
     Individual Object Masks
```

**Important connection to your previous Deep Learning knowledge:**

Mask R-CNN extends Faster R-CNN by adding a mask-prediction branch alongside the classification and bounding-box prediction branches.

You don't need to relearn CNN mathematics here. Focus on understanding how the additional mask branch enables pixel-level object segmentation.

## What You Should Remember 🧠

1. Instance segmentation generates a separate mask for every individual object.
2. Class represents the category; instance ID represents the individual object.
3. Multiple cars receive the same class but different instance masks.
4. Mask R-CNN is a popular instance segmentation architecture.
5. Mask R-CNN extends Faster R-CNN with an additional mask branch.
