# Module 15.1 — Document Classification 🧠


## 1. What is document classification?

Document classification is the task of automatically assigning a predefined category or label to a piece of text.

A document could be a news article, email, customer review, support ticket, or any other text.

For example, given the sentence:

> "The government announced a new economic policy to support small businesses."

A trained model might classify it as Politics or Business, depending on the categories and labeling rules used during training.



The key idea is simple:

\\[ \text{Document} \longrightarrow \text{Model} \longrightarrow \text{Category} \\]

Unlike clustering, where groups are discovered from data, document classification generally learns from labeled examples.

## 2. Four important applications


1\. News category classification

Assign news articles to categories such as Politics, Sports, Business, or Technology.

Example: An article about cricket scores → Sports.



2\. Spam detection

Classify emails or messages as Spam or Not Spam.

Example: "Claim your prize now!" → potentially Spam.



3\. Topic classification

Automatically assign topics to documents, support tickets, or research papers.

Example: An article about neural networks → Artificial Intelligence.



4\. Review classification

Categorize customer reviews by sentiment or subject.

Example: "The delivery was terrible" → Negative sentiment.

## 3. How does it work?

The traditional NLP approach uses the concepts you already learned in Modules 5–14.

    
1. Input: Collect labeled documents.
2. Preprocessing: Clean and normalize the text as appropriate.
3. Feature extraction: Convert documents into numerical features using TF-IDF.
4. Training: Train a classifier using the features and known labels.
5. Prediction: Classify previously unseen documents.
6. Evaluation: Check performance on held-out data.

The essential idea is that the model learns which combinations of words and phrases are useful for distinguishing categories

## 4. Important distinction

| Classification type | Example                                 |
| ------------------- | --------------------------------------- |
| Binary              | Spam vs Not Spam                        |
| Multiclass          | Sports vs Politics vs Business          |
| Multilabel          | One article tagged both AI and Business |

In multiclass classification, an example typically receives one class. In multilabel classification, it can receive multiple labels simultaneously.

## What you should remember 🎯

- Document classification assigns categories to text.
- It is commonly a supervised learning task using labeled examples.
- TF-IDF can convert documents into numerical features.
- Logistic Regression, Naive Bayes, and SVM are useful classical baselines.
- The same general workflow supports news, spam, topic, and review classification.
