```python
import torch
import math

Q = torch.tensor([
    [1.0, 0.0],   # token 1
    [0.0, 1.0],   # token 2
    [1.0, 1.0]    # token 3
])

K = torch.tensor([
    [1.0, 0.0],   # token 1
    [0.0, 1.0],   # token 2
    [1.0, 1.0]    # token 3
])

V = torch.tensor([
    [10.0, 0.0],  # information carried by token 1
    [0.0, 20.0],  # information carried by token 2
    [5.0, 5.0]    # information carried by token 3
])

print(Q.shape)
print(K.shape)
print(V.shape)
```

```python
scores = Q @ K.T
print(scores)
```

```python
d_k = Q.size(-1)

scaled_scores = scores / math.sqrt(d_k)

print(scaled_scores)
```

```python
attention_weights = torch.softmax(scaled_scores,dim = -1)
attention_weights
```

```python
output = attention_weights @ V

print(output)
```

```python
# making function 

def scaled_dot_product_attention(Q,K,V):

    scores = Q @ K.T

    d_k = Q.size(-1)

    scaled_scores = scores / math.sqrt(d_k)

    attention_weights = torch.softmax(scaled_scores, dim = -1)

    output = attention_weights @ V 

    return output , attention_weights
```

```python
output, attention_weights = scaled_dot_product_attention(Q,K,V)
print(output)
print(attention_weights)
```
