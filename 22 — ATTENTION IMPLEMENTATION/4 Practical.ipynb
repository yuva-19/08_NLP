```python
# we will use 

# The boy who won the race thanked his coach.
```

```python
import torch
import math

X = torch.tensor([
    [1., 0., 1., 0.],  # The
    [1., 1., 0., 0.],  # boy
    [0., 1., 1., 0.],  # who
    [0., 1., 0., 1.],  # won
    [1., 0., 0., 1.],  # the
    [1., 1., 1., 0.],  # race
    [0., 0., 1., 1.],  # thanked
    [1., 0., 1., 1.],  # his
    [0., 1., 1., 1.]   # coach
])
```

```python
W_Q = torch.tensor([
    [1., 0.],
    [0., 1.],
    [1., 0.],
    [0., 1.]
])

W_K = torch.tensor([
    [1., 0.],
    [0., 1.],
    [1., 0.],
    [0., 1.]
])

W_V = torch.tensor([
    [1., 0.],
    [0., 1.],
    [1., 1.],
    [0., 1.]
])
```

```python
Q = X @ W_Q
K = X @ W_K
V = X @ W_V

# X (9×4)
#    ×
# W_Q (4×2)
#    ↓
# Q (9×2)
```

```python
scores = Q @ K.T
```

```python
d_k = Q.size(-1)

scaled_scores = scores / math.sqrt(d_k)
```

```python
attention_weights = torch.softmax(
    scaled_scores,
    dim=-1
)
```

```python
output = attention_weights @ V
```

```python
print("Q:\n", Q)
print("K:\n", K)
print("V:\n", V)

print("Attention weights:\n", attention_weights)

print("Output:\n", output)
```

```python
tokens = [
    "The", "boy", "who", "won", "the",
    "race", "thanked", "his", "coach"
]

boy_index = 1

for token, weight in zip(tokens, attention_weights[boy_index]):
    print(f"boy → {token}: {weight.item():.4f}")
```
