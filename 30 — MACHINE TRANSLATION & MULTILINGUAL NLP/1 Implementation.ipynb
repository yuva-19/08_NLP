```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_NAME = "facebook/nllb-200-distilled-600M"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_NAME
)

model = model.to(device)
model.eval()

print("Model loaded successfully!")
```

```python
LANGUAGES = {
    "English": "eng_Latn",
    "Tamil": "tam_Taml",
    "French": "fra_Latn",
    "German": "deu_Latn",
    "Japanese": "jpn_Jpan",
}
```

```python
def translate_text(text, source_language, target_language):
    
    src_code = LANGUAGES[source_language]
    tgt_code = LANGUAGES[target_language]

    # Tell tokenizer which language the input is written in
    tokenizer.src_lang = src_code

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        translated_tokens = model.generate(
            **inputs,
            forced_bos_token_id=tokenizer.convert_tokens_to_ids(tgt_code),
            num_beams=4,
            max_new_tokens=128
        )

    translation = tokenizer.batch_decode(
        translated_tokens,
        skip_special_tokens=True
    )[0]

    return translation
```

```python
text = "I love learning artificial intelligence."

result = translate_text(
    text,
    "English",
    "Tamil"
)

print("Input :", text)
print("Output:", result)
```

```python
result = translate_text(
    "Where is the nearest railway station?",
    "English",
    "German"
)

print(result)
```

```python
result = translate_text(
    "நான் செயற்கை நுண்ணறிவை கற்றுக்கொண்டு இருக்கிறேன்.",
    "Tamil",
    "English"
)

print(result)
```
