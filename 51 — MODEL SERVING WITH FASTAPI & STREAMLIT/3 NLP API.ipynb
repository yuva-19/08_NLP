## 3. NLP API

An **NLP API** exposes NLP model functionality through HTTP endpoints so that other applications can use our models without needing to know how the model works internally.

Instead of directly interacting with a Python model, a client sends a request to an API and receives the NLP result.

For example, a frontend application can send text to our backend, which runs a sentiment classifier, summarizer, translator, or text-generation model.

## 1. NLP API Architecture

**Client Application**

Web UI · Mobile app · Streamlit

**NLP API — FastAPI**

Routes requests to the appropriate endpoint

**NLP Pipeline**

Validation → Preprocessing → Model Inference

**NLP Model**

Classifier · Summarizer · Translator · LLM

**API Response**

Prediction returned as JSON

## 2. Common NLP API Endpoints

| Endpoint          | Purpose                              | Example task                |
| ----------------- | ------------------------------------ | --------------------------- |
| `POST /predict`   | Generate a model prediction          | Sentiment prediction        |
| `POST /classify`  | Assign text to a category            | Spam / topic classification |
| `POST /summarize` | Generate a shorter version of text   | Article summarization       |
| `POST /translate` | Translate text into another language | English → Tamil             |

All these endpoints use the `POST` method because the client sends content in the request body.

## 3. Understanding each endpoint

### A. `POST /predict` — Prediction

Used when a model generates a prediction from the provided input.

Example: Sentiment analysis.

**Request**

```
{
  "text": "This product is excellent!"
}
```

**Response**

```
{
  "prediction": "positive",
  "confidence": 0.98
}
```

### B. `POST /classify` — Text Classification

Used to assign input text to a predefined category.

Applications:

- Spam detection
- News classification
- Intent classification
- Topic categorization
- Toxicity detection

**Request**

```
{
  "text": "Your account has won a free prize!"
}
```

**Response**

```
{
  "category": "spam",
  "confidence": 0.96
}
```

## 4. Implementing NLP endpoints using FastAPI

Here's a simple API structure that exposes all four NLP operations.

```
from fastapi import FastAPI
from pydantic import BaseModel, Field

app = FastAPI(title="NLP API")

# ---------- Request Schemas ----------

class TextRequest(BaseModel):
    text: str = Field(min_length=1, max_length=10000)

class TranslationRequest(BaseModel):
    text: str = Field(min_length=1)
    target_language: str

# ---------- Endpoints ----------

@app.post("/predict")
def predict(request: TextRequest):
    # Connect sentiment model here
    return {
        "prediction": "positive",
        "confidence": 0.98
    }

@app.post("/classify")
def classify(request: TextRequest):
    # Connect text classification model here
    return {
        "category": "technology",
        "confidence": 0.95
    }

@app.post("/summarize")
def summarize(request: TextRequest):
    # Connect summarization model here
    return {
        "summary": "Shortened version of the input text."
    }

@app.post("/translate")
def translate(request: TranslationRequest):
    # Connect translation model here
    return {
        "translated_text": "Translated output"
    }
```

Note: These are placeholder outputs. The endpoints are functional as API routes, but actual NLP models must be connected to produce real predictions, summaries, and translations.

## 5. How the API routes requests

```
                    FastAPI
                       │
          ┌────────────┼────────────┬─────────────┐
          ▼            ▼            ▼             ▼
      /predict     /classify   /summarize    /translate
          │            │            │             │
          ▼            ▼            ▼             ▼
    Sentiment      Classifier   Summarizer   Translator
      Model          Model         Model        Model
          │            │            │             │
          └────────────┴────────────┴─────────────┘
                       │
                       ▼
                  JSON Response
```

Each endpoint can use a different model, pipeline, preprocessing strategy, and response schema while sharing the same FastAPI application.

## 🧠 What you should remember

- An **NLP API** exposes NLP functionality to external clients.
- Each endpoint represents a particular operation.
- `/predict` → Model prediction.
- `/classify` → Assign text to a category.
- `/summarize` → Condense text.
- `/translate` → Translate between languages.
- FastAPI routes requests to the appropriate NLP pipeline.
- Different endpoints can share infrastructure while using different models.
