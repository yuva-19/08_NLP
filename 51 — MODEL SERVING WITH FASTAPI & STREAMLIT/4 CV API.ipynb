## 4. Computer Vision (CV) API

Just like NLP models, **Computer Vision models can also be deployed through APIs**.

Instead of a user running an object detection model locally inside a Python script, a client can upload an image to a backend API. The backend processes that image, runs the CV model, and returns the results.

The API acts as an interface between the client and the CV model.

## 2. Common CV API Endpoints

| Endpoint | Purpose | Example |
|---|---|---|
| `POST /detect` | Detect objects in an image | YOLO object detection |
| `POST /classify` | Classify an image | Cat vs dog |
| `POST /segment` | Segment objects or regions | Instance segmentation |

All three generally use `POST` because the client uploads image data for processing.

## 3. `POST /detect` — Object Detection

Object detection identifies **what objects are present in an image and where they are located**.

The model returns:

- Class labels
- Bounding boxes
- Confidence scores

### Applications

- Medical image classification
- Defect classification
- Plant disease detection

### Classification vs Detection

| Image Classification | Object Detection |
|---|---|
| Predicts image category | Detects individual objects |
| Usually image-level output | Object-level outputs |
| No bounding boxes required | Bounding boxes included |
| "This image contains a dog" | "Dog is located here" |

## 5. `POST /segment` — Image Segmentation

Image segmentation assigns pixels to objects or regions, providing a more detailed understanding of the image.

Instead of merely predicting a bounding box, the model determines **which pixels belong to the object**.

## 6. CV API — Request and Response Design

Unlike text-based NLP APIs, CV APIs commonly receive image files through multipart form uploads.

### Typical request

```http
POST /detect

Content-Type: multipart/form-data

image: uploaded_image.jpg
```

### Typical response

```json
{
    "image_name": "uploaded_image.jpg",
    "detections": [
        {
            "label": "person",
            "confidence": 0.94,
            "bbox": [45, 30, 210, 400]
        }
    ]
}
```

The model's raw output is postprocessed into a structured response that a frontend can display.

## 7. FastAPI implementation pattern for CV

A CV API commonly uses `UploadFile` and `File` to receive image uploads.

```python
from fastapi import FastAPI, UploadFile, File, HTTPException
from PIL import Image
import io

app = FastAPI()

@app.post("/detect")
async def detect_objects(
    file: UploadFile = File(...)
):
    if not file.content_type.startswith("image/"):
        raise HTTPException(
            status_code=400,
            detail="Please upload an image"
        )

    image_bytes = await file.read()

    try:
        image = Image.open(
            io.BytesIO(image_bytes)
        ).convert("RGB")

    except Exception:
        raise HTTPException(
            status_code=400,
            detail="Invalid image file"
        )

    # Run the detection model here
    # results = model(image)

    return {
        "filename": file.filename,
        "message": "Image processed successfully"
    }
```

This is a basic upload and validation example; the actual detection model and its postprocessing need to be connected.

## 8. NLP API vs CV API

| Aspect | NLP API | CV API |
|---|---|---|
| Input | Text, tokens, documents | Images, video frames |
| Request type | JSON | Multipart file upload commonly |
| Preprocessing | Tokenization, masks | Resize, normalize, image conversion |
| Model | NLP / Transformer | CNN, ViT, YOLO, segmentation model |
| Output | Text, labels, scores | Labels, boxes, masks |
| Example | `/summarize` | `/detect` |

## 🧠 What you should remember

- `POST /detect` → Detect objects and their locations.
- `POST /classify` → Predict an image-level category.
- `POST /segment` → Identify object/region pixels.
- CV APIs commonly accept image uploads using `multipart/form-data`.
- FastAPI `UploadFile` is useful for handling uploaded images.
- Image validation and preprocessing occur before model inference.
- Model outputs are postprocessed into structured JSON or visual outputs.
