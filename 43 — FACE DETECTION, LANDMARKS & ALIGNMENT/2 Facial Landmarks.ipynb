## 1. What are Facial Landmarks?

Facial landmarks are specific reference points on a human face that represent important facial structures.

Each landmark is represented using 2D image coordinates:

$$
P_i=(x_i,y_i)
$$

Where:

- \\(x_i\\) = Horizontal coordinate.
- \\(y_i\\) = Vertical coordinate.
- \\(P_i\\) = One facial landmark.

A face landmark detector produces a collection of these coordinates.

### Example

```python
landmarks = {
    "left_eye": (120, 100),
    "right_eye": (190, 105),
    "nose": (155, 145),
    "mouth_left": (130, 185),
    "mouth_right": (180, 185)
}
```

These coordinates describe the positions of important facial features within an image or face crop.

**Important distinction:** A bounding box represents a region; a landmark represents a specific point.

![Facial landmarks representing individual facial reference points](https://images.openai.com/static-rsc-4/Ou5u44FEU_6b7uZTUKS42pWZU6ViP3mONX62q0ZZpwZi0dnufNqIRjEmLjVZ1jhXRa9rWyoUJ62fNZvW988clr58yNzErMPjD8vrHlVjjoHZQS0votrljqtn3KSp0Jfemfcr3JbLfXQpGK8PvBQiYIeN9GG1Ti4CgNiPYmXEZC0?purpose=inline)

### Bounding Box vs Facial Landmark

| Bounding Box | Facial Landmark |
|---|---|
| Represents a rectangular region. | Represents a specific reference point. |
| Contains the face region. | Identifies individual facial features. |
| Uses coordinates describing boundaries. | Uses coordinates describing keypoints. |
| Example: Face location. | Example: Left eye location. |

## 2. Eyes

Eyes are among the most important facial reference structures.

A landmark detector can identify:

- Left eye position.
- Right eye position.
- Inner eye corners.
- Outer eye corners.
- Upper and lower eyelid positions.

![Facial eye landmark detection showing eye reference points](https://images.openai.com/static-rsc-4/Ou5u44FEU_6b7uZTUKS42pWZU6ViP3mONX62q0ZZpwZi0dnufNqIRjEmLjVZ1jhXRa9rWyoUJ62fNZvW988clr58yNzErMPjD8vrHlVjjoHZQS0votrljqtn3KSp0Jfemfcr3JbLfXQpGK8PvBQiYIeN9GG1Ti4CgNiPYmXEZC0?purpose=inline)

### Why are eye landmarks important?

**A. Face alignment**

The relative positions of the eyes help determine the orientation of a face.

**B. Head pose estimation**

Eye positions contribute to estimating whether someone is looking straight, turning or tilting their head.

**C. Eye tracking**

Eye landmarks can help determine gaze direction when combined with iris or pupil localization.

**D. Facial recognition preprocessing**

Eye positions can be used as reference points to align faces before extracting recognition features.

## 5. Face Geometry

This is where the individual landmarks become particularly useful.

Face geometry refers to the spatial relationships between facial landmarks.

Instead of looking at eyes, nose and mouth independently, we analyze their relative positions, distances, angles and proportions.

![Facial geometry and expression analysis using landmark relationships](https://images.openai.com/static-rsc-4/tepR_XuJgiZABr37wuLbvoMLM1wRD8w7RMFJpJB9IWIOvyqSrirNn7m9pDayV3GvL4OYbfRCQyfB5c_UFIXakqT0UDu8BxFVx4AoluhZ3zd3nOUaKXd4iGkSWzZjN5PEc3UXXFhVpOhX3T69OyfCWbdEKx7AmxCD_eT36rBT0yI?purpose=inline)

### A. Euclidean Distance

Suppose we have two eye landmarks:

$$
P_1=(x_1,y_1)
$$

$$
P_2=(x_2,y_2)
$$

The distance between them is:

$$
d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

**Example:**

Left eye = (100, 100)

Right eye = (160, 100)

$$
d=\sqrt{(160-100)^2+(100-100)^2}
$$

$$
d=\sqrt{60^2+0^2}
$$

$$
\boxed{d=60\ pixels}
$$

This gives us the distance between the two eye reference points.

### B. Facial Proportions

We can compare different facial distances.

For example:

$$
R=\frac{\text{Eye-to-eye distance}}{\text{Nose-to-mouth distance}}
$$

Ratios are useful because they can be more robust to changes in image scale than raw pixel distances.

**Why?**

Suppose an image is resized by a factor of 2.

Both the eye-to-eye distance and nose-to-mouth distance will approximately double.

However, their ratio remains unchanged:

$$
R=\frac{2d_1}{2d_2}
$$

$$
R=\frac{d_1}{d_2}
$$

This makes ratios useful for analyzing relative facial geometry across different image scales.

### C. Facial Symmetry

We can compare corresponding landmark positions on the left and right sides of a face.

Examples:

- Left eye vs right eye.
- Left mouth corner vs right mouth corner.
- Left cheek contour vs right cheek contour.

This can help quantify geometric asymmetry.

However, real faces are naturally asymmetric, and camera angle, expression and lighting can affect these measurements.

## 6. Practical Implementation — Calculate Facial Geometry

Let's implement a small example using Python and NumPy.

```python
import numpy as np

# Example facial landmarks
left_eye = np.array([100, 100])
right_eye = np.array([160, 100])
nose = np.array([130, 145])
mouth_left = np.array([110, 180])
mouth_right = np.array([150, 180])

# Calculate eye distance
eye_distance = np.linalg.norm(
    right_eye - left_eye
)

# Calculate nose-to-mouth distance
mouth_center = (mouth_left + mouth_right) / 2

nose_mouth_distance = np.linalg.norm(
    mouth_center - nose
)

# Calculate facial geometry ratio
ratio = eye_distance / nose_mouth_distance

print("Eye distance:", eye_distance)
print("Nose-mouth distance:", nose_mouth_distance)
print("Geometry ratio:", ratio)
```

### Expected Output

```text
Eye distance: 60.0
Nose-mouth distance: 35.0
Geometry ratio: 1.714
```

### Understanding the Implementation

**Step 1 — Define facial landmarks**

We represent each facial feature using a NumPy array containing its X and Y coordinates.

**Step 2 — Calculate eye distance**

```python
eye_distance = np.linalg.norm(
    right_eye - left_eye
)
```

First, NumPy subtracts the coordinates:

$$
(160,100)-(100,100)=(60,0)
$$

Then `np.linalg.norm()` calculates the Euclidean distance:

$$
d=\sqrt{60^2+0^2}=60
$$

**Step 3 — Calculate mouth center**

```python
mouth_center = (mouth_left + mouth_right) / 2
```

The mouth center is calculated by averaging the two mouth corner coordinates.

$$
M=\frac{(110,180)+(150,180)}{2}
$$

$$
M=(130,180)
$$

**Step 4 — Calculate nose-to-mouth distance**

```python
nose_mouth_distance = np.linalg.norm(
    mouth_center - nose
)
```

$$
d=\sqrt{(130-130)^2+(180-145)^2}
$$

$$
d=35
$$

**Step 5 — Calculate facial geometry ratio**

```python
ratio = eye_distance / nose_mouth_distance
```

$$
R=\frac{60}{35}
$$

$$
R\approx1.714
$$

### What did we accomplish?

We extracted geometric information from facial landmark coordinates without training any deep learning model.

This is an important distinction:

- **Landmark detection:** Deep learning/computer vision model predicts coordinates.
- **Geometry analysis:** Mathematical operations extract relationships from those coordinates.

The two can be combined in a complete facial analysis pipeline.

## Technical Clarifications & Additional Engineering Notes

### 1. Landmark Coordinates Depend on the Coordinate System

Landmark coordinates can be represented relative to:

- The original image.
- A cropped face image.
- Normalized image coordinates.
- A 3D face coordinate system.

Always check which coordinate system the model returns before performing geometric calculations.

For example, a landmark at `(120, 100)` in a cropped face image does not necessarily represent `(120, 100)` in the original image.

If the crop starts at `(x_offset, y_offset)`, the original image coordinates are:

$$
x_{original}=x_{crop}+x_{offset}
$$

$$
y_{original}=y_{crop}+y_{offset}
$$

### 2. Facial Geometry Is Not the Same as Facial Identity

Facial geometry describes spatial relationships between facial features.

It can be used in facial analysis, alignment and recognition preprocessing, but a simple collection of geometric distances does not automatically identify a person.

Modern face recognition systems generally use learned feature representations rather than relying exclusively on manually calculated facial proportions.

### 3. Landmark Detection vs Geometry Analysis

Remember this distinction when building practical Computer Vision systems:

| Stage | Input | Output |
|---|---|---|
| Face detection | Image | Face bounding boxes |
| Landmark detection | Image or detected face crop | Facial keypoint coordinates |
| Geometry analysis | Landmark coordinates | Distances, angles and ratios |
| Facial alignment | Landmark coordinates and face image | Standardized face image |

This distinction will become especially important in the next section.
