## 1.1 What is Digital Audio?

Before understanding how machines process speech, we need to understand how sound is represented inside a computer.

### 1. What is sound?

Sound is a mechanical wave produced by vibrations.

For example:

- When we speak, our vocal cords vibrate.
- When a guitar string vibrates, it produces sound.
- When a speaker vibrates, it produces sound waves.

Sound travels through a medium such as air in the form of pressure variations.

A microphone converts these physical sound waves into an electrical signal.

### 2. What is digital audio?

Digital audio is the representation of sound as numerical data that a computer can store, process, and analyze.

Computers cannot directly process continuous physical sound waves. Therefore, we convert sound into digital values.

This conversion involves three major steps:

1. Sampling
2. Quantization
3. Encoding

### 3. Sampling

Sampling means measuring the amplitude of an analog signal at regular time intervals.

Imagine recording someone saying:

"Hello, how are you?"

The microphone produces a continuous electrical signal. The computer takes measurements of that signal at specific time intervals.

Each measurement is called a sample.

![Audio sampling rate illustration](https://images.openai.com/static-rsc-4/27LR4Qb3xgb9LyGeihiAL0Z16TP4OX2vv4TCMhPDB2afJrXCxtWUiFjFphDlOQBC8ywqkiqYmT-61JKL_EGS4JjyhE1c_Cmf7Fk8NUvLOw_2hQ97-6meTpROG5HYqNQNqh-nOMwKRuRIb3UzNkfCloR9p4OdThDzL9F7Ehyc07Y?purpose=inline)

Source: [SFX Engine](https://sfxengine.com/blog/best-sample-rate-for-audio)

**Sampling rate (sample rate)**

The sampling rate represents the number of samples captured per second.

It is measured in Hertz (Hz).

| Sampling rate | Meaning |
|---|---|
| 8,000 Hz | 8,000 samples per second |
| 16,000 Hz | 16,000 samples per second |
| 22,050 Hz | 22,050 samples per second |
| 44,100 Hz | 44,100 samples per second |
| 48,000 Hz | 48,000 samples per second |

For example:

If we record audio at 16 kHz:

$$
16000 \text{ samples per second}
$$

That means the computer captures 16,000 amplitude measurements every second.

**Why is sampling rate important?**

A higher sampling rate can preserve higher-frequency information, but it also increases the amount of data.

According to the Nyquist sampling theorem:

$$
f_s \geq 2f_{\max}
$$

Where:

- \(f_s\) = Sampling frequency
- \(f_{\max}\) = Highest frequency we want to capture

For example, to represent frequencies up to 8 kHz, the sampling rate must be at least 16 kHz, assuming appropriate anti-alias filtering.

### 4. Quantization

After sampling, each sample has an amplitude value.

However, computers represent these amplitudes using a finite number of discrete levels.

The process of mapping each sample's amplitude to one of those levels is called quantization.

![Quantization and digital data representation](https://images.openai.com/static-rsc-4/fDdGI6dQiupUCv3ohdtOqvXDyCifC--uX37ou0uYQ4ze7e690ujVM42pjRIntl2ScXsVEAfUvKLpC-p3lU3jsrf_D-0_bUHhBCtql-VZRJ2Kk4sgGCgAulCQvVR4GgCBTXDrvoNb7g4XkrM2Q32PnJ1hBqXKXRSuArHY7GKXb-E?purpose=inline)

Source: [MasteryMind](https://masterymind.co.uk/study-guides/computer-science-ocr-gcse-data-representation)

**Bit depth**

Bit depth determines how many amplitude levels are available.

$$
\text{Number of levels} = 2^N
$$

Where \(N\) is the number of bits.

| Bit depth | Quantization levels |
|---|---:|
| 8-bit | 256 |
| 16-bit | 65,536 |
| 24-bit | 16,777,216 |
| 32-bit | 4,294,967,296 |

Higher bit depth generally provides greater amplitude precision and a lower quantization noise floor.

### 5. Encoding

Once the signal has been sampled and quantized, the numerical values are represented in binary form.

This allows computers to store and process audio.

Common audio formats include:

| Format | Description |
|---|---|
| WAV | Common uncompressed audio container |
| MP3 | Lossy compressed audio |
| FLAC | Lossless compressed audio |
| AAC | Lossy compressed audio |
| OGG | Audio container commonly using Vorbis or Opus |

### 6. Mono vs Stereo Audio

**Mono audio**

- One audio channel.
- One sequence of samples.

**Stereo audio**

### 7. Calculate Audio File Size

Let's calculate the size of an uncompressed audio recording.

Suppose:

- Duration = 10 seconds
- Sampling rate = 16,000 Hz
- Bit depth = 16 bits
- Channels = 1 (Mono)

**Formula:**

$$
\text{File size (bits)} = T \times f_s \times B \times C
$$

Substitute the values:

$$
= 10 \times 16000 \times 16 \times 1
$$

$$
= 2,560,000 \text{ bits}
$$

**Convert into bytes:**

$$
\frac{2,560,000}{8} = 320,000 \text{ bytes}
$$

**Result:**

Approximately 320 KB (decimal units), excluding file headers and metadata.

## 1.2 Speech Processing vs Audio Processing

These two terms are related, but they are not identical.

### Audio Processing

Audio processing deals with manipulating, analyzing, enhancing, or understanding audio signals in general.

It includes:

- Music processing
- Noise reduction
- Echo cancellation
- Sound classification
- Audio enhancement
- Environmental sound detection

### Speech Processing

Speech processing specifically focuses on human speech signals.

Its objective is to help machines analyze, understand, recognize, or generate spoken language.

It includes:

- Speech recognition
- Speech synthesis
- Speaker identification
- Speech enhancement
- Speech emotion recognition
- Language identification

Source: [xDev Asia](https://xdev.asia/series/ai-machine-learning/speech-audio-ai-xu-ly-giong-noi-am-thanh/)

**Key difference:**

| Audio Processing | Speech Processing |
|---|---|
| Deals with all types of sound | Focuses on human speech |
| Music, noise, environmental sounds | Spoken words and vocal characteristics |
| Music classification | Speech recognition |
| Noise cancellation | Speech enhancement |
| Sound event detection | Speaker recognition |

**Important:** Speech processing is a specialized area within the broader field of audio processing.

## 1.3 Speech Recognition vs Speech Synthesis

These are two important and opposite directions of speech technology.

### Speech Recognition

Speech recognition converts spoken language into machine-readable information, commonly text.

**Example:**

You say:

"Turn on the lights."

The system recognizes the spoken words and produces:

`Turn on the lights.`

This technology is commonly used in voice assistants, dictation systems, and transcription software.

### Speech Synthesis

Speech synthesis does the opposite.

It converts text or other linguistic representations into artificial speech.

**Example:**

Input text:

`Hello, how are you?`

Output: A computer-generated voice speaking those words.

![Voice AI system](https://images.openai.com/static-rsc-4/a-JQYpzcWyUbDmuPpMgTi7HiohL-T1rZF9tCtAwkstlBjXuOBF-7aw40nK7CGbIx34x2ztZ2_-opOQBMb-mKFQe6l5D70SwcpWil5FF497hAhimw2ka28Lt7D-NmGqZuYyIxhqeI0o9GfjfUj5jDrHxLnz8gOouqN6Qg2heRi3I?purpose=inline)

|---|---|
| Speech → Text | Text → Speech |
| Understands spoken input | Generates spoken output |

## 1.4 Automatic Speech Recognition (ASR)

Automatic Speech Recognition (ASR) is the technology that enables computers to automatically recognize human speech and convert it into text.

It is one of the most important applications of speech processing.

### How does an ASR system work?

Consider the sentence:

"Artificial intelligence is interesting."

An ASR pipeline typically follows these stages:

![Automatic Speech Recognition pipeline](https://images.openai.com/static-rsc-4/MuOWJh_LzFcLTLgmP-rx0yaV1AVEb1JjxlBrOZKpb64Dz3-YNxw8BCcBsiH76aNy-kZacqHHThANGiNJHdFnn9tGqGTJ2wY0S_I2-ifDPunzYFWuWtV5LrUt8K1oYHBgJ0uuKZhvw39javnoGAUsVZmjXCc18kWTeqjT7Jur3Rw?purpose=inline)

Source: [Balto](https://www.balto.ai/blog/automatic-speech-recognition-meaning/)

**Step 1 — Audio Input**

The user speaks into a microphone.

The microphone captures the sound and converts it into a digital audio signal.

**Step 2 — Audio Preprocessing**

The system prepares the audio for recognition.

Common operations include:

- Resampling
- Noise reduction
- Volume normalization
- Silence detection
- Audio segmentation

**Step 3 — Feature Extraction / Audio Representation**

The system converts raw audio into a representation that a model can learn from.

Examples:

- Spectrogram
- Mel Spectrogram
- MFCC (Mel-Frequency Cepstral Coefficients)

We will study these in detail later.

**Step 4 — Acoustic or Speech Model**

The model learns patterns in speech and predicts linguistic units, such as phonemes, tokens, or text pieces.

Modern systems may use Transformer-based architectures and directly predict text tokens from audio.

**Step 5 — Language Modeling / Decoding**

The system determines the most likely sequence of words.

For example, it must distinguish between words that sound similar using the audio context and linguistic context.

**Step 6 — Text Output**

The final recognized sentence is:

`Artificial intelligence is interesting.`

### Popular ASR systems

- OpenAI Whisper
- Google Speech-to-Text
- Amazon Transcribe
- Mozilla DeepSpeech
- NVIDIA NeMo

## 1.5 Speech-to-Text Systems

Speech-to-Text (STT) systems convert spoken language into written text.

STT is the practical application of ASR.

**For example:**

**Audio Input**

"I want to learn machine learning."

**Speech-to-Text Model**

Audio processing and recognition.

**Text Output**

I want to learn machine learning.

### Types of STT systems

**1. Offline Speech Recognition**

- Audio is recorded first.
- The complete audio is processed.
- Text is generated after processing.

**2. Real-time Speech Recognition**

- Audio is processed continuously.
- Text appears while the person is speaking.
- Used in live captions and voice assistants.

### Applications

- Voice typing
- Meeting transcription
- YouTube subtitles
- Call-center transcription
- Medical dictation
- Voice-controlled applications

## 1.6 Audio Classification

Audio classification is the process of assigning a category or label to an audio recording.

Unlike ASR, the objective is not necessarily to identify the spoken words.

For example, consider these audio recordings:

| Audio input | Predicted class |
|---|---|
| Dog barking | Animal sound |
| Car horn | Vehicle sound |
| Human laughing | Human vocalization |
| Rain falling | Environmental sound |
| Piano playing | Musical instrument |

![Audio classification using YAMNet](https://images.openai.com/static-rsc-4/4E0QrrX9oXaFo_3Nwl1xEI10VzYRFK2CvieV99klmJ1JKpp1sDpbtaiEOYnrRa6fZFA--Uku39XXHpDIY8lIay3wraKW3x99YvSZsVptO0Xd5dDnP7iK1ta8NJUU035eRstEONAYS_JY33tymh5H5WwfDwAw5Y1caMINYiKLwHo?purpose=inline)

### How does audio classification work?

1. Collect audio recordings.
2. Preprocess the audio.
3. Extract audio features or representations.
4. Feed the representations into a classification model.
5. Predict the sound category.

Common models and architectures include:

- CNNs
- Recurrent Neural Networks (RNNs)
- Transformers
- Audio Spectrogram Transformer (AST)

**Example project:** Build an AI model that identifies whether an audio recording contains a dog bark, siren, car horn, or human speech.

## 1.7 Speech Emotion Recognition (SER)

Speech Emotion Recognition is the process of identifying emotional characteristics from a person's voice.

The objective is to predict the emotion conveyed through speech.

**For example:**

A person says:

"I'm doing really well today!"

The system analyzes vocal characteristics and may predict:

`Happy`

Another person says the same sentence in a sad tone. The model may predict:

`Sad`

### What features can reveal emotion?

- **Pitch:** How high or low the voice sounds.
- **Energy:** The strength or intensity of the voice.
- **Speaking rate:** How quickly someone speaks.
- **Tone and prosody:** Patterns of pitch, stress, and rhythm.
- **Voice quality:** Characteristics such as breathiness or roughness.

![Natural language processing and speech projects](https://images.openai.com/static-rsc-4/ZLCmRjYl7B9_dpj07QGLp6az9a5aCKEywVKJ1H3480qX4JD0hDHwkl1_JsGdhOTkqMced4A_ytOGendIJj2I_baGmddNUu0jKiUlUM6yEmuDs6F0PIJsuxGqrssTLY99SwSKv5MwrtU7Q9KzqSnG8Ikoj0CHWu7s5teFuNh5vQA?purpose=inline)

Source: [Tutort](https://www.tutort.net/blogs/explore-12-natural-language-processing-nlp-projects-for-all-levels)

### Common emotion classes

- Happy
- Sad
- Angry
- Neutral
- Fearful
- Disgust
- Surprised

### Applications

- Customer service analytics
- Call-center quality analysis
- Human-computer interaction
- Voice-based conversational systems
- Research on emotional communication

**Important limitation:** A voice-based emotion classifier estimates patterns associated with emotion. It cannot reliably determine someone's true internal emotional state from their voice alone.
