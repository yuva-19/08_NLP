## 2.1 Sound Waves

### What is a sound wave?

Sound is a mechanical wave produced by the vibration of an object.

When an object vibrates, it creates variations in the pressure of the surrounding medium (usually air).

These pressure variations travel through the medium as sound waves.

For example:

- Vocal cords vibrate when we speak.
- A guitar string vibrates when plucked.
- A speaker diaphragm vibrates when playing music.

![Sound wave propagation](https://images.openai.com/static-rsc-4/3cFob_cv8zUvJo6A2AL-k9DP33eZPvdl_ujHoitFZsowIRUhLxbFagQYaZMiMuSrzByLI8ROmvZBNC2eqayKLxg1nrjepeNN2iUT5Je1GJ2T8vQa7me_SvV1Aej34QJ_aFE_b4d5pqIn6ImCxtMA7wz4mLyK5QDS8imzS0HtNrM?purpose=inline)

Source: [Geniebook](https://geniebook.com/tuition/secondary-4/physics/sound)

### Compression and rarefaction

Sound waves traveling through air consist of two important regions:

**1. Compression**

- Air molecules are pushed closer together.
- Pressure becomes higher.
- Represents a high-pressure region.

**2. Rarefaction**

- Air molecules move farther apart.
- Pressure becomes lower.
- Represents a low-pressure region.

![Compression and rarefaction](https://images.openai.com/static-rsc-4/ODGf1tOAP3dYguJdJI96JLDuFT_uNHcSeVa3vK3V5E3Y8Sjw3-8cV112bH9pRaUsOOODpyCmucuEPbklCJIpHu57pZgujVG8SUUbmCebmAp10Wid6tR1TBERYAZioWaxXr5B3CDcKHUYabiJIhmr3uvYwApqc8meVabZXsxIjWE?purpose=inline)

These alternating pressure variations produce the sound wave.

**Important:** Sound requires a medium to travel. It cannot propagate through a vacuum.

## 2.2 Understanding the Properties of a Sound Wave

Let's understand six fundamental properties:

1. Amplitude
2. Frequency
3. Time period
4. Wavelength
5. Phase
6. Waveform

### 1. Amplitude

Amplitude represents the maximum displacement of a wave from its equilibrium position.

In an audio signal, amplitude is related to the strength of the pressure variation.

![Amplitude illustration](https://images.openai.com/static-rsc-4/5nDWU8U1nVtderOkntd75taf9Fb2bKI3vlcwX8Bs-tg5OBspeHbcduPK9mZx7lPF8ThVSVrOnu9sZwa3YT16iWwCLm_afJ6-h5mBgAtSR0hGV1CPpV83EjFlUhOREZEc9Y03W4wPzJYUiDO6ik_YIWuO6P61F6GSOG1WAnJMaFk?purpose=inline)

Source: [Surety Home](https://suretyhome.com/how-wi-fi-motion-sensing-works/)

**What does amplitude tell us?**

- Higher amplitude → generally perceived as louder sound.
- Lower amplitude → generally perceived as quieter sound.

For a sinusoidal signal:

$$
x(t)=A\sin(2\pi ft+\phi)
$$

Where:

- \(A\) = Amplitude
- \(f\) = Frequency
- \(t\) = Time
- \(\phi\) = Phase

The amplitude is the maximum absolute value of the signal.

**Example:**

Suppose two audio signals have amplitudes:

$$
A_1=0.2
$$

$$
A_2=0.8
$$

The second signal has four times the amplitude of the first.

Note that perceived loudness also depends on frequency, duration, and human hearing sensitivity.

### 2. Frequency

Frequency represents the number of complete cycles of a wave occurring per second.

It is measured in Hertz (Hz).

$$
f=\frac{\text{Number of cycles}}{\text{Time}}
$$

**Example:**

- 100 Hz = 100 cycles per second.
- 500 Hz = 500 cycles per second.
- 1,000 Hz = 1,000 cycles per second.

**Relationship between frequency and pitch:**

| Frequency | Perceived pitch |
|---|---|
| Low frequency | Low-pitched sound |
| High frequency | High-pitched sound |

For example:

- Bass guitar → generally lower frequencies.
- Bird chirping → often higher frequencies.
- Human voice → contains a wide range of frequencies.

The approximate human hearing range for healthy young listeners is 20 Hz to 20 kHz.

### 3. Time Period

Time period represents the time required to complete one full cycle of a periodic wave.

It is represented by \(T\).

$$
T=\frac{1}{f}
$$

Where:

- \(T\) = Time period in seconds.
- \(f\) = Frequency in Hertz.

**Example 1:**

If frequency is 100 Hz:

$$
T=\frac{1}{100}
$$

$$
T=0.01\text{ seconds}
$$

Therefore, one cycle takes 10 milliseconds.

**Example 2:**

If frequency is 1,000 Hz:

$$
T=\frac{1}{1000}
$$

$$
T=0.001\text{ seconds}
$$

One cycle takes 1 millisecond.

![Frequency and time period](https://images.openai.com/static-rsc-4/FP5c96xFWQ_R3lWPMqttNw6MHP3LJtTfYcvTAHODzClxNrnyBbTiZdWza0z8nU8BfRqrlfOS264IabfY0wiNjl6xDdrBhU3J4NZWCtzb--W0GEkTn_nG1d2OrGBMt34EIfqR6KtUjaCejPvGXxJ27lJcSR9I06dDAo-0JVr4DzA?purpose=inline)

Source: [Electrical Volt](https://www.electricalvolt.com/difference-between-period-and-frequency/)

**Important relationship:**

- Higher frequency → shorter time period.
- Lower frequency → longer time period.

### 4. Wavelength

Wavelength is the physical distance covered by a wave during one complete cycle.

It is represented by the Greek letter lambda (\(\lambda\)).

$$
\lambda=\frac{v}{f}
$$

Where:

- \(\lambda\) = Wavelength in metres.
- \(v\) = Speed of sound in the medium (m/s).
- \(f\) = Frequency (Hz).

For sound traveling through air at approximately 343 m/s:

**Example:**

If frequency is 343 Hz:

$$
\lambda=\frac{343}{343}
$$

$$
\boxed{\lambda=1\text{ metre}}
$$

![Wavelength illustration](https://images.openai.com/static-rsc-4/ZIXsDMfTXwCUWJcsDPmCqGjRaIFfTETW-PH8FsScfHQZ5pyAyYPLEhNWBW2svNXv1fCwvznxFhzufLhKNFkf-Cwo-oOtz-RhxcvNvBs-8_8vw5pTNNXhrRFmkgZ4JNgHDpztLkQVUbVoNuhuI-KpjiDAyxKRmu_B37EN9q3soI?purpose=inline)

Source: [Winwinchip](https://www.winwinchip.com/blog/detail/wavelength-explained-definition-formula-spectrum-types-real-world-applications/70.html)

**Important relationship:**

For a constant speed of sound:

- Higher frequency → shorter wavelength.
- Lower frequency → longer wavelength.

### 5. Phase

Phase describes the position of a periodic wave within its cycle relative to a reference point.

It is generally measured in degrees or radians.

![Wave phase illustration](https://images.openai.com/static-rsc-4/nCnauzxg8oRp3mQ68_yhKLBWM733_09oQpxbEmT9ybSRurgLXcOkykm1BZIbqJWM2v2izuA6QOXgFfo0NoDrlQrgquTpceP3DkTk26tAqv_BS1hQXO1SqNaMzXAZACKCeMB-ogEQVQz7Cv9Yp7Fq5cWfvTCbrRpeRvIsNQ_sqDE?purpose=inline)

Source: [Factual Audio](https://factualaudio.com/post/phase/)

**Common phase differences:**

| Phase difference | Meaning |
|---|---|
| 0° | Waves are in phase |
| 90° | Quarter-cycle difference |
| 180° | Half-cycle difference |
| 360° | One complete cycle difference |

Consider two identical sine waves:

$$
x_1(t)=A\sin(2\pi ft)
$$

$$
x_2(t)=A\sin(2\pi ft+\phi)
$$

If:

$$
\phi=180^\circ=\pi\text{ radians}
$$

Then the waves are opposite in phase.

When two identical signals are added, they can cancel each other.

This principle is important in:

- Noise cancellation
- Audio mixing
- Stereo processing
- Signal interference

### 6. Waveform

A waveform is the visual representation of how a signal's amplitude changes over time.

![Different waveforms](https://images.openai.com/static-rsc-4/ZaafNdDQXXuPEi8vqnrnszplsWv5QLHgEtYMitCDA4RrQH3lF5E3qBeFnXmpktC2ZHlf-hHjBh51_V6rJ6Bw0lWwr1Yoq6lG6ZzxuX-BloQ7gDt-irAOHcN-f9K0V2T937C9rBY7Px5pI9CeN2sVv8WxvdAln_65jBEgSHc9VuI?purpose=inline)

Source: [The Science of Music](https://websites.umich.edu/~mejn/tsom/figures.html)

**Common waveform shapes:**

| Waveform | Characteristics |
|---|---|
| Sine wave | Smooth periodic waveform |
| Square wave | Alternates between two levels |
| Triangle wave | Linear rising and falling |
| Sawtooth wave | Repeated linear rise and abrupt fall |
| Speech waveform | Complex, changing signal |

Real human speech is not a simple sine wave. It contains many frequency components combined together.

## 2.3 Analog vs Digital Signals

Now let's connect the physical sound wave to computer representation.

### Analog signal

An analog signal is continuous in time and amplitude.

It can take any amplitude value within its allowed range at any instant.

Examples:

- Microphone's analog electrical output.
- Analog audio from a tape recorder.
- Electrical signals in traditional audio equipment.

![Analog signal](https://images.openai.com/static-rsc-4/nmkgJQyRvCkBOqj3j67RjXebbQqZ4hPm_ys2w9XPU2nYsc7EZRFo5UCeXtiiGmM7hx66DIQIpgO7uLoxx-yA41QnwtxCr_uae-LZUCrYdzYEt6-6UpNEB2oCW3-xOJBjZaKMtRCKIV5dqOXzy00uxNFaY7YOFMHHuzmJo6MsgK8?purpose=inline)

Source: [PF Nicholls](https://pfnicholls.com/Electronics/signals.html)

### Digital signal

A digital signal represents information using discrete numerical values.

In digital audio, the signal is represented by a sequence of samples, with quantized amplitude values when using fixed-point PCM.

![Digital audio sampling](https://images.openai.com/static-rsc-4/CHR_Dy9LJY61E1RdDSr-soreNfk-hdutEHn3kD42zN8ayPjBePW12l1Lzt0O8oBKq7mYBMkO9gAdlhxbg3oHizoQ-nS-LEE6g1oTgzbuCxjS_r_jJl_KtJ9b8Be_zffPjs4wRqyPA8sXnFutj45kr0MlNYPG2ax72o5_ahRiHqs?purpose=inline)

Source: [ACE Studio](https://acestudio.ai/blog/what-is-audio-sample-rate/)

### Comparison

| Analog signal | Digital signal |
|---|---|
| Continuous time | Discrete time |
| Continuous amplitude | Discrete amplitude levels in quantized audio |
| Physical electrical waveform | Numerical representation |
| Can be affected by analog noise | Can be copied without accumulating analog-generation noise |
| Difficult to process directly using digital computers | Convenient for computer processing |

**Conversion process:**

```text
Physical Sound
      ↓
Vibrations in air
      ↓
Microphone
      ↓
Sound → Analog electrical signal
      ↓
ADC
(Analog-to-Digital Converter)
      ↓
Digital Audio
      ↓
Numerical samples for computer processing
```

The reverse conversion uses a DAC (Digital-to-Analog Converter) to produce an analog signal for speakers or headphones.

## 2.4 Continuous vs Discrete Signals

These concepts are fundamental to understanding digital signal processing.

### Continuous-time signal

A continuous-time signal exists at every instant of time.

It is represented as:

$$
x(t)
$$

Here, time \(t\) can take any value within the signal's duration.

For example, an analog microphone signal.

### Discrete-time signal

A discrete-time signal exists only at specific time instants.

It is represented as:

$$
x[n]
$$

Here, \(n\) represents an integer sample index.

For example:

$$
x[0],x[1],x[2],x[3],\ldots
$$

These values represent the samples collected from the original continuous signal.

**Important distinction:**

| Term | Meaning |
|---|---|
| Continuous-time | Defined at every instant of time |
| Discrete-time | Defined at specific time instants |
| Continuous-amplitude | Can take any amplitude value |
| Discrete-amplitude | Can take only certain amplitude levels |

A digital PCM audio signal is both discrete in time and quantized in amplitude.

A discrete-time signal does not necessarily have quantized amplitude; quantization is an additional operation.

## 2.5 Important Audio Parameters

Now let's move into the practical parameters you'll encounter when loading audio datasets in Python.

### 1. Sampling Rate (Hz)

Sampling rate is the number of audio samples captured per second.

$$
f_s=\frac{1}{T_s}
$$

Where:

- \(f_s\) = Sampling rate.
- \(T_s\) = Time between consecutive samples.

**For example:**

At 16 kHz:

$$
f_s=16000\text{ Hz}
$$

$$
T_s=\frac{1}{16000}=62.5\ \mu s
$$

Therefore, one sample is captured every 62.5 microseconds.

#### Common sampling rates

| Sampling rate | Typical applications |
|---|---|
| 8 kHz | Traditional narrowband telephony |
| 16 kHz | Speech recognition and speech datasets |
| 22.05 kHz | Music and general audio processing |
| 44.1 kHz | Audio CDs and music distribution |
| 48 kHz | Video, film, broadcasting and professional audio |

![Digital audio sampling rate](https://images.openai.com/static-rsc-4/zgPelLem1fmSW5ZLg8atjH1ulJcgDZfkbXzwM76Y5WuZJY7SuIWHaaJWoZWkuB9IY9CnHlp7OQ9nXwKfD1WIKGfGf-noX7v1LhaJm3tU_kqWxEDTPslebbS_lVr6nWDYZqatnbNbUyCATh1XqAOx-xvDgTtRK5Lf3p66SPHUne0?purpose=inline)

Source: [Digital Audio Sampling Rate](https://storage.googleapis.com/dtjcciheadcbke/digital-audio-sampling-rate.html)

**Why are different sampling rates used?**

Different applications have different bandwidth and storage requirements.

For example:

- Telephone speech traditionally uses 8 kHz sampling.
- Speech recognition models frequently use 16 kHz audio.
- Music production commonly uses 44.1 or 48 kHz.
- Video production commonly uses 48 kHz.

Remember: a higher sampling rate is not automatically better for every AI model. You should use the rate expected by the model and dataset.

### 2. Bit Depth

Bit depth represents the number of bits used to represent each audio sample in fixed-point PCM audio.

The number of possible quantization levels is:

$$
L=2^N
$$

Where \(N\) is bit depth.

| Bit depth | Quantization levels |
|---|---:|
| 8-bit | 256 |
| 16-bit | 65,536 |
| 24-bit | 16,777,216 |
| 32-bit | 4,294,967,296 |

**Example:**

For 16-bit audio:

$$
2^{16}=65,536
$$

This means there are 65,536 possible quantization codes.

#### Why is bit depth important?

Higher bit depth provides:

- Better amplitude resolution.
- Lower quantization noise under comparable conditions.
- Greater dynamic range.

For ideal uniform PCM quantization, the theoretical signal-to-quantization-noise ratio for a full-scale sine wave is approximately:

$$
SNR=6.02N+1.76\text{ dB}
$$

For 16-bit audio:

$$
SNR\approx98.08\text{ dB}
$$

This is an ideal theoretical value, not a guarantee of real recording quality.

### 3. Number of Channels

An audio channel represents an independent stream of audio samples.

Common configurations:

- Mono: 1 channel.
- Stereo: 2 channels.
- 5.1 surround: 6 channels.
- 7.1 surround: 8 channels.

![Spatial audio channels](https://images.openai.com/static-rsc-4/vUFXx4k0hmixn3Dv1yizsrxgI3NSz0WMFLSZy8PpRwiProkE_InAUS2QD02N2EmgZA_lYWVIP8_EJHwRMD1z3D9S-xwJRpqFj_M7NXX4mQn5AcUuHFyiZUy9BGDOwJAKhmVxrLVOUoHOpT5Jp6K1MxYNfwc879YrnZjs1pm5IpI?purpose=inline)

Source: [IEEE MUJ](https://medium.com/%40ieeemuj/apple-is-revolutionizing-the-way-you-listen-through-spatial-audio-a5a8162fa497)

**Why do channels matter in AI/ML?**

Suppose you load a stereo recording.

Its data may have two separate sequences:

$$
X=
\begin{bmatrix}
L_0 & L_1 & L_2 & \cdots \\
R_0 & R_1 & R_2 & \cdots
\end{bmatrix}
$$

Where:

- \(L\) = Left channel.
- \(R\) = Right channel.

Many speech models expect mono audio. Therefore, you may need to convert stereo audio into mono before inference.

A simple average of left and right channels is one possible downmixing method:

$$
x_{\text{mono}}[n]=\frac{L[n]+R[n]}{2}
$$

However, downmixing should be performed carefully because phase differences can cause cancellation.

### 4. Mono vs Stereo

![Mono and stereo comparison](https://images.openai.com/static-rsc-4/117y_gJo3pQqMCPkyE76Ms8FAkMpx3eXLXjQzNCyT8fz46Gz0kcvMab-Fv4UtMnDEiBTVAEjqLUA4bEowHcdvMfRag8664oBEnW4PxOlxPyRy_zxKeuer2I0KN8Bzpr1KONoknTLEymwq73d7b7LoL09QEGFtZSHpM-McRDDqUw?purpose=inline)

**Mono**

- One channel.
- Single audio stream.
- Simpler storage.
- Common in speech datasets.

**Stereo**

- Two channels.
- Left and right streams.
- Can preserve spatial positioning.
- Common in music.

Important: Stereo does not necessarily mean better speech recognition. A mono signal is often sufficient for speech tasks.

### 5. Audio Duration

Audio duration is the total length of a recording.

For a fixed-rate digital audio signal:

$$
T=\frac{N}{f_s}
$$

Where:

- \(T\) = Duration in seconds.
- \(N\) = Number of samples per channel.
- \(f_s\) = Sampling rate.

**Example:**

Suppose an audio recording has:

- 160,000 samples per channel.
- Sampling rate = 16,000 Hz.

$$
T=\frac{160000}{16000}
$$

$$
\boxed{T=10\text{ seconds}}
$$

This is particularly important when preparing audio datasets.

### 6. Bitrate

Bitrate represents the amount of data transmitted or stored per second.

It is usually measured in bits per second (bps) or kilobits per second (kbps).

For uncompressed PCM audio:

$$
\text{Bitrate}=f_s\times B\times C
$$

Where:

- \(f_s\) = Sampling rate.
- \(B\) = Bit depth.
- \(C\) = Number of channels.

**Example:**

Consider:

- Sampling rate = 44,100 Hz.
- Bit depth = 16 bits.
- Channels = 2.

$$
=44100\times16\times2
$$

$$
=1,411,200\text{ bps}
$$

$$
\boxed{1411.2\text{ kbps}}
$$

#### Bitrate in compressed audio

For formats such as MP3, bitrate describes the compressed data rate.

Common MP3 bitrates:

- 64 kbps
- 128 kbps
- 192 kbps
- 256 kbps
- 320 kbps

Higher bitrate generally means less compression-related information loss at comparable encoding settings, but quality also depends on the encoder and source.

**Important difference:**

- Sampling rate → Samples captured per second.
- Bit depth → Bits representing each PCM sample.
- Bitrate → Bits stored or transmitted per second.

These three parameters are related but are not interchangeable.

### 7. Audio Formats

Audio formats determine how audio data is stored and encoded.

Let's understand the three formats you must know.

**MP3 (.mp3)**

- Lossy compression format.
- Reduces file size.
- Removes some audio information.
- Common for music distribution and streaming.

![FLAC format](https://images.openai.com/static-rsc-4/yCiMJb6qybwQA8baquxgfYNkDd4g-twuhFxnOViL4oeYp5FDAz7fZG8xD_N1MQB-6N3KYtztIEp-4xWejZEnwPwQ8alAFAHSdo_POnVcSXWehvJf6oD7hR1z9vFLRkco7HsSsgQbQqyvhJBvrig09S6qxFrEqzh3c2E20Enkspg?purpose=inline)

- Lossless compression format.
- Retains the original audio samples when decoded.

#### Comparison

| Feature | WAV (PCM) | MP3 | FLAC |
|---|---|---|---|
| Compression | Usually none | Lossy | Lossless |
| Audio information | Preserved | Some information discarded | Preserved |

Technical note: WAV is a container format, not a single audio encoding. It can store different types of audio data, although uncompressed PCM is particularly common.

## 2.7 Python Practical — Understanding Audio Signals

Now let's connect theory with Python.

We'll use:

- `NumPy` — Numerical operations.
- `Matplotlib` — Signal visualization.
- `SoundFile` — Audio file reading and writing.
- `Librosa` — Audio analysis.

### Install the required libraries

```bash
pip install numpy matplotlib soundfile librosa
```

### Example 1 — Generate a sine wave

```python
import numpy as np
import matplotlib.pyplot as plt

# Audio parameters
frequency = 440
sampling_rate = 16000
duration = 1

# Generate time values
t = np.arange(
    int(sampling_rate * duration)
) / sampling_rate

# Generate sine wave
amplitude = 0.5

audio = amplitude * np.sin(
    2 * np.pi * frequency * t
)

# Plot waveform
plt.figure(figsize=(12, 4))

plt.plot(t[:500], audio[:500])

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("440 Hz Sine Wave")

plt.grid()
plt.show()
```

**What are we doing here?**

1. Creating a 440 Hz signal.
2. Sampling it at 16,000 Hz.
3. Generating one second of audio.
4. Plotting the amplitude against time.

### Example 2 — Calculate signal properties

```python
import numpy as np

frequency = 440
sampling_rate = 16000
duration = 2
bit_depth = 16
channels = 1

number_of_samples = sampling_rate * duration

time_period = 1 / frequency
wavelength = 343 / frequency

bitrate = sampling_rate * bit_depth * channels

file_size = (
    number_of_samples
    * bit_depth
    * channels
) / 8

print("Frequency:", frequency, "Hz")
print("Time period:", time_period, "seconds")
print("Wavelength:", wavelength, "metres")
print("Samples:", number_of_samples)
print("Bitrate:", bitrate / 1000, "kbps")
print("File size:", file_size / 1000, "KB")
```

### Example 3 — Read an actual audio file

Place a file named `speech.wav` in your working directory before running this example.

```python
import soundfile as sf
import numpy as np

audio, sample_rate = sf.read("speech.wav")

print("Sampling rate:", sample_rate)
print("Audio shape:", audio.shape)
print("Duration:", len(audio) / sample_rate)
print("Maximum amplitude:", np.max(np.abs(audio)))
```

**Notice:**

- `sample_rate` tells us the sampling frequency.
- `audio.shape` tells us the number of samples and possibly channels.
- `len(audio)` gives the number of frames.
- Duration is calculated by dividing frames by sampling rate.

````python
import numpy as np
import matplotlib.pyplot as plt

# Audio parameters
frequency = 440
sampling_rate = 16000
duration = 1

# Generate time values
t = np.arange(
    int(sampling_rate * duration)
) / sampling_rate

# Generate sine wave
amplitude = 0.5

audio = amplitude * np.sin(
    2 * np.pi * frequency * t
)

# Plot waveform
plt.figure(figsize=(12, 4))

plt.plot(t[:500], audio[:500])

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")

plt.title("440 Hz Sine Wave")
plt.grid()

plt.show()
````

````python
import numpy as np

frequency = 440
sampling_rate = 16000
duration = 2
bit_depth = 16
channels = 1

number_of_samples = sampling_rate * duration

time_period = 1 / frequency

wavelength = 343 / frequency

bitrate = sampling_rate * bit_depth * channels

file_size = (
    number_of_samples
    * bit_depth
    * channels
) / 8

print("Frequency:", frequency, "Hz")
print("Time period:", time_period, "seconds")
print("Wavelength:", wavelength, "metres")
print("Samples:", number_of_samples)
print("Bitrate:", bitrate / 1000, "kbps")
print("File size:", file_size / 1000, "KB")
````
