# 34.10 Speech Recognition Evaluation 🎙️

Bro, in the previous module, we implemented Whisper and converted speech into text. But here's an important question:

**How do we know whether the model has actually recognized the speech correctly?**

A transcription can look grammatically correct but still contain incorrect words.

For example:

**Actual speech:**

> I want to learn machine learning.

**Whisper prediction:**

> I want to learn deep learning.

Both sentences make sense, but the model has recognized one word incorrectly.

Speech recognition evaluation measures how accurately an ASR system converts spoken language into text.

---

## 1. Word Error Rate (WER)

<badge color="success">Must Know</badge>

Word Error Rate is one of the most widely used metrics for evaluating speech recognition systems.

It measures how many word-level errors occur in a predicted transcription compared with the reference transcription.

$$
\boxed{WER=\frac{S+D+I}{N}}
$$

Where:

- $S$ = Substitutions
- $D$ = Deletions
- $I$ = Insertions
- $N$ = Number of words in the reference transcription

Usually, WER is expressed as a percentage:

$$
WER(\%)=\frac{S+D+I}{N}\times100
$$

### 1.1 Substitution

A substitution happens when the model recognizes one word instead of another.

**Reference:**

```text
I like machine learning
```

**Prediction:**

```text
I like deep learning
```

**Comparison:**

| Reference | Prediction | Result |
|---|---|---|
| I | I | Correct |
| like | like | Correct |
| machine | deep | Substitution |
| learning | learning | Correct |

$$
S=1
$$

### 1.2 Deletion

A deletion happens when the model completely misses a word that was spoken.

**Reference:**

```text
I like machine learning
```

**Prediction:**

```text
I like learning
```

| Reference | Prediction | Result |
|---|---|---|
| I | I | Correct |
| like | like | Correct |
| machine | — | Deletion |
| learning | learning | Correct |

$$
D=1
$$

### 1.3 Insertion

An insertion happens when the model generates a word that was not present in the reference transcription.

**Reference:**

```text
I like machine learning
```

**Prediction:**

```text
I really like machine learning
```

| Reference | Prediction | Result |
|---|---|---|
| I | I | Correct |
| — | really | Insertion |
| like | like | Correct |
| machine | machine | Correct |
| learning | learning | Correct |

$$
I=1
$$

## 2. Calculate WER — Complete Example

Let's combine all three types of errors.

**Reference transcription:**

```text
I want to learn machine learning today
```

**Predicted transcription:**

```text
I need to learn deep learning
```

Let's compare the words:

| Reference | Prediction | Result |
|---|---|---|
| I | I | Correct |
| want | need | Substitution |
| to | to | Correct |
| learn | learn | Correct |
| machine | deep | Substitution |
| learning | learning | Correct |
| today | — | Deletion |

Therefore:

- $S=2$
- $D=1$
- $I=0$
- $N=7$

$$
WER=\frac{2+1+0}{7}
$$

$$
WER=\frac{3}{7}=0.4286
$$

<badge color="info">42.86% WER</badge>

This means that the transcription has a word error rate of approximately 42.86%.

**Important:** WER uses the minimum number of edit operations needed to align the reference and predicted word sequences. The simple word-by-word alignment above is illustrative; actual evaluation uses sequence alignment.

Also, WER is not simply the percentage of words that are wrong. It counts substitutions, deletions and insertions, and can exceed 100% when there are many insertions.

## 3. Character Error Rate (CER)

WER measures errors at the word level.

But what happens when we need to evaluate individual characters?

This is where **Character Error Rate (CER)** is useful.

$$
\boxed{CER=\frac{S_c+D_c+I_c}{N_c}}
$$

Where:

- $S_c$ = Character substitutions
- $D_c$ = Character deletions
- $I_c$ = Character insertions
- $N_c$ = Number of characters in the reference

### Example

**Reference:**

```text
CAT
```

**Prediction:**

```text
BAT
```

**Comparison:**

| Reference | Prediction | Result |
|---|---|---|
| C | B | Substitution |
| A | A | Correct |
| T | T | Correct |

Therefore:

$$
CER=\frac{1}{3}=0.3333
$$

$$
CER=33.33\%
$$

### WER vs CER

| Feature | WER | CER |
|---|---|---|
| Evaluation unit | Words | Characters |
| Measures | Word-level errors | Character-level errors |
| Formula denominator | Reference word count | Reference character count |
| Useful for | General transcription | Character-sensitive evaluation |
| Particularly relevant | English and many languages | Languages with complex spelling and tokenization |

### Why is CER important for Tamil?

Tamil words can contain multiple Unicode code points, and spelling differences can occur within a word.

For example, a transcription may contain a small character-level error even when most of the word is correct.

CER can reveal such differences that WER treats as a whole-word error.

One technical detail: Unicode characters may consist of multiple code points. For reliable multilingual evaluation, normalization and the definition of a character must be consistent.

## 4. Transcription Accuracy

Transcription accuracy describes how correctly an ASR system recognizes spoken words.

A simple metric sometimes used is:

$$
Accuracy=1-WER
$$

Or:

$$
Accuracy(\%)=(1-WER)\times100
$$

For example, if WER is 20%:

$$
Accuracy=100-20=80\%
$$

But be careful: This is a derived score, not a universal ASR accuracy metric. Since WER includes insertions and can exceed 100%, the resulting accuracy can become negative. Therefore, when reporting ASR performance, always report WER or CER directly rather than relying only on this accuracy calculation.

### What should we evaluate?

Apart from WER and CER, consider:

- Whether the complete sentence was recognized.
- Whether important technical words were recognized.
- Whether numbers, names and abbreviations were recognized correctly.
- Whether punctuation and capitalization are appropriate.
- Whether the model correctly handles different speakers.
- Whether it produces reasonable results in noisy environments.

For example, in an engineering transcription:

**Actual:**

> The motor operates at 1500 RPM.

**Prediction:**

> The motor operates at 150 RPM.

The sentence looks almost correct, but the numerical error is significant.

This is why evaluating domain-specific terms is important.

## 5. Language-Specific Errors

Speech recognition performance can vary significantly between languages.

A model trained extensively on English may not perform equally well on Tamil, Hindi or Malayalam.

Some common challenges:

### 5.1 Pronunciation Differences

The same word can be pronounced differently depending on the speaker's region and accent.

### 5.2 Similar-Sounding Words

Some words have similar acoustic patterns, making them difficult to distinguish.

### 5.3 Complex Writing Systems

Languages have different writing systems, character combinations and spelling conventions.

For Tamil, Unicode normalization and consistent text processing are important when calculating CER.

### 5.4 Code-Switching

Code-switching means mixing two or more languages within a sentence.

Example:

> "Bro, inniku machine learning class irukku."

A model may recognize the English words correctly but struggle with the Tamil-English combination.

### 5.5 Vocabulary Differences

Regional expressions, names, technical terminology and uncommon words may be poorly represented in training data.

**Important:** Do not compare WER across languages without considering differences in tokenization, normalization and reference transcription conventions.

Common noise sources:

| Noise | Possible effect on ASR |
|---|---|
| Traffic | Masks speech sounds |
| Fan noise | Interferes with quieter speech |
| Wind | Distorts microphone recordings |
| Background conversations | Creates competing speech |
| Reverberation | Makes speech sounds overlap |
| Electrical interference | Adds unwanted signal components |

For example:

**Actual speech:**

> I want to buy a new laptop.

**Prediction in clean audio:**

> I want to buy a new laptop.

**Possible prediction in noisy audio:**

> I want to buy a laptop.

The word "new" has been deleted.

### How can we evaluate noise robustness?

Record the same sentence under different conditions:

1. Quiet room.
2. Fan running.
3. Traffic noise.
4. Background conversation.

Run the same ASR model on all four recordings and compare WER.

This helps us understand how noise affects recognition.

**Important:** Keep the spoken sentence identical across experiments. Otherwise, we won't know whether the difference in WER is caused by noise or by changes in the spoken content.

## 7. Accent-Related Recognition Errors

An accent changes how speech sounds are produced.

For example, speakers may differ in:

- Vowel pronunciation.
- Consonant articulation.
- Speaking speed.
- Intonation.
- Word stress.
- Regional pronunciation.

Consider three speakers saying:

> "I am learning Python."

The words are identical, but their acoustic characteristics may differ.

An ASR model may perform well for one speaker and make errors for another.

### How to evaluate accent robustness

Use multiple speakers:

| Recording | Speaker | Accent |
|---|---|---|
| Audio 1 | Speaker A | Indian English |
| Audio 2 | Speaker B | British English |
| Audio 3 | Speaker C | American English |
| Audio 4 | Speaker D | Different regional accent |

Keep the sentence identical, then calculate WER separately for each recording.

This helps identify whether a model performs consistently across speakers.

One important research principle: use enough speakers and recordings before drawing conclusions about an entire accent group. A single recording cannot represent an accent.

## 8. Practical Lab — Evaluate Whisper Using WER and CER 🧪

Now let's extend our previous Whisper project.

Previously, we used:

```python
result = model.transcribe("speech.wav")

recognized_text = result["text"]

print(recognized_text)
```

Now we'll evaluate that prediction against the actual spoken sentence.

### 8.1 Install the Evaluation Library

```bash
pip install jiwer
```

We will use `jiwer` for WER and CER calculations.

### 8.2 Prepare Reference and Predicted Text

```python
import whisper
from jiwer import wer, cer

model = whisper.load_model("base")

audio_path = "speech.wav"

result = model.transcribe(
    audio_path,
    language="English",
    task="transcribe"
)

reference = """
I want to learn artificial intelligence
and machine learning.
"""

hypothesis = result["text"]

print("Reference:")
print(reference)

print("\nPredicted:")
print(hypothesis)
```

### 8.3 Calculate WER

```python
word_error = wer(
    reference,
    hypothesis
)

print("WER:", word_error)
print("WER Percentage:", word_error * 100, "%")
```

### 8.4 Calculate CER

```python
character_error = cer(
    reference,
    hypothesis
)

print("CER:", character_error)
print("CER Percentage:", character_error * 100, "%")
```

### 8.5 Identify Substitution, Deletion and Insertion Errors

We can also inspect the individual error counts.

```python
from jiwer import process_words

evaluation = process_words(
    reference,
    hypothesis
)

print("Correct Words:", evaluation.hits)
print("Substitutions:", evaluation.substitutions)
print("Deletions:", evaluation.deletions)
print("Insertions:", evaluation.insertions)
print("WER:", evaluation.wer)
```

This gives us a more useful evaluation than looking at the final WER alone.

### 8.6 Normalize Text Before Evaluation

Suppose the reference is:

```text
Hello, I am learning Python!
```

And the prediction is:

```text
hello i am learning python
```

These sentences contain the same words, but punctuation and capitalization differ.

We can normalize the text before evaluating it.

```python
import re
import unicodedata

def normalize_text(text):

    text = unicodedata.normalize("NFC", text)

    text = text.lower()

    text = re.sub(r"[^\w\s]", "", text)

    text = re.sub(r"\s+", " ", text)

    return text.strip()


normalized_reference = normalize_text(reference)

normalized_hypothesis = normalize_text(hypothesis)

print(
    "Normalized WER:",
    wer(normalized_reference, normalized_hypothesis)
)

print(
    "Normalized CER:",
    cer(normalized_reference, normalized_hypothesis)
)
```

This is a simple normalization pipeline. For Tamil and other languages, don't blindly remove all punctuation or characters without considering language-specific conventions. Apply the same documented normalization to both reference and prediction.

## 9. Evaluate Different Recording Conditions

Let's build a reusable evaluation function.

### 9.1 Create an Evaluation Function

```python
from jiwer import wer, cer

def evaluate_asr(reference, audio_path, model):

    result = model.transcribe(
        audio_path,
        task="transcribe",
        language="English"
    )

    prediction = result["text"]

    word_error = wer(reference, prediction)

    character_error = cer(reference, prediction)

    print("\nAudio:", audio_path)
    print("Reference:", reference)
    print("Prediction:", prediction)

    print("WER:", round(word_error * 100, 2), "%")
    print("CER:", round(character_error * 100, 2), "%")

    return {
        "audio": audio_path,
        "prediction": prediction,
        "WER": word_error,
        "CER": character_error
    }
```

### 9.2 Evaluate Multiple Recordings

Now evaluate multiple recordings containing the same sentence:

```python
reference = "I want to learn artificial intelligence."

recordings = [
    "clean.wav",
    "fan_noise.wav",
    "traffic_noise.wav",
    "background_conversation.wav"
]

results = []

for audio_path in recordings:

    evaluation = evaluate_asr(
        reference,
        audio_path,
        model
    )

    results.append(evaluation)
```

### 9.3 Organize Results into a Table

```python
import pandas as pd

df = pd.DataFrame(results)

print(
    df[["audio", "WER", "CER"]]
)
```

This gives you a basic ASR evaluation report.

### Important Evaluation Practices

When conducting a proper ASR evaluation:

1. Use accurate human-verified reference transcripts.
2. Keep the test recordings separate from model training data.
3. Document normalization rules.
4. Use identical evaluation conditions when comparing models.
5. Report WER and CER together when appropriate.
6. Evaluate different languages, accents and noise conditions separately.
7. For a dataset, calculate aggregate WER using total errors divided by total reference words, rather than blindly averaging individual recording WERs.

## What You Should Remember 🧠

- WER is a major metric for evaluating ASR systems.
- CER evaluates transcription at the character level.
- Substitution, deletion and insertion are the three fundamental error types.

- Language, accent, noise and recording conditions can affect recognition.
- Always compare model predictions against reliable reference transcripts.
