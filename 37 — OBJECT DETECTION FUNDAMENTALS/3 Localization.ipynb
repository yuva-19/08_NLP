## 3. Localization

We've already learned how bounding boxes represent object positions. Now let's understand localization, which is the process of determining where an object is located within an image.

In object detection, a model must solve two problems simultaneously:

1. **Classification:** What object is present?
2. **Localization:** Where is that object located?

## 3.1 Object Centre

The object centre represents the geometric centre of its bounding box.

Instead of describing the object's location using its top-left and bottom-right corners, we can describe it using its centre coordinates.

$$
(x_c,y_c)
$$

![GitHub - tejaslodaya/car-detection-yolo: Autonomous driving - car detection using the very powerful YOLO model](https://images.openai.com/static-rsc-4/1LEpQSKgVzfzuoWFcmwaqiEH7t4ZnU8i52WpyVg61Apc0ChgRGolkGQm85L6Li9vqOLqG-Sn1K-UYvWqJ8pPhkEResFrlxmpq_NAnNTykQGHBnX7t0ctpNPCKAmTb3GV5PEQprOsa3qz5TyfMIjVhuQbo_7GzVBv7yJYHRvUfO4?purpose=inline)

For a bounding box represented by:

$$
(x_1,y_1,x_2,y_2)
$$

The centre coordinates are:

$$
x_c=\frac{x_1+x_2}{2}
$$

$$
y_c=\frac{y_1+y_2}{2}
$$

### Practical example

Suppose:

```python
x1 = 100
y1 = 50
x2 = 300
y2 = 200
```

Calculate the centre:

$$
x_c=\frac{100+300}{2}=200
$$

$$
y_c=\frac{50+200}{2}=125
$$

**Object centre**

$$
(x_c,y_c)=(200,125)
$$

This tells us that the centre of the detected object is located at pixel coordinate `(200, 125)`.

**Why is this important?**

Many object detection architectures predict bounding boxes using centre coordinates along with width and height.

## 3.2 Width and Height

The width and height describe the spatial dimensions of the detected object.

![A novel UAV-integrated deep network detection and relative position estimation approach for weeds](https://images.openai.com/static-rsc-4/c0AWAAYyqh-D6-KqRRzqV_L7L5TAh09RoJLYQYHcgMWSJOrAq0oF8cXjjccNLl2N_z9VMwjaMpN0LjIftjlJFTvLN-LSoCNSmJhXIQznQIUtc0q7xiQSxlKLD3S2cfeaMTQr6-pc5LXnhFc78LIrWduKVLUfNqFKa2f8XfwJatY?purpose=inline)

For XYXY coordinates:

$$
w=x_2-x_1
$$

$$
h=y_2-y_1
$$

Using our previous example:

```python
x1, y1 = 100, 50
x2, y2 = 300, 200
```

**Width:**

$$
w=300-100=200
$$

**Height:**

$$
h=200-50=150
$$

Therefore, the complete bounding box representation is:

$$
(x_c,y_c,w,h)=(200,125,200,150)
$$

The centre determines the position, while width and height determine the size.

After locating an object, the model must determine which category it belongs to.

This is called class prediction.

The model might predict:

| Detected object | Predicted class |
|---|---|
| Object 1 | Car |
| Object 2 | Bus |
| Object 3 | Person |
| Object 4 | Motorcycle |

### How does class prediction work?

A detection model extracts image features and uses a classification component to predict object categories.

For example, suppose the model supports three classes:

```python
classes = ["Car", "Person", "Bicycle"]
```

For one detected object, its class prediction scores might be:

| Class | Score |
|---|---:|
| Car | 0.92 |
| Person | 0.05 |
| Bicycle | 0.03 |

The predicted class is:

$$
Class=\arg\max(P(class_i))
$$

Therefore:

$$
Class=\text{Car}
$$

Here, `argmax` selects the class with the highest predicted score.

**Important:** Class prediction and localization are different outputs. A model can identify the correct class but predict an inaccurate bounding box.

## 3.4 Confidence Score

**Definition:** A confidence score represents how strongly a detection model supports a particular predicted detection.

It helps determine whether a prediction should be accepted or discarded.

![YOLO Object Detection on the Raspberry Pi](https://images.openai.com/static-rsc-4/zeQ6iEXhgG1o4zsxQIavXvAjRoMYDXv0zANnd6tKwPn2yoDjN9X5HrXw1dm7r3rtVPHAQe267ZT9AW9mbdmV_1-YiQAj1BT3cUKjNrBeLQST_ikiDiYH5AsY0jQ_KMaLTIa3d44f5tHi9oajcpBnX-ehkRbnt_F_PRAVnzRw6uQ?purpose=inline)

[readmedium.com](https://readmedium.com/yolo-object-detection-on-the-raspberry-pi-6de3629256fa?utm_source=chatgpt.com)

For example:

```text
Car       → 0.98
Person    → 0.94
Bicycle   → 0.42
```

A higher score generally indicates stronger model confidence in the detection.

### Confidence threshold

A confidence threshold is a minimum score used to filter predictions.

Suppose:

$$
Threshold=0.50
$$

| Detection | Confidence | Decision |
|---|---:|---|
| Car | 0.98 | Accept |
| Person | 0.94 | Accept |
| Bicycle | 0.42 | Reject |

The bicycle prediction is rejected because:

$$
0.42 < 0.50
$$

### What happens when we change the threshold?

![napari-nuclephaser · PyPI](https://images.openai.com/static-rsc-4/DW-1Q4av_yUjChvmoVYCQpAhs-_2T8x26cxpPPEoWVeM_lSLkfgnWuLRxmBZ1DZpVQqARnmkyUeCTsUX-1fngN_FXE4kEw8mqnZ0_mLxPgWMDP7WHYO1Du7nDvib7Qa_0drM8jMGm1YV2uGFuACPGjI9hMNMYo8maMvSl_sg8YU?purpose=inline)

- **Low threshold:** More detections are accepted, including potentially incorrect predictions.
- **High threshold:** Fewer detections are accepted, but some genuine objects may be missed.

This creates a trade-off between precision and recall.

### Important distinction: Objectness vs Class Confidence

In some detection architectures, particularly traditional YOLO formulations, the model distinguishes:

- **Objectness score:** How likely the predicted region contains an object.
- **Class probability:** Which class the object belongs to.

In such formulations, the class-specific detection confidence is commonly calculated as:

$$
Confidence(class)=P(Object)\times P(Class\mid Object)
$$

For example:

$$
P(Object)=0.90
$$

$$
P(Car\mid Object)=0.95
$$

Then:

$$
Confidence(Car)=0.90\times0.95=0.855
$$

**Note:** Confidence-score definitions and output structures vary across detection architectures and versions. Do not assume every modern detector uses this exact multiplication.
