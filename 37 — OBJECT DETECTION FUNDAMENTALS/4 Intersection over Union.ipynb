## 4. Intersection over Union (IoU)

This is one of the most important mathematical concepts in object detection.

So far, we've learned how an object detection model predicts bounding boxes. But an important question remains:

**How do we measure whether the predicted bounding box is actually accurate?**

For example, imagine that an image contains a car.

- The dataset provides the correct bounding box (Ground Truth).
- Our model predicts another bounding box (Predicted Box).

Even if both boxes surround the same car, their positions and dimensions might differ.

We need a mathematical method to measure how closely they overlap.

That method is called **Intersection over Union (IoU)**.

## 4.1 What is IoU?

**Definition:** Intersection over Union measures the overlap between two bounding boxes by dividing their intersection area by their union area.

The two boxes are:

1. **Ground Truth Box:** The actual bounding box provided by the dataset annotation.
2. **Predicted Box:** The bounding box generated by the object detection model.

The formula is:

$$
\boxed{
IoU=\frac{\text{Area of Intersection}}{\text{Area of Union}}
}
$$

Let's understand each term.

### A. Intersection

Intersection represents the area shared by both bounding boxes.

![Introduction to Object Detection - Analytics Vidhya](https://images.openai.com/static-rsc-4/IWVWDPEeq9md4hZbTellGP1-mz3KcE3y69CHwc4vd54gCL9gKEkqDVSmilvZoMyAD-tBqi9k_BRtV1i2yXFoI3nwJl1HjCXRTJ7ibywmuUoqNifM0UrpzA6kFEtlgYNzDXKAyl5QQe7kh-1eqQNc1A7bcNCFpsWzaBpMw5iY9rQ?purpose=inline)

Imagine two rectangles overlapping each other.

The common region covered by both rectangles is their intersection.

### B. Union

Union represents the total area covered by both bounding boxes, counting their shared region only once.

![Key Object Detection Metrics for Computer Vision](https://images.openai.com/static-rsc-4/eUvigLD0D-mBS3HqOU_keugbDOZFU1WP0KJk1XmpCpdtS1wwQUHdU-JBqdIR9y1J0FXiLYgKxqR97CLlQZdV9KXnQ3tPKjRELj_DvOyuGPLkNZe9c6WUB5N99CqdFzuaKvyKjWQZcFUVuCCIYVGrMJxsISOvig-jr4mat-vI-tI?purpose=inline)

Mathematically:

$$
Area(A\cup B)=Area(A)+Area(B)-Area(A\cap B)
$$

**Why do we subtract the intersection?**

Because when adding the areas of both boxes, the overlapping region gets counted twice.

We subtract it once to avoid double counting.

## 4.2 Understanding the IoU Formula

Let's combine both concepts.

$$
IoU=\frac{A\cap B}{A\cup B}
$$

Where:

- \\(A\\) = Ground Truth bounding box.
- \\(B\\) = Predicted bounding box.
- \\(A\cap B\\) = Intersection area.
- \\(A\cup B\\) = Union area.

Expanding the formula:

$$
\boxed{
IoU=
\frac{Intersection}
{Area(A)+Area(B)-Intersection}
}
$$

### IoU value range

IoU always lies between 0 and 1.

| IoU | Interpretation |
|---:|---|
| 0.0 | No overlap |
| 0.25 | Small overlap |
| 0.50 | Moderate overlap |
| 0.75 | High overlap |
| 1.0 | Perfect overlap |

## 4.3 Numerical Example — Calculate IoU

Let's calculate IoU manually.

Suppose we have two bounding boxes.

**Ground Truth Box (A):**

```python
A = (100, 100, 300, 300)
```

**Predicted Box (B):**

```python
B = (150, 150, 350, 350)
```

Both use the XYXY format.

### Step 1 — Calculate individual areas

Ground Truth:

$$
Area(A)=(300-100)(300-100)
$$

$$
Area(A)=200\times200=40000
$$

Predicted Box:

$$
Area(B)=(350-150)(350-150)
$$

$$
Area(B)=200\times200=40000
$$

### Step 2 — Calculate intersection

To find the overlapping rectangle:

$$
x_{\text{left}}=\max(100,150)=150
$$

$$
y_{\text{top}}=\max(100,150)=150
$$

$$
x_{\text{right}}=\min(300,350)=300
$$

$$
y_{\text{bottom}}=\min(300,350)=300
$$

Therefore:

$$
Intersection=(300-150)(300-150)
$$

$$
\boxed{Intersection=22500}
$$

### Step 3 — Calculate union

$$
Union=A+B-Intersection
$$

$$
Union=40000+40000-22500
$$

$$
\boxed{Union=57500}
$$

### Step 4 — Calculate IoU

$$
IoU=\frac{22500}{57500}
$$

**Final answer:**

$$
\boxed{IoU=0.3913}
$$

Therefore, the predicted bounding box has approximately 39.13% overlap relative to the combined area.

## 4.4 Why Is IoU Important?

Imagine that we train an object detection model to identify cars.

![Detecting objective through Detectron2 | by Van Nguyen | Medium](https://images.openai.com/static-rsc-4/XAM7rhboABex_Ex8bAO1g9UHH9ZPoruTNhPlpMI2gFRVdYbopCwY2drVuQJ--A4NJwsuAQhpYYgEFn-rwDn-wpumA5zNNI3GoPBqSvpQVfeZybh6C61Ku15AqPTHK6VAqBMLu2kMHEdl14bJsRI_mSquJUOwOJ3yBTv5_5Wu4so?purpose=inline)

The model may correctly identify the object as a car but produce an inaccurate bounding box.

For example:

| Prediction | Class | IoU |
|---|---|---:|
| Prediction A | Car | 0.91 |
| Prediction B | Car | 0.65 |
| Prediction C | Car | 0.28 |

All three predictions identify the correct class, but their localization quality differs.

IoU helps us evaluate how accurately the model places its bounding boxes.

### IoU Threshold

During object detection evaluation, we often use an IoU threshold to determine whether a predicted box sufficiently overlaps a ground-truth box.

For example:

$$
IoU\ threshold=0.50
$$

- IoU = 0.91 → Passes the overlap threshold.
- IoU = 0.65 → Passes the overlap threshold.
- IoU = 0.28 → Does not pass the overlap threshold.

However, passing the IoU threshold alone does not guarantee a True Positive. The class must also match, and evaluation must account for duplicate predictions and matching rules.

## 4.5 Python Implementation — Calculate IoU

Now let's implement IoU from scratch using Python.

We will use NumPy for numerical calculations.

### Import NumPy

```python
import numpy as np
```

### Implement the IoU function

```python
def calculate_iou(boxA, boxB):

    # Intersection coordinates
    x_left = max(boxA[0], boxB[0])
    y_top = max(boxA[1], boxB[1])

    x_right = min(boxA[2], boxB[2])
    y_bottom = min(boxA[3], boxB[3])

    # Intersection area
    intersection = (
        max(0, x_right - x_left) *
        max(0, y_bottom - y_top)
    )

    # Individual box areas
    area_A = (
        (boxA[2] - boxA[0]) *
        (boxA[3] - boxA[1])
    )

    area_B = (
        (boxB[2] - boxB[0]) *
        (boxB[3] - boxB[1])
    )

    # Union area
    union = area_A + area_B - intersection

    # IoU
    if union <= 0:
        return 0.0

    return intersection / union


# Ground Truth
boxA = (100, 100, 300, 300)

# Predicted bounding box
boxB = (150, 150, 350, 350)

iou = calculate_iou(boxA, boxB)

print("IoU:", iou)
```

**Output:**

```text
IoU: 0.391304347826087
```

### Important implementation details

- `max()` finds the leftmost overlapping boundary.
- `min()` finds the rightmost overlapping boundary.
- `max(0, ...)` prevents negative intersection dimensions when boxes do not overlap.
- Union subtracts the intersection to prevent double counting.

This implementation assumes valid XYXY boxes with positive width and height.
