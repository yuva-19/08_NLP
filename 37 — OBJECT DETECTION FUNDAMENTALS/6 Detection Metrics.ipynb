## Section 6: Detection Metrics

So far, we've learned how an object detection model identifies objects, predicts bounding boxes, calculates IoU and removes duplicate predictions using NMS.

Now comes an important question:

**How do we know whether our object detection model is actually good?**

Imagine we train an object detection model to detect cars in 1,000 images.

- It detects 900 cars.
- It also incorrectly identifies 100 background regions as cars.
- It misses some actual cars.
- Some predicted bounding boxes are poorly positioned.

Simply saying the model achieved 90% accuracy would not describe its performance properly.

We need evaluation metrics that measure both object classification and bounding-box localization.

The five concepts in this section are:

1. Precision
2. Recall
3. Average Precision (AP)
4. Mean Average Precision (mAP)
5. IoU thresholds

---

## 1. Precision

Precision answers: **Out of all the objects the model predicted, how many predictions were actually correct?**

$$
Precision = \frac{TP}{TP+FP}
$$

Where:

- TP — True Positive: Correctly detected object.
- FP — False Positive: Incorrect detection; the model predicted an object where it should not have.

### Example

Suppose our model predicts 100 cars.

- 80 predictions are correct.
- 20 predictions are incorrect.

$$
Precision = \frac{80}{80+20}=0.8
$$

Precision = 80%

**Interpretation:** Out of every 100 car detections made by the model, 80 are correct.

High precision means the model makes fewer false-positive predictions.

## 2. Recall

Recall answers: **Out of all the actual objects present, how many did the model successfully detect?**

$$
Recall = \frac{TP}{TP+FN}
$$

Where:

- TP — True Positive: Correctly detected object.
- FN — False Negative: Actual object that the model failed to detect.

### Example

Suppose an image dataset contains 100 actual cars.

Our model detects 80 of them correctly and misses 20.

$$
Recall = \frac{80}{80+20}=0.8
$$

Recall = 80%

**Interpretation:** The model successfully detected 80 out of 100 actual cars.

High recall means fewer missed objects.

## 3. Precision vs Recall

This distinction is extremely important for interviews and practical model evaluation.

| Metric | Main question | Focus |
|---|---|---|
| Precision | How many predictions were correct? | Reducing FP |
| Recall | How many actual objects were detected? | Reducing FN |

Let's use one example.

Suppose there are 100 actual cars in an image dataset.

Our model produces:

- 70 correct detections (TP).
- 30 incorrect detections (FP).
- 30 missed cars (FN).

$$
Precision = \frac{70}{70+30}=70\%
$$

$$
Recall = \frac{70}{70+30}=70\%
$$

Both are 70%, but they measure different things.

Remember:

- Precision penalizes false positives.
- Recall penalizes false negatives.

### Interactive calculation

Let's try changing the model's predictions.

**Experiment: Change detection results**

- True Positives (TP): 70
- False Positives (FP): 30
- False Negatives (FN): 30

**Precision:** 70.0%

**Reset values**

Notice what happens when you increase FP: precision decreases. When you increase FN: recall decreases.

## 4. Average Precision (AP)

Precision and recall measure performance at a particular set of predictions. But a detection model can produce different results depending on its confidence threshold.

For example:

- Confidence threshold = 0.90: The model keeps only highly confident detections.
- Confidence threshold = 0.50: The model keeps more detections.
- Confidence threshold = 0.20: The model keeps even more detections, including potentially incorrect ones.

As we change the confidence threshold, precision and recall also change.

**Average Precision (AP)** summarizes the precision–recall performance of a model across different confidence thresholds for one object class.

### Understanding the Precision–Recall curve

A Precision–Recall (PR) curve plots:

- X-axis: Recall
- Y-axis: Precision

![Precision Recall Curve](https://images.openai.com/static-rsc-4/o8XWiesizDWTLZjI7ioSqvxwQiq9iInhxgHgj7l3RrtzsWemVTbqbdqNJ3LZtyyBx1QQq4uukMl7I1HxN40jFNduR5HaiJNxQWmKdEo08PyYcuSVXGPGnWqsFE-prVSg9c7iuJzC8P-VZ26B-WbU8yUOS4k8-qcDBFLe0IQLfvQ?purpose=inline)

[Source: Medium — The Complete Guide to Object Detection Evaluation Metrics](https://medium.com/%40prathameshamrutkar3/the-complete-guide-to-object-detection-evaluation-metrics-from-iou-to-map-and-more-1a23c0ea3c9d)

Each point on the curve represents precision and recall at a particular confidence threshold.

Generally:

- A good detector maintains high precision even as recall increases.
- A poor detector experiences a significant drop in precision as it detects more objects.

### How is AP calculated?

Conceptually, AP is the area under the Precision–Recall curve:

$$
AP = \int_0^1 P(R)\,dR
$$

Where:

- \\(P\\) = Precision
- \\(R\\) = Recall

In practice, evaluation implementations often use interpolated precision and numerical summation rather than calculating an exact continuous integral. The precise AP calculation can depend on the evaluation protocol.

### Example

Suppose a model produces the following precision–recall performance:

| Recall | Precision |
|---|---|
| 0.0 | 1.00 |
| 0.2 | 0.95 |
| 0.4 | 0.90 |
| 0.6 | 0.80 |
| 0.8 | 0.70 |
| 1.0 | 0.50 |

The area under the resulting PR curve represents AP.

These points are illustrative, not a complete numerical AP calculation.

**Important:** AP is calculated separately for each class. For example, we can calculate AP for cars, AP for buses and AP for pedestrians.

## 5. Mean Average Precision (mAP)

Now imagine an object detection model that identifies three different classes:

- Cars
- Buses
- Pedestrians

We calculate AP separately for each class.

Suppose the results are:

| Object class | Average Precision |
|---|---|
| Car | 0.90 |
| Bus | 0.80 |
| Pedestrian | 0.70 |

Mean Average Precision calculates the average of these class-wise AP values.

$$
mAP = \frac{1}{N}\sum_{i=1}^{N} AP_i
$$

Where:

- \\(N\\) = Number of classes
- \\(AP_i\\) = Average Precision of class \\(i\\)

Substituting:

$$
mAP = \frac{0.90+0.80+0.70}{3}
$$

$$
mAP = 0.80
$$

mAP = 80%

**Interpretation:** The model's mean Average Precision across these three classes is 80%.

### Why is mAP important?

Imagine two models:

| Model | Car AP | Bus AP | Pedestrian AP | mAP |
|---|---:|---:|---:|---:|
| Model A | 0.95 | 0.90 | 0.85 | 0.90 |
| Model B | 0.98 | 0.95 | 0.37 | 0.77 |

Model B performs very well on cars and buses but poorly on pedestrians.

Model A has better overall performance across the three classes.

This is why mAP is widely used to evaluate object detection models.

## 6. IoU Thresholds in Evaluation

This is where our previous IoU and NMS lessons become important.

Remember, IoU measures how much a predicted bounding box overlaps with the ground-truth bounding box.

But when evaluating a model, how much overlap is sufficient for a detection to count as correct?

We define an **IoU threshold**.

### Example

Suppose the ground-truth bounding box represents an actual car.

Our model predicts three boxes:

| Prediction | IoU with ground truth |
|---|---:|
| A | 0.90 |
| B | 0.65 |
| C | 0.35 |

If the evaluation IoU threshold is 0.50:

- A passes the IoU requirement.
- B passes the IoU requirement.
- C fails the IoU requirement.

![IoU Evaluation Example](https://images.openai.com/static-rsc-4/ex6c7ZcDCSM5_aYsXpHXC9uv7B5bbSlgNadoBZGJ1usCKG4RwOhqpITPeZvKiVEMeX7I7DZyoA4QPI_9FtN-rLfzrYl6ZUIxxfoNf2nwXudveIo8RLN1YHQDh8hIQef0UESqLRtWfodMNCm-XVmeqeVcY6BAWxYe3xiDMA1lTA0?purpose=inline)

[Source: 易微帮 — IoU Concept and Python Example](https://www.ewbang.com/community/article/details/1000262051.html)

However, passing the IoU threshold alone does not guarantee a true positive.

The predicted class must also match the ground-truth class, and the evaluation protocol must match predictions to ground-truth objects correctly. Duplicate detections of the same object are generally not all counted as true positives.

### Common IoU evaluation thresholds

| Evaluation setting | Meaning |
|---|---|
| IoU ≥ 0.50 | Relatively lenient localization requirement |
| IoU ≥ 0.75 | Stricter localization requirement |
| IoU ≥ 0.95 | Very strict localization requirement |

As the threshold increases, the model must predict more accurately positioned bounding boxes to receive credit.

### AP@0.50

This means Average Precision is calculated using an IoU threshold of 0.50.

A prediction can qualify as a correct localization when its IoU is at least 0.50, subject to the other matching rules.

### AP@0.75

This means Average Precision is calculated using an IoU threshold of 0.75.

This is more demanding because the predicted bounding box must overlap more closely with the ground-truth box.

### AP@0.50:0.95

This is a commonly used COCO evaluation metric.

It calculates AP at ten IoU thresholds:

$$
0.50,\ 0.55,\ 0.60,\ \ldots,\ 0.95
$$

The results are averaged across these IoU thresholds and the evaluated object classes.

This metric is more comprehensive because it evaluates the detector at multiple localization strictness levels.

## 7. Important: Don't Confuse These Thresholds!

We have now encountered three different thresholds in object detection.

| Threshold | Purpose |
|---|---|
| Confidence threshold | Filters predictions based on confidence scores |
| NMS IoU threshold | Removes overlapping duplicate predictions |
| Evaluation IoU threshold | Determines whether a predicted box sufficiently matches ground truth |

Let's use an example:

- Confidence threshold = 0.50
- NMS IoU threshold = 0.60
- Evaluation IoU threshold = 0.50

These three values serve different purposes. Changing one does not automatically change the others.

This distinction is particularly important when configuring and evaluating models such as YOLO.

## What You Should Remember

| Concept | Remember this |
|---|---|
| Precision | TP / (TP + FP) |
| Recall | TP / (TP + FN) |
| AP | Area under the Precision–Recall curve |
| mAP | Mean of AP values across classes |
| IoU threshold | Minimum overlap required for a detection to qualify as a match |

Three particularly important takeaways:

1. Precision focuses on reducing false positives; recall focuses on reducing false negatives.
2. AP evaluates a single class; mAP averages AP across classes (and, in some evaluation protocols, across IoU thresholds too).
3. AP@0.50:0.95 is a stricter, multi-threshold evaluation than AP@0.50.
