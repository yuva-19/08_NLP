# 2. Bounding Boxes

A bounding box is a rectangular region that surrounds an object in an image.

It tells the model (and the application) where an object is located.

For example, in an autonomous driving system, the detector might identify:

- Car → Bounding box
- Pedestrian → Bounding box
- Motorcycle → Bounding box
- Traffic sign → Bounding box

Each detected object gets its own bounding box.

There are two common ways to represent these boxes mathematically.

## 2.1 Bounding Box Representation — (x, y, width, height)

This representation describes a bounding box using:

$$
B=(x,y,w,h)
$$

Where:

| Parameter | Meaning |
|---|---|
| \\(x\\) | X-coordinate of the top-left corner |
| \\(y\\) | Y-coordinate of the top-left corner |
| \\(w\\) | Width of the box |
| \\(h\\) | Height of the box |

![javascript - OpenCV Rect conventions – What is x, y, width, height? - Stack Overflow](https://images.openai.com/static-rsc-4/47LGbV80_LoZnLYKo4-7rWfYO9aJLYpUlSc_xQuvba-IOXwc-xwWxtJNiV698bo2WgHbrvTeSnNnjTzxZ0jFojz2Twcsk5F5_JXu-2pExxsYmQKJtkjAQ578BR2F5yLpoP7jmUGf-3IigF7GPzxNVAO2fUs-j6YndETlUbfOf50?purpose=inline)

### Example

Suppose an image has the following bounding box:

```python
x = 100
y = 50
width = 200
height = 150
```

This means:

- The box starts 100 pixels from the left.
- The box starts 50 pixels from the top.
- Its width is 200 pixels.
- Its height is 150 pixels.

Therefore:

$$
B=(100,50,200,150)
$$

The bottom-right corner can be calculated as:

$$
x_{\max}=x+w
$$

$$
y_{\max}=y+h
$$

Therefore:

$$
x_{\max}=300
$$

$$
y_{\max}=200
$$

## 2.2 Bounding Box Representation — (x1, y1, x2, y2)

Another common representation specifies the coordinates of two opposite corners.

$$
B=(x_1,y_1,x_2,y_2)
$$

Where:

| Parameter | Meaning |
|---|---|
| \\(x_1\\) | Left boundary |
| \\(y_1\\) | Top boundary |
| \\(x_2\\) | Right boundary |
| \\(y_2\\) | Bottom boundary |

### Example

Consider:

```python
x1 = 100
y1 = 50
x2 = 300
y2 = 200
```

The box starts at `(100, 50)` and ends at `(300, 200)`.

We can calculate its dimensions:

$$
w=x_2-x_1
$$

$$
h=y_2-y_1
$$

Therefore:

$$
w=300-100=200
$$

$$
h=200-50=150
$$

Both representations describe exactly the same bounding box.

Before normalization, understand how image coordinates work.

In computer vision, the coordinate origin `(0, 0)` is generally at the top-left corner of the image.

## 2.4 Image Normalization

This is an important concept in object detection.

**Problem:** Different images have different resolutions.

For example:

- Image A: 640 × 480
- Image B: 1920 × 1080
- Image C: 1280 × 720

If bounding boxes are represented using absolute pixel coordinates, their values depend on image resolution.

To make coordinates independent of image resolution, we use bounding box normalization.

### Normalization formulas

For a bounding box represented by `(x1, y1, x2, y2)`:

$$
x_{1,n}=\frac{x_1}{W}
$$

$$
y_{1,n}=\frac{y_1}{H}
$$

$$
x_{2,n}=\frac{x_2}{W}
$$

$$
y_{2,n}=\frac{y_2}{H}
$$

Where:

- \\(W\\) = Image width.
- \\(H\\) = Image height.
- Subscript \\(n\\) = Normalized coordinate.

The resulting coordinates generally lie between 0 and 1 for boxes contained within the image.

![Training YOLOv11 object detector on a custom dataset | by esteban uri | Medium](https://images.openai.com/static-rsc-4/7EpnSgFhgds9_RFpWUcybDbfsS2_LsJ5frvYIh2PMLm2dIzGE-3O7AHrTOzvO-efvmgRmtB5_yDvDpUDOOvVSI9HTMyzL4Vgy9D2ym_hGVCsL9GTzgbxqtytkEniJyVcZNl-gEjh8wlP0UonN3vb5onbQTBpaynfEkQ0zuvENGA?purpose=inline)

### Practical example

Suppose:

Image dimensions:

```python
W = 640
H = 480
```

Bounding box:

```python
x1 = 100
y1 = 50
x2 = 300
y2 = 200
```

Let's normalize each coordinate.

**Step 1 — Normalize X coordinates**

$$
x_{1,n}=\frac{100}{640}=0.15625
$$

$$
x_{2,n}=\frac{300}{640}=0.46875
$$

**Step 2 — Normalize Y coordinates**

$$
y_{1,n}=\frac{50}{480}=0.10417
$$

$$
y_{2,n}=\frac{200}{480}=0.41667
$$

Normalized bounding box:

```text
(0.15625, 0.10417, 0.46875, 0.41667)
```

Notice that we no longer need to store pixel coordinates.

## 2.5 YOLO Bounding Box Format

Now let's connect normalization to a real object detection framework.

YOLO commonly represents bounding boxes using:

$$
(x_c,y_c,w,h)
$$

All four values are normalized.

Where:

- \\(x_c\\): Normalized centre X.
- \\(y_c\\): Normalized centre Y.
- \\(w\\): Normalized box width.
- \\(h\\): Normalized box height.

![Automated custom Scaled-YOLOv4 object detection model training for total beginners(no coding or AI skills needed) — free GPU | by Christoph Haring | Medium](https://images.openai.com/static-rsc-4/f8xLAn52SoG52JLukcBB8I94ZhBMCoTJ6velUtS6iXm7iS1eS9X1dLKwmPXPsGBC4hwfZLf-TCg0BYaJFudj-ys88mGeL2n0iI692WsX4YguOsMAcCry2xU0pqJdVtHgcznmoDqZCnLfilnBU9oSXcMuoXyhsXFhupv-eQQhF_s?purpose=inline)

### Conversion from XYXY to YOLO format

We already have:

```python
x1, y1 = 100, 50
x2, y2 = 300, 200

W, H = 640, 480
```

First calculate the centre:

$$
x_c=\frac{x_1+x_2}{2}
$$

$$
y_c=\frac{y_1+y_2}{2}
$$

Therefore:

$$
x_c=200,\qquad y_c=125
$$

Calculate width and height:

$$
w=x_2-x_1=200
$$

$$
h=y_2-y_1=150
$$

Now normalize:

$$
x_c=\frac{200}{640}=0.3125
$$

$$
y_c=\frac{125}{480}=0.26042
$$

$$
w=\frac{200}{640}=0.3125
$$

$$
h=\frac{150}{480}=0.3125
$$

**Final YOLO bounding box**

```text
0.3125  0.26042  0.3125  0.3125
```

A YOLO annotation line also includes the class ID:

```text
class_id x_center y_center width height
```

For example:

```text
0 0.3125 0.26042 0.3125 0.3125
```

Here, `0` represents the object's class ID.

## 2.6 Python Implementation

Let's implement bounding box conversion using Python.

No GPU is required for these coordinate calculations because they are lightweight numerical operations.

### XYXY → YOLO

```python
def xyxy_to_yolo(x1, y1, x2, y2, W, H):

    # Calculate centre
    x_center = (x1 + x2) / 2
    y_center = (y1 + y2) / 2

    # Calculate dimensions
    width = x2 - x1
    height = y2 - y1

    # Normalize
    x_center /= W
    y_center /= H
    width /= W
    height /= H

    return x_center, y_center, width, height


bbox = xyxy_to_yolo(
    100, 50, 300, 200,
    640, 480
)

print(bbox)
```

**Output:**

```text
(0.3125, 0.260416..., 0.3125, 0.3125)
```

### Reverse conversion: YOLO → XYXY

```python
def yolo_to_xyxy(xc, yc, w, h, W, H):

    # Convert normalized values to pixels
    xc *= W
    yc *= H
    w *= W
    h *= H

    x1 = xc - w / 2
    y1 = yc - h / 2

    x2 = xc + w / 2
    y2 = yc + h / 2

    return x1, y1, x2, y2
```

This allows us to reconstruct the original bounding box.

## 2.7 Important Comparison

| Representation | Format | Common usage |
|---|---|---|
| XYWH | Top-left X, Y, width, height | General image annotation |
| XYXY | Left, top, right, bottom | TorchVision detection |
| Normalized XYXY | Normalized corner coordinates | Resolution-independent representation |
| YOLO | Centre X, centre Y, width, height (normalized) | YOLO datasets |

**Important:** Different frameworks use different coordinate conventions. Always check the dataset and model documentation before converting annotations.

## 🧠 What You Should Remember

1. Bounding boxes identify the spatial location of objects.
2. XYWH uses the top-left corner and box dimensions.
3. XYXY uses two opposite corners.
4. Image coordinates generally originate at the top-left.
5. Normalization divides X-related values by image width and Y-related values by image height.
6. YOLO commonly uses normalized centre-based XYWH representation.
7. You must convert between formats correctly when preparing detection datasets.
