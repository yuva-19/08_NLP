## 1. Detection vs Classification

### 1.1 Image Classification

**Definition:** Image classification is a computer vision task in which a model predicts the category or class of an entire input image.

For example, suppose we provide an image containing a dog.

![Happy sitting and panting Golden retriever dog looking at camera, wearing a collar and identification tag](https://images.openai.com/static-rsc-4/Q6mH78RscLvvkaIbg2g0-U8vIqrWm2qpTqdyelqYwr7yow6gxZ2w0J_MpPUcld2KHJtPbOG7vvRVnTxB4He-aMfeN9BOTLV0UGC473RZuRgSulR6oIn5-zw7tS24ayA8f43uDu4ua_L4TMj2BRSMIMqozjcbTNleZjmPQro1jV-6nPZSWVZMu-49ZpptFPd1?purpose=inline)

The model produces:

```text
Input Image
     ↓
CNN / Pretrained Backbone
     ↓
Classification Head
     ↓
Prediction: DOG
```

The output might look like:

| Class | Probability |
| ----- | ----------- |
| Dog   | 0.94        |
| Cat   | 0.04        |
| Horse | 0.02        |

The model identifies what is present in the image, but it does not tell us where the dog is located.

### 1.2 Object Detection

**Definition:** Object detection identifies multiple objects in an image, predicts their classes, and determines their spatial locations using bounding boxes.

![Enterprise Computer Vision Solutions & Services | OrcaMinds](https://images.openai.com/static-rsc-4/I5RFgPKmFgS9Yvxqh3c0YUyTWHpqC4Fr30kkI73XAMQBF3lzdCQBZJSLaykU3MdbnWgOPTjW8ViBO79Ye9Y70LreUfgxuwZ1rx0hu2MuR73l4jHaAkMRSj7ed_jA7YqQnEG_0W7ywasmyJY-DxyLndOlP8Fr6OSsDUaYUhgTgQc?purpose=inline)

[orcaminds.in](https://orcaminds.in/orcaminds-computer-vision-services.html?utm_source=chatgpt.com)

The output now contains two important pieces of information:

1. **Class prediction:** What object is present?
2. **Localization:** Where is the object located?

For example:

```text
Input Image
     ↓
Feature Extractor
     ↓
Object Detection Head
     ↓
Detections
     ↓
Car       → Bounding Box
Person    → Bounding Box
Bicycle   → Bounding Box
```

Unlike classification, object detection can identify multiple objects in the same image and locate each one.

### 1.3 Side-by-Side Comparison

| Feature             | Classification                  | Object Detection            |
| ------------------- | ------------------------------- | --------------------------- |
| Main purpose        | Identify object                 | Identify and locate objects |
| Output              | Class label                     | Class + bounding box        |
| Multiple objects    | Depends on classification setup | Yes                         |
| Spatial information | No                              | Yes                         |
| Example             | Dog                             | Dog at (x, y, w, h)         |
| Complexity          | Comparatively simpler           | More complex                |

**Object detection approach:**

### 1.5 What Exactly Is a Bounding Box?

A bounding box is a rectangular boundary drawn around an object to indicate its location in an image.

![A Guide to Bounding Box Formats](https://images.openai.com/static-rsc-4/pKxWVR9M2rlYp5UWOZHaTk3zrVHmOi4VHtfYQJ8gP5HHRtN-M01iSliH4jpQGCbF1fNy9vdGWa-29yVRt7a5wL8TtiFKiL4nWF6xspgGcOD3dfiT3wgwZNzMWqfCz79nMiWM-LjgOmKG0hTnQS17YpQa9__wG6w9U5qPMPcP76g?purpose=inline)

[dragoneye.ai](https://dragoneye.ai/blog/a-guide-to-bounding-box-formats/?utm_source=chatgpt.com)

A commonly used coordinate representation is:

$$
B=(x_{\min},y_{\min},x_{\max},y_{\max})
$$

Where:

- \\(x_{\min}\\): Left boundary.
- \\(y_{\min}\\): Top boundary.
- \\(x_{\max}\\): Right boundary.
- \\(y_{\max}\\): Bottom boundary.

Another representation uses:

$$
B=(x_c,y_c,w,h)
$$

Where:

- \\(x_c,y_c\\): Centre coordinates of the bounding box.
- \\(w\\): Width of the bounding box.
- \\(h\\): Height of the bounding box.

We'll explore these coordinate systems properly in the next section of this module.

## What You Should Remember 🧠

1. Classification identifies what is present in an image.
2. Object detection identifies what is present and where it is located.
3. Object detection supports multiple object predictions in a single image.
4. Bounding boxes represent object locations.
5. Object detection combines classification and localization.
