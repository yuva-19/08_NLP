## 3.5 Word Tokenization



**Tokenization** is the process of splitting text into smaller units called **tokens**. For word tokenization, these units are usually words and punctuation marks. [NLTK](https://www.nltk.org/api/nltk.tokenize.html?utm_source=chatgpt.com)

Example:

```text
Input:
"I love NLP!"

        ↓

Tokens:
["I", "love", "NLP", "!"]
```

---

## 1. Word-Level Tokenization

Word-level tokenization treats **words as the primary units**.

```text
"I love machine learning."

        ↓

["I", "love", "machine", "learning", "."]
```

A simple approach is whitespace splitting:

```python
text = "I love machine learning."

tokens = text.split()

print(tokens)
```

Output:

```text
['I', 'love', 'machine', 'learning.']
```

Notice the problem:

```text
"learning."
```

contains punctuation.

A proper tokenizer can separate punctuation:

```text
["I", "love", "machine", "learning", "."]
```

Libraries such as NLTK and spaCy provide more sophisticated tokenizers that handle punctuation and special cases. [NLTK](https://www.nltk.org/api/nltk.tokenize.html?utm_source=chatgpt.com)

---

# 2. Character Tokenization

Instead of treating complete words as tokens, **character tokenization** treats individual characters as units.

Example:

```text
"cat"

↓

["c", "a", "t"]
```

For:

```text
"I love NLP!"
```

a character tokenizer may produce:

```text
["I", " ", "l", "o", "v", "e", " ", "N", "L", "P", "!"]
```



### Advantage

Character tokenization can represent **unknown or rare words**, because any word can be decomposed into characters.

### Disadvantage

Sequences become much longer.

```text
Word-level:
"playing" → 1 token

Character-level:
"playing" → 7 character tokens
```

---

# 3. Token Boundaries

A **token boundary** is the position where one token ends and another begins.

Consider:

```text
"I love NLP."
```

Possible boundaries:

```text
"I" | "love" | "NLP" | "."
```

The tokenizer must decide:

- Where words start/end
- Whether punctuation becomes its own token
- How hyphenated words are handled
- How contractions are split
- How abbreviations are handled

This is why tokenization is **not simply splitting on spaces**.

### Example

```text
"state-of-the-art"
```

Possible tokenization:

```text
["state-of-the-art"]
```

or:

```text
["state", "-", "of", "-", "the", "-", "art"]
```

The choice depends on the tokenizer and task.

spaCy, for example, uses prefix, suffix, infix, and special-case rules to determine token boundaries. [spaCy](https://spacy.io/usage/linguistic-features?utm_source=chatgpt.com)

---

# 4. Contractions

**Contractions** combine multiple words into a shorter form.

Examples:

```text
don't  → do + not
can't  → can + not
I'm    → I + am
it's   → it + is / it + has
we're  → we + are
```

Consider:

```text
"I don't like this movie."
```

A tokenizer might produce:

```text
["I", "do", "n't", "like", "this", "movie", "."]
```

rather than:

```text
["I", "don't", "like", "this", "movie", "."]
```

The exact behavior depends on the tokenizer.

For example, spaCy's English tokenizer has special rules for contractions such as `don't`, producing `do` + `n't`. [spaCy](https://spacy.io/usage/linguistic-features?utm_source=chatgpt.com)



### Why does this matter?

Consider:

```text
"I don't like it."
```

If the model treats `don't` as one completely independent word, it may learn a separate representation for it.

Splitting it into:

```text
do + n't
```

can expose the **negation component** more explicitly.

However, modern pretrained Transformer tokenizers use their own subword tokenization strategies, so you should **not manually split contractions unless the chosen pipeline requires it**.

# 5. Tokenization Examples

| Input | Possible tokens |
|---|---|
| `I love NLP.` | `I \| love \| NLP \| .` |
| `Hello, world!` | `Hello \| , \| world \| !` |
| `don't` | `do \| n't` |
| `I'm happy.` | `I \| 'm \| happy \| .` |
| `state-of-the-art` | `state-of-the-art` or multiple tokens |
| `U.K.` | `U.K.` |

Different tokenizers can produce **different boundaries for the same text**. [spaCy](https://spacy.io/usage/linguistic-features?utm_source=chatgpt.com)

---

# 6. NLTK Example

```python
import nltk
from nltk.tokenize import word_tokenize

text = "I don't like this movie."

tokens = word_tokenize(text)

print(tokens)
```

A Treebank-style tokenizer can produce tokens such as:

```text
['I', 'do', "n't", 'like', 'this', 'movie', '.']
```

NLTK also provides other tokenization strategies, including regular-expression-based tokenizers. [NLTK](https://www.nltk.org/api/nltk.tokenize.html?utm_source=chatgpt.com)

---

# 7. spaCy Example

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "I don't like this movie."

doc = nlp(text)

tokens = [token.text for token in doc]

print(tokens)
```

Possible output:

```text
['I', 'do', "n't", 'like', 'this', 'movie', '.']
```

spaCy's tokenizer uses language-specific rules and exceptions for cases such as **contractions, abbreviations, punctuation, and hyphenated expressions**. [spaCy](https://spacy.io/usage/linguistic-features?utm_source=chatgpt.com)

# 🧠 Word vs Character Tokenization



| Feature | Word-level | Character-level |
|---|---|---|
| Basic unit | Word | Character |
| Sequence length | Shorter | Much longer |
| Vocabulary | Larger | Small |
| Unknown words | Problematic | Less problematic |
| Semantic meaning per token | Higher | Lower |
| Example | `playing` → 1 | `playing` → 7 |

---

## ⭐ Key Takeaways

### Remember these 4 things:

**1. Word tokenization**

> Splits text into word-like units and often punctuation.

**2. Character tokenization**

> Splits text into individual characters.

**3. Token boundaries**

> Determine exactly where one token ends and another begins.

**4. Contractions**

> Require special handling because forms like `don't` can be represented as one token or multiple tokens depending on the tokenizer.

### Most important point 🔥

> **Tokenization is not simply `text.split()` — a real tokenizer uses rules and exceptions to determine meaningful token boundaries.**
