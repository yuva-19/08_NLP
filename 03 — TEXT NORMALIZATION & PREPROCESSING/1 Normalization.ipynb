# 📘 MODULE 3 — TEXT NORMALIZATION & PREPROCESSING
## 3.1 Normalization



**Text normalization** converts different or inconsistent forms of text into a more **consistent representation** before further NLP processing.

For example:

```text
"  HELLO!!!   Visit https://example.com 😍  "
                         ↓
"hello! visit <URL> 😍"
```

The exact normalization depends on the **NLP task**. There is no universally correct cleaning pipeline. [CloudsPress](https://www.cloudspress.com/text-preprocessing-in-natural-language-processing-with-python-a-task-driven-guide/?utm_source=chatgpt.com)

---

## 1. Lowercasing

Converts uppercase characters to lowercase.

```text
"Natural LANGUAGE Processing"
            ↓
"natural language processing"
```

### Why?

Without normalization, these may be treated as different strings:

```text
NLP
Nlp
nlp
```

Lowercasing can reduce vocabulary duplication. [Krishnaik](https://www.krishnaik.in/ai-tutorials/nlp/text-cleaning?utm_source=chatgpt.com)

### ⚠️ Important

Lowercasing can also remove useful information.

```text
Apple → apple
```

This can matter for **Named Entity Recognition**, where capitalization can help distinguish a company from a common noun.

**Rule:** Use lowercasing when case is not important for the task.

---

## 2. Whitespace Normalization

Removes unnecessary spaces, tabs, and line breaks.

```text
"This    is   an
example."
```

becomes:

```text
"This is an example."
```

Typical operation:

```python
import re

text = re.sub(r"\s+", " ", text).strip()
```

### Why?

It provides a consistent representation and prevents meaningless whitespace differences from creating unnecessary variations. [NPBlue](https://npblue.com/tech/nlp/processing/text-normalization/?utm_source=chatgpt.com)

---

## 3. Punctuation Handling

Punctuation includes:

```text
. , ! ? : ; ' " - ( )
```

Depending on the task, punctuation can be:

- Removed
- Preserved
- Replaced
- Normalized

Example:

```text
"Hello!!! How are you???"
```

Possible normalized form:

```text
"Hello How are you"
```

### ⚠️ Don't blindly remove punctuation

Punctuation can carry meaning:

```text
"Really?"
"Really!"
```

They express different intentions.

Repeated punctuation can also provide useful sentiment information:

```text
"I LOVE IT!!! 😍"
```

Therefore, punctuation handling should be **task-dependent**. [CloudsPress](https://www.cloudspress.com/text-preprocessing-in-natural-language-processing-with-python-a-task-driven-guide/?utm_source=chatgpt.com)

---

## 4. Special-Character Handling

Special characters include symbols such as:

```text
@  #  $  %  &  *  +  =  _
```

Example:

```text
"Follow @yuva #NLP"
```

Possible handling:

```text
"Follow yuva NLP"
```

or preserve:

```text
"Follow @yuva #NLP"
```

### Why preserve sometimes?

In social-media NLP:

- `@username` → user mention
- `#NLP` → hashtag

So removing every special character can cause **information loss**.

---

## 5. Number Handling

Numbers can be handled in different ways depending on the task.

Example:

```text
"I bought 3 phones for ₹50,000."
```

Possible approaches:

### Preserve

```text
"I bought 3 phones for ₹50,000."
```

### Replace

```text
"I bought <NUM> phones for <NUM>."
```

### Remove

```text
"I bought phones for."
```

### ⚠️ Task matters

For financial or medical NLP, numbers can be extremely important:

```text
"Blood pressure is 120/80."
"Revenue increased by 25%."
```

Removing them would destroy useful information.

---

## 6. URL Handling

URLs are common in web and social-media text.

```text
"Read more at https://example.com/article"
```

Possible approaches:

### Remove

```text
"Read more at"
```

### Replace with placeholder

```text
"Read more at <URL>"
```

Replacing with a placeholder is often useful because the **presence of a URL** can itself be meaningful while the exact URL may not be. [VizLearn](https://vizlearn.in/natural_language_processing/text_normalization_pipeline.html?utm_source=chatgpt.com)

For example, URLs can be important in **spam detection**.

---

## 7. HTML Removal

Web-scraped text often contains HTML markup.

```html
<p>This product is <b>amazing</b>!</p>
```

After removing HTML:

```text
"This product is amazing!"
```

HTML tags such as:

```html
<div>
<p>
<br>
<a>
<b>
```

are generally formatting/structural information rather than the natural-language content we want.

For actual HTML parsing, tools such as **BeautifulSoup** are generally preferable to relying only on regex. [GitHub](https://github.com/Abdelrahman-Atef-Elsayed/NLP_Preprocessing_pipeline?utm_source=chatgpt.com)

---

## 8. Emoji Handling

Emojis can contain **meaning and sentiment**.

```text
"I love this movie 😍❤️"
```

Possible approaches:

### Remove

```text
"I love this movie"
```

### Preserve

```text
"I love this movie 😍❤️"
```

### Convert to text

```text
"I love this movie smiling_face_with_heart_eyes red_heart"
```

For sentiment analysis, removing emojis may eliminate useful information.

```text
"Great 😍"
"Great 😡"
```

The words are identical, but the emojis strongly change the meaning.

Therefore:

> **Emoji handling should depend on the NLP task rather than automatically removing emojis.** [VizLearn](https://vizlearn.in/natural_language_processing/text_normalization_pipeline.html?utm_source=chatgpt.com)

# 🔥 Complete Example

Consider this raw text:

```text
"  HELLO!!! 😍 <p>This is AMAZING!!!</p>
Visit https://example.com   100%  "
```

A possible normalization process:

```text
RAW
 ↓
"  HELLO!!! 😍 <p>This is AMAZING!!!</p>
 Visit https://example.com 100%  "
 ↓
HTML Removal
 ↓
"HELLO!!! 😍 This is AMAZING!!! Visit https://example.com 100%"
 ↓
URL Replacement
 ↓
"HELLO!!! 😍 This is AMAZING!!! Visit <URL> 100%"
 ↓
Lowercasing
 ↓
"hello!!! 😍 this is amazing!!! visit <url> 100%"
 ↓
Whitespace Normalization
 ↓
"hello!!! 😍 this is amazing!!! visit <url> 100%"
```

Whether we **remove or preserve punctuation, emojis, and numbers** depends on the downstream task.

---

# 🧠 Key Principle

### ❌ Bad preprocessing mindset

> "Remove everything except words."

### ✅ Better NLP mindset

> **"Keep information that is useful for the task and remove or normalize information that introduces irrelevant variation."**

| Normalization | Example |
|---|---|
| Lowercasing | `HELLO → hello` |
| Whitespace | `hello   world → hello world` |
| Punctuation | `hello!!! → hello` *(if appropriate)* |
| Special characters | `@user → user` *(if appropriate)* |
| Numbers | `₹500 → <NUM>` *(if appropriate)* |
| URLs | `https://... → <URL>` |
| HTML | `<p>Hello</p> → Hello` |
| Emojis | `😍 → preserve/remove/convert` |

**⭐ The most important point:** preprocessing is **task-dependent**. Aggressive normalization can improve consistency but can also destroy useful linguistic information.
