## 3.7 Stemming

## 1. What is Stemming?

**Stemming** is the process of reducing related words to a common **stem** by removing or modifying word endings, usually through rule-based suffix stripping.

```text
connected
connecting
connection
connections
connects
      ↓
   connect
```

The resulting stem **does not necessarily have to be a valid English word**.

For example:

```text
studies
studying
      ↓
   studi
```



This is different from **lemmatization**, which tries to return a valid dictionary/base form.

---

# 2. Why Stemming?

Different grammatical forms of a word can represent the same underlying concept.

```text
connect
connected
connecting
connections
```

Without stemming:

```text
connect
connected
connecting
connections
```

may be treated as separate vocabulary terms.

After stemming:

```text
connect
connect
connect
connect
```

This can reduce vocabulary size and help systems match related word forms.

Stemming is particularly useful in traditional **information retrieval and text classification** systems. [NLTK](https://www.nltk.org/howto/stem.html?utm_source=chatgpt.com)

---

# 3. Porter Stemmer

The **Porter Stemmer** was developed by **Martin Porter** and published in 1980 in *An algorithm for suffix stripping*. It applies a sequence of suffix-stripping rules to English words. [NLTK](https://www.nltk.org/api/nltk.stem.PorterStemmer.html?highlight=porterstemmer\&utm_source=chatgpt.com)

### Example

```text
caresses  → caress
flies     → fli
dies      → die
meeting   → meet
stating   → state
```

These examples are also demonstrated in NLTK's Porter stemmer tests. [NLTK](https://www.nltk.org/howto/stem.html?utm_source=chatgpt.com)



### Python

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["connect", "connected", "connecting", "connections"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

Possible output:

```text
connect      → connect
connected    → connect
connecting   → connect
connections  → connect
```

NLTK's current `PorterStemmer` implementation has multiple modes; its default is `NLTK_EXTENSIONS`, while the original published algorithm is available explicitly. [NLTK](https://www.nltk.org/api/nltk.stem.PorterStemmer.html?highlight=porterstemmer\&utm_source=chatgpt.com)

# 4. Snowball Stemmer

**Snowball** is a framework/language created by Martin Porter for developing stemming algorithms. NLTK provides Snowball stemmers for multiple languages. [NLTK](https://www.nltk.org/_modules/nltk/stem/snowball.html?utm_source=chatgpt.com)

For English, the Snowball stemmer is commonly referred to as **English Snowball / Porter2**.

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

print(stemmer.stem("running"))
```

Output:

```text
run
```

NLTK supports Snowball stemming for languages including English, Spanish, French, German, Italian, Portuguese, Russian, Arabic, and several others. [NLTK](https://www.nltk.org/api/nltk.stem.SnowballStemmer.html?utm_source=chatgpt.com)

### Porter vs Snowball

| Feature | Porter | Snowball |
|---|---|---|
| Original purpose | English stemming | Framework for multiple languages |
| English algorithm | Original Porter | Improved English stemmer / Porter2 |
| Language support | Primarily English | Multiple languages |
| Rule-based | ✅ | ✅ |
| NLTK available | ✅ | ✅ |

For English, NLTK's documentation notes that its Snowball English stemmer can produce better results than the original Porter implementation for some words:

```text
generously
    ↓
Snowball → generous
Porter   → gener
``` :chatgpt-content-reference{index="9"}


---

# 5. Advantages of Stemming

### ✅ 1. Reduces vocabulary size

```text
connect
connected
connecting
connection
connections
```

can map to a smaller number of stems.

### ✅ 2. Faster than linguistic analysis

Stemming generally uses **suffix-stripping rules** rather than requiring a dictionary and detailed linguistic analysis.

### ✅ 3. Useful for information retrieval

Related word forms can be mapped together, increasing the chance of matching relevant documents.

### ✅ 4. Simple to implement

```python
stemmer.stem(word)
```

is enough to apply a traditional stemmer.

---

# 6. Limitations

## ❌ 1. Produces non-words

A stem does not necessarily have to be a valid English word.

```text
studies → studi
```

```text
relational → relat
```

The output is useful computationally, but not necessarily linguistically meaningful.

---

## ❌ 2. Over-stemming

**Over-stemming** occurs when words that have different meanings are incorrectly reduced to the same stem.

Conceptually:

```text
different words
      ↓
same stem
      ↓
meaning distinction lost
```

This can reduce precision.

---

## ❌ 3. Under-stemming

**Under-stemming** occurs when words that are actually related are **not** reduced to the same stem.

```text
related words
      ↓
different stems
      ↓
relationship missed
```

So stemming involves a trade-off between **over-stemming and under-stemming**.

---

## ❌ 4. Language-dependent

A stemming algorithm designed for English should not simply be applied to Tamil, Hindi, German, etc.

Different languages have different morphological structures.

This is one reason Snowball provides language-specific algorithms. [NLTK](https://www.nltk.org/api/nltk.stem.SnowballStemmer.html?utm_source=chatgpt.com)

# 🧠 Stemming Pipeline

```text
Raw Text
   ↓
Tokenization
   ↓
Words
   ↓
Stemming
   ↓
Common Stems
   ↓
BoW / TF-IDF / Search / Classification
```

Example:

```text
"The players are playing football."

             ↓

["The", "players", "are", "playing", "football"]

             ↓ Stemming

["the", "player", "are", "play", "footbal"]
```

Notice that **`footbal` is not a normal English word**. That's perfectly possible with stemming.

---

# ⚡ Porter vs Snowball — Remember This

> **Porter Stemmer:** classic English suffix-stripping algorithm.

> **Snowball Stemmer:** a broader stemming framework with language-specific algorithms; its English stemmer is an improved successor commonly known as Porter2. [NLTK](https://www.nltk.org/_modules/nltk/stem/snowball.html?utm_source=chatgpt.com)

### ⭐ Most important limitation

**Stemming is a heuristic process, not true linguistic understanding.**

It tries to make related word forms look similar; it does **not** necessarily determine the correct dictionary root.
