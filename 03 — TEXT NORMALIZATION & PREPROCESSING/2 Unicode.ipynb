## 1. Unicode Basics

**Unicode** is a universal standard for representing text from different writing systems.

Before Unicode, different character encodings were commonly used for different languages, which could cause compatibility problems.

Unicode assigns each character a unique **code point**.

### Example

```text
Character     Unicode Code Point

A             U+0041
B             U+0042
é             U+00E9
அ             U+0B85
中             U+4E2D
```

The `U+` notation represents a Unicode code point written in hexadecimal.

### Important distinction

```text
Character
    ↓
Unicode Code Point
    ↓
Encoding
    ↓
Bytes
```

For example:

```text
"é"
 ↓
U+00E9
 ↓
UTF-8
 ↓
C3 A9
```

A **code point identifies a character**, while an **encoding determines how that code point is represented as bytes**.

# 2. Unicode Normalization

The same visible text can sometimes have **different underlying Unicode representations**.

For example, `é` can be represented as:

```text
Form 1:
é
→ U+00E9
```

or as:

```text
Form 2:
e + combining acute accent
→ U+0065 + U+0301
```

They look identical to humans but contain different sequences of code points.



This can cause problems with:

- String comparison
- Searching
- Tokenization
- Vocabulary creation
- Duplicate detection

### Unicode normalization

**Unicode normalization converts equivalent Unicode sequences into a standardized form.**

The Unicode Standard defines four normalization forms:

| Form | Meaning |
|---|---|
| **NFC** | Canonical Decomposition + Composition |
| **NFD** | Canonical Decomposition |
| **NFKC** | Compatibility Decomposition + Composition |
| **NFKD** | Compatibility Decomposition | :chatgpt-content-reference{index="2"}


### Most important for NLP

**NFC** is generally a good choice for general text because it provides a consistent representation while preserving canonical equivalence. [Unicode](https://www.unicode.org/faq/normalization.html?utm_source=chatgpt.com)

Example in Python:

```python
import unicodedata

text = "café"

normalized = unicodedata.normalize("NFC", text)
```

### ⚠️ NFKC/NFKD

Compatibility normalization can remove distinctions that may carry formatting or semantic information, so it should **not be applied blindly**. [Unicode](https://www.unicode.org/standard/reports/tr15/tr15-21.html?utm_source=chatgpt.com)

# 3. UTF-8

**UTF-8 (Unicode Transformation Format — 8-bit)** is an encoding used to represent Unicode text as bytes.

It can represent characters using **1 to 4 bytes**.

### Basic idea

```text
Unicode Character
       ↓
Unicode Code Point
       ↓
UTF-8 Encoding
       ↓
Bytes
```

For example:

```text
A
↓
U+0041
↓
01000001
```

For a non-ASCII character:

```text
中
↓
U+4E2D
↓
UTF-8
↓
E4 B8 AD
```



### UTF-8 byte length

| Unicode character range | UTF-8 bytes |
|---|---:|
| ASCII characters | 1 byte |
| Many Latin/European characters | 2 bytes |
| Many Asian characters | 3 bytes |
| Many supplementary characters/emojis | 4 bytes |

UTF-8 is **variable-length encoding**, unlike a fixed one-byte representation for every character.

# 4. Why Unicode Matters in Multilingual NLP

NLP systems need to process text from many languages and writing systems.

For example:

```text
English   → Hello
Tamil     → வணக்கம்
Hindi     → नमस्ते
Chinese   → 你好
Arabic    → مرحبا
Japanese  → こんにちは
```

Without proper Unicode handling, multilingual text can become:

```text
乱码 / corrupted text
```

instead of meaningful characters.

### Unicode enables NLP systems to:

- Process multiple languages
- Store multilingual datasets
- Tokenize non-English text
- Build multilingual vocabularies
- Perform cross-language search
- Train multilingual models
- Handle emojis and symbols

# 🧠 Unicode vs UTF-8 vs Normalization

| Concept | What it does |
|---|---|
| **Unicode** | Defines characters and their code points |
| **Code point** | Identifies a specific Unicode character |
| **UTF-8** | Encodes Unicode code points into bytes |
| **Normalization** | Makes equivalent Unicode sequences consistent |

### Easy way to remember

```text
Unicode
"What characters exist?"

Code Point
"Which character is this?"

UTF-8
"How do I store/transmit it as bytes?"

Normalization
"How do I represent equivalent text consistently?"
```

## ⭐ Key Takeaway

> **Unicode provides a universal representation for characters, UTF-8 encodes those characters into bytes, and Unicode normalization ensures equivalent text can be represented consistently.**

This becomes especially important when your NLP systems move beyond English into **Tamil, Hindi, Chinese, Arabic, multilingual datasets, and emoji-rich text**. [Unicode](https://www.unicode.org/versions/Unicode18.0.0/core-spec/chapter-2/?utm_source=chatgpt.com)
