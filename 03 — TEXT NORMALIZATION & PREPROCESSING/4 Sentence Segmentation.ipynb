## 3.4 Sentence Segmentation



### What is Sentence Segmentation?

**Sentence segmentation** is the process of dividing continuous text into individual sentences.

It is also called:

- **Sentence splitting**
- **Sentence boundary detection (SBD)**
- **Sentence boundary disambiguation**

The goal is to determine **where one sentence ends and the next begins**. [Stanford Web Services](https://web.stanford.edu/~jurafsky/slp3/ed3bookfeb3_2024.pdf?utm_source=chatgpt.com)

### Example

```text
Input:
"I love NLP. It is very interesting. I am learning Transformers."

                    ↓

Sentence 1 → "I love NLP."
Sentence 2 → "It is very interesting."
Sentence 3 → "I am learning Transformers."
```

---

# 1. Sentence Boundaries

A **sentence boundary** is the position where one sentence ends and another begins.

Common boundary markers in English are:

```text
.    → Period
?    → Question mark
!    → Exclamation mark
```

Example:

```text
"Are you learning NLP? Yes! It is fascinating."
                  ↑     ↑
               boundary
```

Question marks and exclamation marks are generally easier to interpret as sentence boundaries, while the period `.` is more ambiguous. [Stanford Web Services](https://web.stanford.edu/~jurafsky/slp3/ed3bookfeb3_2024.pdf?utm_source=chatgpt.com)

---

## Why is `.` difficult?

A period does **not always mean the end of a sentence**.

It can represent:

### Sentence ending

```text
"I am learning NLP."
                  ↑
            sentence boundary
```

### Abbreviation

```text
"Dr. Kumar is here."
    ↑
not a boundary
```

### Decimal number

```text
"The price is 4.5 dollars."
               ↑
         not a boundary
```

### Multiple-period abbreviation

```text
"He lives in the U.S.A."
             ↑
        not each period
```

Therefore, sentence segmentation requires **context**, not simply splitting at every `.`. [NLTK](https://www.nltk.org/book/ch03.html?utm_source=chatgpt.com)

---

# 2. Abbreviations

Abbreviations are one of the major challenges in sentence segmentation.

Examples:

```text
Dr.
Mr.
Mrs.
Prof.
Inc.
etc.
U.S.
U.S.A.
```

Consider:

```text
"Dr. Kumar is an NLP researcher. He works in Chennai."
```

❌ Incorrect:

```text
Sentence 1 → "Dr."
Sentence 2 → "Kumar is an NLP researcher."
Sentence 3 → "He works in Chennai."
```

✅ Correct:

```text
Sentence 1 → "Dr. Kumar is an NLP researcher."
Sentence 2 → "He works in Chennai."
```

The system must recognize that the period after `Dr.` belongs to the abbreviation rather than marking a sentence boundary. [Stanford Web Services](https://web.stanford.edu/~jurafsky/slp3/ed3bookfeb3_2024.pdf?utm_source=chatgpt.com)



---

# 3. Sentence Splitting

The simplest approach is to look for sentence-ending punctuation.

```python
text = "I love NLP. It is fascinating. Let's learn!"

sentences = [
    "I love NLP.",
    "It is fascinating.",
    "Let's learn!"
]
```

Conceptually:

```text
Text
 ↓
Find possible boundary markers
 ↓
Check context
 ↓
Split into sentences
```

A naïve approach such as:

```python
text.split(".")
```

is **not reliable** because it incorrectly splits abbreviations and decimal numbers.

For example:

```text
"Dr. Kumar has 4.5 years of experience."
```

A simple period split can produce incorrect fragments.

---

# 4. Rule-Based Sentence Segmentation

A traditional approach uses **rules and heuristics**.

A simplified rule could be:

```text
IF punctuation = "." / "?" / "!"
AND it represents the end of a sentence
THEN create a sentence boundary
```

But additional rules are needed for cases such as:

```text
Dr.
Mr.
U.S.
4.5
example.com
```

An **abbreviation dictionary** can help identify periods that belong to abbreviations. [Stanford Web Services](https://web.stanford.edu/~jurafsky/slp3/ed3bookfeb3_2024.pdf?utm_source=chatgpt.com)

---

# 5. Modern Sentence Segmentation

More advanced sentence segmenters can use **statistical or machine-learning approaches** to determine whether a punctuation mark represents a true sentence boundary.

For example:

```text
Text
 ↓
Tokenization
 ↓
Context/features
 ↓
Boundary Detection
 ↓
Sentence 1 | Sentence 2 | Sentence 3
```

Libraries such as **spaCy** provide sentence segmentation using approaches including dependency parsing and statistical sentence recognition. [spaCy](https://spacy.io/usage/linguistic-features%23%23tokenization?utm_source=chatgpt.com)
