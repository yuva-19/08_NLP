

# 3.8 Lemmatization

Lemmatization is the process of reducing a word to its **base/dictionary form (lemma)** using linguistic information rather than simply cutting characters.



### WordNet

**WordNet** is a lexical database of English containing words and relationships between their meanings, including different word senses and parts of speech. [NLTK](https://www.nltk.org/howto/wordnet.html?utm_source=chatgpt.com)

Conceptually:

```text
Input word
    ↓
Check morphological form
    ↓
Dictionary / lexical knowledge
    ↓
Find valid lemma
    ↓
Output
```

---

# 4. NLTK WordNet Lemmatizer

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["dogs", "running", "studies", "children"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
```

Example:

```text
dogs      → dog
running   → running
studies   → study
children  → child
```

Notice something important:

```text
running → running
```

Why?

Because the default `WordNetLemmatizer` assumes the word is a **noun** unless a POS is explicitly supplied. NLTK's API allows noun, verb, adjective, adverb, and satellite-adjective POS values. [NLTK](https://www.nltk.org/api/nltk.stem.WordNetLemmatizer.html?utm_source=chatgpt.com)

# 5. POS-Aware Lemmatization

This is one of the **most important concepts** in lemmatization. 🔥

**POS = Part of Speech**

Examples:

```text
Noun
Verb
Adjective
Adverb
```

The same word can have different lemmas depending on its grammatical role.

### Example: `meeting`

```text
I attended a meeting.
       ↓
     noun
       ↓
   meeting
```

But:

```text
They are meeting today.
       ↓
      verb
       ↓
      meet
```

Therefore, knowing the **POS** can improve lemmatization.

---

## NLTK POS-aware example

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("meeting", pos="n"))
print(lemmatizer.lemmatize("meeting", pos="v"))
```

Output:

```text
meeting
meet
```

NLTK's `WordNetLemmatizer` supports:

| POS | Meaning |
|---|---|
| `n` | Noun |
| `v` | Verb |
| `a` | Adjective |
| `r` | Adverb |
| `s` | Satellite adjective | :chatgpt-content-reference{index="5"}

# 6. Why POS Matters

Consider:

```text
The leaves are falling.
```

Here:

```text
leaves → noun → leaf
```

But in another context:

```text
He leaves the room.
```

Here:

```text
leaves → verb → leave
```

So:

```text
Word + POS
     ↓
Lemmatization
     ↓
Correct lemma
```

This is why **POS-aware lemmatization** is more reliable than blindly applying a single transformation.

---

# 7. POS Tagging + Lemmatization Pipeline

A practical pipeline can look like:

```text
Raw Text
   ↓
Tokenization
   ↓
POS Tagging
   ↓
POS Mapping
   ↓
Lemmatization
   ↓
Normalized Tokens
```

Example:

```text
"The children are running."

children → NOUN → child
are      → VERB → be
running  → VERB → run
```

So the normalized representation becomes approximately:

```text
["the", "child", "be", "run"]
```

The exact result depends on the POS tagger and lemmatizer being used.

NLTK also provides mechanisms for mapping POS tags to WordNet's POS categories. [NLTK](https://www.nltk.org/api/nltk.corpus.reader.wordnet.html?utm_source=chatgpt.com)

# 8. Stemming vs Lemmatization



| Feature | Stemming | Lemmatization |
|---|---|---|
| Main idea | Remove affixes | Find valid base form |
| Method | Heuristic/rule-based | Dictionary + linguistic information |
| Output | May be non-word | Generally valid word |
| POS awareness | Usually no | Can be POS-aware |
| Accuracy | Lower | Generally higher |
| Speed | Faster | Usually slower |
| Complexity | Simple | More complex |
| Example | `studies → studi` | `studies → study` |
| Example | `better → better` | `better → good` |

### Mental model

```text
STEMMING
running
   ↓
remove suffix
   ↓
runn / run
```

```text
LEMMATIZATION
running
   ↓
POS + lexical knowledge
   ↓
run
```

NLTK's documentation explicitly describes lemmatization as an additional normalization step that checks whether the resulting form is a known dictionary word, making it slower than stemming. [NLTK](https://www.nltk.org/book/ch03.html?utm_source=chatgpt.com)

# 9. When to Use Lemmatization?

### Prefer lemmatization when:

- Word meaning matters
- Linguistic correctness matters
- You need valid dictionary forms
- POS information is available/useful
- Building a clean vocabulary
- Performing linguistic analysis

### Stemming may be preferable when:

- Speed is more important
- Approximate matching is sufficient
- Working with simple information retrieval systems
- You need a lightweight preprocessing step

---

# 10. Important Limitation

Lemmatization is **not perfect**.

It depends on:

- Dictionary coverage
- Language
- POS information
- Quality of morphological analysis
- Context

For example, NLTK's WordNet lemmatizer may leave a word unchanged if it cannot find a suitable entry. [NLTK](https://www.nltk.org/api/nltk.stem.WordNetLemmatizer.html?utm_source=chatgpt.com)

Modern NLP systems such as Transformer models also often use **subword tokenization** rather than traditional stemming/lemmatization as their primary way of handling word variation.

2. The resulting form is called a **lemma**.
3. It can use **dictionary/lexical knowledge** such as WordNet.
4. **POS information is important** for accurate lemmatization.
5. `studies → study` is lemmatization; `studies → studi` is stemming.
