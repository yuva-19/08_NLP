# AutoTokenizer

```python
from transformers import AutoTokenizer

model_name = "distilbert/distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

text = "Whyy are You beinggg late again brotha."

inputs = tokenizer(text)

tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"])

print(tokens)
print(inputs["input_ids"])
```

```python
for token, token_id in zip(tokens, inputs["input_ids"]):
    print(f"{token:<8}  -->  {token_id}")
```

# AutoModel

```python
from transformers import AutoTokenizer, AutoModel

model_name = "distilbert/distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

text = "I love learning AI."

# Text → tokens/numbers
inputs = tokenizer(text, return_tensors="pt")

# Tokens → Transformer representations
outputs = model(**inputs)

print(outputs.last_hidden_state.shape)
```

```python
print(outputs.last_hidden_state)
```

# AutoModelForSequenceClassification

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

text = "I love this movie!"

# Text → tokens
inputs = tokenizer(text, return_tensors="pt")

# Tokens → predictions
outputs = model(**inputs)

# Get predicted class
prediction = torch.argmax(outputs.logits, dim=1)

print(outputs.logits)
print(prediction)
```

# AutoModelForTokenClassification

```python
from transformers import AutoTokenizer, AutoModelForTokenClassification
import torch

model_name = "dbmdz/bert-large-cased-finetuned-conll03-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name)

text = "Elon Musk founded SpaceX."

inputs = tokenizer(text, return_tensors="pt")

outputs = model(**inputs)

predictions = torch.argmax(outputs.logits, dim=2)

print(predictions)
```

```python
id2label = model.config.id2label

for token_id, prediction in zip(
    inputs["input_ids"][0],
    predictions[0]
):
    token = tokenizer.decode([token_id])
    label = id2label[prediction.item()]

    print(f"{token:7} → {label}")
```

# AutoModelForQuestionAnswering

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch

model_name = "distilbert-base-cased-distilled-squad"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

context = "Python was created by Guido van Rossum in 1991."
question = "Who created Python?"

inputs = tokenizer(
    question,
    context,
    return_tensors="pt"
)

outputs = model(**inputs)

start = torch.argmax(outputs.start_logits)
end = torch.argmax(outputs.end_logits)

answer_tokens = inputs["input_ids"][0][start:end + 1]

answer = tokenizer.decode(answer_tokens)

print(answer)
```

# AutoModelForSeq2SeqLM

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = """
Artificial intelligence is being used in many industries.
Companies use machine learning for prediction, automation,
recommendation systems, and natural language processing.
"""

inputs = tokenizer(
    text,
    return_tensors="pt",
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_new_tokens=50
)

summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(summary)
```
