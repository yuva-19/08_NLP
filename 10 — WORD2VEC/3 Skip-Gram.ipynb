# 10.3 Skip-Gram — Target Word → Context Words 🧠

Skip-gram is a Word2Vec architecture that learns to predict surrounding words from a given target word.

\\[ \boxed{\text{Target word}\longrightarrow\text{Context words}} \\]

Unlike CBOW, which uses surrounding words to predict the target, Skip-gram starts with the target and predicts its neighbors.

## 4. How does Skip-gram learn?

1. Input: Convert the target word into its vocabulary index.
2. Embedding lookup: Retrieve its trainable vector.
3. Prediction: Estimate which words are likely to occur in its context.
4. Loss calculation: Penalize incorrect predictions.
5. Weight updates: Use backpropagation and gradient-based optimization to improve the vectors.

### Positive and negative examples

When using Skip-gram with Negative Sampling (SGNS):

- Positive pair: `(drinks, milk)` — a genuine context pair.
- Negative pair: `(drinks, airplane)` — an illustrative sampled pair.

The model learns to score positive pairs higher and sampled negative pairs lower. Negative sampling is an efficient alternative to predicting a probability distribution over the entire vocabulary for every pair.

## 5. CBOW vs Skip-gram — final revision

| CBOW                        | Skip-gram                               |
| --------------------------- | --------------------------------------- |
| Context → target            | Target → context                        |
| Combines context embeddings | Starts with one target embedding        |
| Often faster to train       | Often effective for less frequent words |
| Predicts the missing word   | Predicts surrounding words              |

Key takeaway: Skip-gram learns word embeddings by using a target word to predict its neighboring words. Its training process adjusts the embedding vectors so they capture useful relationships between words.
