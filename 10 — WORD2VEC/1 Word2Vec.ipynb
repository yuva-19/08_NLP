# Module 10.1 — Word2Vec 🧠

Word2Vec is a word-embedding technique that learns dense vector representations by examining how words appear in context.

The central idea is simple: train a model to predict relationships between words in a sentence, then use its learned weights as word embeddings.

For example:

> “The cat drinks fresh milk.”

Word2Vec learns from the relationships between `cat`, `drinks`, `fresh`, and `milk`. It has two main training architectures: CBOW and Skip-gram.

## 1. CBOW vs Skip-gram

## CBOW — Continuous Bag of Words

Predict the target word from its surrounding context.

the

cat

fresh

milk

Predict: drinks

Context words are inputs; the missing target word is the output.

## Skip-gram

Predict surrounding context words from the target word.

Input: drinks

the

cat

fresh

milk

The target word is the input; nearby words are the prediction targets.

| Feature                   | CBOW                       | Skip-gram                            |
| ------------------------- | -------------------------- | ------------------------------------ |
| Input                     | Context words              | Target word                          |
| Predicts                  | Target word                | Context words                        |
| Typical training behavior | Often faster               | Often more computationally expensive |
| Common intuition          | Combine context to predict | Learn contexts around each word      |

Important: “Bag of Words” means CBOW generally does not use the order of context words in its basic architecture.

## 2. Context Window

The context window determines how many neighboring words the model considers around a target word.

Consider this sentence:

`The cat drinks fresh milk today`

Suppose the target word is `drinks` and the context window size is 2.

The

cat

drinks

fresh

milk

today

Target: drinks · Window size: 2

Context words: the, cat, fresh, milk

- Small window: Usually emphasizes nearby, more local relationships.
- Large window: Can capture broader topical or semantic relationships.

The context window is a distance measured in tokens (here, words), not the total sentence length.

## 3. Positive Examples

A positive example is a genuine target–context pair extracted from the training text.

For the sentence:

`The cat drinks fresh milk`

With a context window of 2, Skip-gram can generate these positive pairs:

| Target word | Context word |
| ----------- | ------------ |
| drinks      | the          |
| drinks      | cat          |
| drinks      | fresh        |
| drinks      | milk         |

These pairs teach the model which words occur near one another.

The precise set of pairs depends on the window convention, sentence boundaries, and whether the window size varies randomly during training.

## 4. Negative Examples

Negative examples are word pairs sampled to represent target–context combinations that the model should generally score lower.

For example, suppose the positive pair is:

`(drinks, milk)`

A training process might sample these negative pairs:

- `(drinks, airplane)`
- `(drinks, mountain)`
- `(drinks, keyboard)`

These are illustrative samples, not guaranteed negatives in every corpus.

Negative sampling trains the model to give higher scores to genuine pairs and lower scores to sampled negative pairs. A sampled negative is not necessarily linguistically impossible; it simply serves as a training contrast.

This technique avoids calculating a probability over the entire vocabulary for every training example.

## 5. Embedding Matrix

Word2Vec learns its word vectors through trainable weight matrices.

Suppose the vocabulary contains 10,000 words and the embedding dimension is 100.

\\[ W\_{\text{in}}\in\mathbb{R}^{10000\times100} \\]


Each row represents a word; each column represents an embedding dimension. The actual matrix would contain 10,000 rows and 100 columns.

- Each word receives a vocabulary index.
- That index selects a row from the embedding matrix.
- Training adjusts the matrix values based on the prediction objective.
- After training, the learned rows provide word embeddings.

In the original Word2Vec architecture, there is also an output weight matrix used to predict context words. The input embedding matrix is commonly extracted as the final word representation.

## 6. How everything connects

Training text corpus

Generate target–context pairs using a context window

Train CBOW or Skip-gram, optionally using negative sampling

Learned embedding matrix

### Quick revision 🎯

- CBOW: Context → target.
- Skip-gram: Target → context.
- Context window: Controls which neighboring words are considered.
- Positive examples: Genuine target–context pairs.
- Negative examples: Sampled pairs used for contrastive training.
- Embedding matrix: Trainable weights that store the learned word vectors.
