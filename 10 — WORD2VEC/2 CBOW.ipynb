# 10.2 CBOW — Continuous Bag of Words 🧠

CBOW predicts a target word using its surrounding context words.

\\[ \boxed{\text{Context words} \longrightarrow \text{Target word}} \\]

For example, consider:

> “The cat \_\_\_ fresh milk.”

Given the context words `the`, `cat`, `fresh`, and `milk`, the model tries to predict the missing word: drinks.

## 1. CBOW architecture

Context words

`the` · `cat` · `fresh` · `milk`

Embedding lookup

Convert each context word into a vector.

Combine context vectors

Typically, average the vectors.

Predict target word

`drinks`

## 2. Mathematical intuition

Suppose each context word has a 3-dimensional embedding:

\\[ \begin{aligned} \mathbf{v}\_{the}&=[1,0,1]\\\ \mathbf{v}\_{cat}&=[0,2,1]\\\ \mathbf{v}\_{fresh}&=[1,1,0]\\\ \mathbf{v}\_{milk}&=[2,1,1] \end{aligned} \\]

CBOW combines these vectors by averaging them:

\\[ \mathbf{h}=\frac{1}{C}\sum\_{i=1}^{C}\mathbf{v}\_{w_i} \\]

Here, \\(C=4\\), the number of context words.

Therefore,

\\[ \mathbf{h}=[1,\ 1,\ 0.75] \\]

This combined representation is used to predict the target word. These values are illustrative; real embeddings are learned during training.

## 3. How does CBOW learn?

1. Provide context words as input.
2. Use their embeddings to predict the target word.
3. Compare the prediction with the actual target.
4. Calculate the prediction error using a loss function, typically cross-entropy with a softmax output.
5. Update the model's weights through backpropagation and gradient descent.

Over many training examples, the model learns useful word representations.

## 4. Important characteristics

- Efficient training: Multiple context words contribute to one prediction.
- Order-independent context: Basic CBOW treats the context as a bag of words.
- Learned embeddings: The embedding vectors are updated during training.
- Full softmax vs negative sampling: Original CBOW can use softmax over the vocabulary; negative sampling is a commonly used alternative training objective.

## 5. CBOW vs Skip-gram — quick revision

| CBOW                                 | Skip-gram                                                        |
| ------------------------------------ | ---------------------------------------------------------------- |
| Context → target                     | Target → context                                                 |
| One prediction from combined context | Predicts surrounding words from target                           |
| Often faster to train                | Often useful for learning representations of less frequent words |
