# 📘 MODULE 1 — INTRODUCTION TO NLP
## 1.1 What Is NLP?



## 1. Natural Language Processing (NLP)

**Natural Language Processing (NLP)** is a field of **Artificial Intelligence (AI)** that enables computers to **process, understand, analyze, and generate human language**.

Human language is difficult for computers because it contains:

- Ambiguity
- Context
- Multiple meanings
- Slang and abbreviations
- Different languages
- Spelling variations
- Implicit meaning

### Examples

- Chatbots
- Google Search
- Spam detection
- Sentiment analysis
- Machine translation
- Text summarization
- Question answering

---

## 2. NLP vs ML vs DL

| Concept | Meaning | Example |
|---|---|---|
| **AI** | Broad field of making machines intelligent | AI assistant |
| **ML** | Algorithms learn patterns from data | Spam classifier |
| **DL** | ML using neural networks | BERT |
| **NLP** | Domain concerned with human language | Sentiment analysis |

### Important distinction

**NLP is primarily a problem/domain area**, while **ML and DL are techniques that can be used to solve NLP problems.**

```text
NLP Problem
     ↓
Sentiment Analysis
     ↓
ML → Logistic Regression / SVM
     OR
DL → LSTM / Transformer / BERT
```

# 3. Human Language → Machine Representation

Humans understand:

> **"The movie was amazing!"**

A machine learning model needs a **numerical representation**.

```text
Raw Text
   ↓
"The movie was amazing"
   ↓
Tokenization
   ↓
["The", "movie", "was", "amazing"]
   ↓
Numerical Representation
   ↓
[125, 842, 37, 921]
   ↓
NLP Model
   ↓
Prediction
```

Common text representations include:

- One-Hot Encoding
- Bag of Words
- TF-IDF
- Word Embeddings
- Contextual Embeddings

**Core idea:**

> Computers cannot directly perform mathematical operations on raw natural language; NLP systems transform language into representations that computational models can process.

---

# 4. Structured vs Unstructured Text

### Structured Data

Data organized according to a predefined format.

```text
Name       Age       City
Yuvanesh   22        Trichy
```

Examples:

- SQL tables
- CSV files
- Database records

### Unstructured Data

Data without a predefined tabular structure.

```text
"I really loved this movie. The acting was fantastic!"
```

Examples:

- Reviews
- Emails
- Social media posts
- Articles
- Documents
- Chat messages

**NLP primarily deals with extracting useful information from unstructured or semi-structured language data.**

---

# 5. NLP Pipeline



A general NLP pipeline can be represented as:

```text
Raw Text
   ↓
Cleaning
   ↓
Normalization
   ↓
Tokenization
   ↓
Representation
   ↓
NLP Model
   ↓
Prediction
   ↓
Evaluation
```

### Example — Sentiment Analysis

```text
"This phone has an amazing camera!"
                ↓
          Preprocessing
                ↓
["This", "phone", "has", "an", "amazing", "camera"]
                ↓
       Numerical Representation
                ↓
             Model
                ↓
            Positive
```

⚠️ **Important:** Not every NLP system uses exactly the same pipeline. Modern Transformer-based systems may perform much of the transformation internally.

# 6. NLP Applications



### 6.1 Sentiment Analysis

Determines the sentiment expressed in text.

```text
"I absolutely loved this movie!"
              ↓
           Positive
```

Used in:
- Product reviews
- Customer feedback
- Social media analysis

---

### 6.2 Text Classification

Assigns text to predefined categories.

```text
"Congratulations! You won a free prize!"
                    ↓
                   Spam
```

Applications:
- Spam detection
- News classification
- Support-ticket classification

---

### 6.3 Search

NLP helps search engines understand user queries and retrieve relevant information.

```text
"best laptop for machine learning"
                ↓
        Relevant results
```

---

### 6.4 Recommendation

NLP can analyze textual information to understand user preferences and improve recommendations.

```text
Reviews + User Text
        ↓
Preference Understanding
        ↓
Recommendations
```

---

### 6.5 Machine Translation

Automatically converts text from one language to another.

```text
"I am learning NLP."
          ↓
"நான் NLP கற்றுக்கொண்டு இருக்கிறேன்."
```

---

### 6.6 Text Summarization

Produces a shorter version of a longer document while retaining important information.

Two major approaches:

- **Extractive:** selects important existing text.
- **Abstractive:** generates a new summary.

```text
Long Document
      ↓
 NLP Model
      ↓
Short Summary
```

---

### 6.7 Question Answering

Answers questions using information from a given context or knowledge source.

```text
Context:
"Python was created by Guido van Rossum."

Question:
"Who created Python?"

Answer:
"Guido van Rossum"
```

---

### 6.8 Chatbots

Systems that interact with humans using natural language.

Examples:

- Customer-support bots
- Virtual assistants
- Conversational AI

Modern chatbots commonly use **Transformer-based language models**.

---

### 6.9 Information Extraction

Extracts useful structured information from unstructured text.

```text
"Apple launched the iPhone in India."

        ↓

Company  → Apple
Product  → iPhone
Location → India
```

---

### 6.10 Named Entity Recognition (NER)

NER identifies and classifies entities mentioned in text.

```text
"Elon Musk founded SpaceX."

Elon Musk → PERSON
SpaceX    → ORGANIZATION
```

Common entity types:

- PERSON
- ORGANIZATION
- LOCATION
- DATE
- MONEY
- PRODUCT

---

### 6.11 Semantic Search

Traditional keyword search mainly focuses on matching words.

**Semantic search focuses on the meaning/context of the query.**

```text
Query:
"How can I make my laptop faster?"

Document:
"Methods to improve computer performance"
```

Even though the words are different, the **meaning is similar**.

Semantic search becomes especially important later when learning:

**Embeddings → Sentence Embeddings → Transformers → Dense Retrieval → RAG**
