# 📘 1.2 Evolution of NLP



## 1. Rule-Based NLP

**Early NLP systems were based on manually written rules.**

Humans explicitly defined linguistic rules that the computer followed.

### Example

```text
IF sentence contains "not good"
        ↓
    Negative
```

### Characteristics

- Hand-written rules
- No learning from data
- Highly dependent on linguistic expertise
- Difficult to maintain for complex language

### Problem

Natural language contains too many exceptions and variations to describe everything with manually written rules.

---

# 2. Statistical NLP

The next stage used **statistical methods** to learn patterns from large collections of text.

Instead of explicitly writing every rule:

```text
Rules → Manually defined
```

the system learns:

```text
Text Data
   ↓
Statistical Patterns
   ↓
Prediction
```

Common approaches included:

- N-gram language models
- Hidden Markov Models (HMMs)
- Probabilistic models

### Example

Given:

> "I want to drink ___"

A statistical model can estimate that:

```text
water → high probability
car   → low probability
```

The major shift was:

> **From manually designed rules → patterns learned from data.**

---

# 3. Classical Machine Learning

NLP then adopted traditional **Machine Learning algorithms**.

Text had to be converted into numerical features such as:

- Bag of Words
- N-grams
- TF-IDF

Then classical ML algorithms could perform NLP tasks.

```text
Text
 ↓
Preprocessing
 ↓
BoW / TF-IDF
 ↓
ML Algorithm
 ↓
Prediction
```

Common algorithms:

- Logistic Regression
- Naive Bayes
- SVM
- Decision Trees

### Example

```text
"I love this phone"
        ↓
     TF-IDF
        ↓
Logistic Regression
        ↓
    Positive
```

### Limitation

These approaches generally depended heavily on **hand-designed features** and had limited ability to capture complex semantic/contextual relationships.

---

# 4. Word Embeddings

A major breakthrough was representing words as **dense numerical vectors**.

Instead of:

```text
"king" → [0,0,0,1,0,0,...]
```

embeddings represent words using meaningful dense vectors:

```text
king → [0.21, -0.43, 0.72, ...]
queen → [0.19, -0.40, 0.70, ...]
```

Words with similar meanings tend to have similar vector representations.

Important approaches:

- **Word2Vec**
- **GloVe**
- **FastText**

### Key idea

> **Meaning can be represented through numerical vectors learned from language data.**

This became the foundation for neural NLP.

---

# 5. RNN / LSTM



Neural sequence models introduced the ability to process text **sequentially**.

### RNN

An RNN processes tokens one after another while maintaining a hidden state.

```text
The → cat → is → sleeping
 ↓      ↓      ↓      ↓
h₁     h₂     h₃     h₄
```

The hidden state carries information from previous tokens.

### Problem with basic RNNs

They struggle with **long-term dependencies** because of issues such as vanishing gradients.

### LSTM

**Long Short-Term Memory (LSTM)** was designed to handle longer dependencies more effectively using gates and memory cells.

This improved tasks such as:

- Language modeling
- Machine translation
- Text classification
- Sequence generation

---

# 6. Seq2Seq + Attention

RNN-based **Sequence-to-Sequence (Seq2Seq)** models enabled one sequence to be transformed into another.

```text
Input Sequence
      ↓
   Encoder
      ↓
Representation
      ↓
   Decoder
      ↓
Output Sequence
```

Example:

```text
English sentence
       ↓
    Encoder
       ↓
    Decoder
       ↓
Tamil sentence
```

### The problem

A basic encoder-decoder architecture attempted to compress the entire input sequence into a limited representation.

This created a **context bottleneck**, especially for long sequences.

### Attention

Attention allowed the decoder to dynamically focus on relevant parts of the input.

```text
Input:
"The cat that was sitting on the mat..."

                ↓
             Attention
                ↓
Focus on relevant tokens
                ↓
             Output
```

This was a major step toward modern Transformer architectures.

---

# 7. Transformers



The **Transformer architecture** introduced a major change in NLP.

Instead of processing sequences strictly one token at a time using recurrence, Transformers rely heavily on **self-attention**.

### Core idea

Each token can interact with other tokens in the sequence.

```text
The ───────┐
cat ───────┤
sat ───────┼──→ Self-Attention
on ────────┤
mat ───────┘
```

This allows the model to capture relationships between distant words more effectively.

### Important components

- Self-attention
- Multi-head attention
- Positional information
- Feed-forward networks
- Residual connections
- Layer normalization

Transformers became the foundation of modern NLP.

---

# 8. Pretrained Transformers

Instead of training a language model from scratch for every task, researchers began **pretraining large Transformer models** on massive text datasets.

Then the pretrained model could be adapted to downstream tasks.

```text
Large Text Dataset
        ↓
   Pretraining
        ↓
Pretrained Transformer
        ↓
 ┌──────┼────────┐
 ↓      ↓        ↓
NER   Classification  QA
```

Important examples:

- **BERT**
- **RoBERTa**
- **T5**

This introduced the idea:

> **Pretrain once → adapt to many NLP tasks.**

---

# 9. Large Language Models (LLMs)




Modern NLP evolved toward **Large Language Models (LLMs)**.

LLMs are large neural language models trained on massive amounts of text and capable of performing many language tasks.

Examples include models in the GPT family and other modern Transformer-based language models.

They can perform:

- Text generation
- Question answering
- Summarization
- Translation
- Classification
- Code generation
- Conversational interaction

The evolution can therefore be summarized as:

```text
Rule-Based
    ↓
Statistical NLP
    ↓
Classical ML
    ↓
Word Embeddings
    ↓
RNN / LSTM
    ↓
Seq2Seq + Attention
    ↓
Transformers
    ↓
Pretrained Transformers
    ↓
LLMs
```

## 🧠 The Big Picture

The fundamental shift across NLP evolution was:

```text
Hand-written Rules
       ↓
Statistical Patterns
       ↓
Hand-designed Features
       ↓
Learned Word Representations
       ↓
Neural Sequence Models
       ↓
Attention
       ↓
Self-Attention + Transformers
       ↓
Large-Scale Pretraining
       ↓
General-Purpose Language Models
```

### ⭐ Key takeaway

> **NLP evolved from manually programmed linguistic rules to statistical learning, feature-based ML, neural representations, attention-based architectures, Transformers, and finally large pretrained language models.**
