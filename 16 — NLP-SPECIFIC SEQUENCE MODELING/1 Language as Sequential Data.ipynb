## 1. Token sequences

Language is sequential because the order of tokens often changes the meaning of a sentence.

For example:

- "The dog chased the cat."
- "The cat chased the dog."

The words are almost identical, but their order changes who performed the action.



A token is a unit processed by an NLP model. Depending on the tokenizer, it might be a word, subword, character, or punctuation mark.

A sentence becomes an ordered sequence:

\\[ X=(x_1,x_2,\ldots,x_T) \\]

where \\(x_t\\) represents the token at position \\(t\\), and \\(T\\) is the sequence length.

For example:

`"I love NLP"` → `["I", "love", "NLP"]`

Why it matters: Sequence models use token order to understand relationships between words. A bag-of-words representation, by contrast, does not inherently preserve word order.

## 2. Variable-length text

Unlike fixed-size images, text samples naturally have different lengths.


Consider these sentences:

| Sentence                                     | Number of word tokens |
| -------------------------------------------- | --------------------- |
| I love NLP                                   | 3                     |
| NLP is interesting                           | 3                     |
| I enjoy learning natural language processing | 7                     |

The length of a sequence depends on its tokenization. A subword tokenizer may produce more tokens than a word-level tokenizer for the same sentence.

Why it matters: Neural networks commonly process examples in batches, but batches of sequences with different lengths require a strategy to make their dimensions compatible.

Common approaches include padding, truncation, and length-based batching.

## 3. Sequence padding

Padding adds special tokens to shorter sequences so that sequences in a batch have a common length.



Suppose the longest sequence in a batch contains five tokens.

| Original sequence      | After padding                |
| ---------------------- | ---------------------------- |
| `I love NLP`           | `I love NLP <PAD> <PAD>`     |
| `NLP is useful`        | `NLP is useful <PAD> <PAD>`  |
| `I enjoy learning NLP` | `I enjoy learning NLP <PAD>` |

The special token `<PAD>` occupies the unused positions.

Important distinction:

- Padding: Adds positions to shorter sequences.
- Truncation: Removes tokens from sequences that exceed a specified maximum length.

Padding makes batching easier, but it introduces positions that do not represent actual text. Models must handle those positions appropriately.

## 4. Attention masks

An attention mask indicates which token positions should be treated as valid and which positions are padding or otherwise excluded from attention.


For example, suppose the sequence is:

`I love NLP <PAD> <PAD>`

Its padding mask might be:

\\[ M=[1,1,1,0,0] \\]

- `1` — real token.
- `0` — padding position.

In Transformer models, the attention mechanism uses a padding mask to prevent attention from being assigned to masked padding positions.

Remember that padding and masking are different: padding changes the sequence representation, whereas masking controls how the model treats particular positions. Attention masks can also serve other purposes, such as preventing a decoder from attending to future tokens.

## 5. Vocabulary

A vocabulary is the collection of tokens recognized by a tokenizer, usually associated with a mapping from each token to an integer ID.

For example:

| Token   | Token ID |
| ------- | -------- |
| `<PAD>` | 0        |
| `<UNK>` | 1        |
| `cat`   | 2        |
| `dog`   | 3        |
| `runs`  | 4        |

The sentence `"cat runs"` becomes:

\\[ [2,4] \\]

These IDs are identifiers, not numerical measurements of semantic meaning. Token ID 4 is not inherently more meaningful than token ID 2.

### What about unknown words?

With a fixed word-level vocabulary, an unseen word may be mapped to an unknown token, often represented by `<UNK>`. Modern subword tokenizers can instead represent many unseen words using smaller known pieces.

Why it matters: The vocabulary determines how text is converted into the discrete inputs consumed by a neural NLP model.

## 6. Embedding lookup

Once tokens have been converted to integer IDs, the model needs useful numerical representations of them.

An embedding maps each token ID to a dense vector of real numbers.



Suppose the vocabulary contains \\(V\\) tokens and each embedding has dimension \\(D\\). The embedding matrix is:

\\[ E\in\mathbb{R}^{V\times D} \\]

For a token with ID \\(i\\), the embedding lookup retrieves row \\(i\\):

\\[ \mathbf{e}\_i=E[i] \\]

For example:

`"cat runs"` → `[2, 4]` → two embedding vectors.

These vectors can be learned during model training or initialized from pretrained embeddings.

Unlike token IDs, embeddings can encode useful learned relationships between tokens. In contextual Transformer models, the representation of a token is further updated using surrounding context.

## What you should remember 🎯

1. Token sequences preserve the order of language units.
2. Text samples have variable lengths.
3. Padding makes sequence lengths compatible within a batch; truncation limits long sequences.
4. Attention masks identify positions that should be excluded from attention.
5. Vocabulary IDs are lookup indices, not meaningful numerical values.
6. Embedding lookup converts discrete token IDs into dense vectors.
