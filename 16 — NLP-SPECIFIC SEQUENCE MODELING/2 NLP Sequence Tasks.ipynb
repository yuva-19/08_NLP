# Module 16.2 — NLP Sequence Tasks 🧠


## 1. The three sequence-task patterns



| Pattern                         | Input              | Output              | Examples                   |
| ------------------------------- | ------------------ | ------------------- | -------------------------- |
| Many-to-One                     | Sequence of tokens | One prediction      | Sentiment classification   |
| Many-to-Many                    | Sequence of tokens | One label per token | POS tagging, NER           |
| Many-to-Many, different lengths | Source sequence    | Target sequence     | Translation, summarization |

These patterns describe the relationship between the input and output sequences. They are not restricted to a single neural architecture.

## 2. Many-to-One — Sentiment classification



Goal: Read an entire text sequence and produce one prediction.

Example:

> "The movie was surprisingly good."

Input tokens:

`The → movie → was → surprisingly → good`

Output:

`Positive`

Conceptually, the model builds a representation of the input sequence and uses it to predict a class.

\\[ (x_1,x_2,\ldots,x_T)\rightarrow \hat{y} \\]

Here, \\(T\\) is the number of input tokens and \\(\hat{y}\\) is the predicted label.

Industry applications

- Customer review sentiment analysis
- Spam detection
- Intent classification in chatbots
- Customer support ticket categorization

Key point: One prediction represents the entire input, even though the input contains multiple tokens.

## 3. Many-to-Many — POS tagging and NER



Goal: Produce a label for each token in an input sequence.

### A. Part-of-Speech (POS) tagging

POS tagging identifies the grammatical role of each word.

Example:

| Token   | POS label  |
| ------- | ---------- |
| The     | Determiner |
| cat     | Noun       |
| runs    | Verb       |
| quickly | Adverb     |

The output sequence has one label per input token.

### B. Named Entity Recognition (NER)

NER identifies entities such as people, organizations, locations, and dates.

Example:

> "Sundar Pichai works at Google."

| Token  | Entity label |
| ------ | ------------ |
| Sundar | Person       |
| Pichai | Person       |
| works  | Other        |
| at     | Other        |
| Google | Organization |

In practical NER systems, labels often use a scheme such as BIO to distinguish the beginning and continuation of an entity.

Mathematically, the task can be represented as:

\\[ (x_1,x_2,\ldots,x_T) \rightarrow (\hat{y}\_1,\hat{y}\_2,\ldots,\hat{y}\_T) \\]

Industry applications

- Extracting names and organizations from documents
- Identifying medical entities in clinical text
- Extracting dates and amounts from invoices
- Analyzing grammatical structure

Key point: The model makes token-level predictions, rather than assigning one label to the whole document.

## 4. Many-to-Many with different lengths — Translation and summarization



Goal: Transform an input sequence into a new output sequence, whose length may differ from the input.

### A. Machine translation

Input:

> "How are you?"

Output:

> "¿Cómo estás?"

The source and target sentences have different tokens and potentially different lengths. The model generates the target sequence based on the source.

### B. Text summarization

Input:

> A long article describing a company's quarterly financial results, challenges, and future plans.

Output:

> A short paragraph containing the article's most important information.

Summarization compresses the source into a shorter text in many cases, although the generated output is not guaranteed to be shorter.

A sequence-to-sequence model represents this transformation as:

\\[ (x_1,\ldots,x_T)\rightarrow(\hat{y}\_1,\ldots,\hat{y}\_U) \\]

Here, \\(T\\) is the source length and \\(U\\) is the generated output length. They do not have to be equal.

Industry applications

- Multilingual translation
- Document summarization
- Conversational response generation
- Text rewriting

Key point: Unlike token labeling, the model generates a target sequence that need not align token-by-token with the input. Encoder-decoder architectures and attention are important approaches for these tasks.

## 5. Compare the patterns

| Question                        | Many-to-One     | Token-level Many-to-Many               | Seq2Seq         |
| ------------------------------- | --------------- | -------------------------------------- | --------------- |
| What does the output represent? | Whole text      | Each input token                       | Generated text  |
| Is token alignment required?    | No              | Yes                                    | No              |
| Can output length differ?       | One output      | Usually aligned to input               | Yes             |
| Example architecture            | LSTM classifier | BiLSTM or Transformer token classifier | Encoder-decoder |
| Typical task                    | Sentiment       | NER                                    | Translation     |

One important nuance: these are task patterns, not strict architecture rules. Transformers can perform all three types of tasks using different prediction heads and decoding strategies.

## What you should remember 🎯

1. Many-to-One maps an input sequence to one label.
2. POS tagging and NER assign labels to individual tokens.
3. Translation and summarization generate a target sequence.
4. Seq2Seq output length can differ from input length.
5. The task determines the output structure; the architecture is chosen to suit it.
