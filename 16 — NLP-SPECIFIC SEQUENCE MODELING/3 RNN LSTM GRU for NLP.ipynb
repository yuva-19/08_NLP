# Module 16.3 — RNN, LSTM & GRU for NLP 🧠

Bro, you've already learned the mathematics and implementation of RNNs, LSTMs, and GRUs in Deep Learning. Here, we'll focus on how to apply these architectures to language tasks, what each task expects as output, and which design choices matter in practice.

## 1. Where do RNNs, LSTMs and GRUs fit in NLP?


All three can process a sequence of token embeddings and maintain a hidden state that carries information from earlier tokens.

- RNN: A basic sequential baseline.
- LSTM: Uses gates to help preserve information across longer sequences.
- GRU: Uses a simpler gating mechanism than LSTM.

You already know how these mechanisms work. The important NLP question is: What should the model predict from the sequence?

| Task                    | Input                | Output                    |
| ----------------------- | -------------------- | ------------------------- |
| Sentiment analysis      | Review               | Positive / Negative       |
| Language modeling       | Previous tokens      | Next-token probabilities  |
| Sequence classification | Sentence or document | One class                 |
| Sequence labeling       | Token sequence       | One label per token       |
| Seq2Seq                 | Source sequence      | Generated target sequence |

## 2. Application 1 — Sentiment analysis


Goal: Predict the sentiment of an entire review.

Example:

> "The acting was excellent, but the story was boring."

The model processes the token sequence and predicts a sentiment label.

\\[ (x_1,x_2,\ldots,x_T)\rightarrow \hat{y} \\]

### NLP-specific design choices

- Input: Token IDs converted into embeddings.
- Architecture: An RNN, LSTM, GRU, or bidirectional variant.
- Output: A classification head produces class scores.
- Challenge: Negation, mixed sentiment, sarcasm, and long reviews.

For example, the phrase "not good" requires the model to account for how not changes the meaning of good.

Industry use: Product reviews, customer feedback, app-store ratings, and social-media analysis.

## 3. Application 2 — Language modeling


Goal: Predict the next token using the preceding tokens.

Given:

> "Natural language processing is ..."

A language model might assign probabilities to `important`, `useful`, or `complex`.

\\[ P(x_t\mid x_1,\ldots,x\_{t-1}) \\]

The model is trained to predict the next token at successive positions in a sequence.

### What matters in practice?

- Training data: Text sequences, often converted into input-target pairs by shifting tokens.
- Output: A probability distribution over the vocabulary.
- Evaluation: Perplexity is a common language-modeling metric.
- Challenge: Maintaining useful context as the sequence grows.

RNN-based language models are historically important and help explain the foundations of neural text generation. Modern large language models primarily use Transformer architectures.

## 4. Application 3 — Sequence classification

Goal: Assign one label to an entire sequence.

Sequence classification is a broader task category; sentiment analysis is one example of it.

Other examples include:

- Spam versus legitimate email
- News-topic classification
- Support-ticket routing
- Intent detection in chatbots

An RNN, LSTM, or GRU can process the tokens and use a sequence representation to predict a class.

Important design choice: Decide how to obtain the sequence representation. Common approaches include the final valid hidden state or pooling over token representations. For bidirectional models, information from both directions can be combined.

The final hidden state must be handled carefully when batches contain padded sequences.

## 5. Application 4 — Sequence labeling


Goal: Predict a label for every token.

Example:

> "Sundar Pichai joined Google."

| Token  | Predicted label |
| ------ | --------------- |
| Sundar | B-PER           |
| Pichai | I-PER           |
| joined | O               |
| Google | B-ORG           |

Here, BIO tags mark the beginning and continuation of named entities, while `O` indicates a token outside an entity.

### Why use a bidirectional LSTM?

The meaning of a token can depend on both preceding and following words. A bidirectional LSTM can incorporate information from both directions when the entire input sequence is available.

Applications: Named Entity Recognition (NER), Part-of-Speech (POS) tagging, and token-level information extraction.

## 6. Application 5 — Seq2Seq


Goal: Transform an input sequence into a generated output sequence.

Examples:

- English → Tamil translation
- Long text → Short summary
- Input message → Conversational response

A traditional recurrent Seq2Seq model uses an encoder to process the source and a decoder to generate the target.

You already know the encoder-decoder architecture and teacher forcing, so focus on the NLP-specific issues:

- Source and target vocabularies or tokenization
- Start-of-sequence and end-of-sequence tokens
- Padding and masking
- Exposure bias during generation
- Information bottlenecks in vanilla Seq2Seq
- Why attention helps the decoder access relevant encoder states

For longer inputs, a single fixed-size context vector can become a bottleneck. Attention addresses this limitation, which leads directly into your upcoming attention modules.

## 7. How to choose the architecture?

| Requirement                                | Reasonable starting point                                   |
| ------------------------------------------ | ----------------------------------------------------------- |
| Simple sequence classification             | GRU or LSTM                                                 |
| Classification using both sides of context | BiLSTM                                                      |
| Next-token prediction                      | RNN, LSTM, or GRU language model                            |
| Token-level labeling                       | BiLSTM or Transformer token classifier                      |
| Translation or summarization               | Encoder-decoder with attention, or a pretrained Transformer |

These are starting points, not universal rules. Dataset size, sequence length, latency, accuracy requirements, and available pretrained models influence the final choice.

## What you should remember 🎯

1. RNNs, LSTMs, and GRUs can support multiple NLP tasks; the output structure determines the task formulation.
2. Sentiment analysis and sequence classification produce one prediction per input.
3. Language modeling predicts the next token.
4. Sequence labeling predicts a label for each token.
5. Seq2Seq generates a target sequence that may differ in length from the source.
6. Attention helps overcome the fixed-context bottleneck of vanilla recurrent Seq2Seq.
