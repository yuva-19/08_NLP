````python
# NLTK 

import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
````

````python
text = "The movies were amazing! The actors were running quickly."

# Sentence tokenization
sentences = sent_tokenize(text)

# Word tokenization
tokens = word_tokenize(text)

# Stopword removal
stop_words = set(stopwords.words("english"))
filtered_tokens = [
    word for word in tokens
    if word.lower() not in stop_words
]

# Stemming
stemmer = PorterStemmer()
stemmed = [stemmer.stem(word) for word in filtered_tokens]

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized = [
    lemmatizer.lemmatize(word.lower())
    for word in filtered_tokens
]

print("Sentences:", sentences)
print("Tokens:", tokens)
print("Without stopwords:", filtered_tokens)
print("Stemmed:", stemmed)
print("Lemmatized:", lemmatized)
````

````python
print(lemmatizer.lemmatize("running"))
print(lemmatizer.lemmatize("running", pos="v"))
````

# Module 4.3 — NLP Libraries: `re` vs NLTK vs spaCy

## 1. Overview

NLP libraries provide tools to clean, process, analyze, and extract meaningful information from human language.

- **Python `re`:** Pattern matching and text manipulation using regular expressions.
- **NLTK:** A collection of NLP tools for learning, experimentation, and linguistic processing.
- **spaCy:** An efficient NLP library designed for integrated processing pipelines and practical applications.

## 2. Comparison Table

| Feature | Python `re` | NLTK | spaCy |
|---|---|---|---|
| Main purpose | Pattern matching and text cleaning | General NLP tasks and experimentation | Efficient, integrated NLP pipelines |
| Tokenization | Must implement or split using patterns | `word_tokenize()` | `nlp(text)` |
| Sentence splitting | Pattern-based, limited | `sent_tokenize()` | `doc.sents` |
| Stopword removal | Manual implementation | Stopword corpus | `token.is_stop` |
| Stemming | Not built in | `PorterStemmer()` | No standard built-in stemming pipeline |
| Lemmatization | Not built in | `WordNetLemmatizer()` | `token.lemma_` |
| POS tagging | Not built in | Available through taggers | Integrated model pipeline |
| Named Entity Recognition (NER) | Must build patterns manually | Available through NLP tools | Integrated model pipeline |
| Installation | Included with Python | `pip install nltk` | `pip install spacy`, plus a model for typical NLP pipelines |
| Processing style | Rule and pattern based | Individual tools | Integrated pipeline |
| Best suited for | Cleaning and extracting patterns | Learning and specialized NLP experiments | Production-oriented NLP applications |

## 3. Python `re` — Regular Expressions

### Purpose
Find, match, extract, or replace text patterns using rules.

### Common applications
- Removing URLs and unwanted patterns.
- Extracting email addresses, phone numbers, and dates.
- Finding numbers, IDs, and specific text formats.
- Cleaning repeated spaces or punctuation.

### Example

```python
import re

text = "Great movie!!! Visit https://example.com"

cleaned = re.sub(r"https?://\S+", "", text)
cleaned = re.sub(r"!+", "!", cleaned)

print(cleaned)
```

### Merits
- Built into Python; no installation required.
- Fast and effective for well-defined text patterns.
- Precise control over matching and replacement.
- Useful for preprocessing before applying NLP models.

### Demerits
- Does not understand meaning or context.
- Complex expressions can be difficult to maintain.
- Cannot independently perform linguistic tasks such as lemmatization, POS tagging, or semantic analysis.
- Pattern-based sentence splitting can fail on abbreviations and ambiguous punctuation.

**Use `re` when:** You need to find, extract, or replace a known text pattern.

---

## 4. NLTK — Natural Language Toolkit

### Purpose
Provide individual tools and linguistic resources for NLP learning, research, and experimentation.

### Common applications
- Tokenization and sentence segmentation.
- Stopword removal.
- Stemming and lemmatization.
- POS tagging and text classification.
- Exploring corpora and linguistic data.

### Example

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = "The workers are running quickly."
tokens = word_tokenize(text)

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print("Tokens:", tokens)
print("Stems:", [stemmer.stem(word) for word in tokens])
print("Lemmas:", [lemmatizer.lemmatize(word, pos="v") for word in tokens])
```

### Merits
- Beginner-friendly and useful for understanding individual NLP operations.
- Provides many algorithms, corpora, and linguistic resources.
- Allows flexible experimentation with different techniques.
- Useful for academic learning and NLP demonstrations.

### Demerits
- Some operations require separate tools, resources, or configuration.
- Downloadable datasets and models may require additional setup.
- A complete NLP pipeline often requires more manual integration.
- Some workflows can be slower than optimized alternatives, depending on the task.

**Use NLTK when:** You want to learn, experiment with, or customize individual NLP techniques.

---

## 5. spaCy

### Purpose
Provide efficient, integrated NLP pipelines for processing text and extracting linguistic information.

### Common applications
- Tokenization and sentence segmentation.
- Lemmatization and POS tagging.
- Named Entity Recognition (NER).
- Dependency parsing.
- Building information-extraction systems and NLP applications.

### Example

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Apple opened a new office in Chennai."
doc = nlp(text)

print("Tokens:", [token.text for token in doc])
print("Lemmas:", [token.lemma_ for token in doc])

print("Entities:")
for entity in doc.ents:
    print(entity.text, entity.label_)
```

### Merits
- Integrates multiple NLP operations into one pipeline.
- Generally optimized for efficient processing.
- Provides convenient access to linguistic features.
- Supports practical applications such as entity extraction and document analysis.
- Offers pretrained language models for several languages and tasks.

### Demerits
- Typical linguistic pipelines require a separately installed language model.
- Model downloads and environment compatibility can cause setup issues.
- Less focused than NLTK on teaching individual linguistic algorithms and exploring corpora.
- Pretrained models may make incorrect predictions, particularly on unfamiliar domains.
- Model size and computational requirements vary by pipeline.

**Use spaCy when:** You want to build efficient NLP pipelines and applications using integrated linguistic analysis.

---

## 6. Merits and Demerits — Quick Summary

| Library | Biggest merit | Biggest limitation |
|---|---|---|
| `re` | Precise pattern-based text manipulation | No built-in linguistic understanding |
| NLTK | Flexible tools for learning and experimentation | More manual pipeline assembly |
| spaCy | Integrated and efficient NLP processing | Requires appropriate models for many tasks |

## 7. Can We Use All Three Together?

**Yes!** They complement one another rather than compete directly.

A practical NLP workflow might use:

1. **`re`** to remove unwanted URL patterns or extract structured information.
2. **spaCy** to tokenize text, identify grammatical features, and extract named entities.
3. **NLTK** when a specific algorithm, corpus, or experimental technique is needed.

You do not need to use all three in every project. Choose tools according to the task.

## 8. Which Should You Choose?

| Requirement | Recommended starting point |
|---|---|
| Remove URLs or extract dates | `re` |
| Learn stemming and compare algorithms | NLTK |
| Perform POS tagging and NER | spaCy |
| Build an integrated text-processing pipeline | spaCy |
| Experiment with different linguistic resources | NLTK |
| Combine pattern matching with NLP | `re` + spaCy or NLTK |

## 9. Important Practical Considerations

- **More preprocessing is not always better.** Preserve information that matters for the task.
- Do not remove negation words such as *not* without checking their effect.
- Stemming and lemmatization are different: stemming applies word-reduction rules, while lemmatization aims to return a valid base form.
- spaCy's `token.lemma_` depends on the loaded model and its predictions.
- Regex is useful for structured patterns, but it should not replace linguistic analysis when context matters.

## Conclusion

`re` handles patterns, NLTK provides flexible individual NLP tools, and spaCy provides integrated NLP pipelines. In practical projects, they can work together to build a suitable preprocessing and analysis workflow.
