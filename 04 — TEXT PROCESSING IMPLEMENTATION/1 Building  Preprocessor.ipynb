```python
# creating the sample text

raw_text = """
    <h1>Hello WORLD!!!</h1>

    NLP is AMAZING 😍😍.
    Visit https://example.com for more information.

    I am learning NLP...     and   Machine Learning!

    Email me at test@example.com.
"""
```

```python
import unicodedata

def normalize_unicode(text):
    return unicodedata.normalize('NFC',text)

# purpose of unicodedata.normalize('NFC', text) is Converts Unicode text into a standard, consistent format.
# e\u0301 ---> é
```

```python
text = normalize_unicode(raw_text)
print(text)
```

```python
# To Remove HTML

from bs4 import BeautifulSoup

def remove_html(text):
    return BeautifulSoup(text,'html.parser').get_text(" ")
```

```python
text = remove_html(raw_text)
print(text)
```

```python
# To Remove the URL's 

import re 

def remove_urls(text):
    pattern = r"https?://\S+|www\.\S+"
    return re.sub(pattern,"",text)
```

```python
text = remove_urls(text)
print(text)
```

```python
# Whitespace Normalization
def normalize_whitespace(text):
    return re.sub(r"\s+"," ",text).strip()
```

```python
text = normalize_whitespace(text)

print(text)
```

```python
# Case Normalization

def normalize_case(text):
    return text.lower()
```

```python
text = normalize_case(text)
print(text)
```

```python
# Sentence Tokenization
# instead of "hello world. nlp is amazing. i am learning nlp."
# its better to have 
#  [
#     "hello world.",
#     "nlp is amazing.",
#     "i am learning nlp."
# ]

from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(text)
print(sentences)
```

```python
# Word Tokenization
# now we will take each sentence and split into tokens 
# instead of     visit for more information.
# ['visit','for','more','information','.']

from nltk.tokenize import word_tokenize

tokens = []

for sentence in sentences:
    tokens.extend(word_tokenize(sentence))

print(tokens)
```

```python
# Stopword Handling

# Now we have tokens such as:
# ['this', 'is', 'a', 'very', 'good', 'movie']

# Depending on the task, words like:
# is
# a
# the
# and
# of

# may not provide much useful information.
# We'll create a function to remove them.

from nltk.corpus import stopwords 

stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return[
        token
        for token in tokens
        if token.lower() not in stop_words
    ]
```

```python
tokens = remove_stopwords(tokens)
print(tokens)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

def stem_tokens(tokens):
    return [
        stemmer.stem(token)
        for token in tokens
    ]
```

```python
stem_final = stem_tokens(tokens)
print(stem_final)
```

```python
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [
        lemmatizer.lemmatize(token)
        for token in tokens
    ]
```

```python
lemm_final = lemmatize_tokens(tokens)
print(lemm_final)
```

```python
print(tokens)
print(stem_final)
print(lemm_final)
```

```python
test_words = ["running", "ran", "studies", "better", "meeting"]

for word in test_words:
    print(
        word,
        "→",
        lemmatizer.lemmatize(word, pos="v")
    )
```

```python
tests = [
    ("running", "v"), # v is verb 
    ("ran", "v"),
    ("studies", "v"),
    ("better", "a"), # a is adjective thats why its changed from better to good
    ("meeting", "v")
]

for word, pos in tests:
    print(word, "→", lemmatizer.lemmatize(word, pos=pos))
```
