
# Module 4 — Emotion-Relevant Feature Groups

## 1. First, understand feature groups

When a person speaks, their emotional expression influences different characteristics of their voice.

For example, when someone becomes angry:

- Their pitch may change.
- Their vocal energy may increase.
- Their speaking rate may change.
- Their voice may become tense or rough.

Instead of relying on one feature, we extract multiple features representing different aspects of speech.

We can organize these into four groups:

1. Pitch and prosody
2. Energy
3. Spectral features
4. Voice quality

## A. Pitch and Prosody

**Must Know**

This group describes how the speaker uses pitch, timing and rhythm while speaking.

### 1. Mean F0

Mean F0 represents the average fundamental frequency of voiced speech.

$$
F0_{\text{mean}}=\frac{1}{N}\sum_{i=1}^{N}F0_i
$$

**Example:**

Suppose a speaker has these F0 estimates:

$$
[120,130,140,150,160]\text{ Hz}
$$

$$
F0_{\text{mean}}=140\text{ Hz}
$$

It gives us an idea of the speaker's average pitch during that recording.

**SER relevance:** Can help capture differences in vocal pitch, but natural speaker differences must be considered.

### 2. Minimum and Maximum F0

- Minimum F0: lowest detected voiced pitch.
- Maximum F0: highest detected voiced pitch.

For example:

$$
F0=[120,130,140,150,160]
$$

$$
F0_{\min}=120\text{ Hz}
$$

$$
F0_{\max}=160\text{ Hz}
$$

These measurements help describe the pitch boundaries.

### 3. F0 Standard Deviation

Standard deviation measures how much pitch varies around its mean.

$$
\sigma_{F0}=
\sqrt{
\frac{1}{N}
\sum_{i=1}^{N}
(F0_i-\mu_{F0})^2
}
$$

- Lower standard deviation: relatively stable pitch.
- Higher standard deviation: greater pitch variation.

For example, compare:

**Speaker A — Relatively stable pitch**

Relatively small pitch variation.

**Speaker B — More variable pitch**

Larger pitch variation.

This is useful because emotion can influence how dynamically a person uses their voice.

But high pitch variation does not automatically mean happiness. It can also occur in anger, fear or surprise.

### 4. Pitch Range

$$
F0_{\text{range}}=F0_{\max}-F0_{\min}
$$

It measures the difference between the highest and lowest detected F0.

Be careful: extreme pitch-tracking errors can make the range misleading.

### 5. Pitch Contour

A pitch contour is the variation of F0 over time.

Unlike mean F0, which gives one number, a pitch contour preserves the temporal behaviour of pitch.

![Pitch Contour](https://images.openai.com/static-rsc-4/hXgVLXidowx67whTDhFqJeA6DkoY8-bGWOuFcP0H5PqpCG7vRoHw64NGY-R0xXPGhovrnzNg0Ze34oaaFDER51TtaCXpxELzc9yvayRN2DPVCjKZ96-975Z3lRsmDNo7H5wJkbFiuMflMNXZF1D6fN75hDMzFf9iTiSEzxNCM2w?purpose=inline)

[Source: K. J. Kohler — Timing and communicative functions of pitch contours](https://www.ipds.uni-kiel.de/kjk/pub_exx/kk2005_2/kk_05.html?utm_source=chatgpt.com)

**Why is this important?**

Imagine two recordings:

- Recording A: pitch stays relatively stable.
- Recording B: pitch rises and falls dramatically.

They might have the same mean F0 but very different emotional expression.

This is why preserving frame-level features can be valuable.

### 6. Speaking Rate and Pauses

These describe the temporal behaviour of speech.

**Speaking rate:** How quickly someone speaks.

It can be measured in words per minute or syllables per second.

**Pause duration:** How long the speaker remains silent between speech segments.

Useful measurements include:

- Average pause duration.
- Maximum pause duration.
- Number of pauses.
- Silence-to-speech ratio.

For SER, pauses may help identify hesitation, emotional interruption or subdued speech.

However, pauses are also influenced by language, sentence structure and speaking habits.

## B. Energy Features

**Must Know**

Energy features describe the strength and variation of the audio signal.

### 1. RMS Mean

You already learned RMS energy.

For a frame containing $N$ samples:

$$
RMS=\sqrt{\frac{1}{N}\sum_{n=1}^{N}x[n]^2}
$$

RMS mean calculates the average energy across all frames.

$$
RMS_{\text{mean}}=
\frac{1}{T}\sum_{t=1}^{T}RMS_t
$$

**SER relevance:**

- May capture increased vocal intensity.
- May help distinguish forceful speech from subdued speech.
- Can capture differences in vocal expression.

### 2. RMS Standard Deviation

$$
RMS_{\text{std}}=
\sqrt{
\frac{1}{T}
\sum_{t=1}^{T}
(RMS_t-\mu_{RMS})^2
}
$$

It measures how much energy varies throughout the recording.

For example:

- Stable energy: relatively consistent loudness.
- Variable energy: changes in vocal intensity.

### 3. Energy Range

$$
RMS_{\text{range}}=RMS_{\max}-RMS_{\min}
$$

This measures the difference between maximum and minimum frame-level RMS.

Like pitch range, it is sensitive to outliers and silence.

### 4. Energy Contour

Energy contour represents the change in signal energy over time.

![Energy Contour](https://images.openai.com/static-rsc-4/RLsA-qbg2_V_pA6Yh32-T5kxjY6jSO39uV1hgWn4k0SIaH_qGd1u8gqTQhFxBTfPtECKOaJ5knQPHAmd5GskqmoB5PKVGvXUAXiEU_jBJXlpNluiGiIsMdoAE2c99ciFmdFHq5UNr4Nng9mYBvUT6J6hfQDEY0jTq1cPfSykOEQ?purpose=inline)

[Source: DisVoice Phonation Documentation](https://disvoice.readthedocs.io/en/latest/Phonation.html?utm_source=chatgpt.com)

For example:

A speaker might begin quietly and suddenly shout.

The RMS contour captures this change.

### 5. Short-Time Energy Statistics

Instead of calculating energy for the entire recording, we calculate it for individual frames.

Then we calculate:

- Mean.
- Standard deviation.
- Minimum.
- Maximum.
- Range.
- Percentiles.

This is the same frame-level to utterance-level aggregation approach you already implemented in Module 35.2.

## C. Spectral Features

**Must Know**

Spectral features describe the frequency characteristics of speech.

You've already studied these in detail, so let's focus on their role in emotion recognition.

### 1. MFCC

![MFCC](https://images.openai.com/static-rsc-4/K8e2AjvopGfEuNyoop0xQ1Vaf_4qzRPJZSsUZMKmOFaizQAXTW-d4-wr8djUM2xkCsFpVSecoL6BIwQm7dFwHqRFouuJtdUCmgBfS2Jfq_z0BiJxcnY33hqQEdNV6uSQEzo3xc5e5SfE0cvaPAHs4tyZ2OsFC7ypdtjkAfv7eoA?purpose=inline)

[Source: MDPI](https://www.mdpi.com/2076-3417/14/17/7604?utm_source=chatgpt.com)

MFCCs represent the broad spectral characteristics of speech using a compact cepstral representation.

They capture information related to:

- Vocal timbre.
- Spectral envelope.
- Changes in vocal articulation and sound characteristics.

For SER, MFCCs are particularly useful because emotional expression can change the spectral characteristics of the voice.

### 2. Delta MFCC

Delta MFCC represents the temporal change in MFCC coefficients.

In simple terms:

- MFCC: What are the spectral characteristics at this moment?
- Delta MFCC: How are those characteristics changing?

### 3. Spectral Centroid

Represents the centre of spectral energy.

A higher centroid can indicate a brighter or more high-frequency-weighted sound.

### 4. Spectral Bandwidth

Measures how widely spectral energy is distributed around the centroid.

### 5. Spectral Rolloff

Represents the frequency below which a specified percentage of spectral energy is concentrated.

For example, at 85% rolloff, 85% of the selected spectral energy lies below that frequency.

### 6. Spectral Contrast

Measures the difference between spectral peaks and valleys across frequency bands.

It provides information about the spectral structure of a sound.

### 7. Spectral Flux

Measures how much the spectrum changes between consecutive frames.

It can help describe rapid spectral changes during speech.

**Important for your project:** Don't assume every spectral feature will improve classification. MFCCs should be your primary spectral baseline; the other spectral features are candidates for experimentation.


## D. Voice Quality Features

**Good to Know — Advanced**

These features attempt to describe how the voice is physically produced and how it sounds beyond its basic pitch and energy.

### 1. Jitter

Jitter measures variations in the duration of consecutive vocal-fold vibration cycles.

Imagine an ideal periodic signal:

$$
T_1=T_2=T_3=T_4
$$

A real voice may have:

$$
T_1=8.0,\quad T_2=8.2,\quad
T_3=7.9,\quad T_4=8.3\text{ ms}
$$

The cycle durations are not perfectly identical.

Jitter measures this cycle-to-cycle frequency-period instability.

**SER relevance:** Some emotional states can influence vocal tension and phonation, which may affect jitter.

### 2. Shimmer

Shimmer measures variations in the amplitude of consecutive vocal-fold vibration cycles.

For example:

$$
A_1=0.8,\quad A_2=0.9,\quad A_3=0.75
$$

It describes cycle-to-cycle amplitude variation.

### 3. HNR — Harmonics-to-Noise Ratio

HNR estimates the ratio between periodic harmonic energy and noise-like energy in the voice.

$$
HNR=10\log_{10}
\left(
\frac{P_{\text{harmonic}}}{P_{\text{noise}}}
\right)
$$

- Higher HNR: more harmonic energy relative to noise.
- Lower HNR: more noise-like energy relative to harmonic energy.

It can help characterize breathiness or roughness, although these voice qualities are more complex than HNR alone.

### 4. Formants

Formants are resonant frequency regions created by the vocal tract.

Commonly studied formants:

- F1
- F2
- F3

![Formants](https://images.openai.com/static-rsc-4/FdCHMakJ__xkSFw_o-E6e51tcoYcl-HJpTPfSAkqcW46wsoNCl9mDq2YSUbNlEkUpQxXHM5Jf2olmLOKnYV--5Q-85Q44G5yDLxHaZrv7dMhuDbG2ePVSjdflrrBrSZ-lm4k78YKUtI7iXoLs7VO2XfpVc0yq7LtqfxetXHgDeo?purpose=inline)

[Source: ResearchGate](https://www.researchgate.net/figure/Spectrogram-of-the-vowels-i-and-u-showing-the-first-three-formants-F1-F2-and_fig1_243126231?utm_source=chatgpt.com)

Remember the distinction:

| F0 | Formants |
|---|---|
| Fundamental frequency | Vocal-tract resonances |
| Related to vocal-fold vibration rate | Related to vocal-tract shape |
| Closely associated with perceived pitch | Important for vowel identity and vocal characteristics |

**SER relevance:** Formants may provide additional vocal-tract and articulation information, but they are also strongly influenced by phonetic content and speaker anatomy.

### Why are these features more difficult?

Jitter, shimmer and HNR are most meaningful when measured on suitable voiced, sufficiently periodic speech.

They can become unreliable when:

- The recording contains background noise.
- Speech is unvoiced.
- The speaker is shouting or crying.
- Pitch tracking is inaccurate.
- The recording is heavily compressed.

So don't spend too much time implementing them before establishing a reliable baseline.

## 3. How to combine these features in your SER project

Think of each feature group as providing a different perspective of the same recording.

**Input audio recording**

16 kHz waveform

**Pitch / Prosody**

F0, pauses, speaking rate

**Energy**

RMS, energy variation

**Spectral**

MFCC, centroid, flux

**Voice quality**

Jitter, shimmer, HNR

**Feature fusion**

Combine useful information from feature groups

**SER classifier**

Predict emotion category

There are two important ways to combine these features:

### Approach 1 — Feature-level fusion

Extract numerical features from each group, aggregate them where appropriate, and concatenate them into one feature vector.

For example:

$$
X=[MFCC,\Delta MFCC,F0,RMS,\text{Spectral}]
$$

This is useful for classical machine-learning models such as:

- SVM
- Random Forest
- XGBoost
- Logistic Regression

### Approach 2 — Representation-level fusion

Keep temporal information, such as MFCC sequences and pitch/energy contours, and combine their learned representations inside a neural network.

This is useful for CNN, LSTM, GRU and attention-based architectures.

We'll study feature fusion properly in its dedicated module rather than implementing it prematurely.

## 4. What you should remember

1. Pitch and prosody describe how the speaker uses pitch and timing.
2. Energy describes vocal intensity and its variation.
3. Spectral features describe the frequency structure of speech.
4. Voice-quality features describe characteristics such as periodicity, breathiness and vocal roughness.
5. MFCC, pitch and energy are your first priorities.
6. Preserve frame-level features when you want a model to learn temporal patterns.
7. Don't assume adding more features automatically improves performance.

**My recommendation:** Build a reliable baseline with MFCC + F0 + RMS first. Then add feature groups one at a time and compare validation performance. This will tell you which features genuinely contribute to emotion recognition.
