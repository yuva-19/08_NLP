
# Module 3 — Emotion Categories and Acoustic Patterns

## 1. First, understand the two dimensions of emotion

Before studying individual emotions, understand these two concepts:

### A. Arousal

Arousal represents the level of psychological activation or intensity of an emotion.

- High arousal: emotionally activated, energetic, or intense.
- Low arousal: emotionally calm, subdued, or less activated.

### B. Valence

Valence represents whether an emotion is generally positive or negative.

- Positive valence: pleasant emotional experience.
- Negative valence: unpleasant emotional experience.

![Valence and Arousal](https://images.openai.com/static-rsc-4/Hhy3X26xeCUTudMgqY_zzB1BfnCw_8bR33re6LpmvY12zTr96DcY4V8BaGVLnsoYl_L8ugeahlsdInYoH3sHvPtkncmLxsKSGuFDVMC-OdHEbvHKX5TwKWwjH_zzlQFMuzDYuEEK6yQKivhlpTLfdMopcAQySnZurrg8Mr42oXs?purpose=inline)

[Source: MDPI](https://www.mdpi.com/1424-8220/23/3/1598?utm_source=chatgpt.com)

For example:

| Emotion | Valence | Arousal |
|---|---|---|
| Happy | Positive | Usually high |
| Sad | Negative | Usually low |
| Angry | Negative | Usually high |
| Fearful | Negative | Usually high |
| Disgust | Negative | Variable |
| Surprised | Variable | Usually high |
| Neutral | Neutral | Usually low to moderate |

**Important:** These are broad tendencies, not universal measurements. Quiet happiness and intense sadness are possible.

Why does this matter?

Because two emotions can have similar arousal levels but different valence.

For example, anger and happiness can both have high energy and high pitch. Their acoustic patterns may overlap even though the emotions are different.

## 2. Acoustic patterns of individual emotions

Let's examine the four important acoustic dimensions:

1. **Pitch (F0):** How high or low the voice sounds.
2. **Energy:** How strong or loud the voice is.
3. **Timing:** Speaking rate, pauses, rhythm and duration.
4. **Voice quality:** How the voice sounds — breathy, rough, tense, strained, etc.

The patterns below are common tendencies, not strict classification rules.

### A. Neutral

![Neutral emotion](https://images.openai.com/static-rsc-4/N5B5Zm8J17qkCBT9BDW4bwCSiEerDl_aEb8wkSBQCV_vUPmqfuvoXaUJe3xVNZ2ZHysbmtIln3nV0DlvShbZpcKpsJhouelxYJIaMYDR03xvQDTMqJr2VS0YA2SFIUWxOqARILX9NckaBDKiQwwPg1eIvg4R9g1GXg3hUnqpnIQ?purpose=inline)

**Neutral emotion**

Emotionally unmarked or relatively ordinary speech.

| Acoustic feature | Typical tendency |
|---|---|
| Pitch | Moderate, relatively stable |
| Energy | Moderate |
| Timing | Regular speaking rate |
| Voice quality | Relatively natural, less expressive variation |

**Why is neutral difficult to classify?**

Neutral speech may not have strong emotional characteristics. It can also overlap with low-arousal sadness, boredom, or calmness.

For example:

- A person speaking normally.
- A person speaking sadly but trying to hide it.
- A person speaking calmly after an emotional event.

All three may have similar acoustic features.

### B. Happy

![Happy emotion](https://images.openai.com/static-rsc-4/9DYvLShbLUWFwEW_1s4B124UMqGyE9O3SH_OvJSY5Z0mBGfOnOg7zirO7t2b-SlxWvs9l56dJrbHTttwoyZQX4DypwSjQkvRdyXspkcWtjb9Nn-CunCfmXu3cc78MmbukE212uJ500u_vTkTxghcBYt021Ew26nb7t_9zXMhCDs?purpose=inline)

**Happy emotion**

Often associated with positive emotional expression.

| Acoustic feature | Typical tendency |
|---|---|
| Pitch | Often elevated, with greater variation |
| Energy | Often increased |
| Timing | Sometimes faster, more dynamic |
| Voice quality | Brighter, more animated |

**Example:**

"That's amazing! I finally got the job!"

The speaker may show:

- Increased pitch variation.
- Stronger vocal energy.
- Faster speech.
- More expressive intonation.

But remember: quiet happiness may have low energy and a relatively stable pitch.

### C. Sad

![Sad emotion](https://images.openai.com/static-rsc-4/U1OeWzKebbL3wuicPBlS4Lq2oRKZ58cvt0YtbpUh6T-sXaxelxoArtEZLc7fkqtwHnb_F83RsKofHvtIunYxN6vYMAYkwafaYBxaJnyf_HjjEuJAWsudf-Abr-fHEJB5cvyDXtQgGEt-TKBgCc4jdOxEannlEI4TZCTq4uMQPVwRxswGli2PFX9O4KO5Kb2f?purpose=inline)

**Sad emotion**

Often associated with low-arousal negative expression.

| Acoustic feature | Typical tendency |
|---|---|
| Pitch | Often lower or with reduced variation |
| Energy | Often reduced |
| Timing | Slower speech, longer pauses |
| Voice quality | Sometimes breathy, weak or subdued |

**Example:**

"I don't think things are going to work out."

Possible acoustic characteristics:

- Lower vocal energy.
- Slower speaking rate.
- Longer pauses.
- Reduced pitch variation.

However, crying or intense sadness can produce irregular pitch and increased vocal energy.

### D. Angry

![Angry emotion](https://images.openai.com/static-rsc-4/f5ZlwktK9BAVHq7k06q7r5B8lEzHQ275xUl1gdJvvaEwYo1gO87rw-70ihGJQuEKL7k2N4yjqcAx7-OlFuoBForKgmlsCeud7qXRpnhuWtPBafy6XTfFsBvdQmKPPxypEZrNcJ4gZbmnBptaLyyEQRSC22RZDxO3HGvyTRWOUzeRCbBfgOs3c-RnuQBygVv8?purpose=inline)

**Angry emotion**

Often associated with high-arousal negative expression.

| Acoustic feature | Typical tendency |
|---|---|
| Pitch | Often elevated, sometimes lower and forceful |
| Energy | Frequently high |
| Timing | Forceful speech, sometimes faster |
| Voice quality | Tense, harsh, pressed or rough |

**Example:**

"How many times do I have to tell you?!"

Possible characteristics:

- High RMS energy.
- Increased pitch variation.
- Strong vocal intensity.
- Tense or harsh voice quality.

Notice something important: anger doesn't always mean high pitch. Some people express anger through a low, controlled, threatening voice.

### E. Fearful

![Fearful emotion](https://images.openai.com/static-rsc-4/cOD4BrkYCnYVfr3eWcaLvgjZHthvWO0K9nNIil5Wwy95vWB_q_CshNPaRjCVJbqDkm8KfsfnkpzPbbL9llpwThSNwHxpVexjI8HE9tH6LTQyM6HOyTsFKdib3hacxmgMCeNq0jMZvShq3Y6P_wzR4Sy-eQcr3vagK_Tk9aDkVK4?purpose=inline)

**Fearful emotion**

Often associated with high-arousal negative expression.

| Acoustic feature | Typical tendency |
|---|---|
| Pitch | Often elevated and unstable |
| Energy | Variable, sometimes high |
| Timing | Hesitation, irregular rhythm |
| Voice quality | Trembling, breathy or tense |

**Example:**

"I... I think somebody is outside."

Possible characteristics:

- Pitch instability.
- Trembling voice.
- Irregular pauses.
- Changes in speaking rate.

Fear and anger can overlap in pitch and energy, but fear may show more hesitation or trembling.

### F. Disgust

![Disgust emotion](https://images.openai.com/static-rsc-4/dHs4kMw0FKXKVwtaooTPcyh6dgA2ZKF01BZi63W_kSlnP4u2W3EeiV3nw0FJLz4TlnK4O7G1tMytTaI_7UHDKEPdug_rOzjFm-hPkzQREV1KA6KPsV3pS5W1dzplHCc60JWmEW8hEIdO3OZFLyo5BHh99Eonbf4KpkGbHS939Vg?purpose=inline)

**Disgust emotion**

Often associated with negative emotional expression.

| Acoustic feature | Typical tendency |
|---|---|
| Pitch | Variable, sometimes lowered |
| Energy | Variable, often moderate |
| Timing | Sometimes slower or emphatic |
| Voice quality | May sound constricted, nasal or tense |

**Example:**

"That's absolutely disgusting."

The speaker might use:

- A lowered or altered pitch.
- A strained or constricted voice.
- Emphasis on particular words.
- Changes in vocal intensity.

Disgust is particularly difficult to identify from acoustic features alone because its vocal expression varies significantly.

### G. Surprised

![Surprised emotion](https://images.openai.com/static-rsc-4/Lm2yQEX10iZDkNprxPybn3IWsibdaHPiwT84vwlQWz9xJRcq_jzcSrR7FB9S9gIS0BAnA0855R_V1icLgZLT7L_Vaxd8to0L-EznK21qtM1LWEgLh2X7Qspx2RtjHTXo_gY49FlC41e2Uyf5ciLf9skX7HSR7S35E6w9KYeztss?purpose=inline)

**Surprised emotion**

Often associated with sudden changes in vocal expression.

| Acoustic feature | Typical tendency |
|---|---|
| Pitch | Sudden pitch elevation or movement |
| Energy | Often increased |
| Timing | Sudden onset, brief pauses |
| Voice quality | More expressive, sometimes breathy |

**Example:**

"Wait... WHAT?!"

Possible characteristics:

- Sudden pitch movement.
- Increased vocal intensity.
- Abrupt changes in speech rhythm.
- Short pauses before or after the expression.

Surprise can be positive or negative. Someone can be surprised by good news or shocking news.



## 3. Comparing all seven emotions

Here is a practical summary for your SER feature-engineering work.

| Emotion | Pitch tendency | Energy tendency | Timing tendency |
|---|---|---|---|
| Neutral | Stable/moderate | Moderate | Regular |
| Happy | Often elevated/variable | Often increased | Dynamic |
| Sad | Often lower/reduced variation | Often reduced | Slower |
| Angry | Variable, often elevated | Often high | Forceful |
| Fearful | Often elevated/unstable | Variable | Irregular |
| Disgust | Variable | Variable | Variable |
| Surprised | Sudden pitch changes | Often increased | Abrupt |

Don't directly translate this table into classification rules. It describes tendencies, not reliable boundaries.

**Angry**

Potentially high pitch, high energy and expressive speech.

Both can have high RMS energy and elevated F0. So those two features alone may not distinguish them.

### Sad vs Neutral

![Sad](https://images.openai.com/static-rsc-4/dO7Uo4QCLqCpqiRryjvjmU4u9rjg8lOCaKOipdPNxr6ao2dvbTdmkO9g6BAba8cjYOBqFj80pYRE8sS7NfNZlTto9Sr4OR9fu9vTt39QzS21ITKuB8dKviqhF3CC4h6KROuevj3fDwaHzBIBQi-fm8lnaxhHepZCnXgWFSpH3oU?purpose=inline)

**Sad**

May have low energy and reduced pitch variation.

![Neutral](https://images.openai.com/static-rsc-4/b8ovmcn6jAtiBngap2T931Jao1iEcFYF01Us6vBfYc6ARGxgIna-eTDZPhisu1jr-2q2tfcBrn7v-tEQbQEnDDVjkN5d7osg9khwE5fQRdpjmq9E6O3WooICy-CF5Z_ymi6va9EO_tEYvJdHtHV-kas3ZamF8QBP7vIXFEL7pdM?purpose=inline)

**Neutral**

May also have low energy and relatively stable pitch.

Their acoustic features may overlap significantly.

This is why SER requires multiple features rather than a single threshold.

For example, a model can combine:

- MFCC: spectral and timbral information.
- Log-Mel spectrogram: time-frequency representation.
- F0: pitch characteristics.
- RMS: energy characteristics.
- ZCR: signal zero-crossing behaviour.
- Prosodic features: pauses, speaking rate and pitch variation.

The model learns patterns and relationships across these features.

## 5. Factors that influence acoustic patterns

An important challenge: the voice does not depend only on emotion.

### A. Speaker differences

Different people have different natural vocal characteristics.

For example:

- One person naturally has a high-pitched voice.
- Another naturally has a low-pitched voice.
- One person speaks loudly.
- Another speaks softly.

Your model must avoid confusing a person's natural voice with their emotional state.

### B. Language and accent

Languages and accents affect:

- Pitch patterns.
- Intonation.
- Speaking rate.
- Pronunciation.
- Rhythm.

A pitch pattern that is common in one language may not carry the same meaning in another.

### C. Acting style

Many SER datasets contain acted emotional speech.

An actor may exaggerate anger, happiness or sadness to make the emotion obvious.

Real-world emotional speech can be much more subtle.

### D. Recording conditions

Microphone distance, background noise, compression and recording equipment can change the extracted acoustic features.

For example, background noise can increase measured energy and interfere with pitch tracking.

**Project implication:** Your model should learn emotional patterns rather than memorizing speakers, accents, recording conditions or acting styles.

## 6. How this connects to the features you've already learned

Let's connect this module directly to your feature extraction pipeline.

| Feature | What it may help capture |
|---|---|
| F0 mean | Average pitch tendency |
| F0 standard deviation | Pitch variation |
| F0 range | Pitch movement |
| RMS mean | Average vocal energy |
| RMS standard deviation | Energy variation |
| MFCC | Vocal spectral characteristics |
| Spectral centroid | Spectral brightness |
| ZCR | Signal characteristics related to noisiness and frequency content |
| Pause duration | Speech timing and hesitation |
| Speaking rate | Temporal characteristics |

Notice how each feature captures a different aspect of speech.

A combination of features gives the model more information than any one feature alone.

## 7. What you should remember

1. Emotions have different acoustic tendencies, but no emotion has a fixed pitch or energy value.
2. Arousal measures emotional activation; valence represents positive or negative emotional direction.
3. Happy and angry speech can both have high energy and elevated pitch.
4. Sad and neutral speech can have similar low-energy characteristics.
5. Speaker, language, accent, acting style and recording conditions influence acoustic features.
6. SER is a pattern-recognition problem, not a collection of fixed acoustic rules.
