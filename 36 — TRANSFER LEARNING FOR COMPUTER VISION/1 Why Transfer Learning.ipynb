## 39.1 Why Transfer Learning?

## 1. What is Transfer Learning?

Transfer learning is a machine learning technique in which a model trained on one task or dataset is reused as the starting point for a different but related task.

Instead of training a deep learning model from scratch, we take an already-trained model and adapt it to our specific problem.

### Example

Imagine you want to build an image classification model to identify defects in electric motor components.

You have only 2,000 labelled images.

Training a large CNN from scratch would require the model to learn everything:

- Basic edges and shapes.
- Textures and patterns.
- Object features.
- Defect-specific characteristics.

Instead, you can use a CNN that has already learned useful visual features from millions of images and adapt it to your defect classification task.

## 2. Why Do We Need Transfer Learning?

There are five important reasons:

1. Limited datasets.
2. Expensive training.
3. Pretrained representations.
4. Feature extraction.
5. Fine-tuning.

Let's understand each one.

### 2.1 Limited Datasets

Deep learning models generally need substantial amounts of data to learn useful patterns.

But collecting and labelling images can be difficult.

- Normal

- Surface defect
- Deformation

Your dataset contains:

Potential problems:

### 2.2 Expensive Training

Training a deep learning model from scratch can be computationally expensive.

Remember that a CNN contains many trainable parameters. During training, the model repeatedly performs:

1. Forward propagation.
2. Loss calculation.
3. Backpropagation.
4. Weight updates.

This process is repeated over multiple epochs.

#### Training from scratch vs. transfer learning

![Transfer learning image classification with PyTorch](https://images.openai.com/static-rsc-4/OL-cV5kc2_jTonb8sdCtJTCRu1iRNcflFB6jXT6a8iieRT11je3fC5o9uxb61ztzRyE0-Gfjrtu5CZ30RN46kSj7Fu23go3ekJrYaMxSjlwFiq4QOULeun8U_y_frkn70B5iWUqLuq05Hx1ZGOU5-BJBY--ieCxi42uP6VqUFv4?purpose=inline)

[Source: Deep View on Transfer Learning with Image Classification — PyTorch](https://purnasaigudikandula.medium.com/deep-view-on-transfer-learning-with-iamge-classification-pytorch-5cf963939575)

| Factor | Training from scratch | Transfer learning |
|---|---|---|
| Initial weights | Randomly initialised | Pretrained weights |
| Feature learning | Learns all features | Reuses learned features |
| Data requirement | Often higher | Can work with smaller datasets |
| Training time | Often longer | Often shorter |
| Computational cost | Can be higher | Often lower |
| Target-task adaptation | Learns everything | Trains a new head and optionally backbone layers |

**Why is transfer learning faster?**

Because the model does not need to learn all visual representations from random initialisation.

For example, if you use a pretrained ResNet18 and train only its new classification head, the backbone's weights remain unchanged.

Only a relatively small number of parameters need to be updated.

The PyTorch transfer learning tutorial demonstrates this using a small ants-and-bees dataset, with approximately 120 training images per class.

### 2.3 Pretrained Representations

This is the most important concept behind transfer learning.

A pretrained model has already learned useful representations from a large source dataset.

#### What is a representation?

A representation is a numerical description of useful information extracted from an input.

In computer vision, these representations are learned by the CNN's layers.

Consider a CNN with several convolutional layers:

![Visualizing different layers of a deep neural network](https://images.openai.com/static-rsc-4/lp8nfLl1t6glsfRI4LFnnW2VIdQ0spcWPeShgGX6P4PbsbeOKtf8W7lOf-Evi6arlMU1kmI_5SD_PrRNA_BLtTbbAi2YUCxAsGMKKBmrBVqiMEuiWfiSkXzkEVc0k1RaLgri5u89YbRptuJa_XXUElxRdQyJStRkwP3Zl5YaQ1E?purpose=inline)

[Source: Visualizing the Layers of DNN — Wikidocs](https://wikidocs.net/138094)

Different layers can learn different levels of visual information.

| CNN depth | Features commonly learned |
|---|---|
| Early layers | Edges, lines, simple colour patterns |
| Middle layers | Textures, curves, shapes, patterns |
| Deeper layers | Object parts and more complex visual structures |

For example, a pretrained CNN might learn:

- Horizontal and vertical edges.
- Circular shapes.
- Surface textures.
- Repeating patterns.
- Complex object structures.

These features can be useful for another image classification task.

#### Example: From ImageNet to motor defect classification

Imagine a model pretrained on ImageNet.

ImageNet contains a large collection of images across 1,000 categories in the commonly used ILSVRC classification task.

The model has learned general visual representations from these images.

Now we want to classify motor components:

```text
Source task:
ImageNet classification
        ↓
Pretrained CNN
        ↓
Learned visual representations
        ↓
Target task:
Motor defect classification
        ↓
Normal / Crack / Surface defect / Deformation
```

The important idea is that the model can reuse useful visual knowledge even though the source and target classes are different.

**However:** A model trained on natural photographs may not automatically recognise industrial defects. It still needs adaptation and target-domain evaluation.

### 2.4 Feature Extraction

Feature extraction is one of the two major ways to use transfer learning.

Here, we take a pretrained CNN and freeze its backbone.

We replace the original classification head with a new classification head for our target task.

![Deep learning for plant species identification](https://images.openai.com/static-rsc-4/82jBENxKfsQbwHLJygEZwdzftF8QmAfwjMQte2EPKVIJBU8yHSSQeUIbvTa9pLikqos5WENOBCQdBkBkJDGi43ClnhkB19SvdypEZJBnMJGlE5I1S5d5hY7pW1ED_l6UvFJEk5Y_qg-bwBn2sm4toqqJpLs4nzjOlJQUHLISXmI?purpose=inline)

[Source: Next Electronics — Deep Learning for Plant Species Identification](https://next.gr/ai/ai-in-education/deep-learning-for-plant-species-identification)

#### How it works

Suppose we use pretrained ResNet18.

Its original classification layer predicts 1,000 ImageNet classes.

We want four motor defect classes.

The architecture becomes:

```text
Input Image
     ↓
Pretrained ResNet18 Backbone
     ↓
Extracted Visual Features
     ↓
New Classification Head
     ↓
4 Output Classes
```

#### What does freezing mean?

Freezing means preventing selected model parameters from being updated during training.

In PyTorch:

```python
for param in model.parameters():
    param.requires_grad = False
```

This freezes all existing model parameters.

Then we replace the classification head:

```python
import torch.nn as nn

model.fc = nn.Linear(
    model.fc.in_features,
    4
)
```

The newly created classification layer has trainable parameters by default.

We train only that layer.

```python
optimizer = torch.optim.Adam(
    model.fc.parameters(),
    lr=0.001
)
```

#### What happens during training?

| Component | Training status |
|---|---|
| Pretrained backbone | Frozen |
| New classification head | Trainable |

The backbone extracts features, and the new head learns how to map those features to the four target classes.

**Advantages:**

- Faster training.
- Fewer trainable parameters.
- Lower GPU memory requirements for gradient computation.
- Useful when the target dataset is small.

**Limitation:** If the target images differ substantially from the source dataset, the frozen features may not be sufficiently specialised.

### 2.5 Fine-Tuning

Fine-tuning is the second major transfer learning approach.

Instead of keeping the entire pretrained backbone frozen, we allow some of its layers to learn from the target dataset.

![Feature extraction vs fine-tuning](https://images.openai.com/static-rsc-4/bWlXwHxnzP9kbTnfWGXQXvKSfvcW4LtLTLIy8vENzqSre9gfCx8UHhNAHPxeaxgGqFBp_48nG5AjGYCyn_fMnVEj7SjIAjIy4C-5E0rQc4FIQMVil_4Vwd0LU7ZamKI2Ps7gNoEk0kI3kHDZAFVv75OUFtPMjl6Nj4sx76SsrVs?purpose=inline)

[Source: Medium — Feature Extraction vs. Fine-Tuning](https://medium.com/@ayasc/transfer-learning-feature-extraction-vs-fine-tuning-3735eff6091e)

#### How it works

Consider the same ResNet18 motor defect classification problem.

Initially:

- Load pretrained weights.
- Replace the original classification head.
- Freeze the backbone.
- Train the new classification head.

After the classification head learns useful target classes, we unfreeze selected deeper backbone layers.

The model can now adapt its learned representations to the new task.

#### Architecture

```text
Input Image
     ↓
Early CNN Layers          → Frozen
     ↓
Middle CNN Layers         → Frozen
     ↓
Deeper CNN Layers         → Trainable
     ↓
New Classification Head   → Trainable
     ↓
4 Output Classes
```

#### Why unfreeze deeper layers?

Early layers often learn general features such as edges and simple textures.

Deeper layers learn more task-specific representations.

Fine-tuning allows the deeper layers to adapt to the target dataset.

For example, the model can become more sensitive to industrial surface patterns and defect characteristics.

#### Feature extraction vs. fine-tuning

| Property | Feature extraction | Fine-tuning |
|---|---|---|
| Backbone | Frozen | Partially or fully trainable |
| New classifier | Trainable | Trainable |
| Training cost | Usually lower | Usually higher |
| Target adaptation | Limited to classifier | Backbone can adapt |
| Dataset requirement | Can work with small datasets | Often benefits from more representative data |
| Best starting point | Small dataset | When more adaptation is needed |

**Important:** Fine-tuning does not mean unfreezing every layer immediately. A common strategy is to train the new head first, then unfreeze selected layers using a smaller learning rate.

The official PyTorch and TensorFlow tutorials demonstrate these two transfer learning strategies.

## 3. Practical PyTorch Example — Load a Pretrained Model

Let's see how transfer learning starts in actual code.

We will use ResNet18 with ImageNet pretrained weights.

```python
import torch
import torch.nn as nn
from torchvision import models

# Check GPU
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

# Load pretrained ResNet18
weights = models.ResNet18_Weights.DEFAULT

model = models.resnet18(weights=weights)

# Freeze pretrained parameters
for param in model.parameters():
    param.requires_grad = False

# Replace original classification head
num_features = model.fc.in_features

model.fc = nn.Linear(
    num_features,
    4
)

model = model.to(device)

print(model.fc)
```

Here:

- `weights=weights` loads pretrained ImageNet weights.
- `requires_grad=False` freezes the existing parameters.
- `model.fc` replaces the original classification head.
- `4` represents our four motor defect classes.

### Verify trainable parameters

```python
total_parameters = sum(
    p.numel() for p in model.parameters()
)

trainable_parameters = sum(
    p.numel() for p in model.parameters()
    if p.requires_grad
)

print("Total parameters:", total_parameters)
print("Trainable parameters:", trainable_parameters)
```

You should see that only the new classification head is trainable.

### Get the correct preprocessing

Pretrained models have specific image preprocessing requirements.

Torchvision provides the recommended transformations through the weights object:

```python
preprocess = weights.transforms()

print(preprocess)
```

This is important because the image size, normalization, and other preprocessing details depend on the pretrained model and its weight version.

## 4. When Should You Choose Each Approach?

| Situation | Recommended starting point |
|---|---|
| Very small dataset | Feature extraction |
| Limited GPU resources | Feature extraction |
| Target images resemble source images | Feature extraction |
| Target images differ from source images | Try fine-tuning |
| Sufficient representative target data | Fine-tuning may help |
| Frozen backbone gives poor results | Experiment with fine-tuning |

For your industrial motor defect classification example, I would start with pretrained ResNet18 feature extraction, evaluate it, and then try fine-tuning if the results show that the backbone needs adaptation.

## 6. What You Should Remember 🧠

1. Transfer learning reuses knowledge from an already-trained model.
2. It is especially useful when target datasets are limited.
3. Pretrained CNNs have already learned useful visual representations.
4. Feature extraction freezes the backbone and trains a new classification head.
5. Fine-tuning allows selected pretrained layers to adapt to the target task.
6. Feature extraction is generally cheaper; fine-tuning offers more flexibility.
7. Always use the preprocessing associated with the pretrained weights.

**The central concept:**

> Feature extraction reuses learned features. Fine-tuning adapts those learned features to your new problem.
