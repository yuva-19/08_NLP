## 1. ResNet — Residual Network

[Source: Second-Order Effects — DeepSeek's mHC Paper Explained](https://rayfdj.dev/deepseek-mhc-explained/)

### 1.1 What is ResNet?

ResNet (Residual Network) is a CNN architecture introduced to make training very deep neural networks easier.

As neural networks become deeper, optimisation can become difficult. Simply stacking more layers does not necessarily improve performance.

ResNet addresses this using **skip connections**, also called residual connections.

### 1.2 Understanding residual connections

In a conventional neural network, information passes through successive layers.

ResNet introduces a shortcut that allows information to bypass some layers.

The residual block can be represented as:

$$
y=F(x)+x
$$

Where:

- \(x\) = Original input.
- \(F(x)\) = Transformation learned by convolutional layers.
- \(y\) = Output after adding the shortcut.

![Residual connections in a neural network](https://images.openai.com/static-rsc-4/iIZN5fsodalRXr5ivusPMR7MMdnQGLzaMa7YSoNAaANmr8iWU6FeURUmEKNp9GaPjvB4Zu1DOfwM5S9Uq52zqO_ISGWAKuXQyHw1AJP3PR2Nn-JLtu_8LGBhFiVvmKHo6ImP7UHlT_-zUaPG_XJfR7hyzduVtDFP8_LOK_3QxUs?purpose=inline)

[Source: Lightning AI — Fine-Tune an Image Classifier](https://lightning.ai/lightning-ai/studios/finetune-an-image-classifier-with-hugging-face)

**Why is this useful?**

- Provides a direct path for information and gradients.
- Makes very deep CNNs easier to optimise.
- Allows layers to learn residual changes instead of complete transformations.

### 1.3 Popular ResNet architectures

| Architecture | Approximate parameters |
|---|---:|
| ResNet18 | 11.7 million |
| ResNet34 | 21.8 million |
| ResNet50 | 25.6 million |
| ResNet101 | 44.5 million |
| ResNet152 | 60.2 million |

These are standard Torchvision model parameter counts, approximately.

### 1.4 Practical use — ResNet18

We'll use ResNet18 as our baseline transfer learning model.

Example: Motor defect classification

```text
Input Image
     ↓
ResNet18 Backbone
     ↓
Feature Extraction
     ↓
New Classification Head
     ↓
4 Defect Classes
```

**PyTorch implementation:**

```python
import torch.nn as nn
from torchvision import models

weights = models.ResNet18_Weights.DEFAULT

model = models.resnet18(weights=weights)

# Freeze backbone
for param in model.parameters():
    param.requires_grad = False

# Replace classification head
model.fc = nn.Linear(
    model.fc.in_features,
    4
)
```

**When should you use ResNet?**

- General-purpose image classification.
- Transfer learning with small or medium datasets.
- As a baseline for comparing other architectures.
- When you want a well-established architecture with straightforward fine-tuning.

## 2. MobileNet — Lightweight CNN

![MobileNetV2 architecture](https://images.openai.com/static-rsc-4/X0aQN3GkL3Hbs0OqAKyFguPt_NCCWggsehRCcSOxjOEL6f-CxNJTbJUWUTgU3PVFTWMlCaC_PDPHSWA2RBz0OBYN-GxZwaBax_6oyxX0zTeswmw3XHasBw1j4Q-uLs3k2-JqZBTJR1RTeideCvgPwMW4GE9Hv6bUXv5SfdIDSMA?purpose=inline)

[Source: DOCSAID — MobileNetV2](https://docsaid.org/papers/lightweight/mobilenet-v2/)

### 2.1 What is MobileNet?

MobileNet is a family of CNN architectures designed for computationally constrained devices.

**Examples:**

- Smartphones.
- Embedded systems.
- Raspberry Pi.
- Edge AI cameras.
- Industrial inspection devices.

The main objective is to achieve useful image recognition performance with fewer parameters and less computation.

### 2.2 Depthwise Separable Convolution

The important idea behind MobileNet is **depthwise separable convolution**.

Recall that a standard convolution applies filters across all input channels.

MobileNet separates this operation into two stages:

1. Depthwise convolution.
2. Pointwise convolution.

![Depthwise separable convolution and lightweight neural networks](https://images.openai.com/static-rsc-4/nmYyS_81IqLtQAUHkbE54tbINCz0MP6nKh-DPkZiUOFGXePCaEOyVMdCMPOHcJMHng3BF-tK9ZNjnsldiGYUEVoNyT0p2I_AMGvJYq70i96g9A8oP6GFAMTvh0rNq0vE6b66dPTVB88euJkEStvDXCwI35lWqfU5o3-bZqUwYLg?purpose=inline)

[Source: MDPI — Lightweight RGB-Guided Neural Network](https://www.mdpi.com/1424-8220/23/15/6927)

**A. Depthwise convolution**

Applies a separate spatial filter to each input channel.

**B. Pointwise convolution**

Uses a \(1 \times 1\) convolution to combine information across channels.

This significantly reduces computation compared with a conventional convolution in suitable layers.

### 2.3 MobileNetV2 — Inverted Residuals

MobileNetV2 introduces:

- Inverted residual blocks.
- Linear bottlenecks.
- Depthwise convolutions.

![MobileNetV2 inverted residual architecture](https://images.openai.com/static-rsc-4/VuLofrjzC9W31Bg1xNL5DDvCmVfiEkdkMZ265d69qKvFm2t3yg1-U12JhYkd1YxHEVhEsWHGUhkMFs4nxalgBGhiTODYuq1r6wqGeV4P6I6Bv1ir_A968FNHJ4ao2jv5wjhOr5-DM28-ILEDNrz36sq0T-0QLqNtc2-aArxE6fU?purpose=inline)

[Source: Springer Nature — Efficient Fine-Tuning with Layer Pruning](https://link.springer.com/article/10.1007/s00521-023-09194-5)

The inverted residual block follows this general sequence:

```text
Input
  ↓
Expansion (1×1 convolution)
  ↓
Depthwise Convolution
  ↓
Projection (1×1 convolution)
  ↓
Output
```

It expands the feature representation internally and projects it back into a narrower representation.

The original paper explains how this architecture balances representational power and computational efficiency.

[MobileNetV2 — Inverted Residuals and Linear Bottlenecks](https://arxiv.org/abs/1801.04381)

### 2.4 Practical use — MobileNetV3

For our four-class motor defect classification task, let's use MobileNetV3 Large.

```python
import torch.nn as nn
from torchvision import models

weights = models.MobileNet_V3_Large_Weights.DEFAULT

model = models.mobilenet_v3_large(
    weights=weights
)

# Freeze backbone
for param in model.parameters():
    param.requires_grad = False

# Replace classification head
in_features = model.classifier[3].in_features

model.classifier[3] = nn.Linear(
    in_features,
    4
)
```

Notice that MobileNetV3 uses `model.classifier[3]`, unlike ResNet's `model.fc`.

**When should you use MobileNet?**

- Real-time image classification.
- Edge AI applications.
- Mobile applications.
- Devices with limited memory.
- Applications requiring a compact model.

**Engineering example:** An inspection camera installed directly on a production machine may need a lightweight model that can operate without sending every image to a cloud server.

## 3. EfficientNet — Balanced Model Scaling

![EfficientNet architecture](https://images.openai.com/static-rsc-4/JfqX1sE1QekZf7gJAaopECZ7ypBGk-nQPXnaSP8D8DE8GQmsGFS5GhUwThnXe69NN3F8FF0lZ9rzMJMU3hS7RbahkyO1AsdzDvAx5Mp2zwqqLT7bCk9Wo8ecG5RjU2rxYvBZd_oG-cz2U6LaAXdcIpxB3iJ45T_OcAkpQte9vXQ?purpose=inline)

[Source: Medium — EfficientNet Implementation Using PyTorch](https://sahiltinky94.medium.com/know-about-efficientnet-implementation-from-scratch-using-pytorch-b8ec0c0e6a33)

### 3.1 What is EfficientNet?

EfficientNet is a CNN family designed to balance accuracy and computational efficiency.

Its central contribution is **compound scaling**.

Instead of increasing only the number of layers or channels, EfficientNet scales three dimensions together:

1. Depth.
2. Width.
3. Resolution.

### 3.2 Understanding Compound Scaling

![EfficientNet compound scaling](https://images.openai.com/static-rsc-4/rrtyqcfXx5U6Fpol9e-OUbeasChUUfZNF4f2cSpj7Z28lQoGahg_PJPLUwk2RWQk7W7K8hrdhDdQvyaT2ZLIse1ZgsZ2YcJEmFKQJF8Cpl3WG2Mmph4XsR2UAFeZxQADoPtMyVRCOtBKGHurCmxCne2vTEqoVWB99tyAdDtCOQE?purpose=inline)

[Source: Medium — EfficientNet: Rethinking Model Scaling](https://medium.com/@nainaakash012/efficientnet-rethinking-model-scaling-for-convolutional-neural-networks-92941c5bfb95)

**Depth:** Number of network layers.

**Width:** Number of channels or filters in the network.

**Resolution:** Spatial size of the input image.

**Why scale all three?**

- Increasing depth allows more complex feature processing.
- Increasing width allows more feature channels.
- Increasing resolution provides more visual detail.

EfficientNet balances these dimensions using a compound scaling method rather than increasing them independently.

### 3.3 MBConv and Squeeze-and-Excitation

EfficientNet uses Mobile Inverted Bottleneck Convolution (MBConv) blocks.

![EfficientNet MBConv and feature channel processing](https://images.openai.com/static-rsc-4/Kg0vwOvtknYPFsu5htuNzBXfgL7_xdoYHFjrAo_41waT1TIsuSy7Yn8GkrRYTNAE7Zrur6WAXjGTqcGO0Sj9lCyEZR6hU8POtwMfttQ8f21QmRz3kWaW2Yl6yg98P1TYkFr7zjXLPq4OuiplL7ZOUIuFLLYWf9oVmHuIkU5Stys?purpose=inline)

[Source: MDPI — EfficientNet and Vision Model Comparison](https://www.mdpi.com/2075-4426/12/10/1707)

The MBConv block commonly involves:

- Expansion convolution.
- Depthwise convolution.
- Squeeze-and-Excitation.
- Projection convolution.

Squeeze-and-Excitation (SE) helps the network adaptively recalibrate the importance of feature channels.

### 3.5 Practical use — EfficientNet-B0

```python
import torch.nn as nn
from torchvision import models

weights = models.EfficientNet_B0_Weights.DEFAULT

model = models.efficientnet_b0(
    weights=weights
)

# Freeze backbone
for param in model.parameters():
    param.requires_grad = False

# Replace classification head
in_features = model.classifier[1].in_features

model.classifier[1] = nn.Linear(
    in_features,
    4
)
```

**When should you use EfficientNet?**

- General-purpose classification.
- When you want a good accuracy-to-computation balance.
- Small and medium-sized datasets with pretrained weights.
- Applications where both model size and predictive performance matter.

**Engineering recommendation:** EfficientNet-B0 is a useful starting point when you want a compact model with stronger ImageNet pretrained accuracy than a small baseline such as ResNet18.

## 4. ConvNeXt — Modernised CNN Architecture

![Modernised ResNet and ConvNeXt architecture](https://images.openai.com/static-rsc-4/5-keClFkKnYHGOEtFiw1Eor1jXsgGuiGzYLr-fwp58jBOu7Pxy3bcYrkIMNJqDTUqM_7OX2W2StN1VfrTiO_VLXLoAEj6HfdhPC3Hy1tZvokYqqgavC4fZiM1OaH9xpQUdhsmfk8z9aNPFghFdmwQVKEY5MCO_R7nOnUmuLQV3I?purpose=inline)

[Source: From ResNet to ConvNeXt — Modernising a Vanilla ResNet](https://tokudayo.github.io/from-resnet-to-convnext-2/)

### 4.1 What is ConvNeXt?

ConvNeXt is a modern CNN family introduced in the paper *A ConvNet for the 2020s*.

It revisits traditional CNN architecture design using several ideas that became successful in modern vision architectures.

**The important point:**

ConvNeXt remains a convolutional neural network. It is not a Vision Transformer.

### 4.2 Why was ConvNeXt introduced?

Vision Transformers had become highly competitive in computer vision.

The ConvNeXt researchers investigated whether a conventional CNN could be modernised using successful design ideas from contemporary architectures.

They updated aspects such as:

- Convolution kernel sizes.
- Network stage design.
- Depthwise convolutions.
- Activation functions.
- Normalisation.
- Downsampling structure.

![ConvNeXt paper review and architecture](https://images.openai.com/static-rsc-4/AIhCPAFeZbEg7u7b2U8vXIpiA2iXfHAZZv6ZZtCnatVgJiPkbkKN0gcqZ3_vYKtGN8EqBv1sVCY4ntLK3cfoagYyLSE0tRnInLxKf3Zl3g6vfEbdwRrGELHEDULZDT1n9xXB_M0JMu6D8W0XE1qOVqKvuAzLnXNCSivZwClVqYg?purpose=inline)

[Source: Speaker Deck — Paper Review of ConvNeXt](https://speakerdeck.com/sengopal/paper-review-of-convnext)

The original research demonstrated that modernised ConvNets could compete strongly with Transformer-based vision models.

### 5.5 Memory Consumption

Memory consumption is the RAM or GPU VRAM required while a model operates.

There are two important cases:

**A. Inference memory**

**B. Training memory**

Memory required for:

- Model weights.
- Gradients.
- Optimizer states.
- Intermediate activations needed for backpropagation.

# Part 3 — Practical Model Comparison in PyTorch

## 6. Build All Four Models Using One Function

Let's make the implementation reusable instead of writing four separate training scripts.

We'll use feature extraction initially, keeping the comparison consistent.

```python
import torch
import torch.nn as nn
from torchvision import models

NUM_CLASSES = 4


def build_model(model_name, num_classes=NUM_CLASSES):

    if model_name == "resnet18":

        weights = models.ResNet18_Weights.DEFAULT

        model = models.resnet18(weights=weights)

        for p in model.parameters():
            p.requires_grad = False

        model.fc = nn.Linear(
            model.fc.in_features,
            num_classes
        )

    elif model_name == "mobilenet":

        weights = models.MobileNet_V3_Large_Weights.DEFAULT

        model = models.mobilenet_v3_large(
            weights=weights
        )

        for p in model.parameters():
            p.requires_grad = False

        model.classifier[3] = nn.Linear(
            model.classifier[3].in_features,
            num_classes
        )

    elif model_name == "efficientnet":

        weights = models.EfficientNet_B0_Weights.DEFAULT

        model = models.efficientnet_b0(
            weights=weights
        )

        for p in model.parameters():
            p.requires_grad = False

        model.classifier[1] = nn.Linear(
            model.classifier[1].in_features,
            num_classes
        )

    elif model_name == "convnext":

        weights = models.ConvNeXt_Tiny_Weights.DEFAULT

        model = models.convnext_tiny(
            weights=weights
        )

        for p in model.parameters():
            p.requires_grad = False

        model.classifier[2] = nn.Linear(
            model.classifier[2].in_features,
            num_classes
        )

    else:
        raise ValueError(
            f"Unknown model name: {model_name}"
        )

    return model, weights
```

**Usage:**

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model, weights = build_model("resnet18")

model = model.to(device)

print(model)
```

Change `"resnet18"` to:

```python
"mobilenet"
"efficientnet"
"convnext"
```

This gives you the same four-class classification task using four different backbones.

## 7. Measure Parameters and Model Size

```python
def get_model_statistics(model):

    total_params = sum(
        p.numel() for p in model.parameters()
    )

    trainable_params = sum(
        p.numel() for p in model.parameters()
        if p.requires_grad
    )

    model_size_mb = sum(
        tensor.numel() * tensor.element_size()
        for tensor in model.state_dict().values()
    ) / (1024 ** 2)

    return {
        "parameters": total_params,
        "trainable_parameters": trainable_params,
        "model_size_MB": model_size_mb
    }
```

This measures the actual model state in memory, including buffers, rather than estimating size from parameter count alone.

## 8. Measure Inference Time and GPU Memory

We should measure actual inference speed on your hardware rather than inventing latency values.

```python
import time
import torch


def benchmark_model(
    model,
    device,
    batch_size=1,
    image_size=224,
    warmup=20,
    iterations=100
):

    model = model.to(device)
    model.eval()

    dummy_input = torch.randn(
        batch_size,
        3,
        image_size,
        image_size,
        device=device
    )

    if device.type == "cuda":
        torch.cuda.reset_peak_memory_stats(device)

    # Warm-up
    with torch.inference_mode():

        for _ in range(warmup):
            model(dummy_input)

    if device.type == "cuda":
        torch.cuda.synchronize(device)

    # Benchmark
    start_time = time.perf_counter()

    with torch.inference_mode():

        for _ in range(iterations):
            model(dummy_input)

    if device.type == "cuda":
        torch.cuda.synchronize(device)

    end_time = time.perf_counter()

    total_time = end_time - start_time

    average_latency_ms = (
        total_time / iterations
    ) * 1000

    images_per_second = (
        batch_size * iterations
    ) / total_time

    if device.type == "cuda":

        peak_memory_mb = (
            torch.cuda.max_memory_allocated(device)
            / (1024 ** 2)
        )

    else:
        peak_memory_mb = None

    return {
        "latency_ms": average_latency_ms,
        "images_per_second": images_per_second,
        "peak_gpu_memory_MB": peak_memory_mb
    }
```

### Run the comparison

```python
model_names = [
    "resnet18",
    "mobilenet",
    "efficientnet",
    "convnext"
]

results = {}

for name in model_names:

    print("Testing:", name)

    model, weights = build_model(name)

    model = model.to(device)

    stats = get_model_statistics(model)

    benchmark = benchmark_model(
        model,
        device
    )

    results[name] = {
        **stats,
        **benchmark
    }

    print(results[name])

    del model

    if device.type == "cuda":
        torch.cuda.empty_cache()
```

**Important benchmarking rules:**

- Use the same hardware.
- Use the same batch size.
- Use the same input resolution when comparing raw inference speed.
- Warm up the model before measuring.
- Synchronise CUDA before recording elapsed time.
- Measure target-dataset accuracy separately.
- Repeat benchmarks to account for variation.

The code above measures batch-one inference latency using a synthetic input. For deployment decisions, also test real images, preprocessing overhead, and your actual production batch size.

## 10. What You Should Remember 🧠

1. ResNet uses residual connections to make deep CNNs easier to optimise.
2. MobileNet uses efficient convolution operations for resource-constrained devices.
3. EfficientNet balances depth, width, and resolution through compound scaling.
4. ConvNeXt modernises conventional CNN design.
5. Parameter count and model size are not the same as inference speed.
6. Accuracy must be measured on your target dataset, not assumed from ImageNet.
7. Actual latency and GPU memory depend on hardware, input size, batch size, and implementation.

**The central engineering principle:**

> The best model is not necessarily the biggest or most accurate benchmark model. It is the model that meets your application's accuracy, speed, memory, and deployment requirements.
