# 📘 2.1 NLP Pipeline



An **NLP pipeline** is a sequence of processing stages that transforms **raw human language into a form that an NLP model can use to produce a prediction**. The exact stages can vary depending on the task and model. [GitHub](https://github.com/anshumandwivediusa/nlp/blob/main/02.1%20nlp-pipelines.md?utm_source=chatgpt.com)

```text
Raw Text
   ↓
Cleaning
   ↓
Normalization
   ↓
Segmentation
   ↓
Tokenization
   ↓
Representation
   ↓
Model
   ↓
Prediction
   ↓
Evaluation
```

---

## 1. Raw Text

The original text collected from sources such as:

- Documents
- Websites
- Reviews
- Emails
- Social media
- Chat messages

**Example:**

```text
"OMG!!! I LOVED this movie 😍!!!"
```

---

## 2. Cleaning

Removes **unwanted noise or artifacts** from the text.

Examples:

```text
HTML tags
URLs
Unnecessary characters
Extra whitespace
Duplicate/repeated characters
```

Example:

```text
"<p>Hello!!!</p>"
        ↓
"Hello!!!"
```

⚠️ Cleaning is **task-dependent**. Removing emojis, punctuation, or other information may actually hurt tasks such as sentiment analysis. [CloudsPress](https://www.cloudspress.com/text-preprocessing-in-natural-language-processing-with-python-a-task-driven-guide/?utm_source=chatgpt.com)

---

## 3. Normalization

Makes text more **consistent and standardized**.

Common operations:

- Lowercasing
- Unicode normalization
- Standardizing whitespace
- Handling spelling variations
- Expanding contractions when appropriate

Example:

```text
"  HELLO   WORLD  "
          ↓
"hello world"
```

---

## 4. Segmentation

Divides a document into meaningful sections, most commonly **sentences**.

Example:

```text
Input:
"I love NLP. It is fascinating."

        ↓

Sentence 1 → "I love NLP."
Sentence 2 → "It is fascinating."
```

This is also called **sentence segmentation**.

---

## 5. Tokenization

Breaks text into smaller units called **tokens**.

Example:

```text
"I love NLP"

        ↓

["I", "love", "NLP"]
```

Depending on the tokenizer, tokens can be:

- Words
- Characters
- Subwords

Modern Transformer models commonly use **subword tokenization**. [Gen-AI.in](https://www.gen-ai.in/learn/nlp-pipeline?utm_source=chatgpt.com)

---

## 6. Representation

The tokens must be converted into a **numerical representation** that the model can process.

Traditional representations:

```text
One-Hot Encoding
Bag of Words
TF-IDF
```

Neural representations:

```text
Word Embeddings
Contextual Embeddings
```

Example:

```text
"cat"
 ↓
Token ID
 ↓
Embedding Vector
 ↓
[0.21, -0.43, 0.67, ...]
```

---

## 7. Model

The numerical representation is given to an NLP model.

Examples:

### Classical NLP

```text
TF-IDF
   ↓
Logistic Regression / SVM / Naive Bayes
```

### Deep Learning

```text
Embeddings
   ↓
LSTM / GRU / Transformer
```

### Modern NLP

```text
Tokens
   ↓
Transformer
   ↓
BERT / GPT / T5 etc.
```

---

## 8. Prediction

The model produces the required output.

Example — Sentiment Analysis:

```text
"I absolutely loved this movie!"

             ↓
           Model
             ↓
          Positive
```

Other predictions can be:

- Spam / Not Spam
- Entity labels
- Translation
- Summary
- Answer to a question
- Generated text

---

## 9. Evaluation

The model's predictions are compared with the expected results using appropriate metrics.

Examples:

| NLP Task | Common Metrics |
|---|---|
| Classification | Accuracy, Precision, Recall, F1 |
| NER | Entity-level Precision, Recall, F1 |
| Translation | BLEU |
| Summarization | ROUGE |
| Language Modeling | Perplexity |

---

# 🔥 Complete Example

Suppose we want to classify:

> **"This phone is AMAZING!!! 😍"**

```text
Raw Text
   ↓
"This phone is AMAZING!!! 😍"
   ↓
Cleaning
   ↓
"This phone is AMAZING"
   ↓
Normalization
   ↓
"this phone is amazing"
   ↓
Segmentation
   ↓
One sentence
   ↓
Tokenization
   ↓
["this", "phone", "is", "amazing"]
   ↓
Representation
   ↓
Numerical vectors
   ↓
Model
   ↓
Prediction
   ↓
Positive
   ↓
Evaluation
   ↓
F1 / Accuracy
```

### 🧠 Important

This is a **conceptual pipeline**, not a mandatory recipe. Modern Transformer/LLM pipelines may skip or combine several traditional preprocessing stages; for example, tokenization and learned representations are handled by the model's tokenizer and architecture. [GitHub](https://github.com/anshumandwivediusa/nlp/blob/main/02.1%20nlp-pipelines.md?utm_source=chatgpt.com)

### ⭐ Remember

> **Raw Text → Prepare → Split → Tokenize → Represent → Model → Predict → Evaluate**
