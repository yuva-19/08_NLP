# 📘 2.2 Text Data Types



Text can be represented at different **levels of granularity**. The level we choose affects vocabulary size, sequence length, and how well a model handles unfamiliar words. [NLP Book](https://nlpbook-e65cc6.pages.mi.hdm-stuttgart.de/01access/08ChunkingTokenisation.html?utm_source=chatgpt.com)

---

## 1. Documents

A **document** is a complete piece of text that is treated as one unit.

Examples:

- News article
- Research paper
- Product review
- Email
- Book chapter

Example:

```text
Document
│
├── Sentence 1
├── Sentence 2
├── Sentence 3
└── Sentence 4
```

A document can contain **one or many sentences**.

### NLP applications

Documents are commonly used for:

- Document classification
- Search
- Information retrieval
- Summarization
- Topic analysis

---

## 2. Sentences

A **sentence** is a linguistic unit containing a sequence of words/tokens.

Example:

```text
"The movie was amazing."
```

A document can be divided into sentences:

```text
Document
   ↓
"The movie was amazing."
"I watched it yesterday."
"I would recommend it."
```

This process is called **sentence segmentation**.

---

## 3. Words

A **word** is a linguistic unit formed from characters.

Example:

```text
"The movie was amazing."

        ↓

The | movie | was | amazing
```

Word-level processing was historically very common in NLP.

### Problem

A vocabulary containing every possible word can become extremely large.

For example:

```text
play
playing
played
player
playful
playground
...
```

Treating every form as a completely separate word increases vocabulary size.

---

## 4. Characters

A **character** is an individual symbol in text.

Example:

```text
"cat"

 ↓

c | a | t
```

Characters include:

- Letters
- Digits
- Punctuation
- Symbols

### Advantage

Character-level representations can handle unfamiliar words because any word can be broken into characters.

### Disadvantage

Sequences become much longer.

```text
Word-level:
"playing" → 1 unit

Character-level:
"playing" → p | l | a | y | i | n | g
```

---

## 5. Tokens

A **token** is a unit produced by a tokenizer.

Importantly:

> **A token is not necessarily the same thing as a word.**

Depending on the tokenizer, a token can be:

- A word
- A subword
- A character
- Sometimes punctuation or special symbols

For example:

```text
"I love NLP!"

→ ["I", "love", "NLP", "!"]
```

Tokenization converts continuous text into discrete units that can subsequently be mapped to IDs and processed by models. [Stanford NLP](https://nlp.stanford.edu/IR-book/html/htmledition/tokenization-1.html?utm_source=chatgpt.com)

### Word vs Token

```text
Word:
"playing"

Possible tokens:
["playing"]

OR

["play", "##ing"]
```

So:

**Every word can become a token, but not every token is a complete word.**

---

# 6. Subwords



**Subwords** are units that are smaller than complete words but larger than individual characters in many cases.

Example:

```text
"playing"

        ↓

["play", "ing"]
```

Another example:

```text
"unhappiness"

        ↓

["un", "happi", "ness"]
```

The exact splits depend on the **tokenization algorithm and vocabulary**.

Modern NLP systems commonly use subword-based tokenization because it provides a useful balance between word-level and character-level representations. [obscrivn.github.io](https://obscrivn.github.io/mynewbook/module2/modern_tokenization.html?utm_source=chatgpt.com)

### Why Subwords?

Consider an unfamiliar word:

```text
"neuralization"
```

A word-level tokenizer might encounter an **unknown word**.

A subword tokenizer can potentially represent it using known pieces:

```text
neural + ization
```

This reduces the **out-of-vocabulary (OOV)** problem.

---

# 🔥 Comparing the Levels

For:

> **"I am playing."**

| Level | Possible representation |
|---|---|
| **Document** | Entire text/document |
| **Sentence** | `"I am playing."` |
| **Word** | `I \| am \| playing` |
| **Character** | `I \| a \| m \| p \| l \| a \| y \| i \| n \| g` |
| **Token** | Depends on tokenizer |
| **Subword** | `I \| am \| play \| ing` |



---

## 🧠 Key Concept

Think of text as a hierarchy:

```text
Document
   ↓
Sentences
   ↓
Words
   ↓
Tokens
   ↓
Subwords / Characters
```

But **tokens are not necessarily a separate level below words**. Rather, *token* is a general term for the units selected by a tokenizer.

### ⭐ Remember

> **Document and sentence describe larger text units, while words, subwords, and characters describe increasingly fine-grained units. A token is the actual unit produced by the tokenizer.**

This distinction becomes **very important later when you study BPE, WordPiece, SentencePiece, BERT, GPT, and Hugging Face tokenizers.** [stanford.edu](https://stanford.edu/~jurafsky/slp3/2.pdf?utm_source=chatgpt.com)
