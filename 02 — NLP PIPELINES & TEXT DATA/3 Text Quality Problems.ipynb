# 📘 2.3 Text Quality Problems

Real-world text is rarely clean. Before applying many NLP techniques, we may need to identify and handle **noise, inconsistencies, and unwanted formatting**.



---

## 1. HTML

Web pages often contain HTML tags that are not part of the actual text.

```text
<p>This is a great product!</p>
```

The NLP system may need:

```text
"This is a great product!"
```

Common examples:

```text
<div>...</div>
<br>
<a href="...">...</a>
```

**Problem:** HTML markup can become unwanted noise during text analysis.

---

## 2. URLs

Text collected from websites, emails, and social media often contains URLs.

```text
"Check this product: https://example.com/product"
```

Depending on the task, the URL may be:

- Removed
- Replaced with a special token such as `<URL>`
- Preserved as useful information

⚠️ **Do not automatically remove URLs.** In some tasks, URLs themselves may contain useful information.

---

## 3. Emojis

Emojis carry **semantic and emotional information**.

```text
"I love this movie 😍❤️"
```

For sentiment analysis:

```text
😍 → positive signal
❤️ → positive signal
```

Therefore, blindly removing emojis can result in **information loss**.

Possible approaches:

```text
"I love this 😍"
        ↓
"I love this"
```

or preserve/convert them into meaningful representations.

---

## 4. Punctuation

Examples:

```text
.,!?;:'"-()
```

Example:

```text
"Wow! This is amazing!!!"
```

Punctuation can sometimes be noise, but it can also carry meaning.

For example:

```text
"Really?"
"Really!"
```

have different intentions.

Similarly:

```text
"Great."
"Great!!!"
```

may express different levels of emotion.

**Therefore, punctuation handling should depend on the NLP task.**

---

## 5. Special Characters

Text can contain symbols that may or may not be useful.

Examples:

```text
@ # $ % & * + = < >
```

Example:

```text
"Follow @user #NLP"
```

Here:

- `@user` may identify a social-media user.
- `#NLP` may represent a topic.

So special characters should **not automatically be deleted**.

---

## 6. Repeated Spaces

Text can contain unnecessary whitespace.

```text
"This    is     NLP."
```

Normalization can convert it to:

```text
"This is NLP."
```

Typical causes:

- Multiple spaces
- Tabs
- Newlines
- Formatting artifacts

This is usually a straightforward normalization step.

---

## 7. Spelling Variations

Users may write the same word in different ways.

```text
"color"   vs   "colour"
```

Typos are another common problem:

```text
"recieve" → "receive"
"definately" → "definitely"
```

Informal text can contain:

```text
"hellooo"
"soooo good"
"gr8"
```

Possible approaches include:

- Spell correction
- Normalization
- Keeping the original form
- Context-aware correction

⚠️ Aggressive correction can sometimes **change the intended meaning**.

---

## 8. Numbers

Text may contain numbers in different forms:

```text
100
1,000
₹50,000
20%
2026
```

Whether numbers should be removed or preserved depends on the task.

### Example

For financial NLP:

```text
"The company earned ₹50 crore."
```

The number is **highly important**.

For some simple text-classification tasks, numbers may contribute little.

**Rule:** Don't remove information just because it looks unusual.

---

## 9. Mixed Casing

The same word can appear with different capitalization:

```text
NLP
nlp
Nlp
```

Lowercasing can normalize them:

```text
NLP → nlp
Nlp → nlp
```

### But casing can carry information.

Compare:

```text
apple
Apple
```

- `apple` → fruit
- `Apple` → potentially the company

Therefore, lowercasing should be **task-dependent**.

---

## 10. Contractions

Contractions combine multiple words.

Examples:

```text
I'm   → I am
don't → do not
can't → cannot
it's  → it is / it has
```

For some NLP tasks, expanding contractions can make processing easier:

```text
"I don't like it."

        ↓

"I do not like it."
```

However, modern tokenizers can often handle contractions directly, so manually expanding every contraction is **not always necessary**.

---

# 🔥 Quick Summary

| Problem | Example | Possible Handling |
|---|---|---|
| **HTML** | `<p>Hello</p>` | Remove/extract text |
| **URLs** | `https://...` | Remove/replace/preserve |
| **Emojis** | `😍` | Preserve/convert/remove |
| **Punctuation** | `!!!` | Task-dependent |
| **Special characters** | `@ # %` | Task-dependent |
| **Repeated spaces** | `"hello    world"` | Normalize |
| **Spelling variations** | `colour/color` | Normalize/correct |
| **Numbers** | `₹50,000` | Preserve/remove based on task |
| **Mixed casing** | `NLP/nlp` | Normalize when appropriate |
| **Contractions** | `don't` | Expand or tokenize directly |

## 🧠 Most Important Principle

> **Text preprocessing is not about making text as clean as possible; it is about making text appropriate for the specific NLP task.**

For example, removing **emojis, punctuation, numbers, URLs, or casing information** may improve one task but hurt another.
